# Нестаціонарна теорія збурень

$\newcommand{\ket}[1]{\left.\left\vert\, #1\,\right. \right \rangle}$
$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$
$\newcommand{\av}[3]{\left\langle \,#1 \left \vert\, #2 \,\right\vert #3 \,\right \rangle}$

Раніше ми розглядали системи, гамільтоніан яких не залежав явно від часу. Однак, зрозуміло, що у загальній ситуації оператор Шредінгера явно залежить від часу. В такому випадку система не має стаціонарних станів. Тому задача ставиться як знаходження
залежності вектора стану чи хвильової функції від часу. У подальшому ми також будемо припускати, що фізична ситуація дозволяє представити оператор Шредінгера у виді суми незбуреного гамільтоніана і нестаціонарного збурення.

\begin{equation} 
\hat{H} = \hat{H}_{0} + \varepsilon \,\hat{V}(t)
\end{equation}

Розглянемо рівняння Шредінгера:

\begin{equation}
\label{18eq2} i\,\hbar \frac{\partial}{\partial \,t}\,\ket{\psi} =
\left( \hat{H}_{0} + \varepsilon \,\hat{V}(t) \right)\,\ket{\psi}
\end{equation}

Оскільки динаміка системи, яка визначається незбуреним гамільтоніаном $H_{0} $, вважається відомою, то для загального опису такої ситуації доцільно використати представлення взаємодії. Нагадаємо, що в рамках такого підходу з еволюції вектора стану "виключається" незбурена частина еволюції:

\begin{equation}
\ket{\tilde{\psi}(t)} = \hat{S}_{0}^{ - 1} (t)\ket{\psi(t)}\,,\quad
\hat{S}_{0} (t) = \exp \left(\frac{t}{i\,\hbar}\,\hat{H}_{0}  \right)
\end{equation}

При цьому оператори спостережувальних величин перетворюються як:

\begin{equation}
\label{18eq4} \tilde{A}_{Int}(t) = S_{0}^{ - 1}(t)\,A_{Sh}\,S_{0}(t)
\end{equation}

де $\tilde{A}_{Int} (t)\,$ і $A_{Sh}$ оператори у представленнях взаємодії і Шредінгера відповідно. В результаті замість (\ref{18eq4}) отримаємо:

\begin{equation}
\label{18eq5} i\,\hbar \frac{\partial \,\ket{\tilde{\psi}}}{\partial\,t} = 
\varepsilon \,\tilde{V}(t)\ket{\tilde{\psi}}
\end{equation}

Розкладаючи функцію $\tilde \psi$ в ряд за власними функціями $\phi_{n}^{(0)} $ стаціонарних станів гамільтоніана $H_{0}$ (з яких виключена гармонічна залежність від часу, тобто еволюція, яка відповідає $S_{0}(t)$, отримаємо:

\begin{equation}
\ket{\tilde{\psi} (t)} = \sum\limits_{n} c_{n} (t)\,\ket{\phi_{n}^{(0)}}   \\
 i\,\hbar \frac{\partial \,c_{n}}{\partial \,t} = \varepsilon
\sum\limits_{m} c_{m}(t)\,\av{\phi_{n}^{(0)}}{\tilde{V}(t)\,}{\phi_{m}^{(0)}}
\end{equation}

Діючи у рамках теорії збурень, представимо:

\begin{equation}
 c_{n} = c_{n}^{(0)} + \varepsilon \,c_{n}^{(1)} + \ldots \\
 i\,\hbar \frac{\partial \,c_{n}^{(0)}}{\partial \,t} = 0 \\ 
 i\,\hbar \frac{\partial \,c_{n}^{(1)}}{\partial \,t} = 
\sum\limits_{m} c_{m}^{(0)}(t)\,\av{\phi_{n}^{(0)}}{
\tilde{V}(t)}{\phi_{m}^{(0)}}
\end{equation}

Припустимо, що у початковий момент часу система знаходилась у одному з можливих стаціонарних станів, скажемо $\phi_{n}^{(0)} $,тоді:
$
c_{k} (0) = \delta_{k\,n} ,
$
тому з \eqref{18eq4} витікає:

\begin{equation}
\label{18eq8} i\,\hbar 
\frac{\partial \,c_{m}^{(1)}}{\partial \,t} = 
\av{\phi_{m}^{(0)}}{\tilde{V}(t)}{\phi_{n}^{(0)}}\,\,\, \Rightarrow \,\,
c_{m}^{(1)} (t) = \frac{1}{i\,\hbar}
\,\int\limits_{-\infty}^{t} V_{m\,n}(t)
\exp\left(i\,\omega_{mn} t\right)\,dt
\end{equation}

тут враховано, що:

\begin{multline}\label{18eq9}
 \tilde{V}_{m\,n}(t)= \av{\phi_{m}^{(0)}}{S_{0}^{- 1}(t)\,V(t)S_{0}(t)}{\phi_{n}^{(0)}} =\\ \av{\phi_{m}^{(0)}}{S_{0}^{\dagger}(t)\,V(t)S_{0}(t)}{\phi_{n}^{(0)}} =
\av{S_{0}(t)\phi_{m}^{(0)}}{V(t)S_{0} (t)}{\phi_{n}^{(0)}} =\\ 
\exp \left(\frac{E_{n} \,t}{i\,\hbar} - \frac{E_{m}
\,t}{i\,\hbar} \right)\left(\phi_{m}^{(0)} ,\,\,V(t)\,\phi
_{n}^{(0)}\right) =\exp \left(i\omega_{mn}\,t \right)\,V_{m\,n}(t)
\end{multline} 

де $ \omega_{mn} = \omega_{m} - \omega_{n} = \frac{E_{m} - E_{n}}{\hbar}\,. $

Розглянемо різні варіанти залежності збурення від часу. Збурення будемо називати __фінітним__ за часом, якщо воно зникає за певний інтервал часу, тобто має моменти "включення" і "вимкнення". Після дії збурення система буде перебувати у суперпозиції стаціонарних станів:

$$
\ket{\psi(\infty )} = \sum\limits_{n} c_{n}(\infty)\,\ket{\phi_{n}^{(0)}}
$$

амплітуди яких визначають ймовірності знаходження системи у відповідному стаціонарному стані. Для такого збурення інтегрування в \eqref{18eq8} може проводитися у інтервалі $(-\infty ,+\infty )$ так, що амплітуда і відповідна ймовірність переходу системи з початкового у кінцевий стан у першому порядку (для амплітуд) теорії збурень визначається фур'є - зображенням матричного елемента збурення:

\begin{equation}\label{eq_cm1}
c_{m}^{(1)} (\infty) = \frac{1}{i\,\hbar}
\int\limits_{ - \infty}^{ +\infty}  V_{m\,n} (t)\exp \left(i\,\omega_{mn} t\right)\,d\,t  = 
\frac{2\pi}{i\,\hbar}V(\,\omega_{mn} ) \\
P_{n \to m} = \left|\,c_{m}(\infty) \right|^{\,2} \approx 
\frac{4\pi^{\,2}}{\hbar^{\,2}}
\left|V(\,\omega_{mn}) \right|^{\,2}\,\,, m \ne n 
\end{equation}

Результат залишається вірним і для нефінітного збурення, якщо існує фур'є - зображення відповідних матричних елементів. Якщо збурення не "вимикається", а з часом матричні елементи асимптотично прямують до певних постійних значень гладким чином,
то вираз \eqref{eq_cm1} може бути перетворений:

\begin{multline}
 c_{m}^{(1)} (t) = \frac{1}{i\,\hbar}\,\int\limits_{ - \infty}^{t}
V_{m\,n} (t)\exp \left(i\,\omega_{mn} t \right)\,d\,t = - \frac{1}{\hbar \,\omega_{mn}}
\int\limits_{ - \infty}^{t} V_{m\,n} (t)\,d\left(\exp \left(i\,\omega_{mn} t \right)\right) = \\
\left. - \frac{1}{\hbar \,\omega_{mn}} V_{m\,n} (t)\,\exp\left( i\,\omega_{mn} t \right)
\right|_{ - \infty}^{t} +
\frac{1}{\hbar \,\omega_{mn}}
\int\limits_{ - \infty}^{t}\exp \left(i\,\omega_{mn} t \right)\,d\left(V_{m\,n} (t)\right)
\end{multline}

Переходячи до границі в \eqref{eq_cm1}, отримуємо, що перший член визначає просто зміну хвильових функцій системи під дією постійної у часі "асимптотичної" компоненти збурення. Вона відповідає результату стаціонарної теорії збурень (наявність експоненти в цьому доданку пов'язано з урахуванням характерної часової залежності хвильових функцій стаціонарних станів). Власне кажучи ймовірності переходу визначаються другим доданком у \eqref{eq_cm1} і дорівнюють:

\begin{equation}\label{18eq12} 
P_{n \to m} = \frac{1}{\hbar^{\,2}\,\omega_{mn}^{\,2}}
\left| \int\limits_{ - \infty}^{ + \infty}\frac{\partial \,V_{m\,n}}{\partial \,t}
\exp \left(i\,\omega_{mn} t\right)\,d\,t \right|^{\,2}
\end{equation}

У граничному випадку нескінченно малого зростання збурення (адіабатичне збурення), як це витікає з (\ref{18eq12}), ймовірності переходів системи у інші стани будуть малими, тому при адіабатичному збуренні система з ймовірністю близькою до 1 буде залишатись у початковому стані. У випадку миттєвого включення взаємодії (у порівнянні з $1/\omega_{m\,n} )$, після якого воно залишається постійним, в (\ref{18eq12}) можна знехтувати зміною експоненціального множника і явно проінтегрувати, так щоб ймовірності переходів в цьому випадку дорівнюють:

\begin{equation}
\label{18eq13} P_{n \to m} = \frac{\left|V_{m\,n}
\right|^{\,2}}{\hbar^{\,2}\,\omega_{mn}^{\,2}}
\end{equation}

## Частота переходу. Золоте правило Фермі.

Розглянемо випадок, коли за певний час на систему діє постійне
збурення. В такому випадку інтегрування в \eqref{18eq8}
проводиться явним чином:

\begin{equation}\label{pmn}
  P_{n \to m} = \left|\,c_{m}(t)\right|^2 \approx
  \frac{1}{\hbar^2}|V_{m\,n}|^2\left|
  \frac{\exp(i\omega_{m\,n}T)-1}{\omega_{m\,n}}\right|^2=
\frac{4}{\hbar^2}|V_{m\,n}|^2
  \frac{\sin^2(\omega_{m\,n}T/2)}{\omega^2_{m\,n}}
\end{equation}
  
Розглянемо функцію
\begin{equation}\label{sin2}
  f(x,T) = \frac{\sin^2(x\,T / 2)}{x^2}
\end{equation}

при $T \gg 1$. Максимум її при $x = 0$ і $f(0,T) = T^{\,2} / 4$.
Практично вся площа під графіком функції (див. Рис.~\ref{2fi1} )
міститься у вузькому інтервалі $ - 2\pi / T < x < 2\pi / T$.

```python
def fsin(x,t):
    return np.sin(x*t/2)**2/x**2
# 
t0=2
x= np.linspace(-10,10,200)
y=fsin(x,t0)
plt.plot(x,y,'k--')
plt.ylim(0,t0**2/3.95)
plt.fill_between(x,y,color=f"{0.9}")#set filling greyscale color
plt.show()
```

\frac{1}{T}
\int\limits_{ -\infty}^{ + \infty}  f(x,T)g(x)dx  = 
\lim \limits_{T\to \infty}  
\int\limits_{ - \infty}^{ + \infty} f(z,1)g(z /
T)dz  = \\ 
g(0)
\int\limits_{ - \infty}^{ + \infty} f(z,1)dz
= g(0) \int\limits_{ - \infty}^{ + \infty} \frac{\sin^{\,2}(z /
2)}{z^{\,2}}dz  =
 g(0) \lim \limits_{\varepsilon \, \to 0} 
 \int\limits_{-\infty}^{ + \infty} \frac{\sin^{\,2}(z / 2)}{z^{\,2} + \varepsilon^{\,2}} dz =
\end{multline*}

\begin{multline*}
  g(0)\lim \limits_{\varepsilon \, \to 0} 
  \int\limits_{ - \infty}^{ + \infty}  \frac{2 - \exp (iz) -

\end{multline*}

Остаточно отримуємо:

\begin{equation}
\label{18eq14} \lim\limits_{T \to \infty} 
\frac{P_{n\to m}}{T} = 
\frac{2\pi} {\hbar^{\,2}}
\left| V_{mn}\right|^{\,2} \delta (\omega_{mn} ) = 
\frac{2\pi}{\hbar}\left| V_{mn}\right|^{\,2}\delta (E_{m} - E_{n} )
\end{equation}

Це і є вираз для ймовірності переходу $n \to m$ у одиницю часу. Аналогічний вивід можливий і у випадку періодичного збурення з частотою $\omega $:

\begin{equation}
V(t) = V^{( + )}\exp (i\omega t) + V^{( - )} \exp (- i\omega t)
\end{equation}

<a id='eq_harmpert'>Вираз (\ref{18eq14}) змінюється відповідним чином</a>:

\begin{equation}\label{eq_harmpert}
\lim\limits_{T \to \infty} \frac{P^{(\pm )}_{n \to m}}{T} = 
\frac{2\pi}{\hbar^{\,2}}\left| V_{mn}\right|^{\,2}\delta (\omega_{mn} \pm \omega) \to
\frac{2\pi}{\hbar} \left|V_{mn}^{(\pm )}\right|^{\,2}
\delta (E_{m} - E_{n} \pm \hbar\, \omega )
\end{equation}

Це відповідає тому, що квантова система може змінити свій стан, лише прийнявши або віддавши "квант" енергії від зовнішнього поля. Дуже часто зустрічається ситуація, коли кінцевий стан належить неперервному спектру (іонізований атом з електроном, розсіяння, фотоефект). При цьому такий рівень є виродженим. В такому випадку треба додати в (\ref{18eq14}) ймовірності переходів у всі такі стани:

\begin{equation}\label{eq_goldrule} 
\Pi_{n \to m} = \int \frac{P^{(\pm )}_{n \to m}}{T}\rho (E)\,dE = \frac{2\pi}{\hbar} \left| V_{mn}^{(\pm )} \right|^{\,2}\rho (E_{m} )
\end{equation}

Тут $\rho $ - енергетична густина станів (число станів на одиничний інтервал енергії). Слід відзначити, що розмірність величини \eqref{eq_goldrule} визначається способом нормування хвильових функцій. Отриманий вираз \eqref{eq_goldrule} називається <a id='fermigoldenrule'>__"золотим правилом Фермі"__ </a> для ймовірності переходу у одиницю часу в стани
неперервного спектру. Цей результат буде застосовано далі.
