Вище було розглянуто наслідки принципу тотожності частинок, для системи невзаємодіючих частинок. Розглянемо тепер випадок взаємодіючих частинок. Покажемо, що врахування правильної симетрії хвильової функції системи тотожних частинок приводить до своєрідної залежності енергії стаціонарних станів від повного спіну системи, навіть якщо гамільтоніан не містить операторів, які явно діють на спінові ступені вільності (або гамільтоніан комутує з оператором повного спіну). Для простоти розглянемо систему двох тотожних частинок. Хвильова функція такої системи, в залежності від типу частинок, може бути або симетричною (__бозони__) або антисиметричною (__ферміони__) відносно переміщення частинок. Можливі значення повного спіна S системи визначаються правилом складення моментів і дорівнюють $S = 0,1,\ldots,2s$, де $s$ - спін частинок. Якщо гамільтоніан не діє на спінові степені свободи, то хвильова функція може бути представлена у вигляді
добутку координатної і спінової частин, тобто у вигляді добутку скалярної функції на вектор спінового простору. Аналогічно, можна просто вказати крім просторових аргументів хвильової функції індексу, спінового стану (проекції спіну на певну вісь): 
$\psi (\mathbf{r}\,;\sigma )$. Якщо частинки не були б тотожними, то у випадку відсутності взаємодії хвильова функція такої системи могла бути представлена у виді добутку хвильових функцій станів кожної з частинок. Однак, у випадку тотожних частинок необхідно додатково накласти умову (анти)симетрії хвильової функції системи.
Розглянемо випадок двох електронів (ферміони зі спіном 1/2). Повний спін такої системи може мати значення $S = 0,\,1$. Нульове значення, очевидно, відповідає компенсації проекцій спінового моменту електронів. Сукупність всіх спінових станів утворює
комплексний простір 4 вимірів (1+3=4). Стан, який відповідає одномірному  підпростору $S=0$ є антисиметричним:
<!--  -->
\begin{equation}
\label{s0} \left|\,S=0\,\right\rangle =
\frac{1}{\sqrt{2}}\left(\,\left|\,\uparrow\,\right\rangle\left|\,\downarrow\,\right\rangle
- \left|\,\downarrow\,\right\rangle\left|\,\uparrow\,\right\rangle
\,\right)\,.\end{equation} Вектора:
\begin{eqnarray}\label{s1}
  \left|\,S_z=+1\,\right\rangle &=&
  \left|\,\uparrow\,\right\rangle\left|\,\uparrow\,\right\rangle\notag\\
  \left|\,S_z=0\,\right\rangle &=&
  \frac{1}{\sqrt{2}}
  \left(\,\left|\,\uparrow\,\right\rangle\left|\,\downarrow\,\right\rangle
  +
  \left|\,\downarrow\,\right\rangle\left|\,\uparrow\,\right\rangle\,\right)\\
\left|\,S_z=-1\,\right\rangle &=&
  \left|\,\downarrow\,\right\rangle\left|\,\downarrow\,\right\rangle\notag
\end{eqnarray}
<!--  -->
є симетричними (по спіновим змінним) станами з $S = 1$ (індекс знизу означає проекцію). Таким чином, симетричні відносно переміщення спинів стани утворюють 3-вимірний простір. Скалярні коефіцієнти це просто координатні частини хвильової функції.

В системі центра мас, двухчастинкова задача перетворюється в одночастинкову для частинки приведеної маси (орбітальний момент зберігається), а переміщеннячастинок еквівалентна операції інверсії, так що для координатної частини хвильової функції, виходячи з загального правила, отримаємо:
<!--  -->
$$ 
\varphi ( - \mathbf{r}) = ( - 1)^{L}\varphi (\mathbf{r})\,\,.
$$
<!--  -->
Враховуючи попередні результати,а також те, що повна хвильова функція є
антисиметричною, отримаємо, що стани з $S = 0$ (синглетні) відповідають парним значенням орбітального моменту. Стани з $S = 1$ (триплетні) мають непарний орбітальний момент. Якщо відомо, що електрони знаходяться у станах з координатними хвильовими функціями 
$\varphi_{1}$ і $\varphi_{2}$, то симетричні і антисиметричні координатні  частини хвильових функцій неважко побудувати:
<!--  -->
\begin{equation}
\label{20eq10}
\begin{array}{l}
 \varphi_{s} (\mathbf{r}_{1} ,\mathbf{r}_{2} ) = \frac{1}{\sqrt{2}}
 \left(\varphi_{1}(\mathbf{r}_{1} )\,\varphi_{2}(\mathbf{r}_{2} ) + \varphi_{1} (\mathbf{r}_{2} )\,\varphi_{2} (\mathbf{r}_{1} ) \right) \\
 \varphi_{a} (\mathbf{r}_{1} ,\mathbf{r}_{2} ) = 
 \frac{1}{\sqrt{2}}\left(\varphi_{1} (\mathbf{r}_{1} )\,\varphi_{2} (\mathbf{r}_{2} ) - \varphi_{1} (\mathbf{r}_{2} )\,
 \varphi_{2} (\mathbf{r}_{1}) \right)
 \end{array}
\end{equation}
<!--  -->
Як видно, у випадку якщо $\varphi_{1}=\varphi_{2}$ остання функція дорівнює 0 завдяки принципу Паулі. Як наслідок того, що гамільтоніан такий системи комутує з оператором спина, витікає, що рівні енергії системи двох електронів можуть бути або синглетними (невиродженими по спіну) або триплетними (3-кратно виродженими по
проекції спіну). В системі невзаємодіючих частинок стани \eqref{20eq10} мають однакову енергію, що легко перевіряється безпосередньо. Це означає, що всі рівні енергії є як мінімум 4-кратно виродженими (по спіну). Наявність слабкої взаємодії, не пов'язаної із дією на спінові степені свободи приведе до того, що рівні енергії розщеплюються на два - один синглетний і один триплетний. Подальше розщеплення можливо тільки для триплетного
при врахуванні спінових взаємодій (магнітного поля, спін-орбітальне і інш.).<br>

## Гамільтоніан та основний стан He 
<!--  -->
Реальною системою, до якої можуть бути застосовані міркування, наведені вище, є атом гелію. Гамільтоніан системи має вид: 
<!--  -->
\begin{equation}\label{hamhe}
 H = H_{1} + H_{2} + \frac{e^{\,2}}{\vert \mathbf{r}_{1} - \mathbf{r}_{2} \vert}  \\
 H_{i} = -\frac{\hbar^2}{2\,m}\Delta_{i} - 2\frac{e^{\,2}}{\vert \mathbf{r}_{\,i}
 \vert}
\end{equation}
<!--  -->
Користуючись результатами які були отримані раніше для атома водня з простих міркувань можна оцінити енергію основного стану атома гелія. Припустимо, що координатна частина хвильової функції основного стану факторизується у добуток хвильових функцій воднеподібних станів 
<!--  -->
$$\Phi_0(\mathbf{r}_1,\mathbf{r}_2) = \varphi_0(\mathbf{r}_1)\varphi_0(\mathbf{r}_2)$$ 
<!--  -->
Тоді з \eqref{hamhe} отримаємо:
<!--  -->
\begin{equation}\label{he_e0}
  E^{(0)}_{He} = \bra{\Phi_0}\hat{H}\ket{\Phi_0} = -8\,E_{0} + e^2
\left\langle\, \frac{1}{r} \,\right\rangle = -8\,E_{0}+2\,E_{0}= -6\,E_{0} = - 81.6 \, \text{eV}
\end{equation}
<!--  -->
Експериментальне значення $E^{(0)}_{He} = -79.1\,\text{eV}$.
Завищення оцінки пов'язане із наближенням для взаімодії яке припускає, що електрон
рухається у полі точкового заряда іншого електрона. Енергія
однократної іонизації може бути розрахована на основі наступного. При іонизації електрон покидає область де він рухався у полі ядра $Z=2$ на нескінченність, де з врахуванням екраніровки іншим електроном $Z=1$. Це можно описати за допомогою певного ефективного значення $1<Z_{eff}<2$ (див. нижче). Оскільки $E\propto Z_{eff}^2$, то для оцінки скористаємось середнім геометричним між двома асимптотиками що враховує те що внесок області сильного екранування ($Z\to 1$) більше ніж ближньої області ($Z\to 2$) завдяки шивдкому зменшенню електронної хвильової функції зв'язаного стану. Тоді отримаємо:
<!--  -->
$$E_{1}\approx -\sqrt{E_{Z=2}\,E_{Z=1}}  = -2E_0= -27.2\,\text{eV}\,.$$
<!--  -->
Експериментальний результат $E^{(exp)}_{1} = -24.6\,\text{eV}$.

## Обмінна взаємодія. Класифікація рівнів атома гелію. Орто- і парагелій.
<!--  -->
Т.ч. основний стан атома гелію є синглетним, триплетний стан заборонений завдяки принципу Паулі. Наступний збуджений стан повинен бути або синглетним або триплетним. З'являється, що найменша енергія при врахуванні кулонівської взаємодії між електронами має триплетний стан. Це можна обгрунтувати наступним чином. У триплетному стані координатна хвильова функція має вузли (див. \eqref{20eq10}), тому ймовірність конфігурацій, у яких електрони опиняються на малих відстанях мала, як наслідок, мало і середнє значення кулонівської енергії відштовхування у порівнянні із її значенням у синглетному стані. Тому триплетні стани лежать нижче синглетних (за виключенням основного, який синглетний). Енергія станів у першому порядку за кулонівською взаємодією електронів дорівнює:

\begin{eqnarray}\label{20eq12}
 E_{S = 0} &=&E_{1} + E_{2} + e^{\,2}\int 
 \frac{\vert \varphi_{s}(\mathbf{r}_{\,1} ,\mathbf{r}_{\,2} )\vert^{\,2}}{\vert \mathbf{r}_{2} - \mathbf{r}_{1} \vert} dV_{1} dV_{2} =
E_{1} + E_{2} + Q + A \nonumber \\ &&\\
 E_{S = 1} &=& E_{1} + E_{2} + e^{\,2}\int \frac{\vert \varphi_{a} (\mathbf{r}_{\,1} , \mathbf{r}_{\,2} )\vert^{\,2}}{\vert \mathbf{r}_{2} - \mathbf{r}_{1} \vert} 
 dV_{1} dV_{2} = E_{1} + E_{2} + Q - A \nonumber
\end{eqnarray}

де 

$$
 Q = e^{\,2}\int \frac{\vert \varphi_{1} (\mathbf{r}_{\,1} )
 \vert^{\,2}\vert \varphi_{2} (\mathbf{r}_{\,2} )\vert^{\,2}}{\vert \mathbf{r}_{2} - \mathbf{r}_{1} \vert} dV_{1} dV_{2}\\
A = e^{\,2}\int \varphi_{2}^*  (\mathbf{r}_{\,1})\varphi_{1}^* (\mathbf{r}_{\,2})\,\frac{1}{\vert \mathbf{r}_{2} - \mathbf{r}_{1} \vert}
\varphi_{1} (\mathbf{r}_{\,1} )\varphi_{2} (\mathbf{r}_{\,2} ) dV_{1} dV_{2}
$$

Доданок, який обумовлений величиною $Q$ має класичну інтерпретацію як енергію взаємодії електронних областей зарядів із відповідними густинами. Інтеграл $A$ навпаки, не має прямої електростатичної інтерпретації і називається обмінним інтегралом, оскільки враховує
можливість обміну станами між електронами. Цей вклад в енергію системи _зазвичай інтерпретують_ як вклад __обмінної взаємодії__ . Зрозуміло, що нічого не зміниться, якщо замість кулонівської ми врахуємо ще якусь взаємодію між електронами. Тоді з'являться відповідні обмінні вклади. Таким чином, __природа обмінної взаємодії визначається вихідною взаємодією тотожних частинок у гамільтоніані__.

Якщо нехтувати спіновими взаємодіями переходи між станами з різними значеннями спінового квантового числа не можливі (спін - інтеграл руху). Враховуючи слабкість таких взаємодій
(релятивистських за своєю природою), можна представити цю ситуацію як наявність двох квазистаціонарних станів гелію: __парагелій__ - синглетні стани, __ортогелій__ - триплетні стани. Вони різняться своїми властивостями: __парагелій є діамагнетиком__ (спін 0), __ортогелій - парамагнетик__ (спін 1) має власний магнітний момент.

### Застосування варіаційного метода для обчислення основного стану атому ${\rm He}$
Продемонструємо теорію обчисленям основного стана атома гелія, яке відповідає парастану з $S=0$ у конфігурації $1s^2$.

Для розрахунків використовуємо хвильову функцію $1s$ стана ($n=1,l=0$) воднеподібної задачі. Тоді симетрична координатна частина хвильової функції є:

$$\Psi = \psi_0(r_1;\alpha)\,\psi_0(r_2;\alpha) = \frac{\alpha^3}{\pi}\,e^{-\alpha(r_1+r_2)}$$

З фізичної точки зору варіаційний параметр $\alpha$ визначає ефективний радіус Бора 

$$
a^{(eff)}_{B}=\frac{1}{\alpha} = \frac{\hbar^2}{Z_{eff}\,m\,e^2}
$$

і враховує часткове екранування поля ядра іншим електроном, так що $1<Z_{eff}<2$.
Оскільки:

$$\Delta \psi_0(r;\alpha) = \frac{1}{r}(r\psi_0(r;\alpha))'' = -\frac{2\,\alpha}{r}\,\psi_0(r;\alpha) + \alpha^2\,\psi_0(r;\alpha)$$

то середнє значення енергії дорівнює (у одиницях $\hbar = 1\,, e=1,\, m=1$):

$$\langle H \rangle = \alpha^2 - 4\,\alpha+ \frac{\alpha^6}{\pi^2}\, \iint \frac{e^{-2\alpha(r_1+r_2)}}{r_{12}}\,dV_1\,dV_2  $$

Розрахунок інтеграла засновано на розкладі по поліномам Лежандра:

$$
\frac{1}{r_{12}} = 
\begin{cases}
\frac{1}{r_1}\,\sum\limits_{n=0}^{\infty} \left(\frac{r_2}{r_1} \right)^n\,P_{n}(\cos\theta)\,, & \mbox{if } r_2<r_1<\infty \\
\frac{1}{r_2}\,\sum\limits_{n=0}^{\infty} \left(\frac{r_1}{r_2} \right)^n\,P_{n}(\cos\theta), & \mbox{otherwise }.
\end{cases}
$$
[нагадаємо, що](https://en.wikipedia.org/wiki/Legendre_polynomials):

$$
\frac{1}{\sqrt{1-2\,y\,x+y^2}} = \sum\limits_{l=0}\,y^{l}\,P_{l}(x)
$$

```python
r1,r2=symbols('r1 r2',positive=True)
x=symbols('x',real=True)
intx=simplify(integrate(1/sqrt(r1**2+r2**2-2*r1*r2*x),(x,-1,1)))
intx
```


з якого видно, що весок в інтеграл дадуть тільки стани з $n=0$. Тому

$$\iint \frac{\psi_{0}(r_1)^2\psi_{0}(r_2)^2}{r_{12}} dV_1dV_2 = \frac{(4\pi)^2}{\pi^2}\int\limits_{0}^{\infty}dr_2 r_2^2e^{-2\alpha r_2}\left(\frac{1}{r_2}\int\limits_{0}^{r_2}
e^{-2\alpha r_1}\,r_1^2\,dr_1 + \int\limits_{r_2}^{\infty}e^{-2\alpha r_1}\, r_1 \, dr_1\right)$$

Простий розрахунок веде до результату:

$$
\iint \frac{\psi_{0}(r_1)^2\psi_{0}(r_2)^2}{r_{12}} dV_1dV_2 = \frac{5}{8}\,\alpha
$$

Мінімізація середнього значення:

$$E(\alpha) =-\frac{27}{8}\,\alpha\,+\alpha^2 $$

по параметру $\alpha$ дає:

$$\alpha_{\text{min}} = \frac{27}{16} = Z_{eff}$$

і

$$E_0 \approx 2.85 $$

що у розмірних величинах відповідає значенню $77.5$ еВ. Експериментальне значення
дорівнює $79.1$ еВ.
