__Задача__ Для сферично симетричного бар'єру типу стінки $V=V_0\,, r\le R_0\,,\,\,V=0\,, r>R_0$ для парціальної s-хвилі ($l=0$) знайти (за точним розв'язком рівняння Шредінгера) фазовий зсув $\delta_0$ та переріз розсіяння 
$\sigma_0$.<br>
__Розвязок__ Запишемо рівняння Шредінгера для $l=0$ у відповідних областях:
<!--  -->
$$
r>R_0\,\,:\,\,\,\frac{1}{r}\,(r\,\psi_0(r))'' = - k^2\,\psi_0(r)\,,\,\,k^2 = \frac{2\,m\,E}{\hbar^2}\\
r<R_0\,:\,\,\,\frac{1}{r}\,(r\,\psi_0(r))'' = - (k^2-q^2_0)\,\psi_0(r)\,,\,\,\,\,q^2_0 =\frac{2\,m\,V_0}{\hbar^2}
$$
<!--  -->
із звичайними умовами неперервності х.ф. та її похідної на границі $r=R_0$.
Спочатку розглянемо випадок $E>V_0\to k>q_0 \Rightarrow \tilde{k}^2 = k^2-q^2_0>0$:
<!--  -->
$$
r>R_0\,\,:\,\,\,\psi_0(r) = \frac{\sin (k\,r+\delta_0)}{r}\\
r<R_0\,:\,\,\,\psi_0(r) = C\,\frac{\sin \tilde{k}\,r}{r}
$$
<!--  -->

```python
k,kt,r,d0, c=symbols('k kt r d0 c')
psiout = sin(k*r+d0)/r
psin = c*sin(kt*r)/r
sol=solve(diff(psiout,r)/psiout-diff(psin,r)/psin,d0,dict=True)
```

Фазу $\delta_0$ одразу можна отримати з граничної умови неперервності логарифмічної похідної:
$$\left.\frac{\psi'}{\psi}\right\vert_{r\to R_0+} =\left. \frac{\psi'}{\psi}\right\vert_{r\to R_0-}$$

```python
simplify(diff(psiout,r)/psiout-diff(psin,r)/psin)
```

$$\delta_0 = -k\,R_0+ {\rm ArcTan}\,\left(\frac{k}{\tilde{k}}\,{\rm Tan} \,\tilde{k}\,R_0\right)$$

```python
fig = plt.figure()
ax = fig.add_subplot(111) 
# 
def delta0(x):
    return -x+np.arctan(x/np.sqrt(x**2-1)*np.tan(np.sqrt(x**2-1)))
# 
x=np.linspace(1.1,15,200)
# 
ax.plot(x, delta0(x))
plt.title('Phase $\\delta_0$',fontsize=14)
plt.show()
```
