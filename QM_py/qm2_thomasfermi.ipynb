## Теория Томаса-Фермі багатоелектронних атомов.

Для елементів, які містять велику кількість електронів розрахунки СУП пов'язані з великими алгоритмічними труднощями, вирішення яких потребує значного машинного часу. Тому виникає необхідність у наближеному методі, який хоча і був би менш точним, але дозволяв би на основі простих фізичних міркувань основні закономірності. Теорія такого роду була запропонована Л. Томасом і Е. Фермі. Основним припущенням є те, що у багатоелектронному атомі більшість електронів знаходяться у станах з великими квантовими числами, що дозволяє застосувати до їх опису квазикласичне наближення для отримання потенціалу СУП. Дійсно, число електронів, які мають імпульс менший певної величини дорівнює (з врахуванням спіну):

\begin{equation}
\label{tfeq1} 
2\,\frac{4\pi \,p^{\,3}}{3\,(2\pi \,\hbar)^{3}}\,dV
\end{equation}

Оскільки повна енергія електрона в атомі від'ємна (зв'язаний стан), то існує граничне значення імпульсу $p_{\max}  $ яке, взагалі кажучи, залежить від координати (квазікласика!). Тоді:

\begin{equation}
\label{tfeq2} 
2\,\frac{4\pi \,p_{\max}^{3}} {3\,(2\pi \,\hbar)^{3}}\,dV = 
n\,dV \Rightarrow p_{\max}  = \hbar \,\left( 3\pi^{\,2}n \right)^{1 / 3}
\end{equation}

В стаціонарному стані максимальне значення енергії електрона не повинно залежати від координат (в протилежному випадку виник би потік електронів), тому:

\begin{equation}
\label{tfeq3} 
\frac{p_{\max}^{\,2}}{2\,m} - e\,\varphi = - e\,\varphi_{0} = const
\end{equation}

(потенціал на $\infty $ дорівнює 0). З (\ref{tfeq2}) і
(\ref{tfeq3}) отримуємо:

\begin{equation}
\label{tfeq4} 
n = \frac{1}{3\pi^{\,2}\hbar^{3}}\left(2m\,e\,(\varphi - \varphi_{0} ) 
\right)^{3 / 2}
\end{equation}

Використовуючи рівняння Пуассона, остаточно отримуємо рівняння,
яке визначає потенціал СУП в теорії Томаса - Фермі:

\begin{equation}
\label{tfeq5} 
\Delta \,\varphi = 4\pi en = \frac{4\,e}{3\pi \hbar^{3}}
\left(2m\,e\,(\varphi - \varphi_{0} ) \right)^{3 / 2}
\end{equation}

Будемо шукати тільки сферично симетричні рішення цього рівняння.
Тоді \eqref{tfeq5} є нелінійним ОДУ 2-го порядку:

\begin{equation} \label{tfeq6}
\frac{1}{r}\,\left( r\,\varphi  \right)^{"} =
\frac{4\,e}{3\pi \hbar^{3}}\left( 2m\,e\,(\varphi - \varphi_{0})\right)^{3 / 2}
\end{equation}

Це рівняння повинно бути доповнено відповідними граничними
умовами:

\begin{equation}
\label{tfeq7}
\begin{array}{l}
 r \to 0\,, \quad \varphi \to \frac{Z\,e}{r} \\
 r \to \infty\,, \quad  r\,\varphi \to q \\
 \end{array}
\end{equation}

Тут $q$ - заряд "атома", тобто іона (у випадку атома $q=0$). З рівняння (\ref{tfeq5}) величина $\varphi_{0} $ визначає межу "атома" в моделі Томаса - Фермі. Дійсно, при $\varphi = \varphi_{0}\,, \to n = 0$, об'ємний заряд відсутній. За неперервністю звідси витікає, що $\varphi = 0$ при $r > r\,_{0}$, де $r\,_{0} $ - розмір "атома". Неважко бачити, що $\varphi_{0} = 0$ відповідає нейтральному атому, в той час як $\varphi_{0} > 0$ - іону. Таким чином, в теорії Томаса - Фермі радіус атома є необмеженим. Розглянемо випадок атома. Внаслідок
того, що (\ref{tfeq7}) містить тільки універсальні стали, як
наслідок, теорія Томаса - Фермі універсальна в том сенсі, що
кількісні характеристики атомів в цьому наближенні відрізняються
тільки масштабними факторами, які залежать від номеру $Z$. Вводячи
безрозмірні змінні:
$$
\varphi (r) = {\frac{{Ze}}{{r}}}\chi \left( {{\frac{{r}}{{r_{0}}} }}
\right)
$$
і підібрав $r_{0} $ таким чином, щоб скоротити всі константи:
\begin{equation} \label{tfeq8} r_{0} = Z^{ - 1 /
3}\left( {{\frac{{3\pi}} {{4}}}} \right)^{2 / 3}{\frac{{\hbar
^{\,2}}}{{2me^{\,2}}}},
\end{equation}
остаточно отримаємо:
\begin{equation}
\label{tfeq9}
\begin{array}{l}
 x^{1 / 2}\chi "(x) = \chi^{3 / 2} \\
 \chi (0) = 1\,,\quad \chi (\infty ) = 0 \\
 \end{array}
\end{equation}

З наведених результатів можна отримати ряд цікавих виводів. Як
витікає з (\ref{tfeq8}) розподіл електронної густини у
багатоелектронних атомах подібні з коефіцієнтом  $\,\,\sim Z^{ - 1
/ 3}$. Це означає, що з ростом атомного номера електрони у
більшості рухаються поблизу ядра. Середня швидкість електронів в
атомі $\,\,\sim Z^{2 / 3}$. Цікаво також відмітити, що в даному
наближенні можна досить точно обчислити, починаючи з якого Z, буде
заповняться дана оболонка. Електрон на орбіталі, яка відповідає
значенню квантового числа $l$ рухається у полі ефективної
потенціальної енергії:

\begin{equation}
\label{tfeq10} U_{eff} (r) = - e\varphi (r) + {\frac{{\hbar
^{\,2}\left( {l + 1 / 2} \right)^{\,2}}}{{2mr^{\,2}}}}
\end{equation}

Перший доданок це потенціальна енергія СУП, друге - енергія
обертання у квазикласичному наближенні (правило квантування
Бора-Зоммерфельда). Заповнення орбіталі можливо тільки якщо:

\begin{equation}
\label{tfeq11} U_{eff} (r) < 0
\end{equation}

і це трапиться вперше якщо

\begin{equation}
\label{tfeq12} U_{eff} (r) = 0,\quad {\frac{{d}}{{d\,r}}}\,U_{eff}
(r) =0
\end{equation}

Тобто орбіталь з числом $l$ починає заповнюватися для порядкового
номеру $Z\propto l^3$. Розрахунок приводить до залежності:

\begin{equation}
\label{tfeq13} Z_{*} = 0.155\,(2l + 1)^{3}
\end{equation}

У наближенні ТФ також можна визначити енергію атома. Енергія буде складатись з трьох внесків (рухом ядра нехтуємо):

- кінетична енергія електронів;
\begin{equation}\label{tfekin}
  E_{k} = 2\int\limits_{|\mathbf{p}|\le p_{max}} \int\limits_{V} \frac{\mathbf{p}^2}{2m}\,
 \frac{dV\,d\,\mathbf{p}}{(2\pi\hbar)^3} = 2\int\limits_{|\mathbf{p}|\le p_{max}} \int\limits_{V} \frac{p^4}{2m}\,4\pi
 \frac{d\,{p}\,dV}{(2\pi\hbar)^3} = \frac{3\hbar^2\left(\,3\pi^2\,\right)^{2/3}}{10\,m}\,\int\limits_{V} n^{5/3}\,dV
\end{equation}

- потенціальна енергія електрон-електронної взаємодії,

\begin{equation}\label{tfee}
  E_{e-e} = -\frac{e}{2}\int\limits_{V}  \varphi_{e}\, n\,dV =
  \frac{Ze^2}{2}\int\limits_{V} \frac{n}{r}\,dV - \frac{\hbar^2\,\left(\,3\pi^2\,\right)^{2/3}}{4m}
  \int\limits_{V} n^{5/3}\,dV
\end{equation}
де
$$\varphi_{e}= \varphi - \frac{Z\,e}{r}$$
  - потенціал електрон-електронної взаємодії;

- потенциальна енергія взаємодії електрон-ядро.

\begin{equation}\label{tfen}
  E_{n-e} = -Ze^2\int\limits_{V}  \frac{n}{r}\,dV
\end{equation}

З цих виразів виводимо, що:

\begin{equation}\label{tf_eee}
E_{n-e}= - 2 E_{e-e} - \frac{5}{3} E_{k}
\end{equation}

Враховуючи теорему віріала:

$$E_{e-e}+ E_{n-e}= - 2 E_{k}\,$$

з \eqref{tfen} остаточно знаходимо:

\begin{equation}\label{vir}
E = -3 E_{ee} = -E_{k}\,,\,\,\,\,\,  E_{ne} = - 7 E_{ee} = \frac{7}{3}\,E\,.
\end{equation}
