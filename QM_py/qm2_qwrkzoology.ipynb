## Квантова зоологія елементарних частинок.
<!--  -->
Класифікація елементарних частинок є чудовим прикладом застосування загальних положень квантової механіки та принципів сметрії. Вивчення закономірностей у взаємоперетвореннях частинок, перерізів розсіяння та реакцій розпаду дозволяє відкривати закони будови матерії. Закони збереження (наближені та точні) відображують структуру внурішнього простору та відповідну групу симетрії.
