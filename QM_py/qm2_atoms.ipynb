# Будова спектра багатоелектронних атомів. Самоузгоджене поле (СУП). Одноелектронні стани.
<!--  -->
Опис багатоелектронного атома є дуже складною задачею внаслідок великої кількості ступенів свободи, які взаємодіють між собою як звичайною електростатичною взаємодією, так і взаємодіями релятивистської природи: спін-орбітальною, спін-спіновою та інш.. Строго кажучи, опис такої системи не зводиться до опису її окремих складових, оскільки атом є цільним квантовим об'єктом. Тим не менш, як буде видно з подальшого, можна ввести поняття про одноелектронні стани у такій системі, застосував, таким чином,
результати, отримані раніше для атома водню. Нехтуючи взаємодіями релятивистської природи (нижче буде указано у порівнянні з чим), які явно залежать від спінових ступенів свободи, гамільтоніан атома має вигляд:
<!--  -->
\begin{equation}
\label{ateq1} H = \sum\limits_{k = 1}^{Z} H^{(0)}_{k}   + 
\frac{e^{\,2}}{2}
\sum\limits_{i \ne k} \frac{1}{\vert r_{i}- r_{k} \vert} 
\end{equation}
<!--  -->
Тут перша сума є сумою одноелектронних гамільтоніанів, друга - електростатична взаємодія електронів. Для покращення результатів теорії збурень можна перебудувати гамільтоніан таким чином щоб включити сферично-симетричу частину електростатичної енергії електрон-електронної взаємодії до незбуренного одночастинкового гамільтоніану:
<!--  -->
\begin{equation}
\label{ateq2} 
H_0 = \sum\limits_{k = 1}^{Z} H^{(0)}_{k}   + \sum_k e\,\Phi_k(r_k)
\end{equation}
<!--  -->
Якщо потенціали $\Phi_k$ підібрати оптимальним чином так щоб спектр енергії такого одночастинкового гамільтоніана був якомога ближче до реального то остаточна (залишкова) взаємодія
<!--  -->
$$H^{(int)} = \frac{e^{\,2}}{2} \sum\limits_{i \ne k} \frac{1}{\vert r_{i}- r_{k} \vert} - 
\sum_k e\,\Phi_k(r_k)$$
<!--  -->
може вважатись малим збуренням.

Гамільтоніан \eqref{ateq2} є інваріантним відносно обертань, що відповідає закону збереження орбітального моменту імпульсу ${\mathbf{L}}$. Крім того, \eqref{ateq1} комутує з оператором інверсії, що відповідає тому, що кожний стаціонарний стан характеризується своєю парністю. Оскільки в \eqref{ateq1}) відсутня спінова взаємодія, то оператор повного спіну ${\mathbf{S}}$ також є інтегралом руху. Таким чином, в даному наближенні \eqref{ateq1}, кожний стаціонарний стан, який
характеризується числами $L,S$ (атомний терм) є виродженим з кратністю виродження $(2L + 1)(2S + 1)$. 

## СУП Хартрі та Хартрі-Фока
<!--  -->
Задача про спектр багатоелектронного атома розв'язується завдяки зведенню її до ефективної задачі про одночастинковий гамільтоніан на основі якого можно ввести одноелектронні стани у певному полі. З фізичної точки зору введення одноелектронних станів означає, що розглядається поведінка даного електрона у поле ядра і інших електронів. Оскільки таким електроном може бути обраний довільний, необхідно, що би дане поле вводилося самоузгодженим чином, враховуючи той факт, що воно само залежить від одноелектронних станів. Історично першою такою процедурою була модель, запропонована Хартрі. Її фізичний зміст дуже простий: електрон рухається в полі ядра і інших електронів. Дійсно, з \eqref{ateq1}) витікає, що енергія одноелектронного стану дорівнює:
<!--  -->
\begin{equation}
\label{ateq5} \varepsilon_{k} = 
\av{\varphi_{k}}{H^{(0)}_{k}}{\varphi_{k}}+ 
e^{\,2}\,\sum\limits_{i \ne k} \int \frac{\vert \varphi_{i}\vert^{\,2}\vert \varphi_{k}\vert^{\,2}}{\vert r_{i} - r_{k}\vert}
dV_{i}\,dV_{k}
\end{equation}
<!--  -->
Другий доданок є не що інше, як енергія електростатичної взаємодії двох об'ємних зарядів. Очевидно, що \eqref{ateq5}) можна записати у вигляді:
<!--  -->
\begin{equation}
\label{ateq6}
\begin{array}{l}
 \left( H^{(0)}_{k} + \bar {U}^{H}_{k} - \varepsilon_{k}  \right)\,\varphi^{n}_{k} = 0  \\
 k = 1..\,N \\
 \end{array}
\end{equation}
<!--  -->
Система рівнянь \eqref{ateq6} є нелінійною системою інтегро-диференціальних рівнянь - рівняннь Хартрі. Хартрі запропонував рекурентну процедуру для її розв'язку. 
На 0-м шагу у якості $\varphi $ вибираються хвильові функції воднеподібних атомів. По ним будується оператор \eqref{ateq7} потенціальної енергії взаємодії 0-го шагу даного електрона з іншими. Розв'язок системи \eqref{ateq6} визначає хвильові функції наступного наближення, а разом з ними і потенціал поля інших електронів і так далі. 
<!--  -->
\begin{equation}\label{ateq7} 
\bar{U}^{(n)}_{k} = e^{\,2} \sum\limits_{i \ne k} 
\int\frac{\vert \varphi^{(n-1)}_{i} \vert^{2}}{r_{ik}} dV_{i}
\end{equation}
<!--  -->
Процес припиняється, якщо потенціал самоузгодженного поля \eqref{ateq7} побудований на хвильових функціях наступного наближення у границях заданої точності, не відрізняється від отриманого на попередному кроці. В результаті визначається самоузгоджене поле (СУП), яке діє на кожний електрон з боку ядра і інших електронів. В.А. Фок показав, що рівняння Хартрі \eqref{ateq6} можуть бути отримані з варіаційного принципу
для функціоналу:
<!--  -->
\begin{equation}
\label{eq:hartry} 
E = \av{\psi}{H}{\psi}\,,
\end{equation}
<!--  -->
де хвильова функція має вигляд:
<!--  -->
\begin{equation}
\label{eq:psihartry} 
\psi = \prod\limits_{i = 1}^{N} \varphi_{i}
\end{equation}
<!--  -->
за умови
<!--  -->
$$
\braket{\phi_{k}}{\phi_{k}} = 1\,,\quad k = 1,\,\ldots\,, N
$$
<!--  -->
Дійсно:
<!--  -->
$$
\delta E =\av{\delta\psi}{H}{\psi}\,\,\,\,,\,\,\,\,\delta \psi^*=
\sum\limits_{k = 1}^{N} \delta \varphi_{k}^{*}
\prod\limits_{i \ne k}^{N} \varphi_{i}^{*}
$$
<!--  -->
і
<!--  -->
\begin{equation}
\label{ateq10} 
\delta E = 
\sum\limits_{k = 1}^{N} 
\int \delta\varphi_{k}^{*} \left(H^{(0)}_{k} + \bar {U}_{k}^{H} - \varepsilon_{k} 
\right)\varphi_{k} \,dV\, = 0
\end{equation}
<!--  -->
Завдяки довільності варіацій з \eqref{ateq10} витікає \eqref{ateq6}. Для повної енергії основного стану в наближенні СУП отримаємо:
<!--  -->
\begin{equation} \label{ateq11} 
E = \sum\limits_{k = 1}^{N} \varepsilon_{k} - 
\frac{1}{2} \sum\limits_{k = 1}^{N} \av{\varphi_{k}}{\bar{U}_{k}^{H}}{\varphi_{k}}
\end{equation}
<!--  -->
Енергії збуджених станів можуть бути побудовані аналогічним чином враховуючи ортогональність їх власних функцій до всіх нижчих станів.
<!--  -->

Запишемо рівняння Хартрі для атома гелія для електронної конфігураціх де електрони мають енергії $\varepsilon_{1,2}$ відповідно:
<!--  -->
\begin{equation}\label{he_hartry}
	\left\{
	\begin{aligned}
		\left(H^{(0)}_{1} + V_{1}(r_1) - \varepsilon_{1}\right)\varphi_{1}(\mathbf{r}_{1}) &= 0 \\
		\left(H^{(0)}_{2} + V_{2}(r_2) - \varepsilon_{2}\right)\varphi_{2}(\mathbf{r}_{2}) &= 0
	\end{aligned}
	\right.
\end{equation}
<!--  -->
де
<!--  -->
$$V_{1}(r_{1}) = e^2\,\int \frac{|\varphi_{2}(\mathbf{r}_2)|^2}{|\mathbf{r}_{1}- \mathbf{r}_2|}\,d\mathbf{r}_2\,,$$
<!--  -->
Якщо одночастинкові стани однакові, як у випадку конфігурації $1s^2$, то самоузгоджені потенціали також одинакові $V_{1} = V_{2}$.

Описаний підхід до побудови СУП на основі \eqref{eq:psihartry}, однак, є не повністю послідовним. Вибір хвильової функції у вигляді простого добутку \eqref{eq:psihartry}, не задовольняє принципу тотожності
і, як наслідок, нехтує обмінними ефектами, які е суттєвими. Процедура побудови СУП з врахуванням кореляцій, які обумовлені принципом Паулі, називається наближенням Хартрі - Фока. Наприклад, визначення
енергії основного стану для атому з повністю заповненими (при даному N) одночастинковими станами (заповнена оболонка), робиться на основі вибору хвильової функції у вигляді:
<!--  -->
\begin{equation}
\label{eq:psifock} 
\psi = \frac{1}{\sqrt {N!}}
\left|
\begin{array}{*{20}c}
 \varphi_{1} (1) \hfill & \ldots \hfill & \varphi_{1} (N) \hfill \\
 \ldots \hfill & \ldots \hfill &\ldots \hfill \\
 \varphi_{N} (1) \hfill & \ldots \hfill & \varphi_{N} (N) \hfill \\
\end{array} 
\right| = 
\frac{1}{\sqrt{N!}}\,\varepsilon^{i_{1}}
,\ldots\,,i_{N}\varphi_{i_{1}}(1)\,\varphi_{i_{2}}(2)\,\ldots\,\varphi_{i_{N}}(N)
\end{equation}
<!--  -->
Тут $\varphi_{{\rm n}}$ позначають хвильові функції станів одноелектронної задачі про рух електрона у сферично симетричному потенціалі. Користуючись \eqref{eq:psifock} замість \eqref{eq:psihartry} отримуємо СУП з урахуванням обмінних вкладів. Останні не мають класичної електростатичної інтерпретації взаємодії об'ємних зарядів (ситуація є аналогічною до розглянутої раніше у атомі гелію, де врахування правильної симетрії координатної частини хвильової функції приводило до різниці у енергіях станів з різним
повним спіном). У загальному випадку існування незаповнених одночастинкових станів (вакансій) розв'язок слід шукати у вигляді суперпозиції хвильових функцій виду \eqref{ateq13}, які
відповідають електронній конфігурації. Так, наприклад, електронній конфігурації $(1s^{1})(2s^{1})$ відповідають атомні терми ${}^{1}S_{0} ,{}^{3}S_{1} $, станів з повним спіном 0 (синглетний
терм) і 1 (триплетний терм). Оскільки електронної конфігурації, взагалі кажучи, відповідає декілька термів, то стан атома характеризується записом електронної конфігурації і того
терму, який їй відповідає. Наприклад, для основного станіу атома
водню H $(1s)\,^{\,2}S_{1 / 2} $, для He $(1s^{\,2})\,^{1}S_{0} $, Li $\left( {1s^{\,2}} \right)\,\left( {2s^{1}} \right)\,{}^{\,2}S_{1 / 2} $, Be $\left( {1s^{\,2}}
\right)\,\left( {2s^{\,2}} \right)\,{}^{1}S_{0} $, B $\left({1s^{\,2}} \right)\,\left( {2s^{\,2}} \right)\left( {2p^{1}} \right)\,{}^{\,2}P_{1 / 2} $ і інш.. Взагалі у випадку не повністю заповненої оболонки, яка відповідає квантовому числу $l$, з числом електронів $t < 2(2l + 1)$, можливі:
<!--  -->
\begin{equation}\label{ateq15} 
C_{2(2l + 1)}^{t} = \frac{(4l + 2)!}{(4l + 2 -t)!\,t!}
\end{equation}
<!--  -->
електронних конфігурацій. Саме це число\, визначає кількість детермінантів, у суперпозиції для пробної хвильової функції в методі Хартрі-Фока для незаповненої оболонки. Зрозуміло, що випадки, коли числа електронів в оболонці дорівнюють $t$ або $4l + 2 - t$ еквівалентні.

Для приклада розглянемо випадок 2-х електронів в стані з $l = 1$ ($p$ - орбіталь). Очевидно, повний орбітальний момент може відповідати $L = 0,1,2$, а спін $S = 0,1$. Число способів дорівнює 15. Одразу можна сказати, що завдяки принципу Паулі терми ${}^{3}S$ і ${}^{3}D$ заборонені. Залишаються терми
${}^{1}P,\,{}^{1}S,\,\,{}^{3}P,\,{}^{1}D$ з числом станів $3,1,9,5$ відповідно. Очевидно, $15 = {}^{1}S + {}^{3}P + {}^{1}D = 1 + 9 + 5$. Тому можливі тільки терми ${}^{1}S,\,\,{}^{3}P,\,
{}^{1}D$.

У випадку 3 електронів на $p$ орбіталі, число способів розміщення \eqref{ateq15} дорівнюють 20. Зрозуміло, що повний орбітальний момент такої системи може приймати значення: $L = 0,1,2$ ($F$-терм
з $L = 3$ заборонений принципом Паулі). Спіновий момент відповідно $S = \frac{1}{2},\,\frac{3}{2}$. Простіше всього визначити терми з $S = 3 / 2$, Оскільки в цьому випадку, існує повністю
симетричне по спіновим змінним стан ${\left. {{\left| { \uparrow}
\right.}} \right\rangle} \,{\left. {{\left| { \uparrow} \right.}}
\right\rangle }\,{\left. {{\left| { \uparrow}  \right.}}
\right\rangle} $, який відповідає максимальної проекції спінового
моменту 3/2. Очевидно, що внаслідок принципу Паулі існує тільки
один варіант розподілу електронів по власними станам проекції
моменту імпульсу: $( - 1;\, 0\,;\,+ 1)$, що відповідає $M = 0$ і
повністю антисиметричної функції (детермінанту $3\times 3)$.
Відповідний терм ${}^{4}S$ (мультиплетність 4). Далі розглянемо
терми з $S = {\frac{1}{2}}$. Їх може бути три:
${}^{\,2}S,\,{}^{\,2}P,\,{}^{\,2}D$. Число станів в кожному
дорівнює відповідно $2,\,6\,,\,10$. Звідки одразу ясно, що $20 =
{}^{4}S\, + {}^{\,2}P\, + {}^{\,2}D = 4 + 6 + 10$. Терм $^{\,2}S$
не існує, оскільки всі стани з $M = 0$ вже є у вказаних термах.

## Врахування не ізотропних внесків до СУП
Взагалі, потенціал СУП \eqref{ateq7} не є сферично симетричним, однак з нього можна виділити сферично симетричну складову, розглядаючи іншу частину як збурення. В результаті гамільтоніан багатоелектроного атома у наближенні Хартри має вигляд:
<!--  -->
\begin{equation}
\label{ateq12} H = 
\sum\limits_{k = 1}^{N} \left( H^{(0)}_{k} + \bar{U}_{k}^{H} (r_{k}) \right)  + W
\end{equation}
<!--  -->
Тут $W$ - позначає оператор взаємодії, який зображує відхилення потенціала СУП від точного потенціалу. Він розглядається як збурення. Оператор збурення у \eqref{ateq12}, таким чином, має структуру:
<!--  -->
\begin{equation}
\label{ateq13}
\begin{array}{l}
 W = W_{q} + W_{s - orb} \\
 W_{q} = \frac{1}{2}
 \sum\limits_{i \ne k}^{N} \frac{e^{\,2}}{\vert\, \mathbf{r}_{i} -
 \mathbf{r}_{k}\vert} - 
 \sum\limits_{k =1}^{N} \bar {U}_{k}^{H}(r_{k} )  \\
 \end{array}
\end{equation}
<!--  -->
де $W_{s - orb} $ - спін-орбітальна взаємодія, $W_{q} $ - відхилення СУП від точного потенціалу електростатичної взаємодії (остаточна взаємодія). Принцип класифікації станів багатоелектронного атома визначається співвідношенням між спін - орбитальною і остаточною взаємодіями. Розглянута раніше класифікація термів побудована на припущені, що ${\mathbf{L}}$ і ${\mathbf{S}}$ є інтегралами руху. Така картина адекватна, тільки якщо $W_{q} \gg W_{s - orb} $ так, що одночастинковий орбітальний момент не зберігається, а зберігається тільки сумарний орбітальний момент і повний спін. Така ситуація називається випадком $LS$-зв'язку або Рассел-Саундерсовським (див. вище). 

Таким чином, гамільтоніан системи представлений у вигляді суми одночастинкових СУП-операторів і збурення до СУП. Тому, в рамках метода СУП, можливо ввести (наближено) поняття про одноелектронні стани у багатоелектронному атомі. Такі стани характеризуються, як відомо, квантовими числами $n,\,l,\,m,\,m_{s} $, причому $n = l + 1,\,\,l + 2,\,\ldots$. Стан такої багаточастинкової системи в рамках СУП можна характеризувати т.з. електронною конфігурацією, тобто вказувати одночастинкові стани які є заповненими. Електрона конфігурація записується у вигляді послідовності чисел (головних квантових чисел) і символів, які відповідають орбітальному квантовому числу. При заданих $n,l$ електрон може знаходитись у $2(2l + 1)$ станах, які відповідають різним проекціями орбітального моменту і спіна ($s=1/2$) на даний напрямок. Наприклад, $(1s^{1})(2s^{1})$ позначає стан атома гелію, у якому один з електронів находиться у основному воднеподібному $n = 1,\,l = 0$ стані, а другий у першому збудженому $n = 2,\,l = 0$. Зрозуміло що слова "перший" і "другий" досить умовні, оскільки електрони є тотожніми. Фактично ж просто вказуються числа заповнення одночастинкових станів, які завдяки ферміоному характеру статистики (принцип Паулі) можуть приймати тільки значення $0$ чи $1$. Повна хвильова функція електронної конфігурації є антисиметричною (див. нижче).

Про сукупність електронів, які заповнюють всі можливі при даних $n,\,l$ стани, говорять як про замкнену оболонку типу $n,\,l$. Повний орбітальний момент і спін такій оболонки дорівнюють 0. Дійсно, оскільки заповнення електронних станів здійснюється згідно з принципом Паулі, виникає компенсація проекцій всіх векторних операторів - інтегралів руху, а саме, повного спіна і орбітального моменту.
Всеможливі терми, які
відповідають даної конфігурації, а також обчислення їх енергії засновано на наступних принципах. Повністю заповненої оболонці $s^{\,2},p^{\,6},d^{\,10},\ldots$ відповідає терм ${}^{1}S_{0} $.
Різниця в енергії для термів з однаковою електронною конфігурацією обумовлена електростатичною взаємодією електронів (обмінна взаємодія). Найменшу можливу енергією має терм з максимально
можливим значенням $S$ при даної конфігурації (не заповнена оболонка) і максимально можливим при даному $S$ значенні $L$ (емпірічне правило Хунда). В нормальному стані при однієї не повністю заповненої оболонці, якщо вона заповнена не більш ніж на половину, $J = \vert L - S\vert $. Якщо оболонка заповнена більш ніж на половину, то $J = L + S$. Цей факт пов'язаний з тією обставиною, що у \eqref{ateq4} $\bar {A} > 0$ і, як було сказано вище, рівність 0 повного спіна і орбітального моменту імпульсу відповідає повністю заповнений оболонці.

При врахуванні релятивистських ефектів типа спін-орбітального і спін-спінового взаємодій дане виродження знімається (тонка структура) і залишається тільки (універсальний) закон збереження повного
моменту імпульсу $\mathbf{J} = \mathbf{L} + \mathbf{S}$, який характеризує стаціонарні стани своїм квантовим числом $J = \vert L - S\vert \,\,\ldots\,\,L + S$. В більшості випадків такі взаємодії можна розглядати як збурення, так що, у першому наближенні хвильові функції будуть визначатись певними лінійними комбінаціями незбурених хвильових функцій з даними $L,S$ ($L\,S$ зв'язок). Тому розщеплені рівні можна характеризувати цими квантовими числами. Це означає, що величина розщеплення менша ніж відстань між різним термами. Таким чином, рівень з даними $L,S$
розщеплюється на:
<!--  -->
$$
 2S + 1\,,\quad L \ge S \\
 2L + 1\,,\quad S > L 
$$
<!--  -->
підрівней. Кожний з цих підрівнів залишається виродженим по проекції повного моменту $J_{z} $. Класифікація атомних рівнів енергії проводиться за наступними правилам:
<!--  -->
\begin{equation}
\label{ateq3} L = 
\left\{
\begin{array}{cccccc}
 0& 1&   2&   3&   4&   5 \\
 S& P&   D&   F&   G&   H \\
 \end{array} 
 \right.
\end{equation}
<!--  -->
і інш. Зверху ліворуч  вказується мультиплетність $2S + 1$ (слід, однак, пам'ятати, що воно співпадає з числом рівнів тонкої структури тільки при $L \ge S)$. Справа внизу вказується значення
$\vert L - S\vert \, \le J \le \,\,L + S$, наприклад ${}^{\,2}S_{1/2} \,,\,{}^{\,2}P_{3 / 2} \,,\,{}^{\,2}P_{\,1 / 2} \,.$ Неважко отримати величину розщеплення за рахунок спін-орбітальної
взаємодії у випадку LS-зв'язку:
<!--  -->
$$\mathbf{L}\cdot\mathbf{S} = \frac{1}{2}\left((\mathbf{L}+\mathbf{S})^2 - \mathbf{L}^2 - \mathbf{S}^2\right)$$

\begin{multline}\label{ateq4}
\av{\psi}{A\,\mathbf{L}\cdot\mathbf{S}}{\psi}_{J} -
\av{\psi}{A\,\mathbf{L}\cdot\mathbf{S}}{\psi}_{J - 1} = 
\frac{\bar{A}}{2}
\left( J(J + 1) - L(L + 1) - S(S + 1)\right)-\\
\frac{\bar{A}}{2}\left( J(J - 1) - L(L + 1) - S(S + 1) \right) = \bar{A}\,J\,.
\end{multline}

Якщо $\bar{A}>0$,  то такий мультиплет називають нормальним. У протилежному випадку -- оберненим. Якщо додатково і взаємодію між електронами розглядати як малу, то можлива подальша класифікація по одноелектронним станам.

У протилежному випадку $W_{q} \ll W_{s - orb} $ мова може йти тільки о збереженні одночастинкового повного моменту імпульсу. Такій випадок називається $jj$-зв'язком. Перший тип зв'язку характерний для легких атомів, другий - для тяжких елементів. Фактично, у реальності, реалізується певний проміжний
випадок.


### Періодичний закон з точки зору квантової механіки атома.
Пояснення природи періодичного закону зміни хімічних властивостей атомів, відкритого Д.І. Менделєєвим, демонструє силу квантової механіки, як фізичної теорії, яка адекватно відображує закономірності всесвіту. Метод СУП, який був розглянутий  вище, дозволяє у певному наближенні, говорити про одноелектронні стани в атомі. Енергія електронної конфігурації атома, в цьому наближенні, є адитивна величина. Заповнення одночастинкових станів визначається принципом Паулі (не більш 2-х електронів в одному орбітальному стані). Оскільки основний стан це стан з мінімальною енергією, то при такому заповнені конкурують два основних ефекту: збільшення заряду ядра, який приводить до к пониження енергії, і перехід у стан з більш високим квантовим числом, який приводе до підвищення енергії. Таким чином, енергія зв'язку додаткового електрона повинна зростати до моменту повного заповнення оболонки, тоді вона різко падає,  а також коли починається заселення якоїсь орбітали, наприклад p-орбітали (див. $Be\to B, Mg \to Al$). Така картина заповнення оболонок характерна для всієї послідовності елементів.
Всі електронні стани можна розподілити по групам які послідовно заповнюються

$$
\begin{array}{*{20}c}
 {1s} \hfill & {2} \hfill \\
 {2s,\,2p} \hfill & {8} \hfill \\
 {3s,3p} \hfill & {8} \hfill \\
 {4s,4p,3d} \hfill & {18} \hfill \\
 {5s,5p,4d} \hfill & {18} \hfill \\
 {6s,6p,5d,4f} \hfill & {32} \hfill \\
\end{array}

Як видно, заповнення s і p оболонок робиться регулярним чином, на відміну від d і f орбіталей. Це пов'язано з тим (див. [теорію Томаса - Фермі](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics2/blob/main/QM_py/qm2_thomasfermi.ipynb)), що середня відстань електрона до ядра для d,f-станів є значно меншою, ніж в s і p (при такому ж n). Тому в деяких випадках заповнення d оболонки є більш енергетично доцільним, ніж перехід на більш високий рівень (наприклад, Cr з конфигурацією $3d^{5}4s^{1}$ після конфігурації $3d^{3}4s^{\,2}$ для V). Крім того, оскільки більшість хімічних властивостей елементів пов'язани із зовнішніми електронами, елементи, в атомах яких відбувається заповнення таких оболонок (група рідкоземельних елементів), є близькими за своїми хімічними властивостями. Терми, які відповідають основним станам електронних конфігурацій атомів, визначаються правилом Хунда (для незаповнених оболонок нормальних термів): стан з найнижчою енергією відповідає найбільном можливому значенню S та найбільному L при даному S.

### Атом у магнітному полі. Ефекти Зеємана і Пашена-Бака.
<!--  -->
Як вже відмічалось, наявність тонкої структури атомних рівнів пов'язано з малістю спінової взаємодії. При цьому, кожний атомний терм залишається виродженим по проекції повного моменту. Наявність такої симетрії пов'язано з обертальної інваріантністю гамільтоніана атома, що, в свою чергу, обумовлено тим, що всі взаємодії в системі залежать тільки від відстані між її елементами. Порушення ізотропії, повинно привести до зняття цього виродження. Оскільки повний момент пов'язаний з магнітними властивостями атома, доречно провести розгляд поведінки атома у слабкому магнітному полі. Для простоти розглянемо випадок однорідного магнітного поля (ефект Зеємана). У випадку $L\,S$-зв'язку аналіз такій системи можна провести на основі законів збереження. 

Гамільтоніан атома в однорідному зовнішньому магнітному поле має вид:

\begin{equation}
\label{tfeq14}
\begin{array}{l}
 H = \frac{1}{2\,m}\sum\limits_{k}^{N} \left[ \hat{\mathbf{p}}_{k} -
\frac{q}{c}\hat{\mathbf{A}}(\mathbf{r}_{k} ) \right]^{2} + U -
\frac{q\,\hbar}{m\,c}\mathbf{H}\cdot\mathbf{S} \\
 {\mathbf{S} = \sum\limits_{k}^{N} \mathbf{s}_{k}\,,\quad
\hat{\mathbf{A}}(\mathbf{r}_{k} ) = \frac{1}{2}}
\mathbf{H}\times \hat{\mathbf{r}}_{k} \\
 \end{array}
\end{equation}


сума проводиться за всіма електронами, $q = - \,e$. Тут ми виділили постійну Планка $\hbar$ з оператора спина, так що оператор спина є безрозмірним. Виділяючи в (\ref{tfeq14})
незбурений гамільтоніан, отримуємо:

\begin{multline}\label{tfeq15}
 H = H_{0} + \frac{e}{m\,c}
 \sum\limits_{k}^{N} \mathbf{A}(\mathbf{r}_{k})\cdot\mathbf{p}_{k}+\frac{e^2}{8\,m\,c^2}\,\mathcal{H}^2\,\sum\limits_{k}^{N} \vert\mathbf{e}_H\times\mathbf{r}_{k}\vert^2
 -
\frac{q\,\hbar }{m\,c}
\mathbf{H}\cdot\mathbf{S}=\\ 
H_{0} +
\mu _{B} \mathbf{H} \cdot \left(\mathbf{L}+
2\,\mathbf{S} \right) =
    H_{0} +
    \mu _{B} \mathbf{H} \cdot \,\left(\mathbf{J} + \mathbf{S}
    \right)\\
\end{multline}

$$ \mu _{B} = {\frac{{e\,\hbar}} {{2\,m\,c}}} $$

Оскільки у подальшому магнітне поле за припущенням є малим, ми опустили квадратичний доданок в (\ref{tfeq15}). Зовнішнє магнітне поле приводить до того, що стани з різними проекціями повного
моменту імпульсу мають різні енергії. Якщо магнітне поле мало настільки, що не виникає перекриття мультиплетів при розщепленні, то збурені рівні можна нумеровати квантовими числами $J,L,S$. У першому порядку теорії збурень розщеплення мультиплету, згідно (\ref{tfeq15}), характеризується середніми значеннями операторів $J_{z} ,\,S_{z} $. Середнє значення оператора $J_{z} $ дорівнює $\hbar \,M_{J} $. Середнє значення оператора проекції спина в станах з певними $J,L,S$ (хвильові функції мультиплету) можна розглядати як оператор, який діє в цьому просторі станів. Його дія визначається тільки квантовим числом $M_{J} $. Оскільки він комутує з оператором $J_{z} $, то:

\begin{equation}\label{tfeq16} 
\av{\psi _{J,L,S,M_{J}}}{S_{z}}{\psi_{J,L,S,M_{J}}} = C\,M_{J} ,
\end{equation}

але, з іншого боку:

\begin{equation}
\label{tfeq17}
\begin{array}{l}
\av{J,L,S,M_{J}}{\mathbf{J}\cdot\mathbf{S}}{J,L,S,M_{J}} = C\,J(J + 1) =\av{J,L,S,M_{J}}{(\mathbf{L} + \mathbf{S})\cdot\mathbf{S}}{\,J,L,S,M_{J}} = \\
\av{J,L,S,M_{J}}{\mathbf{L}\mathbf{S}}{J,L,S,M_{J}} + \av{J,L,S,M_{J}}{\mathbf{S}^{2}}{J,L,S,M_{J}} = \\
 \av{\,J,L,S,M_{J}}{\frac{1}{2}\left(\mathbf{J}^{2} -\mathbf{S}^{2} - \mathbf{L}^{2} \right)}{\,J,L,S,M_{J}} + S(S+1)\, = \\
 \frac{1}{2}\left( J(J + 1) - L(L + 1) + S(S + 1) \right)\,,
 \end{array}
\end{equation}

Тому:

\begin{equation}\label{tfeq18} 
\av{\psi _{J,L,S,M_{J}}}{S_{z}}{\psi_{J,L,S,M_{J}}} = \,M_{J}
\frac{\mathbf{J}\cdot\mathbf{S}}{\mathbf{J}^{\,2}}
\end{equation}

і для енергії розщеплення остаточно отримаємо:

\begin{equation}
\label{tfeq19}
\begin{array}{l}
 \Delta \,E = \mu _{B} g\,M_{J} {\rm H} \\
 g = 1 + {\frac{{J(J + 1) - L(L + 1) + S(S + 1)}}{{2\,J(J + 1)}}} \\
 \end{array}
\end{equation}
$g$ називається __множником Ланде__ або гіромагнітним множником. Фізично зрозуміло, що атоми з повністю заповненими оболонками $L = S = 0$ є діамагнітними. Для таких атомів перший порядок по полю тотожно дорівнює 0 і вклад в енергію дає доданок квадратичний по полю. Сприйнятливість такого атома є від'ємною. Для атомів з $J = 0,\,\,L = S \ne 0$ ($J = |L-S|,\,\ldots L+S$) ефект першого порядку також відсутній, але другий порядок який пов'язаний з власними спіновим моментом перевищує діамагнітний вклад і тому такі атоми є
парамагнітними.

З (\ref{tfeq19}) видно, що магнітне поле повністю знімає виродження по проекції $\mathbf{J}$. Ця обставина відрізняє ефект Зеємана від ефекту Штарка, де анізотропія обумовлена електричним полем. При ефекті Штарка залишається виродження рівнів з проекціями $\pm M_{J} $, Оскільки гамільтоніан є інваріантним відносно віддзеркалень в площині, яка містить напрямок поля (проекція моменту імпульсу на ось змінює знак). У випадку ефекту Зеємана гамільтоніан (\ref{tfeq14}) не є інваріантним при такому
перетворенні, тому таке виродження відсутнє. Якщо магнітне поле таке, що енергія спін-орбітальної взаємодії мала у порівнянні з енергією атома у магнітному поле, то виникає перекриття термів і збереження $L,S$ не має місця. Однак можна говорить про збереження проекцій операторів на напрямок поля. В цьому випадку енергія розщеплення, очевидно, дорівнює:

\begin{equation}
\label{tfeq20} \Delta \,E = \mu _{B} \,{\rm H}{\rm (}M_{L} +
2M_{S} {\rm )}
\end{equation}

Розщеплення у випадку сильних (у порівнянні з інтервалами тонкої структури) полів називається ефектом Пашена -Бака.
