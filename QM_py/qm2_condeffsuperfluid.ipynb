## Конденсация Бозе-Айнштайна у ідеальному бозе-газі.

Нагадаємо основні формули для стат. суми (великого) канонічного ансамблю ((В)КА) і термодинамічних потенціалів. Стат. інтеграл і квантова стат. сума КА:

$$Z_{N}(T,V) = \int e^{-\beta\,H_{N}}\,d\Gamma\,\quad d\Gamma = \frac{\prod_{i=1}^{N}d\mathbf{r}_id\mathbf{p}_i}{N!(2\pi\,\hbar)^{3N}}$$

$$Z_{N}(T,V)= \sum_{i} e^{-\beta\,E_{i}}$$

Вільна енергія:
$$F = -T\,\ln{Z_N}$$

Стат. сума (інтеграл) ВКА

$$\Xi(V,T,\mu) = \sum_{N=0}^{\infty}\frac{1}{N!}\sum_{i} e^{-\beta\,(E_{i}-\mu\,N)} = \sum_{N=0}^{\infty}\frac{e^{\beta\,\mu\,N}}{N!}\,Z_{N}$$

Великий термодинамічний потенціал:

$$\Omega(T,\mu,V) = -T\,\ln{\Xi} = -V\,p(T,\mu)$$

Середня кількість частинок у ВКА:

$$n(T\,\mu)=\frac{1}{V} \left.\frac{\partial \Omega}{\partial \mu}\right|_{T,V} $$

Розрахунок термодинамічних функцій зручно проводити у рамках ВКА. 
Нехай спектр стаціонарних одночастинкових станів відомий - $\varepsilon_n$. Завдяки відсутності взаємодії між частинками енергія $N$-частинкової системи складається з енергій окремих частинок: $$E_{N} = \sum\limits_{i} \, n_i\,\varepsilon_{i}$$ де $n_i$ - кількість частинок у стані з енергією $\varepsilon_{i}$. Це кількість називається числом заповнення одночастинкового стану. Таким чином конфігурація $N$-частинкової системи повністю визначається вказанням чисел
заповнення одночастинкових станів. Вектор стану $N$-частинкову системи можна представити у вигляді послідовності чисел заповнення одночастинкових станів:

$$
\ket{\Psi_N} = \ket{n_1,n_2,\ldots}
$$

Зрозуміло, що для фермі-систем $n_i$ дорівнює 0 або 1, для бозе- $n_i\ge 0$. Повна кількість частинок очевидно дорівнює сумі чисел заповнення одночастинкових станів $n_{i}$: $$N = \sum\limits_{i} \, n_{i}$$ Оскільки заповнення рівнів за відсутності взаємодії здійснюється незалежно, великий термодинамічний потенціал є сумою відповідних потенціалів для кожного стану:

\begin{eqnarray}\label{eq:bigcan_quant}
	\Xi &=& \sum\limits_{n_i = 0}\,e^{-\beta\,E_{N} +\mu\,N} = \prod_{i}\sum\limits_{n_i=0}\,e^{-\beta \, \varepsilon_{i}+\mu}
	\\
	\Omega &=& -T\,\ln{\Xi}  = \sum\limits_{i}\,\Omega_i\,,\\
	\Omega_i &=&
	-T\,\ln\,\sum\limits_{n_i=0}\,\exp\left(\,\frac{\mu -
		\varepsilon_i}{T}\,\right)^{n_i}
\end{eqnarray} 

Найбільш просто це розраховується для Фермі-системи, оскільки $n_i = 0,1$. В результаті

\begin{equation}\label{eq:bigcan_quant_fermi} \Omega_i =
	-T\,\ln\,\left(\, 1+ \exp\left(\,\frac{\mu -
		\varepsilon_i}{T}\,\right)\right) 
\end{equation} 

і відповідно визначаємо середню заповненість рівня: 

\begin{equation}\label{eq:ni_fermi}
	\left\langle\, n_i \,\right\rangle =
	\frac{\partial\, \Omega_i}{\partial\, \mu} =
	\frac{1}{1+\exp\left(\,\frac{\varepsilon_i-\mu}{T} \,\right)}\le 1
\end{equation} 

Для бозе-системи обчислення зводиться до суми геометричної прогресії за умови, коли $\mu\le\varepsilon_0$:

\begin{equation}\label{eq:bigcan_quant_bose} \Omega_i =
	T\,\ln\,\left(\, 1- \exp\left(\,\frac{\mu -
		\varepsilon_i}{T}\,\right)\right) 
\end{equation} 

і середнє заповнення рівня: 

\begin{equation}\label{eq:ni_bose}
	\left\langle\, n_i \,\right\rangle =
	\frac{\partial\, \Omega_i}{\partial\, \mu} =
	\frac{1}{\exp\left(\,\frac{\varepsilon_i-\mu}{T} \,\right)-1}\,.
\end{equation} 

Слід зазначити, що ефекти квантової статистики (різниця між бозонами і ферміонами) несуттєві, якщо
$\exp\left(\,\frac{\varepsilon_i-\mu}{T}\right) \gg 1$. Це повинно відбуватись у всякому випадку за високих температур, коли теплова де бройлівська довжина хвилі мала у порівнянні із середньою відстанню між частинками і тому ефект перекриття хвильових пакетів є малим. Частинки ``розмазані`` по всім станам неперервного спектра, що і обумовлює експоненційно малу ймовірність заповнення стану (хім. потенціал від'ємний і великий за абсолютною величиною). Таким чином,  умова

\begin{equation}\label{eq:lambda_density}
	n\,\lambda^3_{T}\ll 1\,,
\end{equation} 

означає перехід до класичної формули Больцмана: 

$$\left\langle\, n_i \,\right\rangle =
e^{-\frac{\epsilon_i - \mu}{T}}\,.$$ 

Отже квантові ефекти повязані із тотожністю частинок проявляються при низьких температурах.

Явище конденсації у бозе газі обумовлено властивістю бозе-частинок накопичуватись у основному стані при $T\to 0$. При $T = 0 $ всі бозони знаходяться у найнижчому  енергетичному рівні. Такие частинки називаются
бозе-айнштайновским конденсатом (БАк). Для ідеального газа вільних частинок це стан $\varepsilon =0$. Тому очевидно що при
$T=0$: 
<!--  -->
\begin{equation}\label{n_t0}
	n_{0}(\mathbf{p}) = (2\pi \hbar)^3\, \frac{N}{V}\,\delta (\mathbf{p})\,.
\end{equation} 
<!--  -->
C підвищенням температури частина частинок буде
переходить на більш високі енергетичні рівні, однак зрозуміло, що певна доля частинок всі ж буде перебувати у основному стані. При високих температурах зрозуміло що всі квантові ефекти нивелюються і доля частинок з $\varepsilon =0$ дорівнює 0. За непрервністю зрозуміло, що повинна існувати певна температура при якій у стані $\varepsilon =0$ з'являється ненульова доля частинок. Тому кількість заповнення розумно прелставити у вигляді суми відповідних внесків:
<!--  -->
\begin{equation}\label{np_t}
	\left\langle\, n(\mathbf{p}) \,\right\rangle = N\,\frac{(2\pi
		\hbar)^3}{V}\,A(T)\,n_{0}(\mathbf{p}) +
	\frac{1}{\exp\left(\,\frac{\frac{\mathbf{p}^2}{2m} - \mu }{T} \,\right)-1}
\end{equation} 
<!--  -->
Тут функція $A(T)$ визначає долю частинок які знаходяться у конденсаті. Зрозуміло, що при температурі за якої з'являється БА-конденсат хім. потенціал повинен прямувати до 0 з боку відємних значень (нагадаємо, що для ідеального бозе газа у будь якому випадку $\mu\le 0$). Використовуючи нормування на повну кількість частинок 
<!--  -->
$$N = V\int\,\left\langle\, n(\mathbf{p})
\,\right\rangle\,\frac{d^3\mathbf{p}}{(2\pi \hbar)^3}$$ 
<!--  -->
з \eqref{np_t} отримаємо: 
<!--  -->
\begin{equation}\label{n_mu}
	N(1-A(T)) = V\int\,\frac{1}{\exp\left(\,\frac{\frac{\mathbf{p}^2}{2m}}{T}
		\,\right)-1}\,\frac{d^3\mathbf{p}}{(2\pi \hbar)^3}
\end{equation} 
<!--  -->
Умова на температуру конденсації яку було вказано
вище визначається оберненням хім. потенціалу у 0:
\begin{equation}\label{n0_mu}
	N = V\int\,\frac{1}{\exp\left(\,\frac{\frac{\mathbf{p}^2}{2m}}{T}
		\,\right)-1}\,\frac{d^3\mathbf{p}}{(2\pi \hbar)^3} =
	\frac{Vm^{3/2}}{2^{1/2}\,\pi^2\,\hbar^3}\,\int\limits_{0}^{\infty}\,\frac{\sqrt{\varepsilon
	}}{e^{\frac{\varepsilon }{T}}-1}\,d\varepsilon =
	T^{3/2}\,\frac{Vm^{3/2}}{2^{1/2}\,\pi^2\,\hbar^3}\,
	\int\limits_{0}^{\infty}\,\frac{\sqrt{x}}{e^{x}-1}\,dx
\end{equation} 
<!--  -->
з якого чисельним інтегруванням можна отримати
$$T_{BEC}  \approx 3.3\,
\frac{\hbar^2}{m}\,\left(\frac{N}{V}\right)^{2/3}$$ 
<!--  -->
Доля конденсату $A(T)$ визначається з рівняння \eqref{n_mu} при $\mu =0$ і по суті співпадає з обчисленням у \eqref{n0_mu}:
<!--  -->
\begin{equation}\label{a_condens}
	1-A(T) = T^{3/2}\,\frac{Vm^{3/2}}{N\,2^{1/2}\,\pi^2\,\hbar^3}\,
	\int\limits_{0}^{\infty}\,\frac{\sqrt{x}}{e^{x}-1}\,dx =
	\left(\,\frac{T}{T_{BEC}} \,\right)^{3/2}
\end{equation} 
<!--  -->
Звідки отримаємо остаточний результат
<!--  -->
\begin{equation}\label{a_becondens}
	A(T) =1- \left(\,\frac{T}{T_{BEC}} \,\right)^{3/2}\,.
\end{equation} 
<!--  -->
Як видно з \eqref{a_becondens} точка БА конденсації є особою, оскільки друга похідна по температурі від $A(T)$ розбіжна. Цікаво проаналізувати поведінку теплоємності, оскільки зрозуміло, що кількість ступенів вільності задіяних у тепловому русі залежить від долі конденсата. На якісном рівні можна очікувати підвищення теплоємності із зростанням температури вплоть до $T_{BEC}$ у зв'язку із зменшенням долі конденсату і збільшенням числа частинок задіяних у тепловому русі. При високих температурах, зрозуміло що $c_v$ прямує до класичної границі $3/2$. Тому характер поведінки $c_v$ залежить від співвідношення значення теплоємності при $T=T_{BEC}$ і класичного значення 3/2. Якщо $c_v(T_{BEC})>3/2$, то очевидно
що температурна залежність не є монотонною. Саме цей випадок має місце. Дійсно, при температурі
$T_{BEC}$ квантові ефекти тотожності частинок все ще є суттєвими, так як хім. потенціал відрізняється від класичного значення. Квантові властивості
пов'язані з перекриттям хвильових пакетів, тому якісно зрозуміло що ефективна кількість ступенів вільності при $T_{BEC}$ більше такового для класичного числа точкових частинок.

Тому $c_v(T_{BEC})>3/2$ і залежність
теплоємності від $T$ має вигляд изображенний на Рис.
Наявність злому похідної теплоємності при температурі конденсації є наслідком розбіжності другої похідної у
функції $A(T)$, яка дає долю конденсату, при температурі конденсації. Покажемо це. Внутрішня енергія визначається тільки
збужденними частинками у станах $\mathbf{p} \ne 0$. Тому при температурах нижче $T_{BEC}$: 
<!--  -->
\begin{equation}\label{u_bose}
	U = \frac{V}{(2\pi \hbar)^3}\,\int \,\frac{\mathbf{p}^2}{2m}\,
	\left\langle\, n(\mathbf{p}) \,\right\rangle\, d^3\mathbf{p} =
	T^{5/2}\,\frac{Vm^{3/2}}{2^{1/2}\,\pi^2\,\hbar^3}\,
	\int\limits_{0}^{\infty}\,\frac{x^{3/2}}{e^{x}-1}\,dx\,,
\end{equation} 
<!--  -->
і внутрішня енергія пропорційна $T^{5/2}$. Відповідно теплоємність веде себе як: 
<!--  -->
$$c_v = \left. \frac{\partial\, U}{\partial\, T}\right|_V
= \frac{5}{2}\,\frac{U}{T} \sim T^{3/2}\,.$$ 
<!--  -->
Нагадаємо, що теплове випромінювання (бозе газ фотонів) чи фононний внесок має іншу залежність ($U\sim T^4\,, C_V\sim T^3$) завдяки лінійному закону дисперсії $\varepsilon \sim p$.}
Оскільки теплоємність визначається похідною внутрішньої енергії за температурою то питання про поведінку теплоємності зводиться до розгляду поведінки хім. потенціалу і його похідних при $T\to T_{BEC}+0$, оскільки при $T\to T_{BEC} -0$ хім. потенціал і всі його похідні дорівнюють 0. Диференціювання виразу який пов'язує хім. потенціал з густиною і температурою вище $T_{BEC}$:
<!--  -->
\begin{equation}\label{nmut}
n = \int\limits_{0}^{\infty}\,\frac{\varepsilon^{1/2}}{e^{\frac{\varepsilon-\mu}{T}}-1}\,d\varepsilon
\end{equation}
<!--  -->
при $T_{BEC}$, дає вираз для $\left. \frac{\partial\, \mu }{\partial\, T}\right|_{T=T_{BEC}+0}$
\footnote{ми учли що при $\mu(T_{BEC}) =0$.}
<!--  -->
$$ \frac{ \partial\,
	\mu }{\partial\, T}\,\int\limits_{0}^{\infty}\,\frac{\varepsilon^{1/2}
	e^{\frac{\varepsilon}{T}}}{\left(\,
	e^{\frac{\varepsilon}{T}}-1\,\right)^2}\,d\varepsilon=
-\frac{1}{T}\int\limits_{0}^{\infty}\,\frac{\varepsilon^{3/2}
	e^{\frac{\varepsilon}{T}}}{\left(\,
	e^{\frac{\varepsilon}{T}}-1\,\right)^2}\,d\varepsilon $$
<!--  -->
Т.ч. ми получаєо $\left.\frac{\partial\, \mu }{\partial\,T}\right|_{T=T{BEC}}\ne 0$, 
що і означає наявність стрибка у похідній теплоємності по температурі. З точки зору
классификації Эренфеста точка конденсації є прикладом фазового переходу 3-го роду.

## Надплинність
### Надплинність неідеального Бозе-газу. Крітерій Ландау.
<!--  -->
Властивість бозечастинок утворювати конденсат при низьких температурах є визначальним для утворення надплинності у рідкому ${\rm He}$. При цьому суттєвим є взаємодія між частинками, оскільки ідеальний бозе-газ навіть маючи бозе-конденсат не може бути надплинним. Це видно з простих енергетичних міркувань \cite{sc_bogolyubov_dan1958}. Якщо конденсат утворює стійкий надплинний стан, то поява елементарного збудження (випадання частинок з конденсату), наприклад при взаємодії (тертя) зі стінками повинно супроводжуватися підвищенням енергії системи, так щоб народження збудження було б енергетично невигідне. Але енергія надплинного конденсата ідеального бозе-газау який рухається зі швидкістю $u$ при появі (квазі)частинки зменшується, тому що:
<!--  -->
\begin{equation}\label{eq:bosequasi0}
E_0 =N_0\,\frac{m\,u^2}{2} >  (N_0-1)\,\frac{m\,u^2}{2} + \frac{m\,v^2}{2}\,,\quad v<u
\end{equation}
<!--  -->
Відповідно надплинний рух не є стійким і весь конденсат ''випариться''. 
Тому надплинність з т.з. термодинамічної стійкості повинна бути обумовлена взаємодією між частинками. В рамках теорії збурень можна вважати її слабкою. Завдяки взаємодії у  системі змінюється спектр елементарних збуджень. Із зрозумілих причин неідеальний бозе-газ вже уявляє суцільне середовище у якому найменшу енергію збудження мають колективні збудження  типу звукових хвиль - фонони, із законом дисперсії $\varepsilon (p) = c\,p$, де $c$ - швидкість звуку. Довжина хвилі таких збуджень значно більше ніж міжчастинкова відстань. Із збільшенням енергії (імпульсу) колективний харакетр зникає і збудження все більше набуває характер руху окремої частинки з масою $m$ тобто законом дисперсії $\varepsilon (p) = \frac{p^2}{2\,m}$. З \eqref{eq:bosequasi0} зрозуміло, що для надплинності спектр елементарнх збуджень повинен мати такий характер щоб випадіння частинок з конденсату тобто народження ЕЗ було енергетично не вигідно.
<!--  -->
Нехай конденсат рухається зі швидкістю $\mathbf{u}_0$  і в ньому народжується збудження з імпульсом $\mathbf{p}$, тоді імпульс конденсата:
$$\mathbf{P}_0 = M\,\mathbf{u}_0 = \mathbf{P}'_0 + \mathbf{p}$$
$$E_0'= \frac{\mathbf{P}'^2_0}{2\,M} = E_0 - \mathbf{u}_0\cdot \mathbf{p}  + \frac{\mathbf{p}^2}{2\,M} \approx E_0 - \mathbf{u}_0\cdot \mathbf{p}$$
Народження ЕЗ буде не вигідним, якщо:
<!--  -->
\begin{equation}\label{eq:landcrit}
	E_0' + \varepsilon(p) > E_0 \Rightarrow u_0 < \varepsilon(p)/p
\end{equation}
<!--  -->
тобто функція $\varepsilon(p)/p$ повинна мати додатній мінімум для деякого $p$. Це є знаменитий крітерий надплинності Ландау.

М.М. Боголюбов показав яким чином в моделі слабко неідеального газу виникає спектр який задовільняє крітерію Ландау \eqref{eq:landcrit}. Наведемо цей простий метод наближенної діагоналізації гамільтоніану. Гамільтоніан слабко неідеального бозе газу у представленні операторів народження-знищення $\hat{b}_{\mathbf{p}}, \hat{b}^\dagger_\mathbf{p}$ станів вільних частинок з імпульсами $\hbar\mathbf{k}$ 
$$\hat{b}^\dagger_\mathbf{p} \ket{N_{\mathbf{p}}} = \sqrt{N_{\mathbf{p}}+1}\ket{N_{\mathbf{p}}+1}\\
\hat{b}_\mathbf{p} \ket{N_{\mathbf{p}}} = \sqrt{N_{\mathbf{p}}}\ket{N_{\mathbf{p}}-1}$$
має вигляд:
<!--  -->
\begin{equation}\label{hambog0}
	\hat{H} - \mu\,N = \sum_{\mathbf{p}}\left(\frac{\mathbf{p}^2}{2m}-\mu\right)\,\hat{b}^{\dagger}_{\mathbf{p}}\,\hat{b}_{\mathbf{p}}+\frac{1}{2\,V}\,\sum_{\mathbf{p}_1+\mathbf{p}_2-\mathbf{p}'_1-\mathbf{p}'_2 = 0}	\Phi(\mathbf{q})\,\hat{b}^{\dagger}_{\mathbf{p}'_1}\hat{b}^{\dagger}_{\mathbf{p}'_2}\hat{b}_{\mathbf{p}_1}\hat{b}_{\mathbf{p}_2}
\end{equation}
<!--  -->
де $\mathbf{q}$ це вектор зміни імпульсу при взаємодії:
<!--  -->
$$\mathbf{q} = \mathbf{p}'_1-\mathbf{p}_1 = \mathbf{p}_2-\mathbf{p}'_2$$
<!--  -->
Оскільки при низьких темпертурах більшість атомів знаходяться у конденсаті, то найбільший внесок у енергію дають доданки з операторами $b_0\,,b^{\dagger}_0$ які можна вважати комутуючими оскільки $N_0\gg 1$ і просто покласти:
$$\hat{b}_0\sim \hat{b}^{\dagger}_0 = \sqrt{N_0}$$
<!--  -->
Тоді у операторі взаємодії можна обмежитися тільки доданками порядка $N^2_0$ і $N_0$:
<!--  -->
\begin{multline}\label{hambog1}
	\hat{H} - \mu\,N = \frac{N_0^2}{2\,V}\,\Phi(0) - \mu\,N_0 +\\
	\sum_{\mathbf{p}}\left(\frac{\mathbf{p}^2}{2m} + \frac{N_0}{V}\,\Phi(\mathbf{p})+\frac{N_0}{V}\,\Phi(0)-\mu\right)\,\hat{n}_{\mathbf{p}}+\\
	\frac{N_0}{2\,V}\,\sum_{\mathbf{p}}	\Phi(\mathbf{p})\,\left(\hat{b}^{\dagger}_{\mathbf{p}}\hat{b}^{\dagger}_{-\mathbf{p}}+\hat{b}_{-\mathbf{p}}\hat{b}_{\mathbf{p}}\right)
\end{multline}
<!--  -->
хім. потенціал визначаємо з умови мінімума енергії основного стану:
<!--  -->
$$E_0 = \frac{N_0^2}{2\,V}\,\Phi(0) - \mu\,N_0\Rightarrow \frac{\partial \, E_0}{\partial\,N_0} = 0 \Rightarrow \mu = n_0\,\Phi(0)$$
<!--  -->
\begin{equation}\label{hambog2}
	\hat{H} - \mu\,N = -\frac{N_0^2}{2\,V}\,\Phi(0) +
	\sum_{\mathbf{p}}\left(\frac{\mathbf{p}^2}{2m} + \frac{N_0}{V}\,\Phi(\mathbf{p})\right)\,\hat{n}_{\mathbf{p}}+
	\\
	\frac{N_0}{2\,V}\,\sum_{\mathbf{p}}	\Phi(\mathbf{p})\,\left(\hat{b}^{\dagger}_{\mathbf{p}}\hat{b}^{\dagger}_{-\mathbf{p}}+\hat{b}_{-\mathbf{p}}\hat{b}_{\mathbf{p}}\right)
\end{equation}
<!--  -->
Отриманий гамільтоніан має квадратичну структуру за операторами $\hat{b},\hat{b}^\dagger$, Тому його можна діагоналізувати:
<!--  -->
\begin{equation}\label{hambog_diag}
	\hat{H} - \mu\,N = -E_0 +
	\sum_{\mathbf{p}}E(p)\,\hat{c}^{\dagger}_{\mathbf{p}}\,
	\hat{c}_{\mathbf{p}}
\end{equation}
<!--  -->
вводячи оператори:
<!--  -->
\begin{equation}
	\hat{c}_{\mathbf{p}} = u_{\mathbf{p}}\,\hat{b}_{\mathbf{p}} +v_{\mathbf{p}}\,\hat{b}^{\dagger}_{-\mathbf{p}}\,,\quad 
	\hat{c}^{\dagger}_{\mathbf{p}} = u_{\mathbf{p}}\,\hat{b}^{\dagger}_{\mathbf{p}} +v_{\mathbf{p}}\,\hat{b}_{-\mathbf{p}}
\end{equation}
<!--  -->
і накладаючи коммутаційні співвідношення:
<!--  -->
$$ [\hat{c}_{\mathbf{p}},\hat{c}^{\dagger}_{\mathbf{p}}] = 1 \Rightarrow u^2_{\mathbf{p}} - v^2_{\mathbf{p}} =1 $$
<!--  -->
Роблячи необхідні підстановки і прирівнюючи відповідні доданки, отримаємо:
<!--  -->
\begin{gather}
	E_{p}\,u_{\mathbf{p}}\,v_{\mathbf{p}} =\frac{n_0}{2}\,\Phi(\mathbf{p})\,\\
	E_{p}\,\left(u^2_{\mathbf{p}}+
	v^2_{\mathbf{p}}\right) = \frac{\mathbf{p}^2}{2m} + n_0\,\Phi(\mathbf{p})
\end{gather}
<!--  -->
Звідки маємо:
$$
u^2_{\mathbf{p}} =\frac{1}{2\,E_p}\left(
\frac{\mathbf{p}^2}{2m} + n_0\,\Phi(\mathbf{p})+E_p
\right)\,,\quad 
v^2_{\mathbf{p}} =\frac{1}{2\,E_p}\left(
\frac{\mathbf{p}^2}{2m} + n_0\,\Phi(\mathbf{p}) - E_p
\right)$$
$$
E_p = \sqrt{\left(\frac{\mathbf{p}^2}{2m}\right)^2 + \frac{\mathbf{p}^2}{m}\,n_0\,\Phi(\mathbf{p})}
$$
<!--  -->
Неважко бачити, що отриманий спектр задовільняє крітерію Ландау якщо $\Phi(0)>0$.
