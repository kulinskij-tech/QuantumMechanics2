##  Квантовий (цілий) ефект Холла. (Integer Quantum Hall effect)
<!--  -->
Двовимірний електронний газ при дуже низькій температурі (T < 4 K), підданий сильному перпендикулярному магнітному полю $B \sim 1\,T$, демонструє квантовану електропровідність. Це принципово відрізняється від передбачень класичної фізики. Цілочисельний Квантовий Ефект Холла (IQHE) це явище, яке відбувається в двовимірному електронному газі під сильним магнітним полем. Він характеризується квантуванням опору Холла до цілих значень, що є фундаментальною константою природи.

Почнемо з короткого нагадування про класичний ефект Холла.

### Класичний ефект Холла
<!--  -->
Розглянемо 2D (дуже тонкий) провідник зі струмом $\mathbf{j}$, підданий (слабкому) зовнішньому магнітному полю B, яке перпендикулярно до площини. Значення поля відповідає критерію
$$
\omega_B\,\tau \ll 1\,, \\
\omega_B = \frac{q\,B}{m\,c} - \text{cyclotron frequency}
$$

де $\tau$ це середній час вільного пробігу носіїв струму. Залежно від типу носіїв електричного струму різниця потенціалів виникає на протилежних сторонах провідника в напрямку, перпендикулярному до векторів $\mathbf{B}$ і $\mathbf{j}$. Дійсно, якщо струм обумовлений «+» -носіями, то $\mathbf{j}$ || $\mathbf{v}$ і сила Лоренца:

$$\mathbf{F}_{L} = \frac{q}{c}\,\mathbf{v}\times \mathbf{B} = \frac{1}{n\,c}\,\mathbf{j}\times \mathbf{B}$$

буде відхиляти носії, поки електричне поле, створене перерозподілом заряду, не врівноважить цю силу. В результате виникне відповідне електричне поле: 

$$\mathbf{F} = q\, \mathbf{E} = - \frac{1}{n\,c}\,\mathbf{j}\times \mathbf{B} \Rightarrow \mathbf{E} = - \frac{1}{q\,n\,c}\,\mathbf{j}\times \mathbf{B}$$

У випадку негативно заряджених носіїв полярність буде протилежною. Таким чином, вимірюючи полярність, можна визначити заряд носіїв. У напівпровідниках цей ефект широко використовується для перевірки $p-$ або $n-$типу провідності. 

В рамках простої моделі провідності Друде:

$$\frac{d\mathbf{v}}{dt} = \frac{q}{m}\,\mathbf{E} + \frac{q}{m\,c}\,\mathbf{v}\times \mathbf{B} - \mathbf{v}/\tau$$

В стаціонарному стані $d\mathbf{v}/dt = 0$ і

$$\frac{q}{m}\,\mathbf{E} + \frac{q}{m\,c}\,\mathbf{v}\times \mathbf{B} = \mathbf{v}/\tau \Rightarrow 
\mathbf{j} = q\,n\,\mathbf{v} = \frac{q^2 n\,\tau}{m}\,\mathbf{E} + \frac{\tau q}{m\,c}\,\mathbf{j}\times \mathbf{B}
$$
що можна записати у матричному вигляді х використанням тензора питомого опору:
$$
\mathbf{E} = \rho\, \mathbf{j} \Rightarrow  \mathbf{j} = \sigma\, \mathbf{E}\,,\\
\rho =\sigma^{-1} = \begin{pmatrix}
\rho_{xx}&\rho_{xy}\\
-\rho_{xy}&\rho_{yy}
\end{pmatrix}\,,\,\,\rho_{xx} = \rho_{yy} = \frac{m}{q^2 n\,\tau}\,, \,\,\,\rho_{xy} = -\frac{B}{n\,q\,c}
$$
Отже, суть ефекту Холла полягає в появі поперечної (до струму) різниці потенціалів так, що поперечний питомий опір (провідність) не залежить від домішок ($\tau$- залежність відсутня), а залежить тільки від густини носіїв $n$ і магнітного поля $B$:
$$
j = \sigma_{H}\,E\,,\,\,\,\sigma_{H} = \frac{1}{R_{H}\,B}
$$
$R_{H} = \dfrac{1}{n\,q\,c}$ називається константою Холла. Вимірювання напруги Холла дає оцінку густини носіїв. На ефекті Холла працюють датчики магнітного поля та керуючи елементи електронних схем у електродвигунах високих обертів де важлива синхронізація струму обмоток із обертанням магнітного ротору.

## Цілочисельний квантовий ефект Холла (IQHE)
Можна створити (наближено) двовимірний електронний газ із напівпровідникової гетероструктури, наприклад структури GaAs, затиснутої між двома напівпровідниками AlAs. Електрони знаходяться у зоні провідності GaAs, яка має меншу енергію, ніж зона провідності AlAs. Тому вони захоплені шаром GaAs. Цей шар GaAs можна зробити досить тонким, електрони фактично обмежені двовимірною площиною. Потім прикладається постійне перпендикулярне магнітне поле. При сильних магнітних полях $B \sim 1\,T$ і низьких температурах T < 4 K виникає цілочисельний квантовий ефект Холла. Але спочатку згадаємо квантування руху заряду в магнітному полі (рівні Ландау).

### Рівні Ландау
Наявність магнітного поля призводить до квантування  енергії руху заряду у площині перпендикулярній до вектора $\mathbf{B}$. Гамільтоніан для зарядженої частинки в зовнішньому магнітному полі є
$$
\hat{H} = \frac{\left(\hat{\mathbf{p}} - \frac{q}{c}\,\mathbf{A}\right)^2}{2\,m} - \mathbf{m}_s\cdot \mathbf{B}
$$
Магнітне поле вважатимемо спрямованим уздовж осі $z$: $\mathbf{B}= (0,0,B)$.
У такому випадку можна покласти $ \mathbf {A} = (0, B \, x, 0) $ (вибравши певну калібрування!). Рівняння Шредінгера має вигляд:

$$\left(-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial\,x^2}+
\frac{1}{2m}\left(-i\,\hbar\frac{\partial}{\partial\,y}-\frac{q}{c}\,B\,x \right)^2 -\frac{\hbar^2}{2m}\frac{\partial^2}{\partial\,z^2}\right)\psi=E\,\psi$$

Очевидно, можливий поділ змінних:

$$\psi = X(x)\,e^{i\,k_y\,y+i\,k_z\,z}$$

та підстановка дає

$$\left(-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial\,x^2}+
\frac{1}{2m}\left(p_y-\frac{q}{c}\,B\,x \right)^2 \right) X=\varepsilon\,X$$

де $p_y = \hbar\,k_y\,,\,\,\varepsilon = E-\frac{\hbar^2 k^2_{z}}{2\,m}$. Це означає, що енергія $E = E_{\parallel}+\varepsilon^{(\perp)}$ складається із енергії повздожнього (вільного) руху та енергії поперечного руху. Введемо

$$x_B = \frac{p_y\,c}{q\,B}\,,\quad \omega_{B} = \frac{q}{m\,c}\,B $$

так що отримаємо рівняння гармонічного осцилятора (!):

$$\left(-\frac{\hbar^2}{2m}\frac{\partial^2}{\partial\,\tilde{x}^2}+
\frac{m\,\omega^2_{B}}{2}\tilde{x}^2 \right) X=\varepsilon\,X$$

де $\tilde{x} = x-x_B$. Відповідний спектр для поперечних ступенів вільності:

$$\varepsilon^{(\perp)}_{n} = \hbar\,\omega_{B}\,\left(n+\frac{1}{2}\right)$$

Це т.з. рівні Ландау які можна розглядати як квантування класичних колових орбіт руху у магнітному полі (фінітний рух). Рух вздовж напрямка магнітного поля не квантується оскільки він є вільним. Для подальшого важливим є нескінченність кратності виродження рівней Ландау, оскільки всі функції вигляду $X_{n}(\tilde{x})\,e^{i\,k_y\,y}$ відповідають одному і тому ж значенню $\varepsilon_{n}$.

<center> Рівні Ландау є дискретними, але нескінченно виродженими: смуга станів вільної частинки ``втиснута`` на кожен рівень Ландау </center>

Щоб не мати справу з нескінченним виродженням завдяки необмеженості області, розрахуємо ступінь такого виродження для електронів у скінченній області. Припустімо, що у нас є площина розміром $L_x\times L_y$, тоді кількість станів вільної частинки що перейдуть до рівня Ландау дорівнює
$$\Delta\,N_{B} = \frac{DoS}{2}\,\hbar\,\omega_B$$
де $DoS$ - енергетична густина станів, а множник 1/2 враховує те що завдяки магнітному полю розщеплюються стани з протилежними проекціями спінів. Для вільної частинки $E_p = p^2/2m$ у 2D  геометрії: 
$$DoS = 2\,\frac{L_xL_y}{(2\pi\hbar)^2}\frac{2\pi\,p\,dp}{dE_{p}} = \frac{m}{\pi\,\hbar^2}\,L_x\,L_y$$
тому
$$\Delta\,N_{B} = L_x\,L_y\,\frac{e\,B}{h\,c}$$
цей результат має доволі прозорий фізичний зміст. Дійсно, оскільки  $BL_xL_y$ це є повний потік а $\Phi_0=hc/e$ є [квантом магнітного потоку](https://en.wikipedia.org/wiki/Magnetic_flux_quantum) то кратність виродження дорівнює числу таких квантів:
$$\Delta\,N_{B} = \Phi/\Phi_0$$
Зрозуміло, що якщо  $T\ll \hbar\,\omega_{B}$ рівні Ландау з $E \le E_F$ повністю заповнені і вільні якщо $E>E_F$.  У сильному магнітному полі всі електрони розподіляються по дискретних рівнях Ландау, відстань між якими $\sim B$. Тобто зайнятість змінюється стрибкоподібним чином, коли рівень Фермі перетинається зі станом Ландау. Кількість зайнятих рівнів Ландау можна виразити введенням так званого коефіцієнта заповнення $\nu$ як кількість заповнених рівнів Ландау:
$$\nu = \frac{N_{E<E_F}}{\Delta\,N_B}$$
При значеннях 
$$B$$
[коли рівень Фермі потрапляє між рівнями Ландау питомий опір (провідність) має плато](https://drive.google.com/file/d/11Ysjp8g_25vdC38PmfW8uvcxUhXd5YDP/view?usp=drive_link). Чим менше рівнів Ландау лежить нижче $E_F$ тим більший опір (менша провідність). Таким чином, для 2D електронного газу $n_{2D} = \nu \dfrac{eB}{h\,c}$ і відповідно:
$$\rho_{xy}= \frac{h}{\nu\,e^2} \leftrightarrow \sigma_{H} = \nu\,\frac{e^2}{h}$$
Оскільки положення плато визначаються з високою точністю ($10^{-9}$). [Величина опору $R_K = h/e^{2} = 25812.80745 \Omega$ тепер вважається фіксованою, дозволяє визначити сталу Планка і відповідно ввести новий квантовий(!) стандарт кілограму](https://en.wikipedia.org/wiki/International_System_of_Units).

Квант провідності $\frac{e^2}{h}$ що виникає у IQHE є половинним значенням 
[кванту провідності](https://en.wikipedia.org/wiki/Conductance_quantum) балістичного режиму провідності квазі одновимірного провідника. Це пояснюється як результат провідності за рахунок станів на краю зразка, оскільки завдяки сильному магнітному полю обєм виключений з провідності. В результаті утворються два одновимірних канала.

### Застосування IQHE
- Квантова метрологія: Встановлення стандарту опору, визначення сталої Планка і як наслідок еталону кілограму
- Quantum computing: Топологічні властивості IQHE були запропоновані як основа для створення кубітів у квантових комп'ютерах.
- Material science: для вивчення електронних властивостей матеріалів, таких як графен та інші двовимірні речовини
- Condensed matter physics: для дослідження поведінки електронів у системах конденсованого середовища, таких як напівпровідники та надпровідники.
