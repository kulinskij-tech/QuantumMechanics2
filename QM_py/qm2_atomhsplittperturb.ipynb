## Тонке та надтонке розщеплення рівней атома водню в рамках теорії збурень.
<!--  -->

$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$

Виродження рівней атома водню по орбітальному числу $l$ пов'язано зі специфікой [кулонівського потенціалу](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_atomh_theor.ipynb). Зовнішне електричне поле знімає виродження рівней атома водню по орбітальному числу $l$ (ефект Штарка) явним порушенням ізотропности зовнішнім полем. У відсутності зовнішніх полів це виродження знимається завдяки поправкам до електростатичного потенціалу взаємодії електрон-ядро. Такі поправки можна розділити на два класа: спін-(не)залежні взаємодії. Розщеплення завдяки взаємодії магнітного момента електрона (спіну) з орбітальним моментом називається тонкою (fine) структурою. Розщеплення завдяки взаємодії магнітного момента електрона (спіну) магнітним моментом ядра -  надтонкою (hyperfine) структурою.  

Почнемо з розмірного аналізу. Релятивістські поправки виникають завдяки скінченній швидкості електромагнітної взаємодії. Параметром розкладу для рівней енергии є відношення енергії нерелятивістського наближення і енергії спокою:
<!--  -->
\begin{equation}
	\epsilon = \frac{m_{e}\,e^4/\hbar^2}{m_{e}\,c^2}=\frac{e^4}{\hbar^2\,c^2}
\end{equation}
<!--  -->
Як видно, вона є величиною порядка $v^2/c^2$. В цьому наближенні ми і розрахуємо поправку до енергії оскільки
<!--  -->
$$
\alpha = \frac{e^2}{\hbar\,c}\sim \frac{\hbar}{m_{e}\,c\,r_0}
\sim \frac{p}{m_{e}\,c}=v/c$$
<!--  -->
Величина
$$\alpha=\frac{e^2}{\hbar\,c}\approx \frac{1}{137}$$
називается сталою тонкої (fine) структури.

Фактично потрібно врахувати всі внески порядка $\left(\mathbf{p}/m\,c\right)^2\sim \alpha^2$.
Перший такий внесок до нерелятивістського гамільтоніану виникає з розкладу енергії:
<!--  -->
$$E = \sqrt{(m_{e}\,c^2)^2+\mathbf{p}^2\,c^2}-m_{e}\,c^2-\frac{e^2}{r}=
\frac{\mathbf{p}^2}{2\,m_{e}}-\frac{e^2}{r}-
\frac{\mathbf{p}^4}{8\,m_{e}^3c^2}+\ldots
$$
<!--  -->
Внесок від відповідного ''кінетичного'' оператору має порядок:
<!--  -->
$$\frac{\hat{\mathbf{p}}^4}{8\,m_{e}^3c^2}\sim \alpha^2\,E_{0}$$
<!--  -->
Інші поправки такого порядка пов'язані і з наявністю у електрона спіну, який  взаємодіє з магнітним полем, що виникає (у його системі відліку ) завдяки руху електрона у електростатичному полі ядра. Тому  виникає відповідна магнітна взаємодія навіть у класичному підході:
<!--  -->
\begin{equation}\label{ham_spm}
	H_{sp} =-\mathbf{M}\cdot\mathcal{H}'
\end{equation}
<!--  -->
де
<!--  -->
$$\mathbf{M} = -\frac{e}{mc}\,\mathbf{S}\,,\quad
\mathcal{H}' = -\frac{1}{c}\,\mathbf{v}\times \mathbf{E}$$
<!--  -->
і ми скористались формулами перетворення ЕМП: $\mathbf{E}'_{\parallel}=\mathbf{E}_{\parallel}\,,
\,\,\,\mathcal{H}'_{\parallel}=\mathcal{H}_{\parallel}$
<!--  -->
$$\mathbf{E}'_{\perp} =\gamma\,\left( \mathbf{E}_{\perp}+\frac{1}{c}\,\mathbf{v}\times\mathcal{H}\right)\,,\quad
\mathcal{H}'_{\perp} =\gamma\,\left( \mathcal{H}_{\perp}-\frac{1}{c}\,\mathbf{v}\times\mathbf{E}\right)$$
<!--  -->
Додатково, однак слід врахувати, що електрон рухається не вздовж прямої, а по круговій орбіті і має місце додаткова прецесія Томаса. Це призводить до зменшення швидкості прецесії у 2 рази:
<!--  -->
\begin{equation}\label{ham_sp}
H_{sp} =\frac{e}{2\,m^2c^2}\,\mathbf{S}\cdot\left(\mathbf{E}\times \mathbf{p}\right)
\end{equation}
<!--  -->
Це т.з. спін-орбітальна взаємодія, оскільки для центрального поля:
<!--  -->
\begin{equation}\label{eq:spinorb}
\mathbf{S}\cdot\left(\mathbf{E}\times \mathbf{p}\right) = -\frac{1}{r}\frac{d\,\varphi}{d\,r}\mathbf{S}\cdot\left(\mathbf{r}\times \mathbf{p}\right) = -\frac{1}{r}\frac{d\,\varphi}{d\,r}\,
\mathbf{S}\cdot\mathbf{L}
\end{equation}
<!--  -->
Цей внесок має порядок величини:
<!--  -->
$$H_{sp}\sim \frac{e^2\,\hbar^2}{m^2c^2\,r^3_{0}} =
\frac{e^2/r_0}{m\,c^2}\,\frac{\hbar^2}{m\,r^2_{0}}\sim \alpha^2\,E_{0}
$$
<!--  -->
Ще один внесок такого же порядку даеться виразом:
<!--  -->
\begin{equation}\label{ham_sp_center}
H_{sp0} =\frac{e\,\hbar^2}{8\,m^2c^2}\,{\rm div}\,\mathbf{E}
\sim \frac{e^2\,\hbar^2}{m^2c^2\,r^3_{0}}
\end{equation}
<!--  -->
що враховує характер розподілу заряду ядра при спін-орбитальній взаємодії для станів з $l=0$ (див. \cite{book_vakarchukQM4}).

Всі ці оператори вже є діагональними у базисі незбурених станів $\ket{n,l,m,m_s}$, тому поправки є просто середніми від цих операторів:
<!--  -->
\begin{equation}
	E^{(1)}_{p} = -\frac{1}{2m\,c^2}\, \langle n,l,m\vert \left(\hat{H}^{(0)} +
	 \frac{e^2}{r}\right)^2\vert n,l,m\rangle = -
	 \frac{E^{(0)^2}_{n} - 4\,E^{(0)^2}_{n}
	 +\langle\frac{e^4}{r^2}\rangle}{2m\,c^2} =
 -\frac{\langle\frac{e^4}{r^2}\rangle - 3\,E^{(0)^2}_{n}}{2m\,c^2}
\end{equation}
<!--  -->
Далі скористаємось тим, що $n = n_{r}+l+1$ і:
<!--  -->
$$\left\langle\frac{\hbar^2\,(2l+1)}{2\,m\,r^2}\right\rangle = \frac{\partial}{\partial\,l}\,E^{(0)}_{n} = -\frac{2}{n}\,E^{(0)}_{n}\,\, \Rightarrow \,\,
\left\langle\frac{e^4}{r^2}\right\rangle = 4\,
\frac{n}{(l+1/2)}\,E^{(0)^2}_{n}
$$
<!--  -->
так що:
<!--  -->
$$E^{(1)}_{p} = - \frac{2}{m\,c^2}\,E^{(0)^2}_{n}
\left(\frac{n}{l+1/2} - \frac{3}{4}\right) =
\frac{\alpha^{2}}{n^2}\,E^{(0)}_{n}
\left(\frac{n}{l+1/2} - \frac{3}{4}\right)
$$
<!--  -->
Т.ч. вже видно, що відбувається розщеплення рівней з різними $l$. Розрахунок поправки від спін-орбитальної взаємодії заснований на співвідношені:
<!--  -->
\begin{equation}
	\mathbf{S}\cdot \mathbf{L} =
	\frac{1}{2}\left(\mathbf{J}^2 - \mathbf{S}^2 - \mathbf{L}^2\right)
	\label{eq_jls}
\end{equation}
<!--  -->
\begin{equation}
\left\langle\hat{H}_{sp}\right\rangle =
\frac{e^2\,\hbar^2}{4\,m^2c^2}
\left\langle\frac{1}{r^3}\right\rangle\,
\left(j(j+1)-l(l+1)-3/4\right)
\label{eq_esporb}
\end{equation}
<!--  -->
Середнє $1/r^3$ можна порахувати за допомогою
такого прийому:
<!--  -->
$$H_{r} = \frac{p^2_{r}}{2\,m}\,,\quad p_{r} = -\frac{i\,\hbar}{r}\,\frac{\partial }{\partial\,r}\,r$$
<!--  -->
оскільки в стаціонарних станах середні не залежать від часу, то
<!--  -->
$$0=\left\langle\frac{d p_{r}}{d\,t}\right\rangle =\frac{1}{i\,\hbar} \left\langle[p_{r},\hat{H}]\right\rangle =
\frac{1}{i\,\hbar}
\left\langle\left[p_{r},\frac{\hat{\mathbf{L}^2}}{2\,m\,r^2} - \frac{e^2}{r}\right]
\right\rangle\,\Rightarrow\,
\left\langle\frac{\hbar^2\,l(l+1)}{m\,r^3}\right\rangle=
\left\langle\frac{e^2}{r^2}\right\rangle
$$
<!--  -->
$$\left\langle\frac{1}{r^3}\right\rangle=
\frac{m}{\hbar^2}\,
\left\langle\frac{e^2}{r^2}\right\rangle\,\frac{1}{l(l+1)} = 4\,\frac{m}{\hbar^2\,e^2}\,
\frac{n}{l\,(l+1/2)(l+1)}\,E^{(0)^2}_{n}$$
<!--  -->
$$\left\langle\hat{H}_{sp}\right\rangle =
\frac{1}{m\,c^2}\,
\frac{n\left(j(j+1)-l(l+1)-3/4\right)}{l\,(l+1/2)(l+1)}\,E^{(0)^2}_{n}
=-\frac{\alpha^2}{2}\,\frac{\left(j(j+1)-l(l+1)-3/4\right)}{n\,l\,(l+1/2)(l+1)}\,E^{(0)}_{n}$$
<!--  -->
Оскільки $j=l\pm 1/2$, то
<!--  -->
$$
\begin{cases}
  \left\langle\hat{H}_{sp}\right\rangle =
-\frac{\alpha^2}{2}\,\frac{1}{n\,(l+1/2)(l+1)}\,E^{(0)}_{n}>0, & \mbox{if } \quad j=l+1/2 \,,\,\,l>0\\
  \left\langle\hat{H}_{sp}\right\rangle =\hphantom{-}
\frac{\alpha^2}{2}\,\frac{1}{n\,l\,(l+1/2)}\,E^{(0)}_{n}<0, & \mbox{if}\quad  j=l-1/2\,,\,\,l>0
\end{cases}
$$
<!--  -->
Остаточно додаючі всі поправки порядка $\alpha^2$ ми отримуємо вираз
$$
  E^{(1)}_{n} =\frac{\alpha^2}{n^2}\,E^{(0)}_{n}\,\left(\frac{n}{(j+1/2)}-3/4\right)
$$
<!--  -->
Тобто в першому порядку за $E_{0}/mc^2\sim \alpha^2$ розщеплення відбувається для станів з різними значеннями повного моменту $j$, наприклад $2s_{1/2}$ и $2p_{3/2}$. Рівни $2s_{1/2}$ и $2p_{1/2}$ в цьому наближені залишаються вирожденими. Вони розщеплюються завдяки взаємодії з флуктуаціями ЕМП - т.з радіаційні доданки ([Лембівський зсув](#https://en.wikipedia.org/wiki/Lamb_shift)).

Поправка \eqref{ham_sp_center} зсуває (збільшує) енергію тільки $s$-станів ($l=0$):
<!--  -->
\begin{equation}\label{ham_sp_centerenrg}
E^{(1)}_{sp0} =\frac{e^2\,\hbar^2}{8\,m^2c^2}\,\left\langle 4\pi\,\delta(r)\right\rangle =
\frac{\pi\,e^2\,\hbar^2}{2\,m^2c^2}\vert
\psi_{n\,l,m}(0)
\vert^2 =
-\alpha^2\,\frac{1}{n}\,E^{(0)}_{n}\,\delta_{l,0}>0
\end{equation}
