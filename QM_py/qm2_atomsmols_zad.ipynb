__Розв'язок:__ Гамильтониан задачі:

$$\hat{H} = \hat{H}^{(0)}_{1}+\hat{H}^{(0)}_{2}+\frac{e^2}{r_{12}}$$

Для розрахунків використаємо хвильову функцію воднеподібної задачі:

$$\label{wavprob}
\Psi = \psi_0(r_1;\alpha)\,\psi_0(r_2;\alpha) = \frac{\alpha^3}{\pi}\,e^{-\alpha(r_1+r_2)}$$

з параметром $\alpha$ який визначається з варіаційної процедури мінімізації енергії для \eqref{wavprob}.
Знаходимо:

$$\Delta \psi_0(r;\alpha) = \frac{1}{r}(r\psi_0(r;\alpha))'' = -\frac{2\,\alpha}{r}\,\psi_0(r;\alpha) + \alpha^2\,\psi_0(r;\alpha)
$$

Відповідно середнє значення енергії дорівнює (покладемо $\hbar = 1\,, e=1,\, m=1$):

$$\langle H \rangle = \alpha^2 - 4\,\alpha+ \frac{\alpha^6}{\pi^2}\, \iint \frac{e^{-2\alpha(r_1+r_2)}}{r_{12}}\,dV_1\,dV_2  $$

Обчислення останнього інтеграла ґрунтується на тому, що при усередненні за напрямками кутова залежність не дає внеску, тому використаємо відомий розклад за поліномами Лежандра:

$$
\frac{1}{r_{12}} = \begin{cases}
\frac{1}{r_1}\,\sum\limits_{n=0}^{\infty} \left(\frac{r_2}{r_1} \right)^n\,P_{n}(\cos\theta)\,, & \mbox{if } r_2<r_1<\infty \\
\frac{1}{r_2}\,\sum\limits_{n=0}^{\infty} \left(\frac{r_1}{r_2} \right)^n\,P_{n}(\cos\theta), & \mbox{otherwise }.
                     \end{cases}
$$

з якого видно, що внесок в інтеграл дадуть тільки доданки з $n=0$.
Тому

$$\iint \frac{\psi_{0}(r_1)^2\psi_{0}(r_2)^2}{r_{12}} dV_1dV_2 = \frac{(4\pi)^2}{\pi^2}\int\limits_{0}^{\infty}dr_2 r_2^2e^{-2\alpha r_2}\left(\frac{1}{r_2}\int\limits_{0}^{r_2}
e^{-2\alpha r_1}\,r_1^2\,dr_1 + \int\limits_{r_2}^{\infty}e^{-2\alpha r_1}\, r_1 \, dr_1\right)$$

Просте обчислення приводить до результату:

$$
\iint \frac{\psi_{0}(r_1)^2\psi_{0}(r_2)^2}{r_{12}} dV_1dV_2 = \frac{5}{8}\,\alpha
$$

Минимизация среднего значення

$$E(\alpha) =-\frac{27}{8}\,\alpha\,+\alpha^2 $$

за параметром $\alpha$ дає:

$$\alpha_{\text{min}} = \frac{27}{16}$$

і

$$E_0 \approx 2.85 $$

що у розмірних величинах становить $77.5$~еВ. Експериментальне значення
дорівнює$79.1$ еВ.

Реальною задачею, до якої можуть бути застосовані міркування, наведені вище, є атом гелію. Гамільтоніан системи має вигляд:

Основний стан атома гелію є синглетним, триплетний стан заборонений завдяки принципу Паулі. Наступний збуджений стан повинен бути або синглетним або триплетним. Виявляється, що найменшу енергію при врахуванні кулонівської взаємодії між електронами має триплетний стан. Це можна обґрунтувати наступним чином. У триплетному стані координатна хвильова функція має вузли (див. (\ref{20eq10})), тому ймовірність конфігурацій, у яких електрони опиняються на малих відстанях мала, як наслідок, мало і середнє значення кулонівської енергії відштовхування у порівнянні із її значенням у синглетному стані. Тому триплетні стани лежать нижче синглетних (за виключенням основного, який синглетний). Енергія станів у першому порядку за кулонівською взаємодією електронів дорівнює:
\begin{eqnarray}\label{20eq12}
 E_{S = 0} &=&E_{1} + E_{2} + e^{\,2}\int 
 \frac{\vert \varphi_{s}(\mathbf{r}_{\,1} ,\mathbf{r}_{\,2} )\vert^{\,2}}{\vert \mathbf{r}_{2} - \mathbf{r}_{1} \vert} dV_{1} dV_{2} =
E_{1} + E_{2} + K + C \nonumber \\ &&\\
 E_{S = 1} &=& E_{1} + E_{2} + e^{\,2}\int \frac{\vert \varphi_{a} (\mathbf{r}_{\,1} , \mathbf{r}_{\,2} )\vert^{\,2}}{\vert \mathbf{r}_{2} - \mathbf{r}_{1} \vert} 
 dV_{1} dV_{2} = E_{1} + E_{2} + K - C \nonumber
\end{eqnarray}
де 
$$
 K = e^{\,2}\int \frac{\vert \varphi_{1} (\mathbf{r}_{\,1} )
 \vert^{\,2}\vert \varphi_{2} (\mathbf{r}_{\,2} )\vert^{\,2}}{\vert \mathbf{r}_{2} - \mathbf{r}_{1} \vert} dV_{1} dV_{2}\\
C = e^{\,2}\int \varphi_{2}^*  (\mathbf{r}_{\,1})\varphi_{1}^* (\mathbf{r}_{\,2}\frac{1}{\vert \mathbf{r}_{2} - \mathbf{r}_{1} \vert}
\varphi_{1} (\mathbf{r}_{\,1} )\varphi_{2} (\mathbf{r}_{\,2} ) dV_{1} dV_{2}
$$
Доданок, який обумовлений величиною $K$ має класичну інтерпретацію як енергію взаємодії електронних областей зарядів із відповідними густинами. Інтеграл $C$ навпаки, не має прямої електростатичної інтерпретації і називається обмінним інтегралом, оскільки враховує
можливість обміну станами між електронами. Цей вклад в енергію системи інтерпретують як вклад __обмінної взаємодії__ . Зрозуміло, що нічого не зміниться, якщо замість кулонівської ми врахуємо ще якусь взаємодію між електронами. Тоді з'являться відповідні обмінні вклади. Таким чином, __природа обмінної взаємодії визначається вихідною взаємодією тотожних частинок у гамільтоніані__.

Якщо нехтувати спіновими взаємодіями переходи між станами з різними значеннями спінового квантового числа неможливі (спін - інтеграл руху). Враховуючи слабкість таких взаємодій
(релятивістських за своєю природою), можна представити цю ситуацію як наявність двох квазістаціонарних станів гелію: __парагелій__ - синглетні стани, __ортогелій__ - триплетні стани. Вони різняться своїми властивостями: __парагелій є діамагнетиком__ (спін 0), __ортогелій - парамагнетик__ (спін 1) має власний магнітний момент.
