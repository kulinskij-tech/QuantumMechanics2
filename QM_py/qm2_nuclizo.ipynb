Хоча квантова теорія релятивістських полів значно складніша. Ускладнення повязані із еквівалентністю маса-енергія ($E = m\,c^2$) яке призводить до необхідності працювати із системою зі змінною кількістю частинок-квантів - квантовим полем. Ситуація якісно схожа на вивчення квантових явищ у конденсованому стані (надплинність, надпровідність тінш) і між теорією квантованих (релятивістських) полів і квантовою теорією (нерелятивістського) конденсованого стану існує дуже тісний зв'язок. Тут доречно застосування формалізму операторів народження та знищення квантів відповідних полів (приклад - квантування ЕМП). Зрозуміло, що загальні уявлення про суперпозицію станів, оператори фізичних величин, (квазі)стаціонарні стани та їх еволюцію тінш, що було введено і розвинуто у рамках нерелятивістської квантової теорії можуть бути застосовані і для аналізу процесів народження і знищення частинок при високих енергіях. При цьому, аналіз експериментальних даних та характерних часів протікання процесів чи розпаду одних частинок на інші (квазістаціонарні стани) дозволяє провести класифікацію взаємодій по їх енергетичному та просторовому/часовому масштабам. Наприклад, характерний радіус $r_0$ взаємодії може бути повязаний із характерною масою/енергією спокою $m_0$ медіатора взаємодії:

\begin{equation}\label{eq:mass0}
\Delta E\,\Delta t \sim \hbar \Rightarrow m_0\,c\, r_0\sim \hbar \Rightarrow m_0\sim \frac{\hbar}{r_0
\,c}
\end{equation}

Найбільш вивченою є електромагнітна взаємодія, в якій приймають участь заряджені частинки ($e^{\pm},p,\tilde{p}$) а також частинки з магнітним моментом (нейтрон). Медіатором цієї взаємодії є квант електромагнітного полю - фотон з нульовою масою спокою а отже з необмежени радіусом взаємодії $r_0\to \infty$. Електромагнітна взаємодія лежіть в основі переважної більшості явищ що ми спостерігаємо "неозброєним оком" і характеризується сталою тонкої структури $\alpha = \dfrac{e^2}{\hbar\,c} \approx 1/137 $. Наступна взаємодія, яку можна вважати умовно достатньо "наочною" - це слабка взаємодія. Вона спричиняє $\beta$-радіоактивність важких ядер з утворенням легких лептонів (наприклад електронів) та нейтріно. Радіоактивне випромінювання призводить до мутацій в клітинах живих організмів та за достатньої енергії руйнує їх ДНК-структуру. Тому ми дійсно спостерігаємо прояв цієї взаємодіїї так чи інакше на себе. Слабка взаємодія відповідає ща нестабільність вільного нейтрону:

$$n \to p+e^{-}+\tilde{\nu}_{e}$$

Звідси за \eqref{eq:mass0} можна оцінити масу медіатора (т.з. проміжний векторний бозон)

```python
import numpy as np
# import sympy.printing as printing
from scipy import constants
# nuclear size
print(constants.find('radius'))
radn=constants.physical_constants[constants.find('radius')[-1]]
print('M <',f"{constants.hbar*constants.c/(radn[-1]*1.6*10**(-19))/10**9:.0f}", 'GeV')
```

Сильна взаємодія відповідає за стабільність ядер атомів, це взаємодія між [адронами = баріони+мезони](https://en.wikipedia.org/wiki/Hadron). Картина цієї взаємодії залежить від масштабу. Пояснення внутрішньої структури самих адронів вже потребує введення нових частинок - кварків та глюонів у звязку із великою енергією на таких малих відстанях (вони не спостерігаються у вільному стані - т.з. конфайнмент). На відстанях же порядку розмірів ядра $10^{-15} m$ (відносно низька енергія) нуклони обмінюються квантами масою:

```python
print(constants.find('radius'))
print('M <',f"{constants.hbar*constants.c/(10**(-15)*1.6*10**(-19))/10**6:.0f}", 'MeV')
```

які ідентифікуються як $\pi$-мезони ($m_0 \sim 140 MeV$). Найбільш слабкою є гравітаційна взаємодія, але вона поки що не піддається процедурі квантування (гіпотетичний безмасовий квант - гравітон). 

Методи теорії симетрії які були розвинуті в нерелятивістській квантовій теорії також успішно застосовуються в теорії фундаментальних взаємодій, наводячи певний порядок в "зоопарку" різноманітних станів (резонансів) які утворюються при зіткненні високоенергетичних пучків електронів та протонів. Резонанси – це стани з чітко визначеним спіном і парністю, масою, зарядом тощо, і їх слід розглядати нарівні з іншими елементарними стабільним частинками такими як протони, електрони та фотони.
Всі взаємодії крім слабкої зберігають просторову парність ($P$-симетрія право-ліво), зарядову спряженість ($С$-симетрія частинка-античастинка) та обернення часу ($T$-симетрія $t\to -t$).

Також відомо, що ядра у яких число протонів у одному дорівнює числу нейтронів у іншому мають дуже близькі енергетичні рівні. Різниця пов'язана з додатковою електромагнітною взаємодією міжду протонами. Це наводить на думку, що (сильна) взаємодія протон-протон і нейтрон-нейтрон, за виключенням єлектромагнітного внеску, таж сама. З точки зору загальних положень квантової механіки можна припустити, що т.з. ядерних сил протон і нейтрон є різними станами однієї частинки - __нуклона__. Гамільтоніан сильної взаємодії не розрізняє ці стани, тобто має місце виродження. Наявність електричного заряду  протона призводить до зняття виродження за рахунок збурення в якості якого (на ядерних масштабах) виступає електромагнітна взаємодія:
$$\frac{e^2/a_n}{m_0\,c^2} \simeq 10^{-3} $$

Таким чином, стан нуклона крім звичайних координатної та спінової $(s = 1 / 2)$ ступенями вільності, може бути охарактеризований ще одним квантовим числом яке розрізняє
стани які відповідають протону $\vert\, p\,\rangle$ і  $\vert\, n\,\rangle$ нейтрону. За принципом суперпозиції ці стани утворюють базис двовимірного комплексного простору. Гамільтоніан сильної взаємодії інвариантний відносно унітарного перетворення базису:
<!--  -->
\begin{eqnarray}
 U\vert\, p\,\rangle &=& \alpha\, \vert\, p\,\rangle + \gamma \,
 \vert\, n \,\rangle \nonumber\\
 U\vert\, n\,\rangle &=& \beta\, \vert\, p\,\rangle + \delta \,
 \vert\, n\,\rangle
 \label{feq19}
\end{eqnarray}
<!-- % -->
$$U=\begin{pmatrix}
  \alpha & \gamma \\
   \beta & \delta
\end{pmatrix}
\,,\quad U^{\dagger} = U^{- 1}$$
<!--  -->
Математично ця ситуація тотожня з випадком спіну. Тому таку величину назвали 
__ізотопічним спіном - ізоспіном__.  Ізоспін відображає структуру внутрішнього простору нуклонів так само як спін відображає 2-компонентну структуру хвильової функції електрона. Вектор стану нуклона з ізоспіном $I$ має $2I+1$ компонент. Фізична відмінність полягає в тому що перетворенню \eqref{feq19} не відповідають перетворення обертання у реальному просторі, як це було у випадку із спіном. Ізоспін не повязаний з магнітним моментом і не взаємодіє з якимось реальним силовим полем як магнітний момент що породжується спіновим. Однак, для формальних співвідношень це не суттєво. Виходячи з цієї аналогії можна ввести оператор ізоспіна як генератора ізоспінових перетворень:
<!--  -->
\begin{equation}
\label{feq20} U(\varepsilon _{1} ,\varepsilon _{2} ,\varepsilon
_{3} ) = \exp \left( i\varepsilon _{1} I_{1} + i\varepsilon _{2}
I_{2} + i\varepsilon _{3} I_{3} \right)
\end{equation}
<!-- -->
Алгебра операторів $I_{i} $ співпадає з таковою для оператора
спіна:
<!--  -->
$$[I_{i},I_{j}] = i\,\varepsilon_{ijk}\,I_{k}\,,\,\,\,
\mathbf{I}^{2} = I(I + 1)
$$
<!--  -->
<!-- -->
Оскільки ядро це багаточастинкова система ферміонів, то до неї можна застосувати методи які були розвинені у випадку багатоелектронних атомів. Стани які відповідають одному і тому ж самому значенню квантового числа ізоспіна $I$ називаються ізоспіновим мультиплетом ($I=0$ - синглет, $I=1/2$ - дублет, $I=1$ - триплет і т.д.). Протон і нейтрон є ізоспіновим дублетом. За згодою нейтронний стан $\ket{n}$ відповідає $I_{3}= - 1/2$, а протонний $\ket{p} \to  I_{3}= + 1/2$. Для сильної взаємодії ізоспін є інтегралом руху, тобто гамільтоніан сильної взаємодії комутує із оператором $\mathbf{I}$. Гамльтоніан ЕМ взаємодії не комутує з оператором ізоспіна, що і призводить до "розщеплення" дублету $\ket{n},\ket{p}$ на 2 компоненти з різними масами $m_{n}\ne m_{p}$ (енергіями). 
Прикладом ізоспінового триплета з $I=1\,, I_{3}=\pm 1, 0$  можна вважати трійку піонів $\pi^{-},\pi^{0},\pi^{+}$ завдяки малої різниці їх мас (наслідок електромагнітної взаємодії):

```python
from IPython.display import display, Math
mpi0=134.977
mpi1= 139.57061
dm=round((mpi1/mpi0-1),4)
display(Math(r'\Delta m_\pi/m_\pi^0 ={0}'.format(str(dm))))
```

Відмітимо, однак, що на відміну від ситуації із протон-нейтрон де заряджений протон мав меншу масу, тут навпаки заряджені $\pi^{\pm}$ мають більшу масу ніж ії нейтральний ізо-брат $\pi^{0}$, який є більш нестабільним ($\tau_{\pi^0}\sim 10^{-16} s$, $\tau_{\pi^{\pm}}\sim 10^{-8} s$). Це пояснюється в рамках кваркової структури адронів.

Ізоспін складної системи обчислюється за звичайним [правилом складання моментів](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_intsymom_theor.ipynb#momsum). Так ізоспін двох нуклонів може приймати значення 0 і 1. Розглянемо з цієї т.з. зв'язаний стан двох нуклонів дейтрон - ядро дейтерія D$_{2}$. Вектор стану ядра $D_2$ представляється як 
$$\ket{D_2} = \ket{l,m}\otimes \ket{S}_{s}\otimes \ket{I}_{iso}$$
За правилом додавання ізоспіну стан з $I= 1$ повинен включати протон - протонні і нейтрон - нейтронні звязані  стани. Однак такі не спостерігаються, тому єдиним станом може бути протон - нейтронне з $I = 0$. Ізоспінова частина такого вектора стана є антисиметричною:
<!--  -->
$$\ket{I=0}_{iso} = \ket{\uparrow}_{iso}\otimes \ket{\downarrow}_{iso}-\ket{\downarrow}_{iso}\otimes\ket{\uparrow}_{iso}$$ 
<!--  -->
Основний стан, вочевидь відповідає відсутності орбітального моменту ядра $l = 0$, тому спінова частина повинна бути симетричною (протон та нейтрон - ферміони) тобто спін дейтрона $S = 1$ (спіновий триплет) і відповідно дейтрон в основному стані має (ядерний) магнітний момент.

Багато властивостей ядер можуть бути описані у рамках моделі оболоньок аналогичної до такої в теорії багатоелектронних атомів із врахуванням ізоспіну. Стан нуклона у ядрі може бути представлено як стан у сферично симетричному самоузгодженому полі (СУП) інших нуклонів. Таким чином вводяться одночастинкові стани в рамках СУП. Одночастинковий стан характеризується повним моментом  $j$ і просторовою парністю $(-1)^{l}$ по якій визначається орбітальне квантове число $l = j\pm 1 / 2$. Стани с $l,j$ позначаються вказуючи орбітальне квантове число, $l$ як $s,p,d$ і $j$ справа внизу. В такому стані може перебувати не більш ніж $2j + 1$ нейтронів і стількі ж протонів. Найбільш стабільною конфігурацією є спарення частинок з протилежними моментами. Так у парно-парних ядрах моменти компенсуються. В непарно-парних ядрах момент визначається станом незпареного протона чи нейтрона. Як і у випадку атомів конфігурації нуклонів можна розбити на групи в кожній з яких енергія зв'язку на один нуклон зростає, а потім падає аналогічно тому що ми бачили у випадку [багатоелектронних атомів](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics2/blob/main/QM_py/qm2_atoms.ipynb#energbnd). Числа нуклонів, які визначають такі групи називаются __магічними__. Вони залежать від деталей самоузгодженного потенціала, який визначає спектр одночастинкових станів. [В моделі ізотропного гармонічного потенціала СУП такі числа наприклад дорівнюють](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics1/blob/main/QM_py/qm1_intsymom_theor.ipynb#3dosc): 2, 8=2(1+3), 20=8+2(1+5), 40=20+2(3+7), $\ldots$.
[В дійсності, звичайно, є суттєвий внесок релятивістської спін-орбітальної взаємодії і магічні числа відповідають послідовності: $2, 8, 20, 28, 50, 82, 126$. Відомі також подвійно-магінчні ядра із магічним числом протоні та нейтронів, наприклад $\vphantom{He}^{4}He,\vphantom{O}^{16}O,\vphantom{Ca}^{40}Ca$](https://en.wikipedia.org/wiki/Magic_number_(physics)#).

Таким чином, ядра які складаються з одного і того ж числа нуклонів є тотожніми з т.з. ядерної взаємодії. Зокрема, це проявляється для т.з. "дзеркальних" ядер, де число протонів в одному дорівнює числу нейтронів у іншому, як наприклад в ядрах $\vphantom{He}^{3}He$ і $\vphantom{He}^{3}H$. Іншим прикладом можуть слугувати ізотопи $\vphantom{C}^{14}C$ і $\vphantom{C}^{14}O$ із складом $6p+8n$ і $8p+6n$ відповідно. Вони відповідають станам $I_3 =-1$ і $I_3 =+1$ належать ізоспіновому триплету $I=1$. А отже повинен існувати стан с $I=1, I_3 = 0$. Стабільне ядро $\vphantom{N}^{14}N$ в якому всі протони зпарени с нейтронами є синглетним станом з $I=0$ і має мінімальну енергію. Тому стан триплета $I=1, I_3 = 0$ є збудженим станом $\vphantom{N}^{14}N^*$ ядра азоту з симетричною ізоспіновою частиною . Енергії триплетних станів вище ніж енергія стабільного синглетного стану і мало відрізніються один від одного (порівняйте із ситуацією атома гелію). З т.з. сильної взаємодії між цими станами переходи неможливі. Однак триплет є нестабільним відносно слабкої і ЕМ взаємодій і ці ізотопи з часом розпадаються до стабільного ізотопу азоту: $\vphantom{C}^{14}C$ і $\vphantom{C}^{14}O$  в результаті слабкої взаємодії ($\beta^{\pm}$-розпад), а $\vphantom{N}^{14}N^*$ швидко розпадається внаслідок  електромагнітної взаємодії з випромінюванням $\gamma$-кванта.

Таким чином, відносно ізоспінової $SU(2)$-симетрії всі частинки які приймають участь у сильній взаємодії (адрони) утворюють ізоспінові мультиплети які складаються з $2I+1$ частинок. Як вже було сказано, $\pi$-мезони утворюють триплет $I=1$
$$
\ket{\pi^{+}} \leftrightarrow \ket{I=1,I_3 = +1}\,,\,\,\,\ket{\pi^{0}} \leftrightarrow \ket{I=1,I_3 = 0}\,,\,\,\,
\ket{\pi^{-}} \leftrightarrow \ket{I=1,I_3 = -1}
$$
Отже "стабільні" по відношенню до сильної взаємодії частинки, це ті, що розпадаються внаслідок слабкої або електромагнітної взаємодій.
