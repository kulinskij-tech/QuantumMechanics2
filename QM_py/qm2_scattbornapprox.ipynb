## Борнівське наближення у теорії розсіяння.

Точний розв'язок задачі розсіяння можливе лише у високосиметричних випадках потенціальних полів окремого виду (наприклад, потенціали нульового радіуса, кулонівський потенціал). Тому необхідно
використовувати наближені методи розрахунку амплітуди розсіяння. Вважаючи потенціал розсіяння малим збуренням, можна отримати наближені результати у рамках теорії збурень, яка була розвинута у
попередніх лекціях. Розглянемо випадок пружного стаціонарного розсіяння. Запишемо
рівняння Шредінгера:
<!--  -->
\begin{equation}
\label{scateq16} (E - H_{0} )\,\psi = U\,\psi
\end{equation}
<!--  -->
Це рівняння можна розглядати як неоднорідне, представив його
розв'язок як:
<!--  -->
\begin{equation}
\label{scateq17} \psi (\mathbf{r}) = \phi (\mathbf{r}) +
\int {d\,\mathbf{r}'G_{0} (\,\mathbf{r} -
\,\mathbf{r}';E)\,U(\,\mathbf{r}')} \,\psi (\,\mathbf{r}'\,)
\end{equation}
<!--  -->
де

\begin{equation}
\label{scateq18}
\begin{array}{l}
 G_{0} (\,\mathbf{r} - \,\mathbf{r}';E) =
 (E - H_{0} )^{- 1}_{\mathbf{r},\,\mathbf{r}'} \\
 (E - H_{0} )\,G_{0} (\,\mathbf{r} - \,\mathbf{r}';E) =
 \delta (\,\mathbf{r} -
\,\mathbf{r}') \\
 \end{array}
\end{equation}

__функція Гріна__ незбуреного рівняння Шредінгера (рівняння Гельмгольца), $\phi $ - загальне розв'язок однорідного рівняння. У випадку задачі розсіяння монохроматичного
пучка $\phi $ має вид плоскої хвилі у відповідності з асимптотичною умовою:

\begin{equation}\label{scateq19} 
\phi_{\,\mathbf{k}}(\mathbf{r}) =
\exp(i\mathbf{k} \cdot \mathbf{r}\,)\,\,\,,\,\,\,
E = \frac{\hbar^{\,2}\,\mathbf{k}^{\,2}}{2\,m}
\end{equation}

Рівняння (\ref{scateq17}) корисне для застосування теорії збурень. Оскільки:

$$
 \psi(\mathbf{r}) = \sum\limits_{n}\psi^{(n)}(\mathbf{r})\\
 \psi^{(0)}(\mathbf{r}) = \phi_{\,\mathbf{k}}(\mathbf{r})\,\,,\\
 \psi^{(n + 1)}(\mathbf{r}) = \int d\,\mathbf{r}'G_{0} (\,\mathbf{r} -
\,\mathbf{r}';E)\,U(\,\mathbf{r}'\,)\,\psi ^{(n)}(\mathbf{r}')
$$
__Борнівське наближення__ відповідає врахуванню тільки першої поправки. Функція Гріна має вигляд (довести самостійно)

\begin{multline} \label{scateq21}
 G_{0} (\,\mathbf{r} - \,\mathbf{r}';E) =
  \int \frac{d\,\mathbf{k}'}{(2\pi )^{3}}
  \frac{\exp \left( i\,\mathbf{k}' \cdot\,
(\,\mathbf{r} - \mathbf{r}')\right)}{E - E'} =\\
 \frac{m}{2\pi^{2}\hbar^{2}}
 \int\limits_{0}^{\infty}
\int\limits_{ - 1}^{1} k\,'^{2}d\,k\,'d( - \cos \,\,\theta
)
\frac{\exp \left( i\,k'\, \vert\, \mathbf{r} -
\,\mathbf{r}'\vert \cos \,\,\theta  \right)}{k^{2} -
k'^{2}}
=\\\frac{m}{2\pi^{2}\hbar^{2}i}\frac{1}{\vert\, \mathbf{r} -
\,\mathbf{r}'\vert}
\int\limits_{ - \infty}^{ + \infty}
d\,{k}'\frac{k'\exp \left( i\,k'\vert\, \mathbf{r} -
\,\mathbf{r}'\vert\right)}{k^{2} - k'^{2}}
\end{multline}

Інтеграл \eqref{scateq21} не визначений доки не заданий напрямок обходу полюсів на дійсної вісі. Цей напрямок слід вибирати таким чином, щоб на великих відстанях розв'язок
(\ref{scateq17}) задовільняв асимптотичній умові задачі розсіяння (розбіжна сферична хвиля). На рисунку показані різні варіанти обходу полюсів. 
<!--  -->
<img src="./figs/fig_greenfree.png" alt="Image" width="300"/>
Очевидно, що розбіжної сферичної хвилі відповідає випадок c), збіжної - випадок а). Функції Гріна, отримані таким чином, дорівнюють відповідно:

\begin{equation} 
\label{scateq22} G_{0}^{\pm}(\,\mathbf{r} - \mathbf{r}'\,;E) = -
\frac{m}{2\,\pi \,\hbar^{\,2}}
\frac{\exp \left(\pm i\,k\vert\,
\mathbf{r} - \mathbf{r}'\vert \right)}{\vert\, \mathbf{r} - \mathbf{r}'\vert}
\end{equation}

("+" - розбіжна, "-"- збіжна хвиля). Для фінітних чи швидко зникаючих потенціалів асимптотика, яка відповідає розбіжній хвилі (\ref{scateq22}) при $\vert\, \mathbf{r}\vert \,\gg \vert \mathbf{r}'\vert $, має вигляд:

\begin{equation}
\label{scateq23} G_{0}^{+}  (\,\mathbf{r} - \mathbf{r}'\,;E) = -
\frac{m}{2\,\pi \,\hbar^{\,2}}\frac{\exp \left(i\,k\,r\right)}{r}
\exp \left( - i\,\mathbf{k}'\,\cdot\mathbf{r}'
\right)
\end{equation}

де $\mathbf{k}' = \mathbf{n}\,k,\,\,\mathbf{n} =\frac{\mathbf{r}}{r}$. Правила обходу полюсів при обчислені функції Гріна можна також задати формально заміноючи у знаменнику \eqref{scateq21} енергію $E$ на $E + i\varepsilon
\,\,,\,\,\varepsilon > 0$:

\begin{equation}
\label{scateq24} G_{0}^{\pm}  \,(\,\mathbf{r} - \,\mathbf{r}';E)
= \int \frac{d\,\mathbf{k}'}{(2\pi )^{3}}
\frac{\exp \left(i\,\mathbf{k}' \cdot (\,\mathbf{r} - \,\mathbf{r}')
\right)}{E - E'\,\,\pm \,\,i\varepsilon}
\end{equation}

Підставляючи асимптотичний вираз (\ref{scateq23}) до (\ref{scateq17}) для амплітуди і перерізу у __борнівському наближенні__ отримуємо:

\begin{align}
 \psi &= \exp \left( i\,\mathbf{k} \cdot \mathbf{r} \right) +
A(\mathbf{n},\mathbf{n}')\frac{\exp \left(i\,k\,r \right)}{r}\,,
\notag\\
 A(\mathbf{n},\mathbf{n}') &= - \frac{m}{2\pi \,\hbar^{\,2}}
 \int d\mathbf{r}\exp \left( i(\mathbf{k} - \mathbf{k}')\cdot
\mathbf{r}\right)U(\mathbf{r})=\notag \\
&-\frac{m}{2\pi \,\hbar^{\,2}} \int d\,\mathbf{r}\exp \left(-
i\,\mathbf{q}\cdot
\mathbf{r}\right)\, U(\mathbf{r})\,,\notag\\
 d\,\sigma &= \frac{m^2}{4\pi^{\,2}}\,\hbar^{4}
 \left| \int d\,\mathbf{r}\exp\left(- i\,\mathbf{q}\, \cdot \mathbf{r} \right)\,
 U(\mathbf{r})\right|^{\,2}d\Omega\,,\label{scateq25/10}\\
 \mathbf{q}&= \mathbf{k}' - \mathbf{k}\,,\,\,q = 2\,k\,\sin{\theta/2}
 \end{align}
 
Неважко бачити, що цей шлях є еквівалентним стандартному розрахунку першої поправки до хвильової функції неперервного спектру у рамках стаціонарної теорії збурень:

\begin{eqnarray}
\psi &=& \psi_{\rm \mathbf{k}}^{(0)} +
 \psi_{\text{расс}}^{(1)} + \ldots \nonumber\\
 \psi_{\mathbf{k}}^{(0)} &=& \exp \left(i\,\mathbf{ k} \cdot 
\mathbf{r} \right)\,\,,   \nonumber \\ 
\psi_{\text{расс}}^{(1)}
&=& \sum\limits_{\mathbf{k}' \ne \mathbf{k}}
\frac{V_{\mathbf{k}',\mathbf{k}}}{E_{\mathbf{k}}^{(0)} - E_{\mathbf{k}'}^{(0)}}\,
\psi_{\mathbf{k}'}^{(0)} = \int \frac{d^{3}\,\mathbf{k}'}{(2\pi )^{3}} \frac{V_{\mathbf{k}'\,\mathbf{k}} \exp \left( i\,\mathbf{k}\,' \cdot\mathbf{r}\right)}{E_{\mathbf{k}}^{(0)} - 
	E_{\mathbf{k}'}^{(0)}}  \\
 V_{\mathbf{k}'\,\mathbf{k}} &=&
 \int \exp \left( - i\,\mathbf{k}\,' \cdot \mathbf{r} \right)\,
 V(\mathbf{r})\,\exp \left( i\,\mathbf{k} \cdot \mathbf{r} \right)d^{3}\,\mathbf{r}\, \nonumber
\end{eqnarray}

Зазначимо, що крім теореми взаємності амплітуда розсіяння у __борнівському наближенні__ задовольняє 
$$
A(\mathbf{n},\mathbf{n}') = A^*(\mathbf{n}',\mathbf{n})
$$
між амплітудами прямого і процесса при якому імпульси початкового і кінцевого стану міняються місцями без зміни напрямку. Це співвідношення безпосередньо може бути
отримано з (\ref{scateq6}), якщо врахувати, що у першому порядку теорії збурень права частина (\ref{scateq6}) дорівнює 0.

Для центрального поля у виразах \eqref{scateq25/10} можна
обчислити інтеграл по кутовим змінним і отримати:

\begin{equation}
\label{scateq25} A(q) = - \frac{2m}{\,\hbar^{\,2}}
\int\limits_{0}^{\infty} d\,r\,\frac{\sin(qr)}{q}\,r\,U(r)\,.
\end{equation}

Звідки видно, що в границі малих імпульсів частинок що падають
амплітуда розсіяння є ізотропною:

\begin{equation}
\label{scateq26} A(q) = - \frac{2m}{\,\hbar^{\,2}}\int
{d\,r\,r^{\,2}U(\mathbf{r})\,\,}\,.
\end{equation}

Цікаво відмітити, що результат борнівського наближення \eqref{scateq25/10} для
перерізу розсіяння може бути отриманий з використанням ["золотого правила Фермі"](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics2/blob/main/QM_py/qm2_nonstacpertheor.ipynb#fermigoldentule). 
Дійсно, процес розсіяння може бути описаний як перехід у стан неперервного спектра. Початковий і кінцевий стани є плоскими хвилями (вільна частинка):

\begin{equation}\label{scateq27}
 \psi_{fin} = \frac{1}{\sqrt{V}}
 \exp(i\,\mathbf{k}_{f} \cdot \mathbf{r})\,,\,\,\, E_{f} = \frac{p_{f}^{\,2}}{2m} \\
 \psi_{in} = C\exp (i\,\mathbf{k}_{in} \cdot \mathbf{r})\,\,,\, E_{in} = \frac{p_{in}^{\,2}}{2m}\,. \\
\end{equation}

Коефіцієнт $C$ залишимо поки невизначеним. Енергетична густина
числа кінцевих станів дорівнює:

\begin{equation}\label{scateq28} 
\rho(E) = \frac{d\,N}{dE} = 
\frac{d^{3}\mathbf{p}}{(2\pi \,\hbar )^{3}dE}\,V =
\frac{p^{\,2}\,dp\,d\Omega}{(2\pi \,\hbar)^{3}dE}\,V =
\frac{m\sqrt{2mE} \,d\Omega}{(2\pi \,\hbar)^{3}}\,V\,.
\end{equation}

Таким чином:

\begin{equation}\label{scateq29}
 \Pi_{\,in \to f} = \frac{2\pi}{\hbar}
 \left|\,V_{f,\,in}\right|^{\,2}\,\rho \,(E_{f}) = 
 \frac{2\pi}{\hbar}\,V\,\vert C\vert^{\,2}
 \left|\tilde{V}_{\mathbf{k}_{f},\mathbf{k}_{in}} \right|^{2}\,\rho(E_{f}) = 
    \frac{m\,p_{f}}{4\pi^{\,2}\hbar^{4}}\vert C\vert^{\,2}
    \left|\tilde{V}_{\mathbf{k}_{f} ,\,\mathbf{k}\,_{in}} \right|^{\,2}d\Omega\,. 
\end{equation}

Обираючи нормування початкового стану на одиничну густину потоку:

$$
\psi_{in} = \sqrt{\frac{m}{p_{in}}}\exp(i\,\mathbf{k}_{in} \cdot \mathbf{r})\,,
$$

отримуємо:

\begin{equation}
\label{scateq30} d\sigma = \frac{m^{\,2}}{4\pi^{\,2}\hbar^{4}}
\frac{p_{f}}{p_{in}}
\left|\int V(\mathbf{r}) \exp( - i \mathbf{q} \cdot \mathbf{r})\,d^{3} \mathbf{r}\right|^{\,2}\,.
\end{equation}

Пружне розсіяння відповідає випадку $p_{f} = p_{in} $.
