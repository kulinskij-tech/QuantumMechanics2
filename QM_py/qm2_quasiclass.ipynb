# Квазікласичне наближення (WKB) у квантовій механіці.

Відсутність поняття траєкторії у квантової механиці приводить до суттєвої різниці у поведінці квантових систем від класичних. З іншого боку, прояв хвильових властивостей у явищах мікросвіту дає основу для проведення аналогії квантової механики з фізичною оптикою (зазначимо, що для хвилі поняття траєкторії також в загальному випадку відсутнє). Як відомо, фізична оптика у граничному випадку нескінченно малої довжини хвилі переходить у
т.з. геометричну оптику, у якої розповсюдження електромагнітних хвиль описується у термінах променів, тобто об'єктів які аналогічні траєкторіям руху матеріальної частинки. Відома оптико-механічна аналогія, яка заснована на еквівалентності варіаційних принципів Ферма (у геометричної оптиці) і Мопертюі - Ейлера - Лагранжа - Якобі (у механиці) встановлює еквівалентність геометричної оптики і механіки гамільтоновських систем. Згадаємо стисло основні моменти.<br>

Хвильове рівняння для фур'є компоненти якоїсь проекції напруженості електричного поля у оптично неоднорідному, ізотропному, бездисперсійному середовищі, може бути записане у
вигляді:
$$\frac{1}{c^2_{loc}}\,\frac{\partial^2\mathcal{E}}{\partial \,t^2} - \Delta \mathcal{E} = 0\,,\,\,\mathcal{E} = e^{i\,\omega\,t}\,u\,,\,\,\omega = c\,k\,,\,\,c_{loc} = c/n$$
<!--  -->
\begin{equation}\label{helm}
\Delta \,u\, + k_{0}^{2} n^{2}u = 0
\end{equation}
<!--  -->
Оскільки нас цікавить випадок $k_{0} \to \infty $, будемо шукати розв'язок цього рівняння у виді (зміною амплітуди нехтуємо):
<!--  -->
\begin{equation}\label{podst}
  u = \exp \left( {i\,k_{0} \theta}  \right)\,,\,\,\text{if} \,\,n=const\Rightarrow \theta = n\,\mathbf{k}_0\cdot \mathbf{r} 
\end{equation}
<!--  -->
Підставляючи \eqref{podst} у \eqref{helm} отримуємо:
<!--  -->
$$ 
 - k_{0}^{2} (\nabla \theta )^{2} + ik_{0} \Delta \,\theta \, + k_{0}^{2}
n^{2} = 0\,\,.
$$
<!--  -->
Наближене (з точністю до величин порядку $1 /k_{0})$ воно дає:
<!--  -->
\begin{equation}\label{ei}
  (\nabla \theta )^{2} - n^{2} = 0
\end{equation}
<!--  -->
Це співвідношення називається __рівнянням ейконалу__. Воно визначає поверхню постійної фази тобто хвильовий фронт. Лінії що перпендикулярні до всіх таких поверхонь є промені. Для механічних систем у ролі рівняння \eqref{ei} виступає рівняння Гамільтона-Якобi, якщо під $n$ розуміти величину $\sqrt {2m\,(E - U)} $, а під $\theta $ - зкорочену дію. Використовуючи математичну подібність стаціонарного рівняння Шредінгера
$$
\Delta \psi + \frac{2\,m\,E}{\hbar ^2}\left(1 -
\frac{U(\,\mathbf{r}\,)}{E}\right)\psi \, = 0
$$
і \eqref{helm}, можна встановити зв'язок і з'ясувати характер граничного переходу між квантовою і класичною механикою (квазикласичне наближення).<br>

Квазикласичне наближення у квантової механиці засновано на
наближеному рішенні рівняння Шредінгера
<!--  -->
\begin{equation}
\label{16eq1} i\,\hbar \frac{\partial \,\psi} {\partial \,t} =
\left(- \frac{\hbar^{2}}{2\,m}\Delta + U(\,\mathbf{r}\,)
\right)\,\psi
\end{equation}
<!--  -->
При цьому рішення рівняння слід шукати у вигляді (метод Вентцеля-Крамерса-Брилюена):
<!--  -->
\begin{equation}
\label{16eq2} \psi (r,t) = \exp \left( \frac{i}{\hbar}\,S(\,\mathbf{r},t) \right)
\end{equation}
<!--  -->
Підставляючи \eqref{16eq2} до \eqref{16eq1} дістаємо
<!--  -->
\begin{equation}
\label{16eq3}
-\frac{\partial \,S}{\partial \,t} = \frac{(\nabla S)^{2}}{2m} + U(\,\mathbf{r}\,) - \frac{i\,\hbar}{2m}\Delta S\\
S = \int L\,dt = \int\left(p\,\dot{q} - H\right)\,dt = \int p\,dq - \int H\,dt\\
\nabla S = \mathbf{p}\,,\,\,\,-\frac{\partial S}{\partial t} = H
\end{equation}
<!--  -->
У випадку стаціонарної задачі часову залежність в $S$ можна виділити стандартним чином
<!--  -->
\begin{equation}\label{16eq4}
S = - Et + \Sigma
\end{equation}
<!--  -->
Для функції $\Sigma $ маємо рівняння:
\begin{equation}
\label{16eq5} (\nabla \Sigma )^{2} + 2m(U - E) - i\,\hbar \,\Delta
\Sigma = 0,
\end{equation}
<!--  -->
яке можно розвязувати формальним розкладом за величиною
сталої Планка:
<!--  -->
\begin{equation}
\label{16eq6} \Sigma = \sigma _{0} + \frac{\hbar}{i}\sigma_{1} + 
\left(\frac{\hbar}{i} \right)^{2}\sigma _{2} +\ldots
\end{equation}
<!--  -->
Для коефіцієнтів розкладу (\ref{16eq6}) маємо ланцюжок
рівнянь
<!--  -->
\begin{equation}
\label{16eq7} (\nabla \sigma _{0} )^{2} = 2m\,\left( {E - U}
\right)\,\,\,,\,\,\,\nabla \sigma _{0} \nabla \sigma _{1} +
\frac{1}{2}\Delta \sigma _{0} = 0\,\,\,\ldots
\end{equation}
<!--  -->
$$
p(x) = \sqrt{2\,m\,\left(E-U\right)}
$$
Перше рівняння для нульового порядка є класичним рівнянням Гамільтона-Якобі. Таким чином, функція $\sigma_{0} $ є класичною скороченою дією, а її градієнт -- класичним імпульсом, выраженным как функція координат. У одновимірному випадку перші два рівняння в \eqref{16eq7} легко решаются
<!--  -->
\begin{equation}\label{16eq8} 
\sigma_{0} = \pm \int\limits_{x_0}^{x}
p\,(x)\,dx\,\,\,,\,\,\,\sigma _{1} = - \ln \sqrt{p\,(x)} +
const
\end{equation}
<!--  -->
і хвильова функція у квазикласичному наближенні набуває вигляду
<!--  -->
\begin{equation}
\label{eq_qclasspsi} \psi (x) = \frac{C_{1}}{\sqrt{p\,(x)}}
\exp\left( \frac{i}{\hbar} \int\limits_{x_0}^{x} p\,(x)\,dx
\right) + \frac{C_{2}}{\sqrt{p\,(x)}}
\exp \left(-\frac{i}{\hbar}\int\limits_{x_0}^{x} p\,(x)\,dx \right)\,\,.
\end{equation}
<!--  -->
Необхідною умовою застосування квазикласичного наближення
є малість сталої Планка у порівнянні з класичною
дією, оскльки їх відношення є безрозмврною
величиною:
<!--  -->
\begin{equation}
\label{16eq10} (\nabla \Sigma )^{2}\,\gg \,\hbar \,\vert \Delta
\Sigma \vert \,\, \Rightarrow (\nabla \sigma _{0} )^{2}\,\gg
\,\hbar \,\vert \Delta \sigma _{0} \vert \,
\end{equation}
<!--  -->
або у одновимврному випадку:
<!--  -->
\begin{equation}
\label{16eq11} p^{2} \gg \,\hbar \,\left| \frac{\partial
\,p}{\partial \,x} \right|\, 
\Rightarrow 1 \gg
\frac{1}{2\pi} \,\left| \frac{\partial \,\lambda}{\partial
\,x} \right|\,\,.
\end{equation}
<!--  -->
У "корпускулярній формі"\,:
<!--  -->
\begin{equation}
\label{16eq12} \frac{\hbar \,m\,\vert\, F\,\vert}{|\,p\,|^{3}}
\, \ll 1\,\,.
\end{equation}
<!--  -->

## Правила квантування Бора-Зоммерфельда
Точки, в яких $U=E$ (класичний імпульс дорівнює 0) називаються \textbf{точками повороту}. У їх околі квазикласичне наближення не прийнятне. В класично забороненій області імпульс є суто уявним, і розв'язок
має вигляд
<!--  -->
\begin{equation}
\psi(x) = \frac{C_{1}}{\sqrt{\left|p\,(x)\right|}}
\exp \left(\frac{1}{\hbar}\int\limits_{x_0}^{x} \left| p\,(x) \right|\,dx  \right) 
+ 
\frac{C_{2}}{\sqrt{\left| p\,(x)\right|}} 
\exp\left(- \frac{1}{\hbar}\int\limits_{x_0}^{x} 
\left| p\,(x) \right|\,dx \right)
\end{equation}
<!--  -->
константи обираються виходячи з конкретики задачи. В якості такої розглянемо задачу про стаціонарні стани частинки в потенціальній ямі. Класичний рух обмежено двома точками поворота $a\, < \,b$. Оскільки амплітуда хвильової функції повинна зменшуватись при віддалені від точки повороту у класично заборонену область руху
асимптотика хвильової функції там є наступною:
<!--  -->
\begin{equation}\label{eq_qclassaleft} 
\psi (x) = \frac{C}{\sqrt{\vert p\,(x)\vert}}\,
\exp\left(\frac{1}{\hbar}\,\int\limits_{a}^{x} \vert p\,(x)\vert
\,dx \right)\,\,\,,\,\,x < a
\end{equation}
$$
\psi (x) = \frac{\tilde{C}}{\sqrt{\vert p\,(x)\vert}}\,
\exp\left(-\frac{1}{\hbar}\,\int\limits_{b}^{x} \vert p\,(x)\vert
\,dx \right)\,\,\,,\,\,x > b
$$
<!--  -->
Поблизу точки повороту
<!--  -->
\begin{equation}
\label{eq_qclasspx} \vert p\,(x)\,\vert =
\sqrt{2\,m\,\left\vert\,\frac{d\,U}{d\,x}\,\right\vert\,\vert a -
x\,\vert}\,\,(1 + o(\,\vert\, a - x\,\vert^{1/2} \,))\,\,.
\end{equation}
<!--  -->
Якщо інтерпретувати змінну $x$ як комплексну величину, то перехід від класично забороної до класично дозволеної області може бути описаний як обхід точки повороту у комплексній площині, при якому фаза $x$ змінюється на $\pm \pi $ у відповідності з напрямом обходу. Хвильова функція є аналитичною функцією $x$, також як і класичний імпульс. Умови застосування квазикласичного наближення \eqref{16eq11} и \eqref{16eq12} ніде формально не порушуються, якщо радіус кола обходу є достатньо великим. При цьому фаза імпульса змінюється на $e^{\pm \frac{i\,p}{2}}$, тобто:
<!--  -->
\begin{equation}\label{eq_pip} 
(x - a) \to - (x - a) = e^{\pm i \pi}\,(x - a)\,,\quad p \to 
e^{\pm \frac{i\pi}{2}}\,p = \pm i\,p\,.
\end{equation}
<!--  -->
Таким чином, з \eqref{eq_pip} отримаємо:
<!--  -->
\begin{equation}
\label{eq_qclasspsiwella}
\begin{array}{l}
 \psi (x) = \frac{C}{\sqrt{p\,(x)}}
 \exp \left( \frac{i}{\hbar}\int\limits_{a}^{x} p\,(x)\,dx  - 
 \frac{i\pi} {4}\right) + 
 \frac{C}{\sqrt{p\,(x)}}
 \exp\left( -
\frac{i}{\hbar}\int\limits_{a}^{x} p\,(x)\,dx  +
\frac{i\pi}{4} \right) = \\
 2\frac{C}{\sqrt{p\,(x)}}
 \cos \left( \frac{1}{\hbar}
 \int\limits_{a}^{x} p\,(x)\,dx  - 
 \frac{\pi}{4}\right)
 \end{array}
\end{equation}
<!--  -->
Тепер розглянемо ситуацію, відносно іншої точки повороту
$b$, де класично дозволена область є зліва, а
забороена - зправа. Асимптотика хвильової функції
далеко від $b$:
<!--  -->
\begin{equation}\label{eq_qclassbright} 
\psi (x) = \frac{C'}{\sqrt{\vert p\,(x)\vert}}
\exp \left( - \frac{1}{\hbar} 
\int\limits_{b}^{x} 
\vert p\,(x)\vert \,dx \right)\,\,,\,\,\,x > \,b
\end{equation}
<!--  -->
Тими ж міркуваннями що і вишче отримаємо, що асимптотика хвильової функції у області $x < b$ є:
<!--  -->
\begin{equation}
\label{eq_qclasspsiwellb}
\begin{array}{l}
 \psi (x) = \frac{C'}{\sqrt{p\,(x)}}
 \exp \left( -
\frac{i}{\hbar} \int\limits_{b}^{x} p\,(x)\,dx  - 
\frac{i\pi}{4} \right) + \frac{C'}{\sqrt{p\,(x)}}
\exp \left(\frac{i}{\hbar}
\int\limits_{b}^{x} p\,(x)\,dx  +
\frac{i\pi }{4} \right) = \\
 2\frac{C'}{\sqrt{p\,(x)}}
 \cos \left(\frac{1}{\hbar}
 \int\limits_{b}^{x} p\,(x)\,dx  + \frac{\pi} {4}
\right) = 2\frac{C'}{\sqrt {p\,(x)} }\cos \left(  -
\frac{1}{\hbar}
\int\limits_{a}^{b} p\,(x)\,dx  +
\frac{1}{\hbar }
\int\limits_{a}^{x} p\,(x)\,dx  +
\frac{\pi} {4} \right)
 \end{array}
\end{equation}
<!--  -->
Для співпадіння виразів \eqref{eq_qclasspsiwella} та \eqref{eq_qclasspsiwellb} у
певній області проміжка $a < b$ необхідно щоб $|C| = |C'|$, тому:
<!--  -->
\begin{equation}
\label{16eq20} 
\int\limits_{a}^{b} p\,(x)\,dx\, = \hbar \pi \,\left( n + \frac{1}{2} \right)
\,\,\, \Rightarrow \,\frac{1}{2\,\pi \,\hbar} 
\,\oint p\,(x)\,dx\, = n +  \frac{1}{2}
\end{equation}
<!--  -->
Отримане співвідношення називається __правилом квантования
Бора-Зоммерфельда__ . Це співвідношення дозволяє в рамках квазикласичного
наближення розв'язати задачу про стаціонарні рівні частинки у довільній (!) потенціальній ямі. Впишемо правила відповідності асимптотик хвильової функції по різні сторони від
точок повороту, які ми отримали:
<!--  -->
\begin{equation}
\label{eq_quasiclassymptot_trans}
\frac{C}{\sqrt {\vert p\,(x)\vert}}\,
\exp \left( - \frac{1}{\hbar}
\left| \int\limits_{x_0}^{x}
p\,(x)\,dx \right| \right)\,\,\, 
\Rightarrow
\,\,2\frac{C}{\sqrt{p\,(x)}}\,\cos \left( \frac{1}{\hbar}
\left| \int\limits_{x_0}^{x} p\,(x)\,dx \right| -
\frac{\pi} {4} \right)
\end{equation}
<!--  -->
Таким чином, постійна Планка може розглядатись як квант дії у квантовій механіці. На кожний стан припадає об'єм фазового простору $(2\,\pi \,\hbar )^{d}$, де $d$ - кількість просторових ступенів вільності, по яким рух розділяється. За допомогою правила квантування Бора-Зоммерфельда можна отримати оцінку на кількість станів
дискретного спектру, які лежать нижче даного значення енергії $E$. 

## Тунельний ефект
Аналогічно розглядається задача про проникнення крізь потенціальний бар'єр (тунельний ефект). Тут припускається, що зліва з $ - \infty $ на бар'єр $a < b$, який є фінітним чи його дією на $\infty $ можна знехтувати, налітає частинка з фіксованою енергією. Визначимо ймовірність проникненя частинки крізь бар'єр. Асимптотика хвильової функції, яка відповідає квазівільній частинці у наближенні
ВКБ має вигляд:
<!--  -->
\begin{equation}\label{16eq22} 
\psi (x) = \frac{C}{\sqrt{p\,(x)}}
\exp \left(\frac{i}{\hbar }\int\limits_{b}^{x} p\,(x)\,dx  +
\frac{i\,\pi} {4} \right)\,\,,\,\,\,x > \,b
\end{equation}
<!--  -->
Це хвиля, що минула бар'єр (фаза
$\pi $/4 обрана з міркувань зручності). При проникненні під
бар'єр амплітуда такої хвилі повинна монотоно зменшуватись,
тому, на відміну від попередньої задачі є тільки одна
можливість для обходу:
$$ \,\,\,p \to i\,p\,\,.$$
В результаті маємо правило переходу через точку поворота:
<!--  -->
\begin{equation}
\label{16eq23}
\begin{array}{l}
 \psi (x) = \frac{C}{\sqrt {p\,(x)}}
 \exp \left( \frac{i}{\hbar} 
 \int\limits_{b}^{x} p\,(x)\,dx  + \frac{i\,\pi} {4}
\right)\,\, 
\Rightarrow \,\,\,\frac{C}{\sqrt{\vert p\,(x)\vert}}
\exp \left( - \frac{1}{\hbar}\int\limits_{b}^{x} \vert
p\,(x)\vert \,dx  \right)\\
 x < \,b \\
 \end{array}
\end{equation}
<!--  -->
Перепишемо інтеграл в \eqref{16eq23} так щоб
дістатися другої точки поворота $a$:
<!--  -->
$$
\exp \left(- \frac{1}{\hbar} 
\int\limits_{b}^{x} 
\vert p\,(x)\vert \,dx  \right) = 
\exp \left( \frac{1}{\hbar} \int\limits_{x}^{b}
\vert p\,(x)\vert \,dx  \right) = 
\exp \left( \frac{1}{\hbar}\int\limits_{a}^{b} \vert p\,(x)\vert
\,dx  - \frac{1}{\hbar}\int\limits_{a}^{x} \vert
p\,(x)\vert \,dx  \right)
$$
<!--  -->
Тому асимптотика хвильової функції у класично забороненій 
області правіше точки $a$ є:
<!--  -->
\begin{equation}\label{eq_asymptbarrier} 
\psi (x) = \,\,\frac{C}{\sqrt{\vert p\,(x)\vert}} 
\exp \left( \frac{1}{\hbar}\int\limits_{a}^{b} \vert
p\,(x)\vert \,dx \right)\exp \left(- \frac{1}{\hbar}
\int\limits_{a}^{x} \vert p\,(x)\vert \,dx  \right)
\end{equation}
<!--  -->
Використовуючи правило перехода \eqref{eq_quasiclassymptot_trans}, отримуємо
відповідну до \eqref{eq_asymptbarrier} асимптотику:
<!--  -->
\begin{equation}
\label{eq_asymptbarrierout0}
\begin{array}{l}
 \psi (x) = \,\,\frac{2\,C}{\sqrt {p\,(x)} }\exp \left(
\frac{1}{\hbar} \int\limits_{a}^{b} {\vert p\,(x)\vert
\,dx} \right)\cos \left( \frac{1}{\hbar}
\int\limits_{a}^{x} p\,(x)\,dx + \frac{\pi} {4} \right)
\\
 x < a \\
 \end{array}
\end{equation}
<!--  -->
лівіше за точку $a$. Запишемо \eqref{eq_asymptbarrierout0} у дещо іншому вигляді:
<!--  -->
$$
\begin{array}{l}
 \psi (x) = \,\,\frac{C}{\sqrt{p\,(x)}}
 \exp \left( \frac{1}{\hbar}
\int\limits_{a}^{b} \vert p\,(x)\vert \,dx  \right)
\left[
\exp \left( \frac{1}{\hbar} \int\limits_{a}^{x}
p\,(x)\,dx + \frac{\pi} {4} \right) + \exp \left( -
\frac{1}{\hbar} \int\limits_{a}^{x} p\,(x)\,dx  -
\frac{\pi}{4} \right) \right] \\
 x < a
 \end{array}
$$
<!--  -->
Таким чином, пореред бар'єру асимптотика хвильової функції
є суперпозицією падаючої та відбитої хвиль. Відношення
потоків хвилі що пройшла до падаючої (квадрат відношення амплітуд
хвиль) за визначенням є коефіціентом прозорості бар'єра:
<!--  -->
\begin{equation}\label{eq_barriertransp} 
D = \exp \left( - \frac{2}{\hbar}
\int\limits_{a}^{b} \vert p\,(x)\vert \,dx  \right),
\end{equation}
<!--  -->
Як витікає з \eqref{eq_asymptbarrierout0} амплітуди падаючої та відбитої хвиль рівні. Це повязано зі специфікою квазикласичного наближення. Дісно, в рамках класичної механіки при $E < U$ всі частинки повинні відбитися від бар'єра (коефіцієнт відбиття дорівнює 1, коефіцієнт прозорості - 0). У квазикласичному наближенні можуть з'явитись тільки поправки порядка $\hbar^{n}$. Однак, \eqref{eq_barriertransp} не може буто представлено у вигляді степеневого разкладу за $\hbar $, а отже $D = o(\hbar^n)\,,\forall n$, тому в квазикласичному наближенні коефіцієнт відбиття є $1+o(\hbar^n)\,,\forall n$.

## Квазикласичне наближення для осцилятора та атома водню
<!--  -->
Застосуємо метод ВКБ для розрахунку спектра осцилятора та атома водню. Для 1-вимірного осцилятора все робиться досить просто:
<!--  -->
$$
E = \frac{m\,\omega^2\,x^2_0}{2}\,,\,\,\,\,p(x) = m\,\omega\,\sqrt{x^2_0-x^2}\\
\int\limits^{x_0}_{-x_0}p(x)\,dx =  \hbar\,\pi\,\left(n+\frac{1}{2}\right)\,\quad \Rightarrow\,\quad E_{n} = \hbar\,\omega\,\left(n+\frac{1}{2}\right)
$$
<!--  -->
Таким чином ми отримали точну формулу, хоча правила квантування використовують наближення \eqref{eq_qclasspsi} для хвильової функції. Це співпадіння обумовлене квадратичним характером гамільтоніана оскільки площа фазової кривої тут пропорційна до енергії так само як і до числа квантів.<br>

Для атома водню, умова квантування радіального руху (для станів $E<0$):
<!--  -->
\begin{equation}\label{eq_hatomqclass}
	\int\limits^{r_{+}}_{r_{-}} p_{r}\,d\,r =
	\int\limits^{r_{+}}_{r_{-}}
	\sqrt{2m\left(E-\frac{\hbar^2\,(l+1/2)^2}{2\,m\,r^2}+\frac{\gamma}{r}\right)}\,d\,r=\,\hbar\,
	 \pi\,\left(n_{r}+\frac{1}{2}\right)
\end{equation}
<!--  -->
ми скористаємося спрощеним квантовим виразом для відцентрового потенціала, для того щоб отримати точну формулу наприкінці. Великої помилки ми не робимо, очкільки квазикласичний випадок відповідає $l\gg 1$, $L^2 = \hbar^2\,l\,(l+1)\approx \hbar^2\,(l+1/2)^2$. Далі обчислимо інтеграл:
<!--  -->
$$
\int\limits^{r_{+}}_{r_{-}} p_{r}\,d\,r =\sqrt{2 m |E|}\, \int\limits^{r_{+}}_{r_{-}}
\frac{\sqrt{(r-r_{-})(r_{+}-r)}}{r}\,d\,r= \sqrt{2 m |E|}\,\pi\,
\frac{\left(\sqrt{r_{+}}-\sqrt{r_{-}}\right)^2}{2}
$$
<!--  -->
$$r_{\pm} = \frac{\gamma}{2\,|E|}
\left(1 \pm\sqrt{1-\frac{\hbar^2\,(2l+1)^2\,|E|}{2\,m\,\gamma^2}}\right)
$$
<!--  -->
звідки отримаємо:
$$
\hbar\left(2\,n_{r}+1\right) = \gamma\sqrt\frac{2m}{|E|}\left(1-\frac{\hbar(2l+1)}{\gamma}\sqrt\frac{|E|}{2m}\right)\quad\Rightarrow \quad
E = -\frac{m\,\gamma^2}{2\,\hbar^2\left(n_r+l+1\right)^2}
$$
<!--  -->
що співпадає із точним виразом якщо трактувати $n=n_r+l+1$ як головне квантове число.
