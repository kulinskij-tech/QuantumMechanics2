Задача о розсіянні взаємодіючих частинок є однією з центральних у квантової теорії. Вона безпосередньо пов'язана з експериментальною ситуацією, яка виникає при дослідженні структури матерії. У подальшому будемо розглядати 2-х частинкове розсіяння не тотожних
частинок, якщо не обговорено інше. Завдяки трансляційної симетрії задачі (ми як завжди вважаємо, що потенціал взаємодії залежить тільки від відносного положення частинок) оператор повного імпульсу системи:
<!--  -->
\begin{equation}\label{p12} \mathbf{P}_{0} = \mathbf{p}_{1} + \mathbf{p}_{2\,} \,\,,\,\,[H,\mathbf{P}_{0}] = [U(1-2),\mathbf{p}_{1} ] + [U(1-2),\mathbf{p}_{2} ] = 0
\end{equation}
<!--  -->
є інтегралом руху, тому, як і в класичній ситуації, 2-х частинкова задача еквівалентна 1-частинковій для частинки зведеної маси, яка рухається у полі нерухомого центру.

Якщо згадати про вже згадану аналогію квантової механіки з фізичною оптикою, то зрозуміло, що постановка задачі розсіяння у квантовому випадку повинна бути близькою до задачі розсіяння хвилі у неоднорідному середовищі. У останньому випадку частіше розглядається плоска монохроматична хвиля. Додатково припускається , що на нескінченності, асимптотично, впливом неоднорідностей можна знехтувати. В результаті розсіяння, по принципу суперпозиції, сумарне поле випромінювання на великих відстаннях від розсіювача, представляється у вигляді суми полів хвиль, що падає і розсіюється. Ця розсіяна хвиля має вигляд розбіжної (асимптотично) сферичної хвилі, амплітуда якої залежить від напрямку, оскільки містить інформацію про взаємодію з центром розсіяння. Повне поле є результатом розв'язування рівнянь Максвелла з відповідними граничними умовами.

## Парціальний розклад амплітуди та перерізу розсіяння. Фаза розсіяння.  

У квантовому випадку, точна хвильова функція, яка дає повний опис ситуації, є розв'язокм відповідного рівняння Шредінгера з певними граничними умовами. Розсіяння, при якому не виникає зміна стану силового центра, та характеристик частинки, тобто не виникають
нові частинки, називається __пружнім__,   у протилежному випадку - __непружнім__. У подальшому будемо розглядати стаціонарний випадок, тобто коли гамільтоніан системи явно від часу не залежить.

При розгляді достатньо швидко спадаючих з відстанню потенціалів, наприклад фінітних, завдяки принципу суперпозиції асимптотика координатної частини хвильової функції на далеких від розсіювача відстанях повинна мати вигляд:
<!--  -->
\begin{equation}
\label{21eq2} \psi (\,\mathbf{r}\,) \to \,
\exp \left(i\mathbf{k}\cdot \mathbf{r}\right) + 
A\,\frac{\exp \left(i\,\mathbf{k}'\cdot \mathbf{r} \right)}{r}\,,\,\,\,r\to \infty
\end{equation}
<!--  -->
$$d\sigma = \frac{\mathbf{j}_{scatt}\cdot \,d\mathbf{S}}{j_{0}} = |A|^2\,d\Omega$$
<!--  -->
де хвильові вектора $\mathbf{k}$ і $\mathbf{k}'$ рівні за величиною у випадку пружного розсіяння, $A$ називається __амплітудою розсіяння__,
структура якої і властивості будуть розглядатись нижче. Припускаючи центральний характер потенціалу, розглянемо стаціонарне рівняння Шредінгера:
<!--  -->
\begin{equation}\label{21eq3} 
\left( - \frac{\hbar^{\,2}}{2m}\Delta_{r} + \frac{L^{\,2}}{2mr^{\,2}} + (U - E) \right)\psi = 0
\end{equation}
<!--  -->
<img src="./figs/fig_cross_section.svg" alt="Image" width="600"/>
<center>Геметрія розсіяння та диференціальний переріз $d\sigma = 2\pi\rho(\theta)\,d\rho$</center>
Нагадаємо, що в класичній механіці саме прицільна відстань (за фіксації енергії) визначає момент імпульсу:
<!--  -->
$$L_{0} = m\,v_{\infty}\,\rho$$
<!--  -->
Незважаючи на те, що гамільтоніан системи формально має сферичну симетрію, асимптотична умова (\ref{21eq2}) цю симетрію порушує. В результаті симетрія понижується до циліндричної і, як неважко бачити, інтегралом руху є оператор проекції моменту імпульсу $L_{z} $ на напрямок імпульсу нерозсіяної частинки (його власне значення дорівнює 0). Тому розв'язок (\ref{21eq3}) може бути представлено у вигляді розкладу за базисними функціями оператора
$\hat{L}^2$ , при цьому проекції на стани з $m \ne 0$ відсутні:
<!--  -->
\begin{equation} \label{21eq4} 
\psi(\,\mathbf{r}\,) = \sum\limits_{l=0}^{\infty} a_{l}\, R_{k,\,l}(\,\mathbf{r}\,)\,P_{l} (\cos{\theta}),
\end{equation}
<!--  -->
де $\theta $ - кут між $\mathbf{k}$ і $\mathbf{k}'$. Радіальні частини $R$ задовольняють рівняння:
<!--  -->
\begin{equation}\label{21eq5} 
\frac{1}{r}\left( r\,R_{k,\,l} \right)^{''} -
\frac{l\,(l + 1)}{r^{\,2}}\,R_{k,\,l} + \left(k^{\,2} -
\frac{2m}{\hbar^{\,2}}U \right)R_{k,\,l} = 0
\end{equation}
<!--  -->
Введемо безрозмірну змінну: $\tilde {r} = k\,r$. Припускаючи потенціал фінітним, і нехтуючи відцентровим внеском, отримаємо:
<!--  -->
\begin{equation}
\label{21eq6} \left( {\tilde {r}R_{k,\,l}}  \right)^{''} + \tilde
{r}R_{k,\,l} = 0
\end{equation}
<!--  -->
звідки витікає, що:
<!--  -->
\begin{equation}
\label{21eq7} R_{k,\,l} = \frac{2}{\tilde {r}}\,
\sin \left(\tilde {r} - \frac{\pi \,l}{2} + \delta_{l} \right)
\end{equation}
<!--  -->
Величини $\delta $ називаються  __фазами розсіяння__ і є функціями хвильового числа і $l$. Вони визначені таким чином, щоб за відсутності розсіяння $\delta $ дорівнювали 0. Дійсно, розглянемо рівняння для вільного
руху (центробіжний вклад в гамільтоніан залишається, оскільки є частиною кінетичної енергії руху). Вводячи функцію $R = \tilde {r}^{l}\chi $, отримуємо, що асимптотика хвильової функції задовольняє рівняння:
<!--  -->
\begin{equation}
\label{21eq8} \chi_{k,\,l}^{''} + \frac{2\,(l + 1)}{\tilde{r}}\,\chi'_{k,\,l} + \chi_{k,\,l} = 0
\end{equation}
<!--  -->
Взявши похідну від (\ref{21eq8}), отримуємо:
<!--  -->
\begin{equation}\label{21eq9} 
\chi^{'''}_{k,\,l} + \frac{2\,(l + 1)}{\tilde
{r}}\,\chi^{''}_{k,\,l} + \left(1 - \frac{2\,(l + 1)}{\tilde{r}^{\,2}} \right)\,\chi_{k,\,l} = 0
\end{equation}
<!--  -->
Зробимо заміну $u = \frac{\chi^{'}_{k,l}}{\tilde {r}}$, і перепишемо (\ref{21eq9}) у вигляді:
<!--  -->
\begin{equation}\label{21eq10} 
u'' + \frac{2\,(l + 2)}{\tilde {r}}\,u\,' + u = 0
\end{equation}
<!--  -->
що збігається з рівнянням (\ref{21eq8}) для величини $\chi_{k\,,\,l + 1} $. Тому:
<!--  -->
\begin{equation} \label{21eq11} 
\chi_{k,\,l} = \left(\frac{1}{\tilde{r}}
\frac{d}{d\,\tilde {r}} \right)^{l}\chi_{k,0}
\end{equation}
<!--  -->
Рівняння для функції $\chi_{k\,,\,0} $ дає:
<!--  -->
\begin{equation}\label{21eq12} 
\chi_{k,\,0} = \frac{1}{\tilde {r}}\left(
C_{0}^{(1)} \exp (i\,\tilde {r}) + C_{0}^{(2)}\,\exp(-i\tilde{r}) \right) = 
\frac{C_{0} \sin(\tilde {r})}{\tilde {r}}
\end{equation}
<!--  -->
фази амплітуд при експонентах обрані виходячи з умови скінченності розв'язку в 0.
Але, з іншого боку, розв'язок вільного рівняння Шредінгера є плоска
хвиля, тому:
<!--  -->
\begin{equation}\label{21eq13}
\begin{array}{l}
 \exp \left(i\tilde {r}\cos \theta \right) = 
 \sum\limits_{l=0}^{\infty}  C_{l} P_{l} (\cos \,\theta )\,
 \tilde {r}^{l}\left(\frac{1}{\tilde {r}}\,\frac{d}{d\,\tilde {r}}\right)^{l}\,
\frac{\sin \,\tilde{r}}{\tilde {r}} = 
\frac{1}{\tilde {r}}\,\sum\limits_{l = 0}^{\infty}
( - 1)^{l}\,C_{l}\sin\left(\tilde {r} - \frac{\pi \,l}{2} \right)
\,P_{l}(\cos \,\theta ) + o(1 / \tilde {r}) = \\
 \frac{1}{\,2\,i\,\tilde {r}}
 \sum\limits_{l = 0}^{\infty} i^{l}C_{l} \left(\exp(i\,\tilde {r}) +  
 (-1)^{l + 1}\exp ( -i\,\tilde {r}) \right)P_{l} (\cos \,\theta )+ o(1 / \tilde{r})
 \end{array}
\end{equation}
<!--  -->
Тут використане просте співвідношення:
<!--  -->
$$
\left(\frac{d}{d\,\tilde {r}} \right)^{l}\sin \,\tilde{r} =
(- 1)^{l}\sin \left(\tilde {r} -\frac{\pi \,l}{2} \right)
$$
<!--  -->
Таким чином, отриманий розклад плоскої хвилі на розбіжні і збіжні сферичні хвилі. На основі \eqref{21eq7} для асимптотики хвильової функції маємо:
<!--  -->
\begin{equation}
\label{21eq14}
\begin{array}{l}
\psi (\,\mathbf{r}\,) = \frac{2}{\tilde r}
\sum\limits_{l=0}^{\infty} a_{l} P_{l} (\cos \,\theta )
\sin \left(\tilde {r} - \frac{\pi \,l}{2} + \delta_{l}\right) = \\
\frac{1}{\,i\,\tilde {r}}
\sum\limits_{l = 0}^{\infty} (-i)^{l}\exp(- i\delta_{l} )\,
a_{l} P_{l} (\cos \,\theta) \left(\exp \left(i\,\tilde {r} + 2i\delta_{l} \right) + 
( - 1)^{l + 1}\exp \left(- i\,\tilde {r} \right) \right)
 \end{array}
\end{equation}
<!--  -->
Як видно, у розкладі \eqref{21eq14} з'явився вклад, пов'язаний із збіжними сферичними хвилями. Завдяки асимптотичної умови \eqref{21eq2} це може бути пов'язано тільки із відповідними вкладами у плоскій хвилі (початковий стан без розсіяння). Тому, коефіцієнти $a_{l} $ слід обрати такими:
<!--  -->
\begin{equation}\label{21eq15} 
a_{l} = \frac{(- 1)^{l}}{2}\,C_{l}\,\exp(i\,\delta_{l})
\end{equation}
<!--  -->
для того, щоб різниця \eqref{21eq13} і \eqref{21eq14} містила тільки розбіжні сферичні хвилі. Константи $C_{l}$ розкладу плоскої хвилі \eqref{21eq14} обчислюються за допомогою співвідношень для поліномів Лежандра і дорівнюють:
<!--  -->
\begin{equation}
\label{21eq16} C_{l} = ( - i)^{l}(2\,l + 1)
\end{equation}
<!--  -->
Звідки отримуємо остаточний вираз для амплітуди розсіяння $A$ (у розмірних змінних) у вигляді суми т.з. парціальних амплітуд $A_{l}$:
<!--  -->
\begin{equation}
\label{21eq17}
 A(\theta ) = \sum\limits_{l = 0}^{\infty}  (2\,l + 1)\,A_{l} P_{l} (\cos\,\theta )  = \frac{1}{2\,ik}\,\sum\limits_{l = 0}^{\infty} (2\,l + 1)\,(S_{l} - 1)
 P_{l} (\cos \,\theta)\\
A_{l} = \frac{1}{2\,ik}\left(\exp (2\,i\,\delta_{l} ) - 1 \right)\,\,\,,\,\,\,\,
S_{l} = \exp (2\,i\,\delta_{l} ) 
\end{equation}
<!--  -->
Ця формула виражає амплітуду розсіяння через фази $\delta $. Диференціальний переріз розсіяння визначається стандартним чином (див. Рис.) як відношення числа частинок, розсіяних у елемент тілесного кута d$\Omega $ до густини числа частинок у падаючому потоці:
<!--  -->
\begin{equation}
\label{21eq18} d\,\sigma (\theta ) = \vert A(\theta )\vert^{\,2}d\Omega = 2\pi \,\sin \theta \,\vert A(\theta )\vert^{\,2}d\theta
\end{equation}
<!--  -->
Використовуючи \eqref{21eq17}, отримуємо інтегральний переріз у вигляді суми парціальних:
<!--  -->
\begin{equation}\label{21eq19}
 \sigma = 2\pi \int\limits_{0}^{\pi}  
 \vert A(\theta)\vert^{\,2}
 \sin\theta \,d\theta   = 
 \sum\limits_{l = 0}^{\infty}
 \sigma_{l} 
 \end{equation}
<!--   -->
$$
\sigma_{l}  = 4\,\pi \,(2\,l + 1)\vert A_{l} \vert^{\,2}  = 
\frac{4\,\pi}{k^{\,2}}\,(2\,l + 1)\sin^{\,2}\delta_{l}
$$
<!--  -->
При виводі була використана умова нормування поліномів Лежандру:
<!--  -->
$$
2\pi {\int\limits_{0}^{\pi}  {P_{l} (\cos \theta )P_{l\,'} (\cos
\theta )\sin \theta \,d\theta}}   = \frac{4\pi} {2l + 1}
\delta_{l,\,l\,'}
$$

Таким чином, формули \ref{21eq17}-\eqref{21eq19} встановлюють зв'язок спостережувальних величин - перерізів розсіяння з фазами, які несуть всю інформацію про процес розсіяння. Зазначимо, що безпосередньо з \eqref{21eq17} і \eqref{21eq19} витікає зв'язок між повним перерізом розсіяння \eqref{21eq19} і мнімою частиною розсіяння "вперед", тобто при куті розсіяння $\theta = 0$:
<!--  -->
\begin{equation}\label{opt}
  \sigma = \frac{4\,\pi}{k}\, {\rm Im}\, A(\theta = 0)\,.
\end{equation}
<!--  -->
Це співвідношення носить назву <a id="optheorem">__оптичної теореми__</a>. З фізичних міркувань зрозуміло, що величини фаз $\delta $ повинні прямувати до 0 з ростом $l$, оскільки рух при великих квантових числах є квазикласичним, і суттєво квантові інтерференційні ефекти слабкі. Тому повний переріз розсіяння (\ref{21eq17}) скінчений, тільки якщо збігається ряд:
<!--  -->
$$
\sum\limits_{l \gg 1}^{\infty}  l\,\delta_{l}^{\,2}  < \infty
$$
<!--  -->
Оскільки мова йде про великі $l$, доцільно використати квазикласичний вираз для фаз:
<!--  -->
\begin{equation}\label{21eq20} 
\delta _{l} \approx \int\limits_{\tilde {r}_{0}}
^{\infty}  \sqrt{1 - \frac{l^{\,2}}{\tilde {r}^{\,2}} -
\frac{2m\,U(\tilde {r})}{\hbar^{\,2}}} \,d\,\tilde{r} \, -
\,\int\limits_{\tilde {r}_{0}}^{\infty} \sqrt{1 -
\frac{l^{\,2}}{\tilde {r}^{\,2}}} \,d\,\tilde {r}  \approx -
\frac{1}{\hbar^{\,2}}\int\limits_{\tilde{r}_{0}}^{\infty}
\frac{m\,U(\tilde {r})}{\sqrt{1 - \frac{l^{\,2}}{\tilde{r}^{\,2}}}}\,d\,\tilde{r}
\end{equation}

Фази скінченні, якщо потенціал зникає швидше, ніж $\frac{1}{r}$. Оскільки

\begin{equation}
\label{21eq21} \sum\limits_{l \gg 1}^{\infty}  {l\delta_{l}
^{\,2}} < \int\limits_{1}^{\infty} x\,\delta^{\,2}(x)\,dx
\approx \int\limits_{1}^{\infty}  x^{3}U^{\,2}(x)\,d\,x \,
\end{equation}

то повний переріз є скінченим якщо

\begin{equation}
\label{21eq22} \,\,U(x) = \frac{1}{x^{n}}\,,\quad n\, > \,2
\end{equation}

Ці результати дозволяють кількісно охарактеризувати клас "швидкозникаючих" потенціалів, для яких вказані наближення про асимптотичну поведінку хвильової функції мають сенс.

## Загальні властивості амплітуди розсіяння.

Амплітуда пружного розсіяння у довільному полі задовольняє певним співвідношенням, які витікають з загальних фізичних міркувань. Запишемо асимптотику хвильової функції при пружному розсіянні (у безрозмірних змінних $\tilde{k} = k\,r\,,\,\,\tilde{A}= A\,k$):
<!--  -->
\begin{equation}\label{scateq1} 
\psi (\,\mathbf{r}\,) \to \,\exp \left(i\,\tilde{r}\,\mathbf{n}' \cdot \mathbf{n}
\right) + \tilde{A}(\mathbf{n},\mathbf{n}')
\frac{\exp\left( i\,\tilde{r} \right)}{\tilde {r}}
\end{equation}
<!--  -->
де $\mathbf{n}$ і $\mathbf{n}'$ - одиничні вектори напрямків
падаючої і розсіяної хвиль відповідно. Узагальнимо ситуацію,
уявив, що падаюча хвиля є суперпозицією станів з фіксованою
енергією але різними напрямками імпульсу, тобто
<!--  -->
\begin{equation}
\label{scateq2} \psi (\,\mathbf{r}\,) \to \,\frac{1}{2\,\pi\,i}\int
F(\mathbf{n})
\left( \exp \left(i\,\tilde{r}\,\mathbf{n}' \cdot \mathbf{n} \right) + 
\tilde{A}(\mathbf{n},\mathbf{n}')
\frac{\exp \left( i\,\tilde{r} \right)}{\tilde {r}} \right)\,d\,\mathbf{n}
\end{equation}
<!--  -->
Множник попереду інтегралу обраний з міркувань зручності у подальшому представленні. На
великій відстані перший член швидко осцилює, тому основний вклад
дають тільки точки стаціонарної фази $\mathbf{n} \cdot \mathbf{n}'
= \pm 1$:
<!--  -->
\begin{equation}
\label{scateq3} \psi (\,\mathbf{r}\,) \to
\,F(-\mathbf{n}')\,\frac{\exp \left(- i\,\tilde {r}
\right)}{\tilde {r}} - F(\mathbf{n}')\,\frac{\exp \left(i\,\tilde {r} \right)}{\tilde{r}} - \frac{i\exp \left( i\,\tilde {r}\right)}{2\pi \,\tilde {r}}
\int \tilde{A}(\mathbf{n},\mathbf{n}'\,)F(\mathbf{n})\,d\mathbf{n}
\end{equation}
<!--  -->
У розмірних змінних, введемо оператор:
<!--  -->
\begin{equation}
\label{scateq4} \hat S = 1 + 2\,i\,k\,\hat{f}\,,\quad \hat{f}\,[F(\mathbf{n}')] = 
\frac{1}{4\,\pi} \int A(\mathbf{n},\mathbf{n}'\,)F(\mathbf{n})\,d\,\mathbf{n}
\end{equation}
<!--  -->
за допомогою якого (\ref{scateq3}) можна записати у вигляді
<!--  -->
\begin{equation}
\label{scateq5} \psi (\,\mathbf{r}\,) \to \,F(-\mathbf{n}')
\frac{\exp \left(- i\,\tilde {r} \right)}{\tilde{r}}\, - 
\frac{\exp \left( i\,\tilde {r} \right)}{\tilde{r}}
\hat S (\,F(\mathbf{n}'))
\end{equation}
<!--  -->
Оператор $\hat S$ був введений Гейзенбергом і називається __оператором (матрицею) розсіяння__ . Його фізичний зміст витікає з його означення \eqref{scateq4} - це оператор, який переводить стан падаючої хвилі (початковий in-стан) у суперпозицію станів падаючої і розсіяної хвилі (кінцевий out-стан). Оператор
<!--  -->
\begin{equation}\label{t}
\hat T =\hat S - 1
\end{equation}
<!--  -->
називають __оператором переходу__. Завдяки збереженню числа частинок оператор розсіяння є унітарним:
<!--  -->
\begin{equation} \label{scateq6} 
\hat S^{\dagger} \hat S = 1\,\, \Rightarrow \,\,f -
f^{\dagger}  = 2i\,k\,f\,f^{\dagger}
\end{equation}
<!--  -->
В матричних елементах \eqref{scateq6} має вигляд:
<!--  -->
\begin{equation}\label{scateq7}
f(\mathbf{n},\mathbf{n}') - f^*(\,\mathbf{n}',\mathbf{n}\,) = 
\frac{i\,k}{2\,\pi}\,\int f(\mathbf{n},\mathbf{n}')\,f^*
(\mathbf{n}',\mathbf{n}'')d\,\mathbf{n}''
\end{equation}
<!--  -->
або
<!--  -->
\begin{equation}\label{scateq8} 
{\rm Im}f(\mathbf{n},\mathbf{n}) =
\frac{\,k}{4\,\pi}\,\int \vert f(\mathbf{n},\mathbf{n}'')\vert
^{\,2}d\,\mathbf{n}'' = \frac{\,k}{4\,\pi}
\sigma
\end{equation}
<!--  -->
Звісно це є та ж сама [__оптична теорема__](#optheorem) \eqref{opt}, яка була отримана раніше. Інше важливе співвідношення витікає з требования симетрії відносно обернення часу. Як відомо, якщо існує хвильова функція $\psi $, яка описує певний квантовий процес, то хвильова функція $\psi^*$ описує процес звернений у часі, якщо гамільтоніан є дійсним оператором, тобто $H^{*} = H$. Тому:
<!--  -->
\begin{equation}\label{scateq9}
 \psi^*(\,\mathbf{r}\,) \to \,F^*( -\mathbf{n}')
 \frac{\exp\left(i\,\tilde {r}\right)}{\tilde {r}}\, - 
 \frac{\exp \left(- i\,\tilde {r} \right)}{\tilde{r}} 
 \hat S^* [F^* (\mathbf{n}')] \\ 
 \hat S^* = 1 - 2\,i\,k\,\int A^*(\mathbf{n}',\mathbf{n})\,\ldots\,d\mathbf{n}\,.
\end{equation}
<!--  -->
Введемо позначення
<!--  -->
\begin{equation}
\label{scateq10}
\begin{array}{l}
 \,\hat S^*
 [\,F^*(\mathbf{n}')] = - \,\Phi ( -\mathbf{n}')\,\,\,
 \Rightarrow \,\,\,F^* (\mathbf{n}') = -
\hat S^*{^{- 1}}\Phi ( -\mathbf{n}') = - \left( {\hat S^{\dagger}}
\right)^*\,\Phi ( -\mathbf{n}') \\
 \hat S^* = 1 - 2\,i\,k\,
 \int A^* (\mathbf{n}',\mathbf{n})\,\ldots\,d\mathbf{n} \\
 \end{array}
\end{equation}
<!--  -->
(крапки стоять замість функція вектора $\mathbf{n}$ - аргументу інтегрального оператора) і введемо оператор інверсії відносно начала координат (центр розсіяння):
<!--  -->
$$
F( - \mathbf{n}) = P\,F(\mathbf{n})
$$
<!--  -->
тоді
<!--  -->
\begin{equation}
\label{scateq11} \psi^* (\,\mathbf{r}\,) \to \,\frac{
\exp\left(-i\,\tilde {r}\right)}{\tilde{r}}\,
\Phi ( -\mathbf{n}') -
\frac{\exp \left( i\,\tilde {r} \right)}{\tilde{r}}\,\,P\left(\hat S^{\dagger} \right)^* P\,\Phi \,(\mathbf{n}')
\end{equation}
<!--  -->
По суті \eqref{scateq11} аналогічно \eqref{scateq5} і, як
наслідок, також описує певний процес розсіяння, якщо покласти
<!--  -->
\begin{equation}
\label{scateq12} \,\,P\,\,\left( \hat{S}^{\dagger}\right)^* P\, =
\hat S
\end{equation}
<!--  -->
Записуючи \eqref{scateq12} через матричні елементи отримуємо
співвідношення:
<!--  -->
\begin{equation}
\label{scateq13} S(\mathbf{n},\mathbf{n}') = S( -\mathbf{n}',
-\mathbf{n})\,\,\, \Rightarrow A(\mathbf{n},\mathbf{n}') = A( -
\mathbf{n}', - \mathbf{n})
\end{equation}
<!--  -->
яке називається __теоремой взаємності__ і встановлює рівність амплітуд розсіяння для процесів, обернених у часі друг до друга. Відмітимо, що для розсіяння у центральному полі співвідношення \eqref{scateq13} виконується, оскільки всі величини залежать тільки від кута між напрямками падаючої і розсіяної хвилі.
