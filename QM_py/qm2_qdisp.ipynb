Нагадаємо деякі факти з класичної електродинаміки про дисперсію світла.

Дисперсією називається залежність діелектричної проникності (показника заломлення) від частоти ЕМХ. Визначення цієї залежності засноване на представленні про мікроскопічну структуру середовища, оскільки зводиться до обчислення поляризовності атома. У класичній ЕД, виходячи з уявлення про електрон як осцилятор певної частоти отримано наступне співвідношення між індукованим дипольним моментом і полем:
$$
 \mathbf{\ddot {r}} = - \omega_{0}^{2}{\mathbf{r}} -
  \frac{e}{m}\,\mathbf{E}_{\omega}\,e^{i\,\omega \,t}\, \Rightarrow \mathbf{d}_{\omega}  = -e\,\mathbf{r}_{\omega}\\ 
  \mathbf{d}_{\omega}=
\frac{e^{2} / m}{\omega_{0}^{2} - \omega^{2}}{\mathbf{E}}_{\omega}\,, 
$$
Для системи таких осциляторів з резонансними частотами $\omega_k$ будемо мати:

$$ \alpha_{\omega} = \frac{e^{2}}{m}\sum\limits_{k}\,\frac{n_{k}}{\omega_{k}^{2} - \omega^{2}}$$

де $\alpha $ - поляризованість системи електронів, які розглядаються як незалежні, $n_{k} $ - сила осцилятора (число електронів, ``резонуючих`` на частоті $\omega_{{\rm k}}$. 
З врахуванням процесів втрати енергії більш адекватною є модель осцилятора із ''тертям''

$$
\mathbf{\ddot {r}} = - \omega_{0}^{\,2} {\mathbf{r}} - \gamma\,\mathbf{\dot {r}} - 
\frac{e}{m}\,\mathbf{E}\,e^{i\omega \,t} \Rightarrow
\alpha_{\omega}  = \frac{e^2}{m}\,\sum \frac{n_k}{\omega^2_k - \omega^2 -i\,\gamma\,\omega}
$$

За поляризацією визначається діелектрична проникність $\varepsilon$. Наприклад, для розріджених систем (газ):
$$
\varepsilon = 1 + 4\pi \,n\,\alpha
$$
так що наприклад
$$
{\rm Re}\,\varepsilon = 1 + 4\pi \,\frac{e^2}{m}\,\sum \frac{n_k\,(\omega^2_k - \omega^2)}{(\omega^2_k - \omega^2)^2 +\gamma^2\,\omega^2}
$$
При більш високих густинах використовується формула Клаузіуса-Мосоті:
$$
\frac{\varepsilon - 1}{\varepsilon + 2} = 
\frac{4\pi}{3}\,n\,\alpha
$$

Коли частота падаючого випромінювання далека від резонансних частот $\frac{d\,\varepsilon}{d\,\omega}>0$ - нормальна дисперсія, яка спостерігається у області частот де речовину можна вважати прозорою. Коли $\omega\to \omega_k$ $\frac{d\,\varepsilon}{d\,\omega}<0$ то говорять про аномальну дисперсію. Це означає, що речовина непрозора і значним чином поглинає ЕМ хвилі. 

Порівняння з експериментом показало, що класична формула якісно правильно передає залежність $\alpha$ від частоти світлової хвилi $\omega$, зокрема у граничних випадках $\omega \to 0\,, \omega \to \infty$.  Однак, для кількісно задовільних результатів виявилося необхідним вводити т.з. ``сили осциляторів`` $n_k \to n_k\,f_k$ із значеннями $f_k<1$ що означало певну долю осциляторів. В деяких випадках i взагалі необхідно було покладати $f_k < 0$. Пояснення і обчислення таких величин надала квантова теорія.

Обчислення поляризованості квантової системи доцільно проводити методами нестаціонарної теорії збурень. Припустимо, що у початковий момент часу на квантову систему (атом), який знаходиться у стаціонарному стані, падає монохроматична ЕМХ. У першому порядку теорії збурень зміна стану визначається як:

\begin{equation}
	\label{poleq1} \psi (t) = \varphi_{0} \exp(- i\omega_{0} t) + \sum\limits_{n}{c^{(1)}_{n} (t)\,\varphi
		_{n} \exp ( - i\omega_{n} t)}
\end{equation}
тут $\varphi_{0}$, наприклад, х.ф. основного стану і $\varphi_n$ - координатні  частини хвильових функцій стаціонарних станів. При цьому оператором збурення є:
де $\varphi_n$ - координатні  частини хвильових функцій стаціонарних станів. При цьому оператором збурення є:

\begin{equation}
	\label{poleq2}
	\begin{array}{l}
		U_{int} = \frac{e}{m\,c}\mathbf{A}(t) \cdot \mathbf{p} =
		\,W\cos (\omega \,t) \\
		W = \frac{e}{m\,\omega}\,\mathbf{E}_{0} \cdot \mathbf{p} \,,\,\,
		\mathbf{A}(t) =  \frac{c}{\omega}\,\mathbf{E}_{0} \cos
		(\omega \,t) 
	\end{array}
\end{equation}

Тут ми скористалися стандартним електродинамічним співвідношенням:
$$
\mathbf{E} = - \frac{1}{c}\,\frac{\partial
	\,\mathbf{A}}{\partial \,t} = {\mathbf{E}}_{0} \sin
\,(\omega \,t)
$$

Використовуючи вираз для амплітуд $c_{n} $ в лінійному наближенні за оператором збурення (див. \eqref{eq:nonstac_an}), отримуємо:

$$
c^{(1)}_{n}(t) =\frac{1}{i\,\hbar} \int_{0}^{t} \av{\varphi_n}{U_{int}(t')}{\varphi_0}\,e^{i\,\omega_{n,0}t'}dt'
$$
Для середнього значення індукованого дипольного моменту:

\begin{equation}
	\label{poleq4}
	\av{\psi}{\mathbf{d}_{ind}}{\psi} =  
	- e(\av{\psi}{\mathbf{r}}{\psi} -
	\av{\varphi_0}{\mathbf{r}}{\varphi_0})
\end{equation}

отримуємо: 
$$
\av{\psi}{\mathbf{d}_{ind}}{\psi} = -e\, \sum\limits_{n\ne 0}\,\left(c^{(1)}_n\, \av{\varphi_0}{\mathbf{r}}{\varphi_n} e^{i\,\omega_{0,n}t}+ \text{conj} \right) =
$$
<!--  -->
$$
\frac{e^2}{2\,\hbar\,m\,\omega}\sum\limits_{n} \left(\frac{\av {\varphi_{n}}{\mathbf{E}_0\cdot \mathbf{p}}{\varphi_{0}}\,\av{\varphi_0}{\mathbf{r}}{\varphi_n}}{\omega_{n,0}-\omega} -\frac{\av {\varphi_{0}}{\mathbf{E}_0\cdot \mathbf{p}}{\varphi_{n}}\,\av{\varphi_n}{\mathbf{r}}{\varphi_0}}{\omega_{n,0}+\omega}\right)
e^{-i\,\omega\,t}+\text{conj}
$$

Оскільки:

$$
\av{\varphi_{n}}{\mathbf{p}}{\varphi_{0}}  = \frac{m}{i\,\hbar}\av{\varphi_{n}}
{\left[\mathbf{r},H_{0} \right]}{\varphi_{0}} = 
-i\,m\,\omega_{0,n}\,\av{\varphi_{n}}{\mathbf{r}}{\varphi_{0}} 
$$
то остаточно маємо для поляризованості вздовж х наприклад :
$$
\langle d_{x}\rangle = \alpha_{xx}\, E_{x}
$$
$$
\alpha_{xx} = \frac{e^2}{\hbar}\sum\limits_{n}\frac{\omega_{n,0}}{\omega}\,|\av{\varphi_0}{x}{\varphi_n}|^2\left(\frac{1}{\omega_{n,0}-\omega}-\frac{1}{\omega_{n,0}+\omega}\right) =\frac{2\,e^2}{\hbar}\sum\limits_{n}\frac{\omega_{n,0}\,|\av{\varphi_0}{x}{\varphi_n}|^2}{\omega^2_{n,0}-\omega^2} 
$$

Порівнюючи із класичним визначенням, введемо величину
$$
f_{n,0} = \frac{2\,m\,\omega_{n,0}}{\hbar}\,|\av{\varphi_0}{x}{\varphi_n}|^2
$$
яка представляє ''силу осцилятора'' у квантовій теорії.
