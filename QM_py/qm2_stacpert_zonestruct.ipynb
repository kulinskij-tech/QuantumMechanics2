## Зонна структура спектру частинкі у періодичному полі у рамках теорії збурень
<!--  -->

$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$

З точки зору теорії збурень можливий також розгляд зонної структури спектра електрона у кристалі. Періодичне поле кристалічної гратки возглядаємо як збурення.

 \begin{equation}\label{hcryst}
   \hat{H} = \hat{H}_0 + V(\mathbf{r})\,,\qquad \hat{H}_0 =
   \frac{\mathbf{p}^2}{2\,m}\,,\quad V(\mathbf{r}) = V(\mathbf{r} + \mathbf{a})\,,
 \end{equation}

де $\mathbf{a}$ - період кристалічної гратки. Такий підхід, із зрозумілих причин називається наближенням слабкого зв'язку. Трансляційна симетрія кристала накладає певні  умови, які відображають симетрію відносно зсувів по періодах гратки. Оскільки:
 
 $$\left[\,\hat{H}, \hat{T}_{\mathbf{a}}
 \,\right] = 0\,,$$ 
 
то для стаціонарних станів $\ket{E}$ має місце:

 \begin{equation}\label{wb}
 \left\langle\,\mathbf{r}\,\right\vert
 \hat{T}_{\mathbf{a}}\left|\,E\,\right\rangle = \left\langle\,\mathbf{r}
 -\mathbf{a}\,\right.\left|\,E\,\right\rangle\,.
 \end{equation}

\begin{equation}\label{wb1}
 \left\langle\,\mathbf{r}\,\right. \left|\,E\,\right\rangle = \int
 \left\langle\,\mathbf{r}\,\right.\left|\,\mathbf{k}\,\right\rangle\,
 \left\langle\,\mathbf{k}\,\right.\left|\,E\,\right\rangle\,d\mathbf{k}
 \end{equation}
 
Базисом незбурених хвильових функцій є набір хвильових функцій частинки у ящику з періодичними граничними умовами на границах. Гамільтоніан задачі має вигляд:
 
$$\hat{H} = -\frac{\hbar^2}{2\,m}\,\frac{ d^2}{d\,x^2}+\hat{V}(x)\,,\quad V(x+a) = V(x)$$
 
Завдяки періодичности:
 
$$\av{x}{\hat{T}_{a}}{E} = \braket{x-a}{E} = \psi_{E}(x-a) = e^{-i\,k\,a}\psi_{E}(x)$$
 
що дозволяє представити хвильову функцію стаціонарного стану $\psi_{E}(x)$ як (теорема Блоха):
 
$$\psi_{E(k)}(x) = e^{i\,k\,x}\,u_{k}(x)\,,\quad u_{k}(x) = u_{k}(x+a)$$
 
Величину $k$ називають квазіімпульсом. Очевидно, енергія є періодичною функцією $k$ з періодом оберненої гратки $k\to k+q_n\,,\,\,\,q_n=\frac{2\,\pi}{a}\,n$. Якщо $\varepsilon(k)$ - енергія стаціонарного стану, то функція $u_{k}(x)$ задовільняє рівнянню:
 
$$\left(-\frac{\hbar^2}{2\,m}\,\left(\frac{ d}{d\,x}+ i\,k\right)^2 - \varepsilon(k)+V(x)\right)u_{k}(x)=0$$
 
яке можна розвязувати (при кожному $k$) у фіксованої ячійці, наприклад $0<x<a$. Покладемо $V(x) = 0$, тоді:
<!--  -->
$$u_{n}(x) = \frac{1}{\sqrt{a}}\,e^{i\,q_n\,x}\,,\quad \varepsilon^{(0)}_{n}(k) =\frac{\hbar^2
}{2\,m}\left(k+q_n\right)^2 $$

З цього результата видно що в точках $k=0,q_n$ має місце виродження рівнів оскільки значеню енергії $\varepsilon^{(0)}_{m}(q_n)$ відповідають хвильові функції з різними
значеннями квазіімпульсу, в частности для $k=0$ это стани с противоположними значеніями $q_n$ (см. Рис.~\ref{fig_zone}). Тепер враховуємо потенціал за теорії збурень. Для невирожденного випадка:
<!--  -->
$$\varepsilon_{n}(k) = \varepsilon^{(0)}_{n}(k)+\av{k+q_n}{\hat{V}}{k+q_n}+
\sum\limits_{m\ne n} \frac{\left|\av{k+q_n}{\hat{V}}{k+q_m}\right|^2}{\varepsilon^{(0)}_{n}(k) - \varepsilon^{(0)}_{m}(k)}$$
<!--  -->
Перший внесок це просто сталий зсув рівней. Другий - демонструє тенденцію до "відштовхування" рівней з близькими значеннями енергії на краях зон. В результаті цього на краях зон виникає енергетичні щілини - заборонені інтервали енергій, які межують дозволені зони енергій. Визначимо розщепленя в рамках теорії збурень для вирожденного рівня. Розглянемо наприклад край зони Бриллюена $k=\pi/a$. Існує 2-кратне виродженя з незбуреними (блохівськими) хвильовими функціями (нормуемо на елементарну ячійку):
<!--  -->
$$u^{(0)}_{0,k}(x) = \frac{1}{\sqrt{a}}\,,\quad u^{(0)}_{1,k}(x)=\frac{e^{i\,q_1\,x}}{\sqrt{a}}\,,\,\,q_1 = -\frac{2\pi}{a}$$
<!--  -->
Матриця оператора збурення на незбурених станах має вигляд:
<!--  -->
\begin{equation}
	V = \begin{pmatrix}
	V_0&V_1\\
	V_1&V_0
	\end{pmatrix}\,,\label{periodpotmtrx}
\end{equation}
<!--  -->
де
<!--  -->
$$V_{0} = \frac{1}{a}\int\limits_{0}^{a}\,V(x)\,dx
\,,\quad
V_{1} = \frac{1}{a}\int\limits_{0}^{a}\,V(x)\,e^{-i\,q_{1}\,x}\,dx
$$
первші коефіцієнти Фур'є-розклад періодичної функції потенціала. Обчислюючи власні значення \eqref{periodpotmtrx} отримаємо:
<!--  -->
$$E^{(1)}_{\pm}(k) = \varepsilon^{(0)}_{n}(\pi/a)+V_0\pm V_{1}\,.$$
<!--  -->
Т.ч. у першому порядку розщеплення зон виникає за рахунок відповідних фур'є-гармонік потенціала періодичної структури.
