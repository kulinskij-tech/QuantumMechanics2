## Зонна структура спектру частинки у періодичному полі у рамках теорії збурень
<!--  -->
$\newcommand{\ket}[1]{\left.\left\vert\, #1\,\right. \right \rangle}$
$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$
$\newcommand{\av}[3]{\left\langle \,#1 \left \vert\, #2 \,\right\vert #3 \,\right \rangle}$
<!--  -->

З точки зору теорії збурень можливий також розгляд зонної структури спектра електрона у кристалі. Періодичне поле кристалічної ґратки розглядаємо як збурення.

 \begin{equation}\label{hcryst}
   \hat{H} = \hat{H}_0 + V(\mathbf{r})\,,\qquad \hat{H}_0 =
   \frac{\mathbf{p}^2}{2\,m}\,,\quad V(\mathbf{r}) = V(\mathbf{r} + \mathbf{a})\,,
 \end{equation}

де $\mathbf{a}$ - період кристалічної ґратки. Такий підхід, із зрозумілих причин називається наближенням слабкого зв'язку. Трансляційна симетрія кристала накладає певні  умови, які відображають симетрію відносно зсувів по періодах ґратки. Оскільки:
 
 $$\left[\,\hat{H}, \hat{T}_{\mathbf{a}}
 \,\right] = 0\,,$$ 
 
то для стаціонарних станів $\ket{E}$ має місце:

 \begin{equation}\label{wb}
 \left\langle\,\mathbf{r}\,\right\vert
 \hat{T}_{\mathbf{a}}\left|\,E\,\right\rangle = \left\langle\,\mathbf{r}
 -\mathbf{a}\,\right.\left|\,E\,\right\rangle\,.
 \end{equation}

\begin{equation}\label{wb1}
 \left\langle\,\mathbf{r}\,\right. \left|\,E\,\right\rangle = \int
 \left\langle\,\mathbf{r}\,\right.\left|\,\mathbf{k}\,\right\rangle\,
 \left\langle\,\mathbf{k}\,\right.\left|\,E\,\right\rangle\,d\mathbf{k}
 \end{equation}
 
Базисом незбурених хвильових функцій є набір хвильових функцій частинки у ящику з періодичними граничними умовами на границах. Гамільтоніан задачі має вигляд:
 
$$\hat{H} = -\frac{\hbar^2}{2\,m}\,\frac{ d^2}{d\,x^2}+\hat{V}(x)\,,\quad V(x+a) = V(x)$$
 
Завдяки періодичности:
 
$$\av{x}{\hat{T}_{a}}{E} = \braket{x-a}{E} = \psi_{E}(x-a) = e^{-i\,k\,a}\psi_{E}(x)$$
 
що дозволяє представити хвильову функцію стаціонарного стану $\psi_{E}(x)$ як (теорема Блоха):
 
$$\psi_{E(k)}(x) = e^{i\,k\,x}\,u_{k}(x)\,,\quad u_{k}(x) = u_{k}(x+a)$$
 
Величину $k$ називають квазіімпульсом. Очевидно, енергія є періодичною функцією $k$ з періодом оберненої ґратки $k\to k+q_n\,,\,\,\,q_n=\frac{2\,\pi}{a}\,n$. Якщо $\varepsilon(k)$ - енергія стаціонарного стану, то функція $u_{k}(x)$ задовольняє рівняння:
 
$$\left(-\frac{\hbar^2}{2\,m}\,\left(\frac{ d}{d\,x}+ i\,k\right)^2 - \varepsilon(k)+V(x)\right)u_{k}(x)=0$$
 
яке можна розвязувати (при кожному $k$) у фіксованої комірці, наприклад $0<x<a$. Покладемо $V(x) = 0$, тоді:
<!--  -->
$$u_{n}(x) = \frac{1}{\sqrt{a}}\,e^{i\,q_n\,x}\,,\quad \varepsilon^{(0)}_{n}(k) =\frac{\hbar^2
}{2\,m}\left(k+q_n\right)^2 $$
<!--  -->
<img src="./figs/fig_zone.png" alt="Image" width="400">
<img src="./figs/fig_zonesplit.png" alt="Image" width="400">
<!--  -->
З цього результату видно, що в точках $k=0,q_n$ має місце виродження рівнів оскільки значенню енергії $\varepsilon^{(0)}_{m}(q_n)$ відповідають хвильові функції з різними
значеннями квазіімпульсу, зокрема для $k=0$ це стани з протилежними значеннями $q_n$ (див. Рис.~\ref{fig_zone}). Тепер враховуємо потенціал за теорії збурень. Для невиродженого випадку:
<!--  -->
$$\varepsilon_{n}(k) = \varepsilon^{(0)}_{n}(k)+\av{k+q_n}{\hat{V}}{k+q_n}+
\sum\limits_{m\ne n} \frac{\left|\av{k+q_n}{\hat{V}}{k+q_m}\right|^2}{\varepsilon^{(0)}_{n}(k) - \varepsilon^{(0)}_{m}(k)}$$
<!--  -->
Перший внесок це просто сталий зсув рівнів. Другий - демонструє тенденцію до "відштовхування" рівнів з близькими значеннями енергії на краях зон. В результаті цього на краях зон виникають енергетичні щілини - заборонені інтервали енергій, які межують дозволені зони енергій. Визначимо розщеплення в рамках теорії збурень для виродженого рівня. Розглянемо наприклад край зони Бриллюена $k=\pi/a$. Існує 2-кратне виродження з незбуреними (блохівськими) хвильовими функціями (нормуємо на елементарну комірку):
<!--  -->
$$u^{(0)}_{0,k}(x) = \frac{1}{\sqrt{a}}\,,\quad u^{(0)}_{1,k}(x)=\frac{e^{i\,q_1\,x}}{\sqrt{a}}\,,\,\,q_1 = -\frac{2\pi}{a}$$
<!--  -->
Матриця оператора збурення на незбурених станах має вигляд:
<!--  -->
\begin{equation}
	V = \begin{pmatrix}
	V_0&V_1\\
	V_1&V_0
	\end{pmatrix}\,,\label{periodpotmtrx}
\end{equation}
<!--  -->
де
<!--  -->
$$V_{0} = \frac{1}{a}\int\limits_{0}^{a}\,V(x)\,dx
\,,\quad
V_{1} = \frac{1}{a}\int\limits_{0}^{a}\,V(x)\,e^{-i\,q_{1}\,x}\,dx
$$
перші коефіцієнти Фур'є-розкладу періодичної функції потенціалу. Обчислюючи власні значення \eqref{periodpotmtrx} отримаємо:
<!--  -->
$$E^{(1)}_{\pm}(k) = \varepsilon^{(0)}_{n}(\pi/a)+V_0\pm V_{1}\,.$$
<!--  -->
Т.ч. у першому порядку розщеплення зон виникає за рахунок відповідних фур'є-гармонік потенціалу періодичної структури.

## Графен: симетрія, зонна структура та точки Дірака

У попередньому прикладі періодичне поле змішує два вироджені стани на межі
зони Бриллюена та відкриває щілину. Для загального комплексного матричного
елемента $V_1$ ермітового матриця має позадіагональні елементи $V_1,V_1^*$,
а величина розщеплення становить $2|V_1|$; відповідним вибором фаз базису
$V_1$ можна зробити дійсним. Але чи завжди періодичність означає наявність
щілини між сусідніми зонами?

Приклад графену показує, що відповідь залежить від внутрішньої структури
елементарної комірки та симетрії. Розглянемо послідовність
**ґратка → двокомпонентний стан Блоха → дві зони → точки Дірака → відкриття щілини**.

Тут використаємо **наближення сильного зв'язку**: базис утворюють локалізовані
$p_z$-орбіталі атомів вуглецю, а перехід електрона між сусідніми атомами
задається матричним елементом $-t$, де $t>0$. Це інший вихідний базис, ніж
плоскі хвилі в попередньому наближенні майже вільних електронів. Спільною
математичною ідеєю є діагоналізація малої матриці в потрібному підпросторі;
нижче її локальний розклад поблизу виродження дасть ефективний гамільтоніан.

**Припущення моделі.** Ідеальний нескінченний моношар, ортонормований базис
з однією $p_z$-орбіталлю на атом, однакові найближчі переходи $t$ і рівні
локальні енергії A/B. Нехтуємо перекриттям орбіталей, далекими переходами,
взаємодією електронів та спін-орбітальним зв'язком. Реальний спін лише подвоює
кількість станів і в матриці не виписаний. Наведені матриці Паулі діють
**на індекс підґратки A/B**, а не на реальний спін.

### 1. Геометрія: дві підґратки, а не дві незалежні зони

Стільникова структура не є ґраткою Браве: її можна подати як трикутну ґратку
Браве з двома атомами A і B у комірці. Позначимо через $a$ **відстань між
найближчими атомами**, а не період трикутної ґратки ($\sqrt{3}a$).
Оберемо вектори від атома A до трьох сусідніх атомів B:

$$
\boldsymbol{\delta}_1=a(0,1),\quad
\boldsymbol{\delta}_2=a(-\sqrt3/2,-1/2),\quad
\boldsymbol{\delta}_3=a(\sqrt3/2,-1/2).
$$

Тут і далі координати мають одну й ту саму орієнтацію. Вектори трансляцій
та оберненої ґратки можна вибрати як

$$
\mathbf a_1=\boldsymbol\delta_1-\boldsymbol\delta_2,
\quad \mathbf a_2=\boldsymbol\delta_1-\boldsymbol\delta_3,
\qquad
\mathbf b_1=\left(\frac{2\pi}{\sqrt3 a},\frac{2\pi}{3a}\right),
\quad
\mathbf b_2=\left(-\frac{2\pi}{\sqrt3 a},\frac{2\pi}{3a}\right),
$$

де $\mathbf a_i\cdot\mathbf b_j=2\pi\delta_{ij}$. Дві нееквівалентні
вершини першої зони Бриллюена:

$$
\mathbf K=\frac{\mathbf b_1-\mathbf b_2}{3}
=\left(\frac{4\pi}{3\sqrt3 a},0\right),\qquad
\mathbf K'=-\mathbf K.
$$

Інші чотири вершини еквівалентні K або K' з точністю до вектора оберненої
ґратки. Для графіків нижче використовуємо безрозмірні координати
$x/a,y/a$ та $ak_x,ak_y$, тобто в коді задаємо $a=1$.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

a = 1.0  # одиниця довжини: відстань C–C
delta = a * np.array([[0, 1], [-np.sqrt(3)/2, -0.5], [np.sqrt(3)/2, -0.5]])
a1, a2 = delta[0] - delta[1], delta[0] - delta[2]
b1 = np.array([2*np.pi/np.sqrt(3), 2*np.pi/3]) / a
b2 = np.array([-2*np.pi/np.sqrt(3), 2*np.pi/3]) / a
K = (b1 - b2)/3
Kp = -K
Gamma = np.zeros(2)
M = (b1 + b2)/2
vertices = np.array([K, [K[0]/2, M[1]], [-K[0]/2, M[1]],
                     Kp, [-K[0]/2, -M[1]], [K[0]/2, -M[1]]])
plt.rcParams.update({'font.size': 11, 'figure.dpi': 120})
fig, (ax, bx) = plt.subplots(1, 2, figsize=(10, 4.5), constrained_layout=True)
sites = np.array([i*a1 + j*a2 for i in range(-2, 3) for j in range(-2, 3)])
for r in sites:
    for d in delta:
        ax.plot([r[0], r[0]+d[0]], [r[1], r[1]+d[1]], color='0.75', lw=1, zorder=0)
ax.scatter(*sites.T, s=38, color='#0072B2', label='A')
ax.scatter(*(sites+delta[0]).T, s=38, color='#D55E00', marker='s', label='B')
ax.add_patch(Polygon([np.zeros(2), a1, a1+a2, a2], fill=False,
                     edgecolor='black', linestyle='--', linewidth=1.2))
for v, label, offset in [(a1, r'$\mathbf{a}_1$', (0.05, 0.1)),
                          (a2, r'$\mathbf{a}_2$', (-0.4, 0.1))]:
    ax.annotate('', xy=v, xytext=(0, 0), arrowprops=dict(arrowstyle='->', lw=1.8))
    ax.text(*(v+offset), label)
ax.set(xlim=(-2.2, 2.2), ylim=(-1.2, 3.6), xlabel=r'$x/a$', ylabel=r'$y/a$',
       title='Стільникова структура: A та B')
ax.set_aspect('equal'); ax.legend(loc='upper right')
bx.add_patch(Polygon(vertices*a, closed=True, facecolor='#EAF3F8', edgecolor='black'))
path_vertices = np.array([Gamma, M, K, Gamma])*a
bx.plot(*path_vertices.T, '--', color='#009E73', lw=1.4, label=r'$\Gamma\to M\to K\to\Gamma$')
for point, label, offset in [(Gamma, r'$\Gamma$', (0.10, -0.27)),
                             (M, r'$M$', (0.08, 0.10)),
                             (K, r'$K$', (0.08, 0.08)),
                             (Kp, r"$K'$", (-0.38, 0.08))]:
    bx.scatter(*(point*a), color='black', s=25)
    bx.text(*(point*a+offset), label)
bx.set(xlim=(-3, 3), ylim=(-2.7, 2.7), xlabel=r'$ak_x$', ylabel=r'$ak_y$',
       title='Перша зона Бриллюена')
bx.set_aspect('equal'); bx.legend(loc='lower center', fontsize=9)
plt.show()
```

### 2. Матриця Блоха і закон дисперсії

У базисі блохівських сум $|A,\mathbf k\rangle,|B,\mathbf k\rangle$, що
містять фази фактичних положень атомів, стан задається стовпцем
$\Psi_{\mathbf k}=(c_A,c_B)^{\mathsf T}$. За нуль енергії приймемо спільну
локальну енергію атомів. Матриця гамільтоніана дорівнює

$$
H(\mathbf k)=-t\begin{pmatrix}0&f(\mathbf k)\\f^*(\mathbf k)&0\end{pmatrix},
\qquad f(\mathbf k)=\sum_{j=1}^{3}e^{i\mathbf k\cdot\boldsymbol\delta_j}
=e^{iak_y}+2e^{-iak_y/2}\cos\frac{\sqrt3 ak_x}{2}.
$$

Отже, з $\det(H-EI)=0$ отримуємо дві $\pi$-зони:

$$
E_\pm(\mathbf k)=\pm t|f(\mathbf k)|,
\qquad
|f|^2=1+4\cos^2\frac{\sqrt3 ak_x}{2}
+4\cos\frac{\sqrt3 ak_x}{2}\cos\frac{3ak_y}{2}.
$$

Підґратки не задають дві незалежні зони: саме перехід A$\leftrightarrow$B
змішує базисні стани та утворює обидві власні енергії.
При $f\ne0$ і $f=|f|e^{i\phi}$ можна вибрати нормовані власні вектори
$u_+=(1,-e^{-i\phi})^{\mathsf T}/\sqrt2$ та
$u_-=(1,e^{-i\phi})^{\mathsf T}/\sqrt2$. У точці $f=0$ фаза $\phi$
не визначена: там слід працювати з усім двовимірним виродженим підпростором.
За електронейтральності та нульової температури нижня зона заповнена,
верхня порожня, а рівень Фермі проходить через точки їх дотику.

```python
t = 1.0  # одиниця енергії; результати графіків наведено в E/t

def structure_factor(k, hopping=(1.0, 1.0, 1.0)):
    """Сума переходів; k має форму (..., 2), hopping — t_j/t."""
    phase = np.einsum('...d,jd->...j', np.asarray(k), delta)
    return np.sum(np.asarray(hopping)*np.exp(1j*phase), axis=-1)

def graphene_H(k, Delta=0.0, hopping=(1.0, 1.0, 1.0)):
    """Delta — енергія A; енергія B дорівнює -Delta."""
    f = structure_factor(k, hopping)
    H = np.zeros(np.shape(f)+(2, 2), dtype=complex)
    H[..., 0, 0], H[..., 1, 1] = Delta, -Delta
    H[..., 0, 1], H[..., 1, 0] = -t*f, -t*np.conj(f)
    return H

def positive_energy(k, Delta=0.0):
    return np.sqrt(Delta**2+t**2*np.abs(structure_factor(k))**2)

for label, point in [(r'Γ', Gamma), ('M', M), ('K', K), ("K'", Kp)]:
    print(f'{label}: E/t = {np.linalg.eigvalsh(graphene_H(point))/t}')
```

### 3. Точки Дірака: вироджений підпростір і локальний розклад

У точці $\mathbf K$ фази трьох еквівалентних переходів дають

$$ f(\mathbf K)=1+e^{-2\pi i/3}+e^{2\pi i/3}=0. $$

Аналогічно $f(\mathbf K')=0$. Тому дві зони торкаються: $E_+=E_-=0$.
Це **точки Дірака**, а околи двох нееквівалентних точок називають долинами.

Нехай $\mathbf k=\mathbf K_\xi+\mathbf q$, де $\mathbf K_\xi=\xi\mathbf K$,
$\xi=\pm1$ і $a|\mathbf q|\ll1$. Для обраних вище осей

$$
f(\mathbf K_\xi+\mathbf q)
=-\frac{3a}{2}(\xi q_x-iq_y)+O(a^2|\mathbf q|^2),
$$

звідки в першому порядку

$$
H_\xi(\mathbf q)=\hbar v_F(\xi q_x\sigma_x+q_y\sigma_y),
\qquad v_F=\frac{3ta}{2\hbar},
\qquad E_\pm(\mathbf K_\xi+\mathbf q)\simeq\pm\hbar v_F|\mathbf q|.
$$

Це й є лінійний закон дисперсії — конус Дірака. Знаки біля матриць Паулі
залежать від осей і фаз базису; спектр та швидкість від такого вибору не
залежать. Для $a=0.142$ нм і **ілюстративного** $t=2.7$ еВ отримуємо
$v_F\approx8.7\cdot10^5$ м/с; точне значення залежить від параметрів моделі.
Це швидкість квазічастинок, а не швидкість світла.

**Зв'язок із теорією збурень.** Матриця $H(\mathbf K_\xi)=0$ має двовимірне
виродження. Малий параметр тут — $a|\mathbf q|$, а збуренням є зміна матриці
Блоха при відхиленні від $\mathbf K_\xi$. Потрібно спочатку діагоналізувати
всю матрицю першого порядку в цьому підпросторі. Звичайна невироджена
формула другого порядку із знаменником $E_+^{(0)}-E_-^{(0)}=0$ непридатна.
Дисперсія $|\mathbf q|$ не має звичайного лінійного скалярного розкладу в
самій вершині, хоча матричний гамільтоніан розкладається гладко.

```python
from matplotlib.path import Path as PlotPath
nodes = [Gamma, M, K, Gamma]
segments = [np.linspace(left, right, 160, endpoint=False)
            for left, right in zip(nodes[:-1], nodes[1:])]
k_path = np.vstack(segments+[Gamma[None, :]])
distance = np.r_[0, np.cumsum(np.linalg.norm(np.diff(k_path, axis=0), axis=1))]*a
ticks = np.r_[0, np.cumsum([np.linalg.norm(v-u)*a for u, v in zip(nodes[:-1], nodes[1:])])]
ep = positive_energy(k_path)/t
fig = plt.figure(figsize=(11, 4.5), constrained_layout=True)
ax = fig.add_subplot(121)
ax.plot(distance, ep, color='#D55E00', label=r'$E_+$')
ax.plot(distance, -ep, color='#0072B2', label=r'$E_-$')
ax.set(xticks=ticks, xticklabels=[r'$\Gamma$', '$M$', '$K$', r'$\Gamma$'],
       ylabel=r'$E/t$', title='Дві зони вздовж симетричного шляху', xlim=(0, ticks[-1]))
ax.grid(alpha=0.25); ax.legend()
bx = fig.add_subplot(122, projection='3d')
gx = np.linspace(-2.5, 2.5, 121)/a
gy = np.linspace(-2.2, 2.2, 121)/a
kx, ky = np.meshgrid(gx, gy)
inside = PlotPath(vertices).contains_points(np.c_[kx.ravel(), ky.ravel()], radius=1e-9).reshape(kx.shape)
z = np.where(inside, positive_energy(np.stack([kx, ky], axis=-1))/t, np.nan)
bx.plot_surface(kx*a, ky*a, z, color='#D55E00', alpha=0.8, linewidth=0, rcount=90, ccount=90)
bx.plot_surface(kx*a, ky*a, -z, color='#0072B2', alpha=0.8, linewidth=0, rcount=90, ccount=90)
bx.scatter([K[0]*a, Kp[0]*a], [0, 0], [0, 0], color='black', s=18)
bx.set(xlabel=r'$ak_x$', ylabel=r'$ak_y$', zlabel=r'$E/t$', title='Зони у першій зоні Бриллюена')
bx.view_init(elev=23, azim=-62)
plt.show()
```

### 4. Що саме визначає симетрія?

Трансляційна симетрія дозволяє позначати стани вектором Блоха $\mathbf k$.
Три еквівалентні зв'язки та трикратна обертальна симетрія пояснюють
скасування внесків у $f(\mathbf K)$ і положення дотиків у K/K' для
ізотропної моделі. Обернення часу пов'язує дві долини: для реальних
переходів $H(-\mathbf k)=H^*(\mathbf k)$.

Будь-яку ермітову двозонну матрицю можна записати як
$H=d_0I+d_x\sigma_x+d_y\sigma_y+d_z\sigma_z$. Для еквівалентних A/B та
збережених інверсії й обернення часу член, що розрізняє енергії підґраток,
заборонений. У вибраному базисі обмеження від їх комбінації має вигляд
$\sigma_xH^*(\mathbf k)\sigma_x=H(\mathbf k)$, отже $d_z=0$.
Для дотику залишається виконати дві умови $d_x=d_y=0$ у двовимірному
просторі $\mathbf k$. Маленьке збурення, що зберігає ці обмеження, може
зсунути простий дотик, але не відкриває в ньому щілину одразу.

У найближчому A–B наближенні також виконується
$\sigma_zH\sigma_z=-H$ (підґраткова, або кіральна, симетрія), що забезпечує
симетрію спектра відносно нуля. Це додаткове обмеження моделі:
однакові далекі переходи всередині A та B можуть додати $d_0(\mathbf k)I$,
зсунувши енергію дотику без відкриття щілини.

**Порушимо еквівалентність A/B.** Нехай локальні енергії дорівнюють
$+\Delta$ та $-\Delta$. Такий потенціал зберігає трансляції та обернення
часу, але порушує інверсію, яка переставляє підґратки:

$$
H_\Delta(\mathbf k)=
\begin{pmatrix}\Delta&-tf\\-tf^*&-\Delta\end{pmatrix},\qquad
E_\pm=\pm\sqrt{\Delta^2+t^2|f|^2}.
$$

Поблизу K/K' це доданок $\Delta\sigma_z$ до ефективного гамільтоніана,
а повна заборонена щілина дорівнює $E_g=2|\Delta|$.
$\Delta$ тут має розмірність енергії. Вона не є масою електрона; за
аналогією з релятивістським спектром можна визначити
$m_{\mathrm{eff}}=|\Delta|/v_F^2$.

**Межі аргументу.** Самої стільникової геометрії недостатньо. Якщо змінити
один перехід $t_1$, залишивши $t_2=t_3=t$, нулі задовольняють
$t_1+2t\cos(\sqrt3 ak_x/2)=0$ на $k_y=0$. За $0<t_1<2t$ два нулі
зміщуються; за $t_1=2t$ вони зливаються, а за $t_1>2t$ відкривається
щілина навіть за еквівалентних A/B. Отже, захист простого дотику є
локальним: пару дотиків можна усунути, звівши їх разом. Див. джерело 3.

Симетрія визначає допустимі члени матриці та властивості вироджень,
але не задає числове значення $t$ чи $v_F$. Ігнорований тут спін-орбітальний
зв'язок також змінює набір можливих членів; висновок про нульову щілину
належить саме сформульованій моделі.

```python
qx = np.linspace(-0.3, 0.3, 401)/a
cut = K + np.c_[qx, np.zeros_like(qx)]
fig, (ax, bx) = plt.subplots(1, 2, figsize=(10, 4), constrained_layout=True)
for D, color in [(0.0, '#0072B2'), (0.2*t, '#D55E00')]:
    energy = positive_energy(cut, Delta=D)/t
    ax.plot(qx*a, energy, color=color, label=fr'$\Delta/t={D/t:g}$')
    ax.plot(qx*a, -energy, color=color)
ax.plot(qx*a, 1.5*np.abs(qx*a), 'k:', lw=1, label='Лінійне наближення')
ax.plot(qx*a, -1.5*np.abs(qx*a), 'k:', lw=1)
ax.annotate('', xy=(0, 0.2), xytext=(0, -0.2),
            arrowprops=dict(arrowstyle='<->', color='black'))
ax.text(0.025, 0, r'$E_g=2|\Delta|$', va='center')
ax.set(xlabel=r'$aq_x\ (q_y=0)$', ylabel=r'$E/t$', title='Різні енергії A/B відкривають щілину')
ax.legend(fontsize=9); ax.grid(alpha=0.25)
kx_line = np.linspace(-2*np.pi/np.sqrt(3), 2*np.pi/np.sqrt(3), 901)/a
line = np.c_[kx_line, np.zeros_like(kx_line)]
for ratio, color in [(1.0, '#0072B2'), (1.5, '#009E73'), (2.0, '#E69F00'), (2.3, '#CC79A7')]:
    energy = np.abs(structure_factor(line, hopping=(ratio, 1, 1)))
    bx.plot(kx_line*a, energy, color=color, label=fr'$t_1/t={ratio:g}$')
bx.set(xlabel=r'$ak_x\ (k_y=0)$', ylabel=r'$E_+/t$',
       title='Анізотропія: рух і злиття точок Дірака', ylim=(-0.08, 4.5))
bx.legend(fontsize=9); bx.grid(alpha=0.25)
plt.show()
```

```python
# Числові контрольні перевірки моделі (не замінюють виведення).
rng = np.random.default_rng(7)
sample = rng.uniform(-3/a, 3/a, size=(100, 2))
H = graphene_H(sample)
assert np.allclose(H, H.conj().swapaxes(-1, -2))
assert np.allclose(np.linalg.eigvalsh(H)[:, 1], positive_energy(sample))
assert np.allclose(structure_factor(np.array([K, Kp])), 0, atol=1e-12)
assert np.allclose(np.vstack([a1, a2]) @ np.vstack([b1, b2]).T, 2*np.pi*np.eye(2))
assert np.allclose(positive_energy(sample+b1), positive_energy(sample))
h = 1e-6/a
for xi in (-1, 1):
    point = xi*K
    grad = np.array([(structure_factor(point+h*d)-structure_factor(point-h*d))/(2*h)
                     for d in np.eye(2)])
    assert np.allclose(grad, [-1.5*a*xi, 1.5j*a], atol=1e-8)
for D in (-0.2*t, 0.0, 0.2*t):
    levels = np.linalg.eigvalsh(graphene_H(K, Delta=D))
    assert np.isclose(levels[1]-levels[0], 2*abs(D), atol=1e-12)
for ratio in (1.0, 1.5, 1.9):
    moving_K = np.array([2*np.arccos(-ratio/2)/(np.sqrt(3)*a), 0.0])
    assert np.isclose(structure_factor(moving_K, (ratio, 1, 1)), 0, atol=1e-12)
assert np.isclose(abs(structure_factor([2*np.pi/(np.sqrt(3)*a), 0], (2.3, 1, 1))), 0.3)
hbar_eVs = 6.582119569e-16
vF_SI = 3*2.7*0.142e-9/(2*hbar_eVs)
print('Перевірки пройдено: ермітовість, спектр, K/K′, обернена ґратка,')
print('періодичність енергії, локальний розклад, щілина та рух дотиків.')
print(f'Для a=0.142 нм, t=2.7 еВ: v_F = {vF_SI:.3g} м/с.')
```

   щілину й енергії в K. Чи зміняться енергії при $\Delta\to-\Delta$?
   А які підґраткові стани мають додатну енергію в K?

   збереження еквівалентності A/B не гарантує безщілинність за довільної
   анізотропії переходів.

<summary>Орієнтири для самоперевірки</summary>

1. Два атоми A/B в елементарній комірці дають дві базисні орбіталі.
   Матриці $\sigma_i$ діють на A/B; реальний спін дає незалежне подвоєння.
2. $E^2=t^2|f|^2$; у $\Gamma$: $E=\pm3t$, у M: $E=\pm t$,
   у K: $E=0$ (двократне виродження без урахування спіну).
3. $\partial_{k_x}f|_{K_\xi}=-3a\xi/2$,
   $\partial_{k_y}f|_{K_\xi}=3ia/2$, $v_F=3ta/(2\hbar)$.
   У невиродженій формулі знаменник між двома рівнями в K дорівнює нулю.
4. $E_g=0.20$ еВ, $E_\pm(K)=\pm0.10$ еВ. Знак $\Delta$ не змінює
   спектр, але переставляє A/B: за $\Delta>0$ стан A має додатну енергію.
5. За $t_1/t=1.5$ нулі є при
   $ak_x=\pm2\arccos(-0.75)/\sqrt3$, $k_y=0$ (з точністю до оберненої

   $E_g=2(t_1-2t)=0.6t$. Відсутність локального масового члена не виключає
