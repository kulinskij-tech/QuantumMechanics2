## Спектр молекул. Адіабатичне наближення.

Молекула є ще більш складною систему ніж атом. Гамільтоніан такій системи складається з взаємодій ядро-ядро, ядро-електрон і електрон-електрон. Однак, той факт, що маси ядер у тисячі разів більше ніж маса електронів, приводить до того, що у нульовому наближенні можна розділити ядерні поступові і електронні степені свободи. Таке наближення, при якому ядра розглядаються як нерухому а електрони рухаються у поле нерухомих ядер, називається __адіабатичним__ (у застосуванні до молекул - наближення __Борна-Оппенгеймера__). В такому наближенні молекулярну систему можна розглядати як складену з двох підсистем: ядерної і електронної. Гамільтоніан молекули має вигляд:

\begin{equation}
\label{feq1} H = - {\sum\limits_{k} {{\frac{{\hbar
^{\,2}}}{{2\,M_{k}}} }}} \Delta _{{\mathbf{R}}_{k}}  -
{\sum\limits_{k} {{\frac{{\hbar^{\,2}}}{{2\,m}}}}} \Delta
_{{\mathbf{r}}_{k}}  + V({\mathbf{R}},{\mathbf{r}})
\end{equation}

Тут змінні ${\mathbf{R}},{\mathbf{r}}$ служать для позначення координат ядер і електронів відповідно. Оскільки $m / M \ll 1$ оператор кінетичної енергії ядер можна розглядати як збурення, так що незбурений гамільтоніан буде залежати від координат ядер як від параметрів. Незбурене рівняння Шредінгера має вид:

\begin{equation}
\label{feq2}
\begin{array}{l}
 \left( {H_{0} - \varepsilon _{n} ({\mathbf{R}})} \right)\,\varphi _{n}
({\mathbf{r}}\,;{\mathbf{R}}) = 0 \\
 H_{0} = - {\sum\limits_{k} {{\frac{{\hbar^{\,2}}}{{2\,m}}}}}
 \Delta _{{\mathbf{r}}_{k}}  + V({\mathbf{R}},{\mathbf{r}}) \\
 \end{array}
\end{equation}

і визначає енергію електронної підсистеми при фіксованому положенні ядер. Припустимо, що з (\ref{feq2}) отримані хвильові функції і рівні енергії електронної підсистеми. Тоді точні хвильові функції для оператора (\ref{feq1})

\begin{equation}
\label{feq3} \left( {H - E} \right)\Psi
({\mathbf{R}},{\mathbf{r}}) = 0
\end{equation}

можуть бути представлені у вигляді:

\begin{equation}
\label{feq4} \Psi ({\mathbf{R}},{\mathbf{r}}) = {\sum\limits_{n}
{\Phi _{n} ({\mathbf{R}})\,\varphi _{n}
({\mathbf{R}},{\mathbf{r}})}}
\end{equation}

Підставляючи (\ref{feq4}) в (\ref{feq3}) з врахуванням (\ref{feq1}), отримаємо:

\begin{equation}
\label{feq5} \left( { - {\sum\limits_{k} {{\frac{{\hbar
^{\,2}}}{{2\,M_{k}}} }\Delta _{{\mathbf{R}}_{k}}} }   +
\varepsilon _{n} ({\mathbf{R}}) - E} \right)\Phi _{n}
({\mathbf{R}}) = {\sum\limits_{k} {C_{nk} \Phi _{k}
({\mathbf{R}})}}
\end{equation}

де

\begin{equation}
\label{feq6}
\begin{array}{l}
 C_{nm} = - {\sum\limits_{k} {{\frac{{\hbar^{\,2}}}{{M_{k}}} }\int {\varphi
_{n}^{*}  ({\mathbf{R}},{\mathbf{r}})\nabla _{{\mathbf{R}}_{k}}
\varphi _{m} ({\mathbf{R}},{\mathbf{r}})\,{\prod\limits_{i}
{d\,{\mathbf{r}}_{i}}} }  \nabla _{{\mathbf{R}}_{k}}} }   + \\
 {\sum\limits_{k} {{\frac{{\hbar^{\,2}}}{{2\,M_{k}}} }}} \int {\varphi
_{n}^{*}  ({\mathbf{R}},{\mathbf{r}})\Delta _{{\mathbf{R}}_{k}}
\varphi _{m} ({\mathbf{R}},{\mathbf{r}})\,{\prod\limits_{i}
{d\,{\mathbf{r}}_{i}}} }  \\
 \end{array}
\end{equation}

($C_{mn} $ - оператори!) з (\ref{feq5}) витікає, що енергія електронної конфігурації відіграє роль потенціальної енергії для ядер. Адіабатичне наближення відповідає нульовому порядку теорії збурень по оператору, якій стоїть в правій частині (\ref{feq5}). В цьому наближенні:

\begin{equation}
\label{feq7} \left( { - {\sum\limits_{k} {{\frac{{\hbar
^{\,2}}}{{2\,M_{k}}} }\Delta _{{\mathbf{R}}_{k}}} }   +
\varepsilon _{n} ({\mathbf{R}}) - E_{n\alpha}^{(0)}} \right)\Phi
_{n\alpha}^{(0)} ({\mathbf{R}}) = 0
\end{equation}

Тут індекс $\alpha $ нумерує стани ядерної підсистеми при даній електронній конфігурації. Таким чином, в адіабатичному наближенні:

\begin{equation}
\label{feq8} \Psi _{n\alpha}  ({\mathbf{R}},{\mathbf{r}}) = \Phi
_{n\alpha}^{(0)} ({\mathbf{R}})\,\varphi _{n}
({\mathbf{R}},{\mathbf{r}})
\end{equation}

рівноважна конфігурація ядер для даного стану електронної підсистеми визначається з умови мінімуму для енергії останньої:

\begin{equation}
\label{feq9} \nabla_{\mathbf{R}_{k}}\varepsilon
({\mathbf{R}}_{k}^{0} ) = 0\,\,\,\,,\forall \, k
\end{equation}

Зрозуміло, що при невеликих збудженнях основного стану (рівні з малими квантовими числами) енергія молекули в такому наближенні може бути представлена як сума енергій електронної підсистеми і енергії руху ядер. Остання складається з енергії коливальних і обертальних ступенів свободи (у випадку малих енергій ці степені свободи можна враховувати як незалежні):

\begin{equation}
\label{feq10} E = \varepsilon _{m} ({\mathbf{R}}_{k}^{0} ) +
{\sum\limits_{s} {\hbar\, \omega_{m\,s} (n_{s} + 1 / 2)}}  +
E_{rot}
\end{equation}

Обчислення енергії обертального руху засновується на розгляді гамільтоніана:

\begin{equation}
\label{feq11} H_{rot} = {\frac{{K_{1}^{\,2}}} {{2\,I_{1}}} } +
{\frac{{K_{2}^{\,2} }}{{2\,I_{2}}} } + {\frac{{K_{3}^{\,2}}}
{{2\,I_{3}}} }
\end{equation}

де $1,\,2,\,3$ позначають компоненти відповідних величин на головні вісі інерції (система відліку, яка пов'язана з молекулою). Комутаційні співвідношення для компонент оператора моменту імпульсу мають вид:

\begin{equation}
\label{feq12} {\left[ {K_{m} ,K_{n}}  \right]} = -
i\hbar\,\varepsilon_{mnl} K_{l}
\end{equation}

Знак мінус відповідає тому, що $K_{m} $ є генераторами обертання навколо осей у системі відліку яка пов'язана з молекулою, що відповідає протилежному напрямку обертання у лабораторній системі. Обчислення власних значень (\ref{feq11}) тривіально у випадку $I_{1} = I_{2} = I_{3} $:

\begin{equation}
\label{feq13} E_{rot} = \frac{\hbar^{\,2}\,K(K + 1)}{2\,I}
\end{equation}

Оцінимо відстань між відповідними рівнями по безрозмірному параметру $m / M$. Очевидно, що відстань між електронними рівнями асимптотично не залежить від цього параметра. Відстань між коливальними рівнями пропорційна $\Delta E_{osc} \sim \omega \sim \frac{1}{\sqrt{M}}\,\sim \sqrt{\frac{m}{M}} $. Для обертальних рівнів маємо $\Delta E_{rot} \sim {\frac{1}{{M}}}\sim
{\frac{{m}}{{M}}}$. Таким чином:

\begin{equation}
\label{feq14} \Delta \varepsilon \gg \Delta E_{osc} \gg \Delta
E_{rot}
\end{equation}

Застосування вказаного наближення визначається малістю амплітуди коливань ядер у порівнянні з розмірами молекули:

\begin{equation}
\label{feq15} \frac{\sqrt {\langle\,r^{\,2}\,\rangle}}{d}\sim
\left(\frac{m}{M}\right)^{1/4} \ll 1
\end{equation}

Це означає, що малим параметром адіабатичного наближення є величина $\left(\frac{m}{M}\right)^{1/4}$. Аналіз молекулярних термів на основі інтегралів руху можливий тільки для молекул, конфігурації яких мають елементи симетрії (оси, площини симетрії, віддзеркалення і інш.). Продемонструємо такий підхід на прикладі двухатомної молекули. Такі молекули мають очевидну вісь симетрії. Цьому елементу (неперервної) симетрії відповідає збереження проекції моменту імпульсу електронної підсистеми на цю вісь. Він позначається $\Lambda $. Нехтуючи спіновими взаємодіями повний спін електронної підсистеми також є інтегралом руху, тому молекулярні терми характеризуються значенням $S$. Значення проекції моменту $\Lambda = 0,1,2,\ldots$ позначаються по аналогії з атомними термами літерами $\Sigma \,,\,\Pi \,,\,\Delta \,,\,\ldots$ вказуючи мультиплетність $2S + 1$, наприклад: $^{3}\Sigma\,,\,^{1}\Pi$ і інш.. Крім того, гамільтоніан системи є інваріантним відносно віддзеркалень в площині яка містить вісь
симетрії. При цьому проекція моменту змінює знак (аксиальний вектор), тому має місце двократне виродження енергії, оскільки стани з протилежними проекціями мають однакову енергію, виключаючи
$\Sigma $ терм ($\Lambda $=0). Однак, хвильова функція такого стану повинна бути власною функцією оператора віддзеркалення, тому стани $\Sigma $ терму можуть бути або парними $\Sigma^{+} $ або
непарними $\Sigma^{ -} $. Якщо молекула складається з однакових атомів, то з'являється додатковий елемент симетрії - перетворення інверсії координат електронів, відносно центра симетрії (проекція
моменту не змінює знак завдяки аксиальності). Цьому перетворенню відповідає ще одне квантове число типа парності: $\Lambda _{g} $ - парні, $\Lambda _{u} $ - непарні терми. Основний стан для
більшості двохатомних молекул є повністю симетричним з рівним 0 повним спіном ${}^{1}\Sigma^{+} $ чи ${}^{1}\Sigma _{g}^{+} $ для однакових атомів. Виключеннями є О$_{2}$ - терм ${}^{3}\Sigma
_{g}^{ -} $ і NO - терм ${}^{\,2}\Pi $.

Таким чином молекулярні спектри разом з переходами між електронними станами (оптичний діапазон) містять також лінії що відповідають переходам між [коливальними станами - коливальна спектроскопія](https://en.wikipedia.org/wiki/Infrared_spectroscopy) (інфрачервоний спектр), [станами обертання - обертальна спектроскопія](https://en.wikipedia.org/wiki/Rotational_spectroscopy) (мікрохвильове випромінювання) та переходами що обумовлені їх взаємодією [ротовібронна спектроскопія](https://en.wikipedia.org/wiki/Rotational%e2%80%93vibrational_spectroscopy).

В результаті молекулярний спектр складається з відносно чітких ліній електронних переходів  між якими йдуть смуги вібраційних переходів  між ними (ще більш щільно) смуги обертальних переходів. Молекулярний спектр - лінійчасто-смугастий на відміну від атомарного (лінійчастий).

Молекулярні коливання взаємодіють з електромагнітним випромінюванням завдяки зміні дипольного моменту молекули. Чим більше зміна дипольного моменту, тим сильніше речовина поглинає падаюче випромінювання. Правило відбору для коливальних переходів у гармонічному наближенні $\Delta n = \pm 1$. З врахуванням ангармонізмів можливі переходи із  $\Delta n = \pm 2, \pm 3\, \ldots$. Отже сполуки з полярними зв'язками типа O-H і C-O, мають лінії випромінювання-поглинання у інфрачервоному спектрі за якими їх ідентифікують. З цього зрозуміло, що неполярні  молекули N$_2$ і O$_2$ мають нульовий дипольний момент і є "інфрачервоно пасивними".

Суто обертальні переходи (без зміни стана електонної системи) здійснюються між станами $|J-J'|=1$ ядерної підсистеми. Відповідні частоти наприклад переходу випромінювання:
$$\omega = \frac{E_{rot}(J+1)-E_{rot}(J)}{\hbar} = \frac{\hbar}{2\,I}\,(J+1)$$

Крім загальної характеристики молекулярних спектрів квантова механіка дозволила з'ясувати механізм утворення молекул на основі поняття валентності. Необхідною умовою утворення молекули є взаємне доповнення електронних оболонок атомів, тобто вільні одночастинкові стани одного атома повинні заповнюватися за рахунок наявності заповнених у іншому. Зрозуміло, що при цьому основну роль знову відігріє принцип Паулі. Продемонструємо це на прикладі молекули водню, яка дає приклад гомополярного зв'язку (однакові атоми), на відміну від гетерополярного зв'язку (різні атоми). Розглянемо два атома водню в основному стані ($1s^{1}\,^{\,2}S)$. При їх зближенні, згідно правила складення моментів можуть бути отримані терми основного стана ${}^{1}\Sigma _{g}^{+}  $ чи ${}^{3}\Sigma _{u}^{+}  $. Дійсно, спінова частина хвильової функції синглетного стани є антисиметричною, тому координатна частина повинна бути парною (g) і не мати вузлів. Спінова частина хвильової функції триплетного навпаки симетрична, а координатна антисиметрична (u). Очевидно, що основний стан відповідає терму ${}^{1}\Sigma _{g}^{ + } $. Дійсно, у даній задачі ми маємо справу з типовим, для систем, які складаються з однакових частинок,
обмінним виродженням. Координатні частини хвильових функцій, правильного нульового наближення, які побудовані на незбурених хвильових функціях дорівнюють:

\begin{equation}
\label{feq16}
\begin{array}{l}
 \psi_{\pm}  = C\left(\varphi_{1}(\mathbf{1})\,\varphi_{2}(\mathbf{2})\pm \varphi_{1}(\mathbf{2})\varphi_{2} (\mathbf{1})\right) \\
 C = \frac{1}{\sqrt{2\left(1\pm N^{2}\right)}}\\
 N = \int\varphi_{1}({\mathbf{r}})\varphi_{2}({\mathbf{r}})\,d\mathbf{r} \\
 \end{array}
\end{equation}

Тут враховано те, що хвильові функції основних станів можна обрати дійсними. Симетрична (координатна) хвильова функція відповідає синглетному стану, антисиметрична - триплетному. Рівні енергії таких станів визначаються (у першому порядку теорії збурень) як середні від оператора кулонівської взаємодії:

\begin{equation}
\label{feq17} E_{\pm}  = 2E_{0} + \left( {{\frac{{e^{\,2}}}{{R}}} +
Q} \right)\pm A \mp N^{\,2}{\frac{{Q\pm A}}{{1\pm N^{\,2}}}}
\end{equation}

де

\begin{equation}
\label{feq18}
\begin{array}{l}
 Q(R) = e^{\,2}\int {\varphi _{1}^{\,2}(\mathbf{1})\,\varphi _{2}^{\,2} (\mathbf{2}){\left[ {{\frac{1}{{\vert \mathbf{1} - \mathbf{2}\vert}} } - {\frac{1}{{\vert \mathbf{1} - R\vert}} } -
{\frac{1}{{\vert \mathbf{2} - R\vert}} }}
\right]}\,d\,\mathbf{1}\,d\,\mathbf{2}} \\
 A(R) = e^{\,2}\int {\varphi _{1}(\mathbf{1})\,\varphi _{2}(\mathbf{2}){\left[ {{\frac{1}{{\vert \mathbf{1} - \mathbf{2}\vert}}
} - {\frac{1}{{\vert \mathbf{1} - R\vert}} } - {\frac{1}{{\vert
\mathbf{2} - R\vert}} }} \right]}\varphi _{2} (\mathbf{1})\,\varphi
_{1} (\mathbf{2})\,d\,\mathbf{1}\,d\,\mathbf{2}}
\\
 \end{array}
\end{equation}

Другий доданок у дужках (\ref{feq17}) дає ван дер ваальсівську взаємодію (притягнення) на великих відстанях, де обмінні доданки експоненціально малі. Наявність обмінного вкладу є суттєвим на близьких відстанях і приводить до того, що для енергії синглетного терму є глибокий мінімум, який відповідає зв'язаному стану, тобто молекулі водню (див Рис). Енергія триплетного терму в цієї області монотонна. Тому утворення молекули ${\rm H}_{2}$ обумовлене типово квантовим обмінним ефектом (утворення пари електронів з протилежними спінами).

До тепер ми ніяк не враховували спін ядер. Однак спін ядер впливає на енергетичні рівні молекули знову внаслідок умови правильної симетрії хвильової функції ядерної підсистеми. Для розглянутої вище молекули водню врахування тотожності ядер (для атома водню, ядро - протон, який є ферміоном) приводить до того, що існують два типу станів - синглетних і триплетних по ядерному спину. Синглетним станам відповідає симетрична координатна функція положень ядер, а триплетним антисимметрична. Як вже було показано раніше, синглетні стани відповідають парним значенням орбітального моменту імпульсу, а триплетні - непарним. Основний стан молекули водню є станом з $l = 0$. [Як і у випадку з атомом гелію (див. Лекцію)](./), завдяки слабкості спінових взаємодій між ядрами переходи між такими станами малоймовірні і вони ведуть себе практично як незалежні, так що спостерігається два типа водню - орто-
(триплетні терми) і пара- (синглетні терми) водород. В стані теплової рівноваги внаслідок виродження по ядерним спинам кількості пара- і ортоводню відносяться як 1:3.
