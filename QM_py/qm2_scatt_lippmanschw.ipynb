## Рівняння Ліппмана-Швінгера. Непружне розсіяння.

Раніше ми розглянули випадок пружного розсіяння. Розглянемо тепер
постановку задачі при непружному розсіянні, коли внутрішній стан
взаємодіючих частинок змінюється. Зрозуміло, що при цьому
необходимо враховувати "координати", внутрішніх степенів
свободи. Для простоти (виключаючи обмінні ефекти) припустимо, що у
системі падаюча частинка - центр розсіяння (мішень) немає
тотожних частинок. Падаючу частинку будемо розглядати як
безструктурну (тобто яка не має внутрішної будови, наприклад
електрон чи фотон). Сукупність ступенів свободи, які визначають
стан розсіювача, будемо позначати я $n$. Рівняння Шредінгера для
такій системи має вид:

\begin{equation}
\label{lipeq1} \left( {E - H_{c} (R) + {\frac{{\hbar
^{\,2}}}{2\,m}\Delta}  \right)\Psi (\,\mathbf{r},R) =
U(\,\mathbf{r},R)\,\Psi (\,\mathbf{r},R)
\end{equation}

Тут $H_{c}$ - гамільтоніан мішені. Очевидно, спектр незбуреного гамільтоніана (системи є незалежними):

\begin{equation}
\label{lipeq2} H_{0} = H_{c} (R) - {\frac{{\hbar
^{\,2}}}{2\,m}\Delta \,\,\, \Rightarrow
\,\,\text{спектр}\,\,\,\, \Rightarrow \,\,E = E_{c} +
{\frac{{\hbar^{\,2}k^{\,2}}}{2\,m}
\end{equation}

Хвильові функції будуються як добуток відповідних хвильових
функцій підсистем:

\begin{equation}
\label{lipeq3} \Phi _{n,\mathbf{k}} (\,\mathbf{r},R) = \varphi
_{n} (R)\,\exp \left( {i\,\mathbf{k}\, \cdot \mathbf{r}} \right)
\end{equation}

і нормуються наступним чином:
\begin{equation}
\label{lipeq4} {\frac{{1}}{{(2\pi )^{3}}}}{\sum\limits_{n} {\int
{\Phi^*_{n,\mathbf{k}} (\,\mathbf{r}',R')\,\Phi _{n,\mathbf{k}}
(\,\mathbf{r},R)\,\,d^{3}\mathbf{k}\,}}}   = \delta (\mathbf{k} -
\mathbf{k}')\,\delta (R - R\,')
\end{equation}
Оскільки ми розглядаємо стаціонарне розсіяння, повна енергія
системи є інтегралом руху, тому виконується рівність:
\begin{equation}
\label{lipeq5} {\frac{{\hbar^{\,2}k_{f}^{\,2}}} {2\,m} = E_{i} -
E_{f} + {\frac{{\hbar^{\,2}k_{i}^{\,2}}} {2\,m}
\end{equation}
для енергії відносного руху після розсіяння. Різні кінцеві стани
системи після розсіяння називаються \textbf{каналами розсіяння}.
Канал розсіяння називається відкритим
\index{розсіяння!канал!откритий}, якщо:
\begin{equation}
\label{lipeq6} 0\,\, \le \,\,E_{i} - E_{f} + {\frac{{\hbar
^{\,2}k_{i}^{\,2}}} {2\,m}
\end{equation}
тобто коли енергія відносного руху не нульова (підсистеми
віддаляються одна від одної на $\infty )$. У випадку:
\begin{equation}
\label{lipeq7} E_{i} - E_{f} + {\frac{{\hbar^{\,2}k_{i}^{\,2}}}
{2\,m} < 0
\end{equation}
канал розсіяння називається
\textbf{закритим}\index{розсіяння!канал!закритий}.

Як і у випадку пружного розсіяння, нас цікавлять рішення рівняння
Шредінгера (\ref{lipeq1}), які мають на нескінченності вид
суперпозиції падаючої плоскої і розбіжної сферичної хвиль.
Доцільно і в цьому випадку застосувати формалізм функцій Гріна.
Для (\ref{lipeq1}) функція Гріна визначається як рішення рівняння:
\begin{equation}
\label{lipeq8} \left( {E - H_{0}}  \right)G(\,\mathbf{r},R\vert\,
\mathbf{r}\,',R\,') = \delta (\,\mathbf{r}\,' -
\mathbf{r}\,)\delta (R - R\,')
\end{equation}
Завдяки (\ref{lipeq4}) а також раніше отриманого результату для
функції Гріна вільної частинки, яка описується розбіжною сферичною
хвилею отримуємо:
\[
G_{0}^{( + )} (\,\mathbf{r},R\vert\, \mathbf{r}\,',R\,') =
{\frac{{1}}{{(2\pi )^{3}}}}\left( {E - H_{0}}  \right)^{-
1}{\sum\limits_{n} {\int {\Phi^*_{n,\mathbf{k}}
(\,\mathbf{r}',R')\,\Phi _{n,\mathbf{k}}
(\,\mathbf{r},R)\,\,d^{3}\mathbf{k}\,}}} =
\]
\begin{equation}
\label{lipeq9} \,\,\,{\frac{{1}}{{(2\pi )^{3}}}}{\sum\limits_{n}
{\int {{\frac{{\Phi _{n,\mathbf{k}}
(\,\mathbf{r},R)\Phi^*_{n,\mathbf{k}} (\,\mathbf{r}',R')}}{{E -
E_{n\,,\,\mathbf{k}} + i\,0}}}\,\,d^{3}\mathbf{k}\,}}} =
\,\,{\sum\limits_{n} {g_{n}^{( + )}}}
\end{equation}
Для відкритих каналів (реальне розсіяння):
\begin{align}
 g_{n}^{( + )} = {\frac{{2m}}{{(2\pi )^{3}\hbar^{\,2}}}}\varphi
_{n}^*  (R\,')\varphi _{n} (R)\int {{\frac{{\exp \left(
{i\,\mathbf{k}(\,\mathbf{r} - \mathbf{r}\,')}
\right)}}{{k_{f}^{\,2} - k^{\,2} + i\,0}}}\,\,d^{3}\mathbf{k}\,}
=&\notag \\
     - {\frac{{m}}{{2\,\pi \,\hbar^{\,2}}}}\varphi _{n}^*  (R\,')\,\varphi
_{n} (R){\frac{{\exp \left( {i\,k_{f} \vert\, \mathbf{r} -
\mathbf{r}\,'\vert} \right)}}{{\vert\, \mathbf{r} -
\mathbf{r}\,'\vert}} } & \label{lipeq10}
\end{align}
За допомогою правильно обраної, з асимптотичних умов, функції
Гріна (\ref{lipeq9}) представимо рівняння Шредінгера у операторний
формі:
\begin{equation}
\label{lipeq11} \Psi^{( + )} = \Phi _{i} + G_{0}^{( + )} \,U\,\Psi
^{( + )}\,\,,\,\,\,G_{0}^{( + )} = {\frac{{1}}{{E - H_{0} + i\,0}}}
\end{equation}
Це рівняння називається \textbf{рівнянням
Ліппмана-Швінгера}\index{рівняння!Липпмана-Швингера}. У явній
інтегральній формі воно має вигляд
\[
\begin{array}{l}
 \Psi^{( + )}(\,\mathbf{r}\,,\,R\,) = \Phi _{i} (\,\mathbf{r}\,,\,R\,) + \Psi _{op}
+ \Psi _{cl} \\
 \,\,\Psi _{op} = - {\frac{{m}}{{2\,\pi \,\hbar^{\,2}}}}{\sum\limits_{op}
{\varphi _{op} (R\,)\int {\varphi _{op}^*  (R\,'){\frac{{\exp \left(
{i\,\mathbf{k}_{op} (\,\mathbf{r} - \mathbf{r}\,')}
\right)}}{{\vert\, \mathbf{r} - \mathbf{r}\,'\vert}}
}U(\,\mathbf{r}\,',\,R\,')\,\Psi ^{( + )}\,\,d^{3}\mathbf{k}_{op}
\,}}}   \\
 \,\Psi _{cl} = {\sum\limits_{cl} {\int {g_{cl}^{( + )}
 (\,\mathbf{r}\,,\,R\vert\, \mathbf{r}\,',\,R\,')\,U(\,\mathbf{r}\,',\,R\,')\,\Psi^{( +
)}\,\,d^{3}\mathbf{k}_{cl} \,} dR\,'}}  \\
 \end{array}
\]
Зрозуміло, що вклад в амплітуду розсіяння дають тільки відкриті
канали розсіяння. Використовуючи те, що асимптотика функції Гріна
$G_{0}^{( + )} $ на великих відстанях є:
\begin{equation}
\label{lipeq12} G_{0}^{( + )} \propto - {\frac{{m}}{{2\,\pi \,\hbar
^{\,2}}}}{\frac{{\exp \left( {i\,k_{f} r} \right)}}{{r}}}\exp \left(
{- i\,\mathbf{k}_{f} \mathbf{r}} \right)
\end{equation}
з (\ref{lipeq9}) - (\ref{lipeq12}) отримуємо вираз для амплітуди
розсіяння:
\begin{multline}
 A_{f,\,i} = - {\frac{{m}}{{2\,\pi \,\hbar^{\,2}}}}\int {\varphi _{f}^{*}
(R)\exp \,\left( {- i\,\mathbf{k}_{f} \mathbf{r}}
\right)\,U(\,\mathbf{r},\,R\,)\,\Psi _{i}^{( + )}
\,\,d^{3}\mathbf{r}\,\,d\,R\,} =\\ - {\frac{{m}}{{2\,\pi \,\hbar
^{\,2}}}}\left( {\Phi _{f} ,U(\,\mathbf{r},\,R\,)\,\Psi _{i}^{( +
)}} \right)\,,
\end{multline}
\begin{equation}
 \Psi^{( + )}(\,\mathbf{r},R) =\, \Phi _{i} (\,\mathbf{r},R) + {\sum\limits_{f}
{A_{f,\,i} \,\varphi _{n} (R)}} {\frac{{\exp \left( {i\,k_{f} r}
\right)}}{{r}}} \,.\label{lipeq13}
\end{equation}
У випадку $E_{f} = E_{i} $ отримуємо результат для пружного
розсіяння. Диференціальний переріз розсіяння визначається через
потік розсіяних частинок звичайним чином:
\begin{equation}
\label{lipeq14}
\begin{array}{l}
 \left( {\varphi _{i} (R\,),\Psi _{i}^{( + )} (\,\mathbf{r}\,,\,R\,)} \right) =
\exp \left( {i\,\mathbf{k}_{i} \mathbf{r}} \right) + A_{ii}
{\frac{{\exp (ikr)}}{{r}}} \\
 \left( {\varphi _{f} (R\,),\Psi _{i}^{( + )} (\,\mathbf{r}\,,\,R\,)} \right) =
A_{fi} {\frac{{\exp (ik_{f} r)}}{{r}}} \\
 \end{array}
\end{equation}
\begin{equation}
\label{lipeq15} d\,\sigma _{f\,i} = {\left| {{\frac{{j_{sc}}}
{{j_{in}}} }} \right|}d\,S = {\left| {{\frac{{j_{sc}}} {{j_{in}}} }}
\right|}r^{\,2}d\,\Omega = {\frac{{k_{f}}} {{k_{i}}} }{\left|
{A_{f\,i} (\mathbf{k}_{f} ,\mathbf{k}_{i} )} \right|}^{\,2}d\Omega
\end{equation}
В борнівському наближенні\index{наближення!борновское} з
(\ref{lipeq15}) отримуємо:
\begin{equation}
\label{lipeq16} d\,\sigma _{f\,i} = {\frac{{k_{f}}} {{k_{i}}}
}{\left| {A_{f\,i} (\mathbf{k}_{f} ,\mathbf{k}_{i} )}
\right|}^{\,2}d\Omega = \left( {{\frac{{m}}{{2\pi \hbar^{\,2}}}}}
\right)^{\,2}{\frac{{k_{f}}} {{k_{i}}} }{\left| {\left( {\Phi _{f}
,U\,\Phi _{i}}  \right)} \right|}^{\,2}\,,
\end{equation}
що, очевидно, співпадає з \eqref{scateq30}. Асимптотика хвильової
функції системи частинок є суперпозицією розсіяних хвиль з певною
амплітудою розсіяння по різним каналам розсіяння. Вона, також як і
аналогічна величина у випадку пружного розсіяння, може бути
розкладена на збіжні та розбіжні сферичні хвилі (у припущенні що
потенціал сферично-симетричний):
\begin{equation}
\label{lipeq17} \psi(\,\mathbf{r}\,) =
{\frac{{1}}{{\,2ik\,r}}}{\sum\limits_{l = 0}^{\infty} {(2l + 1)P_{l}
(\cos \,\theta ){\left[ {S_{l} \exp \left( {i\,kr} \right) + ( -
1)^{l + 1}\exp \left( {- i\,kr} \right)} \right]}}}
\end{equation}
Однак, тут $\vert S_{l} \vert \, < 1$, що пов'язано із наявністю
непружних каналів. Для повного перерізу пружного розсіяння
отримуємо:
\begin{equation}
\label{lipeq18}
\begin{array}{l}
 A(\theta ) = {\sum\limits_{l = 0}^{\infty}  {(2\,l + 1)\,A_{l} P_{l} (\cos
\,\theta )}}  = {\frac{{1}}{{2\,ik}}}{\sum\limits_{l = 0}^{\infty}
{(2\,l + 1)\,(S_{l} - 1)P_{l} (\cos \,\theta )}} \,\,\,,\,\, \\
 \sigma _{el} = {\frac{{\pi}} {{k^{\,2}}}}{\sum\limits_{l = 0}^{\infty}
{(2\,l + 1)\,\vert S_{l} - 1\vert^{\,2}}} \,\,\,,\, \\
 \end{array}
\end{equation}
Для знаходження перерізу непружного розсіяння необходимо врахувати
те, що інтенсивність розбіжної хвилі в цьому випадку зменшена у
$\vert S_{l} \vert^{\,2}\,\,\,$ раз у порівнянні з інтенсивністю
збіжної, тому:
\begin{equation}
\label{lipeq19} \sigma _{nel} = {\frac{{\pi}}
{{k^{\,2}}}}{\sum\limits_{l = 0}^{\infty} {(2\,l + 1)\,(1 - \vert
S_{l} \vert^{\,2})}}
\end{equation}
Повний переріз має вигляд:
\begin{equation}
\label{lipeq20} \sigma _{tot} = \sigma _{el} + \sigma _{nel} =
{\frac{{2\pi }}{{k^{\,2}}}}{\sum\limits_{l = 0}^{\infty}  {(2\,l +
1)\,(1 - {\rm Re}S_{l})}}
\end{equation}
Важливо зазначити, що пружне розсіяння може виникати за
відсутністю непружних каналів ($\vert S_{l} \vert = 1)$. Наявність
непружного розсіяння можливо тільки при існуванні пружного каналу.

\subsection{Зіткнення однакових частинок}

При розгляді зіткнень тотожних частинок із спіном $S$ необхідно
прийняти до уваги додаткові умови симетрії хвильової функції, які
пов'язані з принципом тотожності частинок (див. Лекцію~\ref{21}).
Це приводить до залежності амплітуди і перерізу розсіяння від
спінового стану. Тут також можна ввести поняття обмінної взаємодії
як і розглянутому раніше випадку зв'язаних станів 2-х частинкової
системи з тотожних частинок. В системі центра мас розсіяння двох
тотожних частинок описується хвильовою функцією, яка залежить від
радіус-вектора відносного положення цих частинок і
(анти)симетричної функцією їх координат в залежності від типу
частинок. Оскільки аналіз властивостей симетрії хвильової функції
повністю аналогічний тому що був проведений раніше на прикладі
двох-електронної системи одразу можна виписати результат для
асимптотики координатної  частини хвильової функції:
\begin{equation}
\label{lipeq21} \psi = \exp \left( {i\,k\,z} \right)\pm \exp
\left( {- i\,k\,z} \right) + {\frac{{\exp \left( {i\,k\,r}
\right)}}{{r}}}{\left[ {A(\theta )\pm A(\pi - \theta )} \right]}
\end{equation}
Відповідно диференціальний переріз розсіяння визначається як:
\begin{equation}
\label{lipeq22} d\sigma _{\pm}  = {\left[ {A(\theta )\pm A(\pi -
\theta )} \right]}^{\,2}d\Omega
\end{equation}
При цьому , зрозуміло, оператор спіна комутує з гамільтоніаном
системи (наприклад, оператор взаємодії не діє на спінові степені
свободи), тобто спин є інтегралом руху. Тому хвильова функція
береться у вигляді добутку координатної (\ref{lipeq21}) і спінової
частин:
\begin{equation}
\label{lipeq23} \Psi ({\rm \mathbf{1}} - {\rm \mathbf{2}}) =
\psi ({\rm \mathbf{1}} - {\rm \mathbf{2}})\Sigma (1,2)
\end{equation}
Звідси витікає, що парна координатна частина відповідає бозонам, а
непарна ферміонам, при умові парності спінової частини хвильової
функції. Якщо спінова частина хвильової функції непарна, тоді
ситуація з бозонами і ферміонами повністю протилежна: бозонам
відповідає непарна, а ферміонам парна координатна частина
хвильової функції. Покажемо, що парність координатної частини
хвильової функції розсіяння визначається парністю повного спина
системи двох тотожних частинок. Дійсно, повний спін системи
\textbf{$\sigma $} двох тотожних частинок може бути або парним,
або непарним числом. При перестановці частинок хвильова функція
перетворюється як:
\begin{equation}
\label{lipeq24} \Psi ({\rm \mathbf{2}} - {\rm \mathbf{1}}) = ( -
1)^{2\,S}\Psi ({\rm \mathbf{1}} - {\rm \mathbf{2}})
\end{equation}
Трансформаційні властивості відносно обертання для станів з цілим
спіном $\sigma $ аналогічни таким для орбітального моменту $l =
\sigma $. Тому:
\begin{equation}
\label{lipeq25}
\begin{array}{l}
 \psi ({\rm \mathbf{2}} - {\rm \mathbf{1}}) =
 \psi ( - \,({\rm \mathbf{1}} - {\rm
\mathbf{2}})) = ( - 1)^{\sigma} \psi ({\rm \mathbf{1}} - {\rm
\mathbf{2}})
\\
 \Sigma (2,1) = ( - 1)^{2S - \sigma} \Sigma (1,2) \\
 \end{array}
\end{equation}
Це значить, що парність повного спина системи співпадає з парністю
координатної  частини хвильової функції відносно перестановки
частинок. Тобто ``+``\, відповідає парному, а $``-``$\, непарному
значенню повного спіну системи в (\ref{lipeq21}).
