# Взаємодія атома та ЕМП

$\newcommand{\ket}[1]{\left.\left\vert\, #1\,\right. \right \rangle}$
$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$
$\newcommand{\av}[3]{\left\langle \,#1 \left \vert\, #2 \,\right\vert #3 \,\right \rangle}$

Розглянемо питання про квантування ЕМП. Строго кажучи, послідовний розв'язок цієї задачі можливе у рамках квантової теорії поля, оскільки ЕМП є релятивистським об'єктом так, що суттєві ефекти, які пов'язані із скінченністю швидкості розповсюдження взаємодії. Тем не менш, завдяки лінійності рівнянь Максвелла, довільний стан ЕМП може бути представлено у вигляді суперпозиції нормальних мод. Завдяки цьому, ЕМП є аналогом системи гармонічних осциляторів. Квантування гармонічного осцилятора з точки зору нерелятивистської квантової теорії є стандартною задачею. Таким чином, необхідно мати
гамільтонівську форму рівнянь ЕМП в термінах канонічно спряжених
змінних ("координат" і "імпульсів"). В класичної електродинаміці показано, що густина функції Лагранжа ЕМП має вид:

\begin{equation}\label{emf_lag}
L = \frac{1}{8\pi}\left(\mathbf{E}^{\,2} - \mathbf{H}^{\,2}
\right) = \frac{1}{8\pi}\left(\left(\frac{1}{c}
\frac{\partial \mathbf{A}}{\partial \,t}
\right)^{\,2} - \left( {\rm rot}\,\mathbf{A} \right)^{\,2}
\right)  \\
 \mathbf{E} = - \frac{1}{c}\frac{\partial\,\mathbf{A}}{\partial \,t}\,\,,\,\,
 {\rm div}\,\mathbf{A} =
 0\,,\,\,\varphi = 0
\end{equation}

так що рівняння поля у термінах $\mathbf{A}$ мають вид
хвильового рівняння:

\begin{equation}
\label{emeq16} \frac{1}{c^2}\frac{\partial^{\,2}\,\mathbf{A}}{\partial \,t^{\,2}} - \Delta
\,\mathbf{A} = 0
\end{equation}

Завдяки своєї структурі, величина:

\begin{equation}
\label{emeq17} \mathbf{E} = - \frac{1}{c}
\frac{\partial\,\mathbf{A}}{\partial \,t}
\end{equation}

аналогічна узагальненій швидкості (з точністю до коефіцієнта).
Густина функції Гамільтона, яка відповідає \eqref{emf_lag} є:

\begin{equation}
 \mathbf{P} = \frac{\partial \,L}{\partial 
 \left(\frac{\partial \mathbf{A}}{\partial \,t} \right)} = 
\frac{1}{4\pi\,c^{\,2}}\frac{\partial \,\mathbf{A}}{\partial \,t} \\
 H = \frac{1}{8\pi}\left(\mathbf{E}^{\,2} + \mathbf{H}^{\,2}
\right) = 2\pi c^{\,2}\mathbf{P}^{\,2} + \frac{1}{8\pi}\left( {\rm rot}\,\mathbf{A} \right)^{\,2} 
\end{equation}

Польовий спосіб опису веде до необхідності мати справу з нескінченним числом ступенів вільності. Щоб уникнути цього розглянемо ЕМП у "скринці" з періодичними граничними умовами. У такій ситуації, як відомо, кількість незалежних мод (ступенів вільності) ЕМП є скінченим. Математично більш доцільно перейти до фур'є - представлення для таких мод, у ролі яких виступають плоскі хвилі (розв'язки \eqref{emeq16}):
<!--  -->
\begin{equation}
\mathbf{A}(\mathbf{r},t) = c\,\sqrt{\frac{\pi}{V}}
\sum\limits_{\mathbf{q},\alpha}  {\rm A}_{\mathbf{q},\alpha}(t)\,\mathbf{e}_{\alpha} (\mathbf{q})\exp \left(i\,\mathbf{q}
\cdot \mathbf{r}\right)+{\rm A}_{-\mathbf{q},\alpha}(t)\,\mathbf{e}_{\alpha} (\mathbf{q})\exp \left(-i\,\mathbf{q}
\cdot \mathbf{r}\right)
\end{equation}
$$A_{-\mathbf{q}} = A^{*}_{\mathbf{q}}$$
Тут враховано те, що кожна плоска хвиля може мати два незалежних станів поляризації. Їх можна охарактеризувати векторами

\begin{equation}
 \mathbf{e}_{\alpha}  (\mathbf{q})\,\,,\,\,\alpha = 1\,,\,2 \\
 \left(\mathbf{e}_{\alpha}  (\mathbf{q}),\mathbf{e}_{\beta}
(\mathbf{q}) \right) = \delta_{\alpha \,\beta}
\,\,,\,\,\,\,\,\left(\mathbf{q}\,,\mathbf{e}_{\beta}
(\mathbf{q}) \right) = 0 
\end{equation}

У подальшому, будемо вважати, що дані вектора описують стани з різною лінійною поляризацією. Завдяки (\ref{emeq16}) фур'є - моди змінюються з часом гармонічним чином:

\begin{equation}
 \frac{d^2}{d\,t^2}{\rm A}_{\mathbf{q},\alpha}  (t)\, +
  \omega_{\mathbf{q},\alpha}^{\,2} \,{\rm A}_{\mathbf{q},\alpha}  (t) = 0\, \\
 \,\omega_{\mathbf{q},\alpha}  = c\,\left| \mathbf{q}\right|\\
 {\rm \dot {A}}_{\mathbf{q},\alpha}  (t) = -i\,\omega_{{\mathbf{q}},\alpha}\,{\rm A}_{\mathbf{q},\alpha}  (t)\,, \,\,\,
{\rm \dot {A}}_{-\mathbf{q},\alpha}  (t) = i\,\omega_{{\mathbf{q}},\alpha}\,{\rm A}_{\mathbf{q},\alpha}  (t)
\end{equation}

Звідси легко знаходимо:

\begin{equation}
{\rm H} = \int\limits_{V} H\,dV  =
\frac{1}{2}\sum\limits_{\mathbf{q},\alpha} \omega_{\mathbf{q},\alpha }^{\,2} \left| A_{\mathbf{q},\alpha} \right|^{\,2}
\end{equation}

Переходячи до дійсних величин:

\begin{equation}
 {\rm Q}_{\mathbf{q},\alpha}  = {\rm Re}\,
 \left({\rm A}_{\mathbf{q},\alpha}\right) \\
 {\rm P}_{\mathbf{q},\alpha}  = \omega_{\mathbf{q},\alpha}
{\rm Im}\,\left({\rm A}_{\mathbf{q},\alpha} \right)
\end{equation}

отримаємо функцію Гамільтона "осциляторного" типу:

\begin{equation}
\label{emeq24} 
{\rm H} =\frac{1}{2} \sum\limits_{\mathbf{q},\alpha} 
\left({\rm P}_{\mathbf{q},\alpha}^{\,2} + \omega_{\mathbf{q},\alpha}^{\,2}
\,{\rm Q}_{\mathbf{q},\alpha}^{\,2} \right)
\end{equation}

Змінні (\ref{emeq23}) є канонічно спряженими. Тому при побудові квантового гамільтоніана, замість їх повинні бути поставлені оператори, які задовольняють канонічним співвідношенням:

\begin{equation}
\label{emeq25} 
\left[ {\rm Q}_{\mathbf{q},\alpha}  ,{\rm P}_{\mathbf{q}',\alpha \,'} \right] = i\,\hbar \,\delta_{\mathbf{q}\,\mathbf{q}'} \,
\delta_{\alpha \,\alpha \,'}
\end{equation}

Вводячи оператори:

\begin{equation}
 C_{\mathbf{q},\alpha}  = \frac{1}{\sqrt{2\,\hbar
 \omega_{\mathbf{q},\alpha}}}
 \left( \omega_{\mathbf{q},\alpha}  \,{\rm Q}_{\mathbf{q},\alpha}  + 
 i\,{\rm P}_{\mathbf{q},\alpha} \right)
,\,\, \\
 \,C_{\mathbf{q},\alpha}^{\dagger}  = 
 \frac{1}{\sqrt{2\,\hbar\, \omega_{\mathbf{q},\alpha}}}
 \left( \omega_{\mathbf{q},\alpha}
\,{\rm Q}_{\mathbf{q},\alpha}  - i\,{\rm P}_{\mathbf{q},\alpha}
\right) \\
\left[C_{\mathbf{q},\alpha}  ,C_{\mathbf{q},\alpha}^{\dagger} \right] = 1
\end{equation}

як і у випадку осцилятора, отримаємо:

\begin{equation}
\label{emeq27}
\hat{H} = \sum\limits_{\mathbf{q},\alpha} 
\hbar\,\omega_{\mathbf{q},\alpha}\,\left( N_{\mathbf{q},\alpha}  +
\frac{1}{2} \right) ,\,\,\,N_{\mathbf{q},\alpha}  =
C_{\mathbf{q},\alpha}^{ + } C_{\mathbf{q},\alpha}
\end{equation}

В результаті, стан (хвильова функція) ЕМП характеризується набором невід'ємних цілих чисел $N_{\mathbf{q},\alpha}  $ - "чисел заповнення". Хоч енергія основного стану формально є
нескінченною, для спостережувальних величин це не суттєво, оскільки при взаємодії важливо тільки зміна енергії. Енергія (імпульс) представляється у вигляді суми енергій (імпульсу)
одночастинкових станів (квантів) ЕМП - фотонів. При цьому зв'язок між енергією і імпульсом одночастинкового стану:

\begin{equation}\label{emeq28} 
\varepsilon_{q} = \hbar\, \omega_{q} = cp =
c\,\hbar \,q
\end{equation}

характерна для безмасової релятивистської частинки. У подальшому
будемо користуватись гейзенберговським представленням для ЕМП,
тобто стан системи не залежить від часу, а вся часова залежність
переноситься на оператори. При цьому, згідно з \eqref{emeq25}:
$$
i\,\hbar \frac{\partial \,C}{\partial \,t} =\left[C,H\right] = \hbar\, \omega \,C
$$

### Випромінювання і поглинання світла квантовою системою. Спонтанне і вимушене випромінювання.
<!--  -->
Застосуємо результати, які були отримані вище, у випадку взаємодії
квантової системи і ЕМП. При цьому ми будемо нехтувати
релятивистськими ефектами (в тому числі запізнення і спін -
магнітним взаємодіями). Вважаючи поле слабким, гамільтоніан може
бути представлений у вигляді:
$$
H = H_{p} + H_{em} + U_{int}
$$
де $H_{p} $ - гамільтоніан частинки, $H_{em} $ - гамільтоніан
вільного ЕМП,

\begin{equation}
\label{izeq1} U_{int} = - \frac{q}{m\,c}\mathbf{A} \cdot \mathbf{p}
\end{equation}

Хвильові функції 0-го наближення вибираються у вигляді добутку хвильових функцій систем. Оператор вектор-потенціалу виражається через оператори "народження-знищення" квантів ЕМП поля:

\begin{equation}
\mathbf{A} = c\sqrt{\frac{2\pi \,\hbar}{V}} 
\sum\limits_{\mathbf{q},\alpha}  
\frac{
C_{\mathbf{q},\alpha}^{\dagger}  \mathbf{e}_{\alpha}
(\mathbf{q})\, \exp\left( - i\,\mathbf{q} \cdot \mathbf{r}\right) + 
C_{\mathbf{q},\alpha} \mathbf{e}_{\alpha}(\mathbf{q})\, 
\exp\left(i\,\mathbf{q} \cdot \mathbf{r} \right)
}
{\sqrt{\omega_{\mathbf{q},\alpha}}}
\end{equation}
<!--  -->
У першому порядку теорії збурень для ймовірності переходу у одиницю часу отримаємо ([золоте правило Фермі](https://colab.research.google.com/github/kulinskij-tech/QuantumMechanics2/blob/main/QM_py/qm2_nonstacpertheor.ipynb#fermigoldenrule)):
<!--  -->
$$
W_{i \to f}=\frac{P_{i \to f}}{T} = 
\frac{2\pi}{\hbar }\left|\av{\psi_{f}}{U_{int}}{\psi_{i}}\right|^{\,2}\,\rho (E_{f})
$$
<!--  -->
де $\rho $ - густина числа кінцевих станів. Розглянемо процес, який пов'язаний з випромінюванням фотона (дія оператора народження). В цьому випадку число фотонів у кінцевому стані більш
ніж у початковому:
<!--  -->
\begin{equation}
\label{izeq3} \av{\psi_{f}}{U_{int}}{\psi_{i}} = -
\frac{q}{m}\sqrt \frac{2\pi \,\hbar}{V\omega_{\mathbf{q},\alpha}} \sqrt{N_{\mathbf{q},\alpha}  + 1}
\,\,\mathbf{e}_{\alpha}  (\mathbf{q}) \cdot \av{\phi_{f}}{\exp \left(  - i\,\mathbf{q} \cdot \mathbf{r} \right)\,\mathbf{p}}{\phi_{i}}
\end{equation}
<!--  -->
Для більшості систем довжина хвилі значно більше характерних розмірів $a$ системи, так що можна положить:
<!--  -->
\begin{equation}\label{eq_exp1}
\exp \left(  - i\,\mathbf{q} \cdot \mathbf{r}\right) \approx 1 + o(|\mathbf{q}|)
\end{equation}
<!--  -->
-довгохвильове наближення. Але
<!--  -->
\begin{equation}
\av{\phi_{f}}{\mathbf{p}}{\phi_{i}} = i\,m\,\omega_{f\,i} \av{\phi_{f}}{\mathbf{r}}{\phi_{i}} = \frac{i\,m\,\omega_{f\,i}}{q} \mathbf{d}_{f\,i}
\end{equation}
<!--  -->
$$\frac{\mathbf{p}}{m} = \frac{1}{i\hbar}[\mathbf{r},\hat{H}_0]$$
<!--  -->
що витікає з загальних комутаційних співвідношень. Тому дане наближення відповідає врахуванню електричного дипольного випромінювання. Для переходу випромінювання, отримаємо:
<!--  -->
\begin{equation}
W_{i \to f}^{( + ),(\alpha)}= 
\frac{(2\pi)^2\,\omega_{\mathbf{q},\alpha}}{V}
\,\left(N_{\mathbf{q},\alpha}  + 1 \right)\,\left|
\mathbf{e}_{\alpha}  (\mathbf{q}) \cdot\mathbf{d}_{f\,i}\right|^{\,2}
\rho (E_{f}) =
\left( N_{\mathbf{q},\alpha}  + 1 \right)\,\,
\left| \mathbf{e}_{\alpha}  (\mathbf{q})
\cdot \mathbf{d}_{f\,i} \right|^{\,2}\frac{\omega_{f\,i}^{3}}{2\pi c^{3}\hbar}\, d\Omega
\end{equation}
<!--  -->
Тут
<!--  -->
$$
\rho (E_{f}) = \frac{d\,N}{dE} = 
\frac{d\,N}{c\hbar dk} =
V\frac{k^{\,2}}{c\hbar \,\left( {2\pi} \right)^{3}} \,d\Omega\,,
\\
k = \frac{\omega_{fi}}{c}
$$
<!--  -->
енергетична густина станів для випромінювання. Інтенсивність поляризованого випромінювання:
<!--  -->
\begin{equation}
\label{izeq7}
\begin{array}{l}
 I_{i \to f}^{( + ),(\alpha)} = \hbar\, \omega_{f\,i}
 W_{i \to f}^{( + ),(\alpha)} =
 \left(N_{\mathbf{q},\alpha}  + 1 \right)\,\,
\left| \mathbf{e}_{\alpha}  (\mathbf{q})
 \cdot \mathbf{d}_{f\,i}
\right|^{\,2}\frac{\omega_{f\,i}^{4}}{2\pi c^{3}}\,d\Omega \\
 E_{i} = E_{f} + \hbar\, \omega_{\mathbf{q},\alpha}  \\
 \end{array}
\end{equation}
<!--  -->
як видно полягає з двох частин: __вимушеного__ випромінювання, яке зникає разом з числами заповнення станів ЕМП, і __спонтанного__, яке відмінно від 0 навіть у випадку відсутності квантів ЕМП. Результат для інтенсивності поглинання може бути записаний за допомогою матричного елемента понижуючого оператора:
<!--  -->
\begin{equation}
 I_{i \to f}^{( - ),(\alpha)} = \hbar\,
 \omega_{\,f\,i} W_{i \to f}^{( - ),(\alpha)} =
 N_{\mathbf{q},\alpha}  \,\left|
 \mathbf{e}_{\alpha}(\mathbf{q}) \cdot \mathbf{d}_{f\,i}
\right|^{\,2}\frac{\omega_{\,\,f\,i}^{4}}{2\pi c^{3}}\,d\Omega
\\
 E_{i} = E_{f} - \,\hbar\, \omega_{\mathbf{q},\alpha}
 \end{equation}
<!--   -->
Очевидно, що поглинання відбувається тільки якщо $N_{\mathbf{q},\alpha} \ne 0$, тобто є вимушеним і не може бути спонтанним із зрозумілих причин.

Повна інтенсивність __електричного дипольного випромінювання__ (поглинання) знаходиться інтегруванням по повному тілесному куту і сумою по станам поляризації $\alpha = 1,2$:
<!--  -->
\begin{equation}
\label{izeq9}
\begin{array}{l}
 {\rm I}_{i \to f}^{( + )} = \frac{\omega_{f\,i}^{4}}{2\pi\,
c^{3}}\,\sum\limits_{\alpha}
 \int I_{i \to f}^{( + )} \,d\Omega = \frac{\omega_{f\,i}^{4}}{2\pi
c^{3}}\,\sum\limits_{\alpha}\left( {N_{\mathbf{q},\alpha}  + 1}
\right)\,\,\int\left|\,\mathbf{e}_{\alpha}(\mathbf{q}) \cdot
{\,{\mathbf{d}}_{f\,i}}\, \right|^{\,2}\,d\,\Omega\,.\\
 \end{array}
\end{equation}
<!--  -->
Аналогічно можна одержати результат для інтенсивності поглинання.
Для спонтанного випромінювання \eqref{izeq9} дає
<!--  -->
\begin{equation}
\label{1izeq9}
 {\rm I}_{i \to f}^{( + ),(s)} = \frac{\omega_{f\,i}^{4}}{2\pi\,
c^{3}}\,\left|\,\mathbf{d}_{f\,i}\,\right|^{\,2}\,\int
\sin^2\theta \,d\,\Omega\,=
\frac{4\,\omega_{f\,i}^{4}}{3\,c^{3}}\left|\,\mathbf{d}_{f\,i}\,\right|^{\,2}\,.
\end{equation}
<!--  -->
Матричні елементи оператора дипольного моменту між певними станами можуть бути рівними 0 завдяки симетрії. У такому випадку говорять, що переходи між такими станами __дипольно заборонені__. Виходячи з загальних міркувань симетрії (правило моментів і парність)
отримаємо наступні __правила відбору__ для електричних дипольно дозволених переходів:
<!--  -->
\begin{equation}
l_{f} = l_{\,i} \pm 1\,,\quad   m_{f} = \{m_{i} \pm
1\,,\,\,m_{\,i} \}\,.
\end{equation}
<!--  -->
Разом із електричними дипольними переходами мають місце і інші __мультипольні__ (електричні і магнітні) переходи. Вони відповідають більш високим порядкам розкладу \eqref{eq_exp1} по $a_{B}/\lambda$. Наприклад, у лінійному наближенні:
<!--  -->
\begin{equation}
\exp \left(- i\,\mathbf{q} \cdot \mathbf{r}\right) 
\approx 1 - i\,\mathbf{q} \cdot \mathbf{r}
\end{equation}
<!--  -->
користуючись тим що:
<!--  -->
\begin{equation}
\av{\phi_{f}}{\mathbf{q} \cdot\mathbf{r} \,\mathbf{e}_{\alpha}(\mathbf{q}) \cdot \mathbf{p}}{\phi_{i}} \propto 
\av{\phi_{f}}{Q^{(lk)}}{\phi_{i}} + \av{\phi_{f}}{L^{(m)}}{\phi_{i}}
\end{equation}
<!--  -->
ми отримаємо т.з. магніто-диполні та квадруполні переходи, оскільки $Q^{(lk)},L^{(m)}$ - компоненти операторів квадрупольного і орбітального моментів. Дійсно, наприклад:
<!--  -->
$$
\av{\phi_{f}}{x\,p_{y}}{\phi_{i}}=
\frac{1}{2}\left( \av{\phi_{f}}{(x\,p_{y} + \,y\,p_{x})}{\phi_{i}}\right) + \left(\av{\phi_{f}}{(x\,p_{y} - 	\,y\,p_{x} )}{\phi_{i}}  \right) = \\
\frac{1}{2}\left(\av{\phi_{f}}{(x\,p_{y} + \,y\,p_{x})}{\phi_{i}}  + \av{\phi_{f}}{L_{z}}{\phi_{i}}
\right)
$$
<!--  -->
але
<!--  -->
$$
2\,m\,[x\,y,H_{0} ] = [x\,y,p_{x}^{\,2} + p_{y}^{\,2} ] = [x\,y,p_{x}^{\,2}
] + [x\,y,p_{y}^{\,2} ] = 2\,i\,\hbar \,\left( y\,p_{x} + x\,p_{y}
\right)
$$
<!--  -->
Випромінювання (поглинання), пов'язане з кожним з цих операторів називається __електричним
квадрупольним__ і __магнітним дипольним__ відповідно. При цьому, для квантових систем з центрально симетричним гамільтоніаном, наприклад атом водню, магніто-дипольне випромінювання, яке пов'язане з дією оператора $L_z$ дорівнює 0, завдяки тому, що цей оператор є інтегралом руху. Дія операторів двох других проекцій, як витікає з алгебри операторів моменту, приводить до переходів без випромінювання (оскільки енергія при сферично симетричної взаємодії не залежить від квантового числа проекції моменту імпульсу) між станами в яких азимутальні квантові числа відрізняються на 1.

Для електричного квадрупольного випромінювання правила відбору
мають вигляд:
<!--  -->
\begin{align}
 l_{f} =& \left\{
 {\begin{array}{l}
 {\{l_{\,i} \pm 2\,,\,l_{\,i} \},l_{\,i} \ne 0} \\
 {2,    l_{\,i} = 0} \\
 \end{array}}\right.\notag\\
 m_{f} =& \{m_{i} \pm 2\,,m_{i} \pm 1\,,\,m_{\,i} \}\label{izeq13}
\end{align}
<!--  -->
Аналогічно вводяться інтенсивності вищих мультипольних моментів і
правила відбору для них.

Якщо потенціал $H_{p} $ є центрально симетричним то, завдяки незалежності енергії від азимутального квантового числа, магнітні дипольні переходи не супроводжуються випромінюванням в таких системах. При наявності магнітного поля (ефект Зеємана), чи врахуванні спін-орбітальної взаємодії (тонкої структури) виродження знімається, і такі переходи відбуваються із зміною чисел заповнення стану ЕМП. Реалізація того чи іншого переходу обмежується законами збереження таких величин як повний момент імпульсу (інваріантість відносно обертань) і парність стана (інваріантість відносно перетворення інверсії). Якщо $J_{in} $ і $j$ квантові числа моменту імпульсу початкового стани і фотона, то квантове число $J_{fin} $ моменту імпульсу кінцевого стану може бути, згідно правилу складення моментів, тільки таким:
<!--  -->
\begin{equation}
\label{izeq14} \vert J_{in} - j\vert \, \le J_{fin} \le J_{in} + j
\end{equation}
Правило відбору по парності:
\begin{equation}
\label{izeq15} P_{in} \,P_{ph} = P_{fin}
\end{equation}
<!--  -->
Оскільки вказані закони збереження є точними, то переходи, в яких порушуються ці закони, називають __строго забороненими__. __Парність__ фотонного стана з квантовим числом $j$ визначається як $( - 1)^{j}$ для фотона електричного (EJ) типу і $( - 1)^{j + 1}$ для фотона магнітного (MJ) типу.

## Вивід закону Планка
Отримані вище результати для взаємодії ЕМП і атомів можна застосувати до задачі про спектр рівноважного теплового випромінювання. Зрозуміло, що в рівновазі кількість переходів поглинання дорівнює кількості переходів випромінювання для кожного переходу $m\leftrightarrow n\,, E_{m}<E_{n}$. Нехай кількість атомів у відповідних станах $N^{(a)}_{m}, N^{(a)}_{n}$ (принцип детального балансу):
$$
N^{(a)}_{n} W_{n \to m}^{(+)} = N^{(a)}_{m} W_{m \to n}^{(-)}
$$
Враховуючи вирази $ W_{n \to m}^{(\pm)}$ отримані вище і інтегруючи по всіх напрямках поляризації і росповсюдження випромінювання можна записати вводячи коефіцієнти Айнштайна:
$$
W_{m \to n}^{(-)} = B_{m,n}\,\rho(\omega_{mn})\,,
W_{n \to m}^{(+)} = B_{m,n}\,\rho(\omega_{mn}) + A_{mn}\\
$$
де $\rho(\omega_{mn})$ - спектральна густина випромінювання. Тоді
$$
N^{(a)}_{n} \left(B_{m,n}\,\rho(\omega_{mn}) + A_{mn}\right) = N^{(a)}_{m} B_{m,n}\,\rho(\omega_{mn}) \Rightarrow 
\rho(\omega_{mn}) = \frac{A_{mn}}{B_{m,n}}\frac{1}{\frac{N^{(a)}_m}{N^{(a)}_n} - 1}
$$
Приймаючи до уваги, що в рівновазі розподіл атомів за енергією це розподіл Больцмана $N^{(a)}_n = N_0\,e^{-\frac{E_n}{T}}$ отримаємо:
$$
\rho(\omega_{mn}) = \frac{A_{mn}}{B_{m,n}}\frac{1}{e^{\frac{E_n-E_m}{T}} - 1}
$$
Відношення $\frac{A_{mn}}{B_{m,n}}$ визначається з відповідності до класичної асимптотикі - закону Релея-Джинса:
$$
\frac{A_{mn}}{B_{m,n}} = \frac{\hbar\,\omega^3}{\pi^2\,c^3}
$$
В результаті остаточно приходимо до формули Планка:
$$
 \rho(\omega) = \frac{\hbar\,\omega^3}{\pi^2\,c^3}\,\frac{1}{e^{\frac{\hbar\,\omega}{T}} - 1}
$$

## Квазістаціонарні стани. Природна ширіна спектральних ліній.
З попередного можна зробити висновок, що збуджені рівні атомів не є стаціонарними, оскільки з часом вони еволюціонують до основного стану (висвітлюються). Якщо такий процес розпаду відбувається в середньому повільно (час перебування у стані значно перевищує $1/\omega$, $\omega$- частота переходу), то можна говорити про середній час життя таких станів, які називаються квазістаціонаними. Зрозуміло, що такий час можна оцінити виходячи із величини швидкості переходів $W^{(+)}_{i\to f}$ як 
$$\gamma = \sum\limits_{f} W^{(+)}_{i\to f}\,,\,\,T = 1/\gamma$$
Розподіл енергії у квазістаціонарних станах характеризується (напів)шириною. Тут можна навести аналогію із радіоактивним розпадом. Розпад ядра обумовлений його нестабільністю, тобто стан не є стаціонарним. Ймовірність спостереження ядра у початковому стані монотонно зменшується за відомим законом:
$$P(t) = \left|\braket{\psi_0}{\psi_t}\right|^2 = e^{-t/T}$$
Із загальної т.з. для ймовірності спостереження системи у початковому нестаціонарному стані ми можемо написати:
$$\ket{\psi_0} = \int c(E)\ket{E}\,dE\,,\,\,\ket{\psi_t} = \int c(E)\,\ket{E}\,e^{\frac{E\,t}{i\,\hbar}}\,dE$$
відповідно:
$$\left|\braket{\psi_0}{\psi_t}\right|^2 = \int |c(E)|^2\,e^{\frac{E\,t}{i\,\hbar}}\,dE$$
Тобто така ймовірність є фурє-образом функції енергетичної густини $|c(E)|^2$. В теорії ймовірності це називається характеристичною функцію розподілу. Отже 
$$P(t) = \langle e^{-i\,\omega\,t} \rangle_{\omega}\,,\,\,\,t>0$$
З відповідних властивостей Фур' перетворення витікає, що  
$$\lim\limits_{t\to \infty}P(t) = 0$$ 
якщо функція розподілу є неперервною, тобто не містить $\delta$-образних внесків від станів дискретного спектру. Зокрема, для функції розподілу:
$$|c(E)|^2 = \frac{1}{2\,\pi}\,\frac{\gamma}{(E-E_0)^2+\gamma^2/4}$$
за умови (квазістаціонарності) $E_0\gg \gamma$ отримуємо
$$\left|\braket{\psi_0}{\psi_t}\right|^2 = \left|\frac{\gamma}{2\pi}\,\int \frac{e^{\frac{E\,t}{i\,\hbar}}}{(E-E_0)^2+\gamma^2/4}\,dE\right|^2 \approx e^{-\gamma\,t}$$
Якщо б початковий стан а отже енергетична густина мала внесок від дискретного спектру, то (внесок від неперервного спектру ми опускаємо):
<!--  -->
$$\ket{\psi_0} = \sum\limits_{k} c_k\ket{E_k}+\ldots\,,\,\,\ket{\psi_t} = \sum\limits_{k} c_k\,\ket{E_k}\,e^{\frac{E_k\,t}{i\,\hbar}}+\ldots\\
\left|\braket{\psi_0}{\psi_t}\right|^2 = \sum\limits_{m,n}|c_m|^2\,|c_n|^2\,\cos{\frac{(E_n-E_m)\,t}{i\,\hbar}}+\ldots
$$
Як можна зрозуміти з осцилюючого характеру такого доданка, поведінка ймовірності $P(t)$ не є монотонною, а отже система не розпадається, оскільки при $t>0$ існують моменти часу де ймовірність спостереження початкового стану приймає певне максимальне значення. Зокрема для системи із обмеженною кількістю станів (наприклад спінових рівнів) таке значення дорівнює 1.

Застосовуючи ці факти до системи атом+ЕМП, ми бачимо, що завдяки неперервному енергетичному спектру ЕМП збуджені стани атомів є квазістаціонарними і характеризуються певним часом життя, яке повязане із енергетичною шириною такого рівня $\Delta E\, T \sim \hbar$ і збуренням яке обумовлює такі переходи.
