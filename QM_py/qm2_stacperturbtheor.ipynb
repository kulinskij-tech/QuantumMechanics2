## Теорія збурень для станів невиродженого дискретного спектру.
<!--  -->
$\newcommand{\ket}[1]{\left.\left\vert\, #1\,\right. \right \rangle}$
$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$
$\newcommand{\av}[3]{\left\langle \,#1 \left \vert\, #2 \,\right\vert #3 \,\right \rangle}$
<!--  -->
Кількість точно задач квантової теорії які можуть бути розв'язані точно дуже обмежена. Тому при розгляді реалістичних моделей дуже важливими є надійні наближені методи розрахунку спектра гамільтоніана системи. Дуже важливими є випадки коли реальну фізичну ситуацію вдається представити як трохи ускладнену ідеальну модель. Сама ідеальна модель (незбурений гамільтоніан) припускає точний розв'язок, а відхилення від цієї ідеальної моделі, яке пов'язане з додатковою взаємодією можна розглядати у певному сенсі малим. Тоді можна очікувати, що якісно спектр вихідного гамільтоніана не зміниться, а зміняться лише чисельні значення і хвильові функції, причому ці зміни будуть малими. Більш конкретно,
розглянемо випадок, коли у незбуреного оператора Шредінгера є тільки дискретний невироджений спектр. Нехай гамільтоніан задачі має вигляд:
<!--  -->
\begin{equation}\label{eq_ham} 
H = H_{0} + \varepsilon \,V
\end{equation}
<!--  -->
де $\varepsilon \ll 1$ - мала безрозмірна величина, яка утворюється з констант вихідного гамільтоніана і оператора збурення. Хвильові функції і рівні енергії гамільтоніана
\eqref{eq_ham}
<!--  -->
\begin{equation}\label{eq_hampsi}
\hat{H}\,\ket{\psi_{E}} = E\,\ket{\psi_{E}}
\end{equation}
<!--  -->
будемо будувати у вигляді формальних рядів за малим параметром $\varepsilon$:
<!--  -->
\begin{equation}\label{eq_enrgpsi}
 E_{n} = E_{n}^{(0)} + \varepsilon \,E_{n}^{(1)} + \varepsilon
\,^{\,2}E_{n}^{(2)} + \ldots \\ 
\ket{\psi_{n}} = \ket{\psi_{n}^{(0)}} + \varepsilon \,\ket{\psi_{n}^{(1)}} + \varepsilon^{\,2}\ket{\psi_{n}^{(2)}} +\ldots
 \end{equation}
<!--   -->
$$H_{0}\, \ket{\psi_{n}^{(0)}} = E_{n}^{(0)}\,\ket{\psi_{n}^{(0)}}$$ 
<!--  -->
де індекси зверху відповідають порядку теорії збурень. 
<!--  -->
$$
\hat{H}_0\,\ket{\psi^{(1)}_{n}} + \hat{V}\,\ket{\psi^{(0)}_{n}} = E^{(0)}_n\,\ket{\psi^{(1)}_{n}}+E^{(1)}_n\,\ket{\psi^{(0)}_{n}}
$$
<!--  -->
Підставляючи \eqref{eq_enrgpsi} до \eqref{eq_hampsi} і  збираючи величини одного порядку по $\varepsilon$, отримаємо:
<!--  -->
$$\varepsilon^{k}:\,\,H_{0} \ket{\psi_{n}^{(k)}} + V\ket{\psi_{n}^{(k - 1)}} =
\sum\limits_{m = 0}^{k} E_{n}^{(m)} \ket{\psi_{n}^{(k - m)}}
\,\,\,,\,\,\,k = 1,\,2,\,\ldots\,,$$
<!--  -->
\begin{equation}\label{17eq3}
 \braket{\psi_{n}}{\psi_{n}}= 2{\rm Re}\left(\sum\limits_{i,k = 0}
 \varepsilon^{i + k}\braket{\psi_{n}^{(i)}}{\psi_{n}^{(k)}} \right) = 1\,\,\,
\Rightarrow \sum\limits_{i + k = 1} \varepsilon^{i +
k}\braket{\psi_{n}^{(i)}}{\psi_{n}^{(k)}}  = 0\,.
\end{equation}
<!--  -->
Тут ми cкористалися тим, що хвильові функції дискретного спектра можуть бути нормовані. Наприклад, перший порядок дає:
<!--  -->
\begin{equation}
\label{17eq4}
\begin{array}{l}
H_{0} \ket{\psi_{n}^{(1)}} + V \ket{\psi_{n}^{(0)}} = E_{n}^{(0)} \ket{\psi_{n}^{(1)}} 
 + E_{n}^{(1)} \ket{\psi_{n}^{(0)}} \,,\, \\
 \braket{\psi_{n}}{\psi_{n} } = 1 + 2\,\varepsilon {\rm Re}\,\braket{\psi_{n}^{(1)}}{\psi_{n}^{(0)}} + \,\ldots \\
 \end{array}
\end{equation}
<!--  -->
Першу поправку до хвильової функції можна розкласти в ряд по повній системі власних функцій незбуреного гамільтоніана:
<!--  -->
\begin{equation}
\label{17eq5} \ket{\psi_{n}^{(1)}} = \sum\limits_{m} 
\braket{\psi_{m}^{(0)}}{\psi_{n}^{(1)}}\,
\ket{\psi_{m}^{(0)}}\,.
\end{equation}
<!--  -->
Тому знаходження першої поправки зводиться до знаходження коефіцієнтів ряду \eqref{17eq5}. Знаходячи проекцію першого рівняння \eqref{17eq5} на $\ket{\psi^{(0)}_{m}}$ маємо:
<!--  -->
\begin{equation}
\label{17eq6} E_{m}^{(0)} 
\braket{\psi_{m}^{(0)}}{\psi_{n}^{(1)}} + \bra{\psi_{m}^{(0)}}V\ket{\psi_{n}^{(0)}}
= E_{n}^{(0)} \braket{\psi_{m}^{(0)}}{\psi_{n}^{(1)}} + E_{n}^{(1)} \delta _{mn}\,.
\end{equation}
<!--  -->
Використаємо те, що стани визначені з точністю до фазового множника. Тому фазу $\ket{\psi_{n}^{(0)}}$ можна обрати таким чином, щоб:
<!--  -->
\begin{equation}\label{17eq7} 
\braket{\psi_{n}^{(1)}}{\psi_{n}^{(0)}} = 0\,.
\end{equation}
<!--  -->
Користуючись цим і \eqref{17eq6}, ми можемо записати:
<!--  -->
\begin{align}\label{17eq8}
 E_{n}^{(1)} &= V_{nn} \notag \\
 \braket{\psi_{m}^{(0)}}{\psi_{n}^{(1)}}& = 
 \frac{V_{mn}}{E_{n}^{(0)} - E_{m}^{(0)}}\,\,,\,\,\,m \ne n \\
 V_{mn} \,& = \,\av{\psi_{m}^{(0)}}{V}{\psi_{n}^{(0)}}\notag\,.
 \end{align}
<!--   -->
Необхідна умова застосування теорії збурень витікає з вимоги малості першої поправки до хвильової функції. Ця вимога буде очевидно виконана, якщо:
$$
\frac{V_{mn}}{E_{n}^{(0)} - E_{m}^{(0)}}\, \ll \,1\,,\,\,\,\forall\, m,\,n\,.
$$
За індукцією неважко зрозуміти, що для знаходження поправки до енергії $n$-го порядку достатньо знати хвильову функцію з точністю до $n-1$ - го порядку. Знайдемо друге наближення для енергії та хвильової функції. Згідно \eqref{17eq3} для цього порядку маємо:
<!--  -->
\begin{equation}\label{17eq9}
\begin{array}{rl}
 H_{0} \psi_{n}^{(2)} + V\,\psi_{n}^{(1)} =& E_{n}^{(0)} \psi_{n}^{(2)} +
 E_{n}^{(1)} \psi_{n}^{(1)} + \,E_{n}^{(2)} \psi_{n}^{(0)}\\
 \braket{\psi_{n}}{\psi_{n}} =& 1 + \,\varepsilon^{2}\,\left(2{\rm Re}\,
 \braket{\psi_{n}^{(2)}}{\psi_{n}^{(0)}} + \braket{\psi_{n}^{(1)}}{\psi_{n}^{(1)}}\right) +
\,\ldots
 \end{array}
\end{equation}
<!--  -->
Діючи аналогічно, отримуємо:
<!--  -->
$$
\,E_{m}^{(0)} \braket{\psi_{m}^{(0)}}{\psi_{n}^{(2)}}+ 
\av{\psi_{m}^{(0)}}{V}{\psi_{n}^{(1)}} = E_{n}^{(0)} \braket{\psi_{m}^{(0)}}{\psi_{n}^{(2)}} + E_{n}^{(1)} \braket{\psi_{m}^{(0)}}{\psi_{n}^{(1)}} + \,E_{n}^{(2)} \delta _{mn}\,.
$$
<!--  -->
Якщо підставити отримані раніше результати для першого порядку, то для другої поправки до енергії знаходимо
<!--  -->
\begin{equation}\label{eq_2ndorder} 
E_{n}^{(2)} =  \sum\limits_{m \ne n} \frac{\left|V_{mn}\right|^{\,2}}{E_{n}^{(0)} - E_{m}^{(0)}}
\end{equation}
<!--  -->
Для поправки другого порядку хвильової функції з \eqref{17eq3} і
\eqref{17eq9} витікає, що:
<!--  -->
\begin{equation}
\label{17eq11}
\left\langle \psi_{m}^{(0)} \vert \psi_{n}^{(2)} \right\rangle = 
-\frac{V_{nn} V_{mn}}{\left( {E_{n}^{(0)} - E_{m}^{(0)}}
\right)^{2}} - \sum\limits_{k \ne n} \frac{V_{mk} \,V_{kn}}
{\left( {E_{n}^{(0)} - E_{k}^{(0)}} \right)\left( {E_{m}^{(0)} -
E_{n}^{(0)}}  \right)}
\end{equation}
<!--  -->
$$
2\,{\rm Re}\,\braket{\psi_{n}^{(0)}}{\psi_{n}^{(2)}} = -
\,\,\sum\limits_{m \ne n\,} \frac{\left| {V_{mn}}
\right|^{2}}{\left(E_{n}^{(0)} - E_{m}^{(0)}
\right)^{2}}
$$
Фазу $\ket{\psi_{n}^{(2)}}$ можна обирати довільно, оскільки
все залежить тільки від дійсної частини проекції $\ket{\psi_{n}^{(2)}}$ на $\ket{\psi_{n}^{(0)}}$. Тому оберемо її так:
<!--  -->
\begin{equation}
{\rm Re}\,\braket{\psi_{n}^{(0)}}{\psi_{n}^{(2)}} = 
\braket{\psi_{n}^{(0)}}{\psi_{n}^{(2)}} = - 
\frac{1}{2}\,\sum\limits_{m \ne n\,}
\frac{\left| V_{mn} \right|^{2}}
{\left(E_{n}^{(0)} - E_{m}^{(0)}\right)^{2}}
\end{equation}
<!--  -->
Це остаточно дає:
<!--  -->
\begin{eqnarray}\label{17eq12}
 \ket{\psi_{n}^{(2)}} &=& \sum\limits_{m \ne n} \sum\limits_{k \ne n}
\frac{V_{mk} \,V_{kn}}{\left(E_{n}^{(0)} - E_{k}^{(0)}
\right)\left(E_{n}^{(0)} - E_{m}^{(0)}  \right)}
\ket{\psi_{m}^{(0)}}\, - \,V_{nn} \sum\limits_{m \ne n}
\frac{V_{mn}}{\left(E_{n}^{(0)} - E_{m}^{(0)}\right)^{2}}
\ket{\psi_{m}^{(0)}}  - \nonumber\\
    &&\frac{1}{2}\sum\limits_{m \ne n} 
    \frac{\vert V_{mn} \vert^{2}}{\left( E_{n}^{(0)} - E_{m}^{(0)}
\right)^{2}}\,\ket{\psi_{n}^{(0)}}
 \end{eqnarray}

### Природа сил ван дер Ваальса
З \eqref{eq_2ndorder} витікає важливий загальний факт: поправка другого порядку до основного стану $E^{(2)}_{0}$ завжди(!) від'ємна. Цей факт дозволяє обґрунтувати залежність ван-дер-Ваальсових сил від відстані між атомами, які знаходяться у основному стані. Оператор диполь-дипольної взаємодії, 
<!--  -->
\begin{equation}\label{eq_dipdip}
	\hat{V}_{d} = \frac{\mathbf{d}_{1}\cdot \mathbf{d}_{2}}{R^3} - 3\frac{(\mathbf{d}_{1}\cdot \mathbf{R})\,(\mathbf{d}_{2}\cdot \mathbf{R})}{R^5}
\end{equation}
<!--  -->
який є головним вкладом на великих відстанях, у першому порядку теорії збурень не дає внеску, оскільки середнє значення дипольного моменту у довільному стаціонарному стані у воднеподібного атома дорівнює 0. Тому основний вклад в енергію дає
другий порядок. Потенціал диполь - дипольної взаємодії убиває з відстанню як $1 / r^{3}$. Зрозуміло, що поправка другого порядку приведе до закону $\Delta E\sim - 1 / r^{6}$, що і відповідає слабкому притягненню. Наведене квантово-механічне пояснення виникнення таких дисперсійних сил було дано [Ф. Лондоном (F. London)](https://phy.duke.edu/about/history/historical-faculty/fritz-london) разом з В. Гайтлером. 

У незбуренному стані хвильові функції системи двох атомів мають вигляд добутку (тут ми нехтуємо обмінними ефектами, які пов'язані з тотожністю частинок, оскльки відстань між атомами значно більше характерного масштабу  $\sim a_{B}$ локалізації електронів), тому 

Покажемо, як обчислюється поляризуемость атома водню у основному стані $\ket{1,0,0}$. 
Враховуючи ізотропний характер цього стану, достатньо обчислити матричний елемент вздовж напрямка $z$:

$$
\alpha =-2\, \sum\limits_{n}\frac{\left\vert \av{1,0,0}{ d_{z}}{n,1,0} \right\vert^2}{E^{(0)}_{1}- E^{(0)}_{n}}
$$

Тут враховано те, що завдяки симетрії, єдиними ненульовими матричними елементами є $\av{n,1,0}{d_{i}}{1,0,0}$ (ортогональність сферичних гармонік і правило складання моментів).

Завдяки точному розвязку, вираз для поляризовності основного стану може бути отримано аналітично. Введемо допоміжний оператор:

$$
\frac{1}{i}[\hat{Z},\hat{H}] = z
$$
так що
$$
\sum\limits_{n}\frac{\left\vert \av{1,0,0}{z}{n,1,0} \right\vert^2}{E^{(0)}_{1}- E^{(0)}_{n}}\sim
\sum\limits_{n}\av{1,0,0}{ z \vphantom{\hat{Z}} }{n,1,0} \av{n,1,0}{ \hat{Z} }{1,0,0} =
\av{1,0,0}{z\,\hat{Z}}{1,0,0}
$$
Відповідно:

$$\label{polazalpha}
\alpha = -2\,i\,\,\av{1,0,0}{z\,\hat{Z}}{1,0,0}
$$

За означенням:

$$\label{hamz}
\hat{H}\,\hat{Z}\ket{1,0,0} = \hat{Z}\hat{H}\ket{1,0,0}-i\,z\,\ket{1,0,0} =
E_0\,\hat{Z}\ket{1,0,0}-i\,z\,\ket{1,0,0}
$$

Оскільки $z=r\,\cos{\theta} \sim r\,Y_{1,0} $ у координатному представленні можна записати дію оператора у вигляді:

$$\av{r\,\theta,\phi}{\hat{Z}}{1,0,0} = f(r)\,\cos{\theta}\,\psi_{1,0,0}(r)$$

Подставляючи це у \eqref{hamz} отримаємо:

$$-\frac{(r\,f)''}{2r}+\frac{f}{r^2}-f'\frac{\psi'_{1,0,0}}{\psi_{1,0,0}} =-i\,r $$

```python
a,b,r,f=symbols('a b r f')
f = Function('f')
psi0=2*exp(-r)
eq=(-diff(r*f(r),(r,2))/(2*r)+f(r)/r**2-diff(f(r),r)*diff(psi0,r)/psi0+r)
simplify(eq.subs(f(r),a*r**2+b*r))
```

таким чином, знаходимо вираз:
$$
f(r) = -i\,r\left(1+\frac{r}{2}\right)
$$
В результаті інтегрування в \eqref{polazalpha} отримаємо:

```python
x  = symbols('x')
2*integrate((psi0)**2*(r*(1+r/2))*r**3,(r,0,oo))*integrate(x**2/2,(x,-1,1))
```

$$
\alpha = \frac{9}{2}\,a^3_{B}\,,
$$
що трохи більше ніж проста оцінка $\alpha = \frac{4\,\pi}{3}\,a^3_{B}$.

__Розв'язок__ Гамільтоніан системи:
<!--  -->
$$
\hat{H} = -\frac{\hbar^2}{2\,m\,R^2}\,\frac{\partial^2}{\partial\,\varphi^2} -q\,\mathcal{E}\,R\,\cos{\varphi}
$$
<!--  -->
Незбурені стани відповідають магнітному квантовому числу:
<!--  -->
$$
E^{(0)}_{m} = \frac{\hbar^2\,n^2}{2\,m\,R^2}\,,\,\,\,\ket{n} \to \psi^{(0)}_{n}(\varphi) = \frac{e^{i\,n\,\varphi}}{\sqrt{2\pi}} 
$$
Знову завдякі симетрії перша поравка до рівнів енергії дорівнює 0 (самостійно).  
Друга поправка взначає поляризуємість:
<!--  -->
$$
\alpha_{n} = -2\,q^2\,R^2\sum\limits_{k}\,\frac{\left\vert \av{k}{\cos{\varphi}}{n} \right\vert^2}{E^{(0)}_{n} - E^{(0)}_{k}} = -\frac{m\,q^2\,R^4}{\hbar^2}\,
\left(\frac{1}{n^2 - (n+1)^2}+\frac{1}{n^2 - (n-1)^2}\right) = 
R^3\,\frac{A}{1-4\,n^2}
$$
<!--  -->
$$A =  \frac{2\,q^2/R}{\frac{\hbar^2}{2\,R^2}}$$

### Варіаційний метод
Варіаційний метод є важливим наближеним методом розв'язання задач в квантовій механіці. Він заснований на тому, що стаціонарне рівняння Шредінгера може бути виведено як
необхідна умова екстремуму функціонала середнього знчення енергії. Дійсно маємо функціонал середнього значення енергії з умовою нормування:
\begin{equation}
\label{avenrgfunc} 
\av{\psi}{\hat{H}}{\psi}\,\,,\,\vert\vert\,\psi
\vert\vert\, = 1\,.
\end{equation}
Роль невизначеного множника Лагранжа відіграє енергія основного
стану. Тоді задача на умовний екстремум для \eqref{avenrgfunc}:
\begin{equation}
\label{stache} \delta \left(\av{\psi}{\hat{H}}{\psi}\, - E\braket{\psi}{\psi}\right)= 0\,.
\end{equation}
Хвильові функції і рівні енергії вищих станів визначаються
аналогічно з додатковими умовами ортогональності до всіх нижчих
станів. Практичне застосування полягає у виборі певного сімейства хвильових функцій яке залежить від параметрів. Ці параметри визначаються з пошуку мінімума середнього значення енергії. Комбінація теорії збурень із варіаційною процедурою дає дуже добрі результати. Продемонструємо це прикладами. <br>

__Задача__ Побудуємо варіаційним методом енергію основного стану для електрона у атомі водню. В якості пробної хвильової функції візьмемо $\psi_0(r) = C\,e^{-a\,r}$. <br>

Запишемо гамільтоніан задачі:

\begin{equation}\label{ham_Z}
\hat{H} = \frac{\hat{p}^2}{2\,m}-\frac{Z\,e^2}{r}
\end{equation}

Визначимо константу нормування хвильової функції:

$$  1= \int |\psi_0|^2\,dV = 4\,\pi\,C^2\,
\int\limits_{0}^{+\infty} r^2e^{-2\alpha\,r}\,dr =\pi\,C^2\,\frac{d^2}{d \alpha^2}\int\limits_{0}^{+\infty}e^{-2\alpha\,r}\,dr=\frac{\pi\,C^2}{\alpha^3}
\rightarrow\, C =\sqrt{\frac{\alpha^3}{\pi}}
 $$

Обчислимо середнє значення гамільтоніана \eqref{ham_Z} у стані $\psi$:

\begin{equation}\label{hamZ_aver}
  \langle \hat{H}  \rangle_{\psi} =
  4\pi\frac{\hbar^2C^2}{2 m}\int\limits_{0}^{+\infty}r\,e^{-\alpha r}\,(r\,e^{-\alpha r})''\,dr - 4\,\pi\,Z e^2\,C^2\,\int\limits_{0}^{+\infty}r\,e^{-2\alpha r}\,dr
\end{equation}

В результаті кінетична енергія:

$$\left\langle \frac{\hat{p}^2}{2m}\right\rangle_{\psi} = \frac{\hbar^2}{2\,m}\,\alpha^2$$

Потенціальна енергія:

$$-Ze^2\left\langle \frac{1}{r}\right\rangle_{\psi} =
-Z e^2 \alpha$$

Повна енергія:

$$E(\alpha) = \frac{\hbar^2}{2\,m}\,\alpha^2 - Ze^2\alpha\, \rightarrow \,E'(\alpha) = 0\rightarrow \,\alpha_{\text{min}} = m\,\frac{Z\,e^2}{\hbar^2}$$

боровський радіус

$$a_{B} = \frac{\hbar^2}{Z\,e^2\,m}$$

<a id="wv0">Хвильова функція основного стану:</a>

\begin{equation}\label{psi0}
\psi_{0}(r) = \frac{1}{\sqrt{\pi\,a^3_{B}}}\,e^{-\frac{r}{a_{B}}}
\end{equation}

Внескі в енергію:

\begin{align*}
  E_{\text{kin}} =& \frac{m Z^2 e^4}{2 \hbar^2} \\
  E_{\text{pot}} =& -\frac{m Z^2 e^4}{\hbar^2}= -2 E_{\text{kin}}\\
  E_{\text{min}} =& - \frac{m Z^2 e^4}{2 \hbar^2} = -\,Z^2\,13.6 \,\text{эВ}
\end{align*}

Бачимо, що для середніх значень теорема віріала виконується. 
Код SymPy (у безрозмірних змінних) див. нижче

```python
a=symbols('a', positive=True)
psi0=exp(-a*r)# 1,0,0 wf
cnorm=4*pi*integrate(psi0**2*r**2,(r,0,+oo))# int norm wf
psi0=psi0/sqrt(cnorm) # norm wf
ekin=4*pi*integrate((r*diff(psi0,r))**2,(r,0,+oo))/2 # ekin
epot=-4*pi*integrate((psi0)**2*r,(r,0,+oo)) # epot
enrg0=ekin+epot # energy
# minimize
amin=solve(diff(enrg0,a),a,dict=True)
amin
```

##  Теорія збурень для станів виродженого дискретного спектра.

Розглянемо тепер вироджений випадок. Тут одразу виникає проблема вибору функцій нульового наближення, яка не виникала у попередньому випадку. З'ясуємо ситуацію за допомогою аналогії з моментом інерції.

Для високо симетричних тіл, які мають осі симетрії третього порядку і вище (наприклад правильні багатогранники), всі три головних моменту інерції збігаються, а головні напрямки  можуть бути обрані довільно. Тому з точки зору кінематики і динаміки обертального руху такі тіла еквівалентні кулі з відповідним моментом інерції. Свобода у виборі головних напрямків може бути обмежена, якщо симетрію тіла порушити, наприклад, причепив додаткові малі точкові маси). У випадку __загального положення__, таких мас, свобода у виборі головних осей зникне і сфера інерції перетворюється у звичайний еліпсоїд з різними напівосями. Виродження всіх напрямків осей обертання у тілі зникне. Збурення виділяє з всіх вироджених напрямків ті, які неперервним чином переходять у ті напрямкі відносно яких матриця
збурення діагональна. Таким чином, поперед усе матрицю оператора збурення слід діагоналізувати у базисі власних станів виродженого рівня. Припустимо, що маємо $s$-кратно вироджений рівень. У першому порядку отримаємо:
<!--  -->
\begin{equation}\label{pert1}
\hat{V}\,\ket{\phi^{(0)}} = E^{(1)}\,\ket{\phi^{(0)}}\\
\ket{\phi^{(0)}}  = \sum\limits_{k=1,s} c^{(0)}_{k}\,\ket{\psi^{(0)}_{k}}\\
{\rm det} |\hat{V} - E^{(1)}\,I| = 0
\end{equation}
<!--  -->
Тут $\ket{\psi^{(0)}_{k}}$ - стани, які відповідають виродженому рівню кратності $s$. Рівняння \eqref{pert1} визначає правильну функцію нульового наближення і є рівнянням на власні значення матриці збурення у базисі $\ket{\psi^{(0)}_{k}} \,\,,\,\,\,k = 1\ldots s$. Як відомо з курсу лінійної алгебри необхідною умовою існування нетривіальних рішень такого рівняння є рівність 0 детермінанта цієї системи. Отримане поліноміальне рівняння степени $s$ має не більш ніж $s$ різних коренів, які визначають поправки до незбудженого рівня:
<!--  -->
\begin{equation}\label{pert2}
{\rm det} |\hat{V} - E^{(1)}\,I| = 0
\end{equation}
<!--  -->
Таким чином, як говорять, виникає розщеплення виродженого рівня на не більш ніж $s$ підрівней (зняття виродження). Хвильові функції, які відповідають цим підрівням є власними векторами оператора збурення, обмеженого на підпростір станів незбуреного рівня.

В якості ілюстації цих результатів розглянемо ефект Штарка в атомі водню та воднеподібних атомах. Під цим ефектом розуміють розщеплення атомних рівнів у слабкому, у порівнянні з внутрішньоатомним, однорідному електростатичному полі. Це розщеплення має різний характер у випадку  атома водню і воднеподібних атомів лужних металів. Ефект Штарка для атома водню є ефектом першого порядку за напруженністю поля. Для воднеподібних атомів розщеплення виникає 
тільки у другому порядку теорії збурень.

Дійсно, гамільтоніан атома у однорідному електростатичному полі має вигляд:

\begin{equation}
\label{hamstark} H = H_{0} - \mathbf{d}\,\cdot\,\mathbf{\mathcal{E}}
\end{equation}

Однорідне електричне поле порушує сферичну симетрію задачі. Однак повного зникнення виродження не відбувається. Стани з протилежними проекціями моменту імпульсу мають однакову енергію. Це пов'язано з тим що гамільтоніан системи є інваріантним відносно віддзеркаленя у площині яке містить електричне поле $\mathbf{\mathcal{E}}$. При цьому стани з $\hbar \,m$ і $ - \hbar \,m$ переходять один до одного (момент імпульсу - псевдовектор!):

$$
\{x,y,z\}\to - \{x,-y,z\} \Rightarrow \ket{k,m}\leftrightarrow \ket{k,-m}
$$

Тому ці стани мають у зовнішньому аксиальному електричному полі одинакову енергію.
Матриця оператора збурення для станів виродженого $n$-го рівня (розміром $n^2 \times n^2$) буде містити ненульові елементи (правило складання моментів):

\begin{equation}
\label{stark_matrix} 
\av{\psi_{n\,l\,m} }{\mathbf{d}}{\psi_{n\,l'\,m\,'}} = A_{l,m} \,\delta_{l,\,l'\,\pm 1}\,\delta_{m,m'}\\
\psi_{n\,l\,m} = R_{n\,l}(r)\,Y_{l\,m}(\theta,\varphi)
\end{equation}

що відповідає розщеплення в першому порядку теорії збурень. Таким чином, ефект Штарка в атомі водню це
ефект першого порядку (лінійний по полю). Наприклад, при $n=2$, є 4-х кратне виродження рівня енергії (ми не враховуємо вироджденя за проекцією спіна, т.к. електричне поле не впливає на проекцію спіна у нерелятивистському наближені (надтонке розщеплення). Так що в базисі таких станів $\ket{l,m}:\ket{0,0},\ket{1,0},\ket{1,\pm 1}$  матриця гамільтоніана має вигляд:
<!--  -->
\begin{equation}
\ket{2,0,0}= (1-r/2)\,\frac{e^{-r/2}}{\sqrt{2}}\\
\ket{2,1,m=\pm 1,0} = \frac{r}{2\,\sqrt{6}}\,e^{-r/2}\,Y_{1,m=\pm 1, 0}(\theta,\varphi)
\end{equation}
<!--  -->
\begin{equation}\label{starkl1}
  \hat{H}_{n=2} = 
  \begin{pmatrix}
E^{(0)}_2 & 0 & 0 & 0 \\
0 & E^{(0)}_2 & 0 & 0 \\
0 & 0 & E^{(0)}_2 & 0 \\
0 & 0 & 0 & E^{(0)}_2
\end{pmatrix}
+ 
\begin{pmatrix}
0\vphantom{E^{(0)}_2} & V_{1,0} & 0 & 0 \\
V_{1,0} & 0\vphantom{E^{(0)}_2} & 0 & 0 \\
0 & 0 & 0\vphantom{E^{(0)}_2} & 0 \\
0 & 0 & 0 & 0\vphantom{E^{(0)}_2} 
\end{pmatrix}
\end{equation}

тобто один 4-х кратно вироджений рівень розщеплюється на 3 рівня, один з яких є 2-кратно виродженим. Розщеплюється рівні $\ket{n=2,l=0,m=0}$, $\ket{n=2,l=1,m=0}$. Стани $\ket{n=2,l=1,m=\pm1}$ не розщеплюються електростатичним полем (див. Рис.).

Отже енергія розщеплення дорівнює:
<!--  -->
$$
E^{(1)}_{2} = 2\vert V_{1,0}\vert = 6\,a^3_{B}\,\mathcal{E}
$$
<!--  -->
Може однак виявитися, що матриця оператора збурення дорівнює 0 на всіх незбурених вироджених векторах виродженого рівня $E^{(0)}_{n}$. В такому випадку розщеплення буде другого порядку, а для знаходження функцій нульового наближення треба врахувати внески від інших станів:
<!--  -->
\begin{equation}\label{pertdeg0}
\left(\hat{H}_0+\hat{V}\right)\,\ket{\phi} = (E^{(0)}_{n}+E_n^{(2)})\,\ket{\phi}\\
\ket{\phi}  = \sum\limits_{n'=1,\ldots,s} c^{(0)}_{n'}\,\ket{\psi^{(0)}_{n'}}+\sum\limits_{k} c^{(1)}_{k}\,\ket{\psi^{(0)}_{k}}
\end{equation}
Підставляючи у відповідних порядках малості маємо:
$$
\sum\limits_{k} c^{(1)}_{k}\,E^{0}_{k}\ket{\psi^{(0)}_{k}} +\sum\limits_{n'=1,\ldots,s} c^{(0)}_{n'}\,\hat{V}\,\ket{\psi^{(0)}_{n'}} = E^{(0)}_{n}\,\sum\limits_{k} c^{(1)}_{k}\,\ket{\psi^{(0)}_{k}}\\ 
\sum\limits_{k} c^{(1)}_{k}\,\hat{V}\,\ket{\psi^{(0)}_{k}} = E_n^{(2)}\,\sum\limits_{n'=1,\ldots,s} c^{(0)}_{n'}\,\ket{\psi^{(0)}_{n'}}
$$
З першого рівняння витікає:
$$
c^{(1)}_{m} = \sum\limits \frac{V_{mn'}}{E^{(0)}_{n}-E^{(0)}_{m}}\,c^{(0)}_{n'}
$$
підставляючи цей результат до другого отримаємо рівняння на власні вектори і власні значення:
$$
\sum\limits_{m,n'} \frac{V_{nm}V_{mn'}}{E^{(0)}_{n}-E^{(0)}_{m}}\,c^{(0)}_{n'} = E^{(2)}\,c^{(0)}_{n} \Rightarrow \left(\hat{V}^{(2)} - E^{(2)}\right)\,c^{(0)} = 0 \rightarrow \text{det}\left(\hat{V}^{(2)} - E^{(2)}\right)=0
$$
