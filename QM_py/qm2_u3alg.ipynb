Група $SU(3)$ відіграє важливу роль у фізиці елементарних частинок та теорії сильної взаємодії (кваркова структура адронів). Її алгебру генераторів можна побудувати за аналогією з випадком $SU(2)$-симетрії розглянувши матриці, які утворюють $SU(3)$-алгебру. Таких матриць є 8 і у звичайному просторі $\mathbb{C}^3$ вони мають вигляд:

```python
x1=Matrix([[0, 1,0], [1,0, 0], [0,0,0]])/2
x2=Matrix([[0, -I,0], [I,0, 0], [0,0,0]])/2
x3=Matrix([[1, 0,0], [0,-1, 0], [0,0,0]])/2
x4=Matrix([[0, 0,1], [0,0, 0], [1,0,0]])/2
x5=Matrix([[0,0,-I], [0,0, 0], [I,0,0]])/2
x6=Matrix([[0, 0,0], [0,0, 1], [0,1,0]])/2
x7=Matrix([[0,0,0], [0,0, -I], [0,I,0]])/2
x8=Matrix([[1, 0,0], [0,1, 0], [0,0,-2]])/3
```

В цьому випадку вже є 2 матриці $X_3, X_8$ які є діагональними і лінійно-незалежні. Аналогічно до $SU(2)$-випадку введемо сходинкові оператори:

```python
x1+I*x2
```

```python
x4+I*x5
```

```python
x6+I*x7
```

Їх комутатори із спряженими:

```python
(x1+I*x2)*(x1-I*x2)-(x1-I*x2)*(x1+I*x2)
```

```python
(x1*x2-x2*x1)/I
```

Кожна пара операторів $(X_1,X_2)$, $(X_4,X_5)$, $(X_6,X_7)$ породжує відповідні $SU(2)$-підалгебри. Зрозуміло, що аналогом z-компоненти для $X_1,X_2$ є $X_3$. Знайдемо відповідні $z$-оператори для інших:

```python
sz45=(x4*x5-x5*x4)/I
sz45
```

```python
sz67=(x6*x7-x7*x6)/I
sz67
```

```python
(x4+I*x5)*(x4-I*x5)-(x4-I*x5)*(x4+I*x5)==2*sz45
```

```python
(x6+I*x7)*(x6-I*x7)-(x6-I*x7)*(x6+I*x7)==2*sz67
```

Оператори $X_3,X_8$ є лінійними комбінаціями z-операторів "спіну":
$$\label{x384567}
X_3 = X^{(45)}_{3}-X^{(67)}_{3}\,,\,\,\,\
X_8 = 2/3\left(X^{(45)}_{3}+X^{(67)}_{3}\right)\\ 
X^{(45)}_{3} =3/4\,X_8 + X_3/2\,,\,\,\,\, X^{(67)}_{3} = 3/4\,X_8 - X_3/2
$$

```python
(sz45-sz67)
```

```python
2*(sz67+sz45)/3
```

Отже неприводимий простір станів, на яких діє група симетрії буде визначатися вже двома власними значеннями (у $SU(2)$-випадку, було орбітальне квантове число $l$) які задають власні значення операторів $X_3, X_8$, які комутують, отже мають спільний базис. Власні вектори оператора $X_3$ відповідають власним значенням $0,\pm 1/2,\pm 1, ...$. Відповідно у оператора $X_8$: $0,\pm 1/3,\pm 2/3$, тобто з кроком 1/3.

Як бачимо, всі базисні вектори представлень повинні належати певній гексагональній підґратці вихідної ґратки. Не всі гексагональні підґратки які відповідають елементарним зсувам вихідної ґратки сумісні з \eqref{x384567}. Відповідно до цих співвідношень тільки якщо гексагональна ґратка буде включати вузли $(0,0)$ або $(0,2/3)$ або $(0,-2/3)$ вона складається з векторів, на яких діє SU(3)-алгебра.

Як і у SU(2)-випадку оберемо вектор "максимального рангу" тобто який має максимальне власне значення оператора $X_8$ і при цьому максимальне можливе значення $X_3$. Простір такого представлення $R^{(\lambda,\mu)}$ зручно нумерувати максимальними проекціями $\lambda = 2X_3,\mu = 2X^{(67)}_{3}$, тоді:

$$X^{(45)}_{3} \to \frac{\lambda+\mu}{2}\,,\,\,\,X_8 \to \frac{\lambda+2\mu}{3}$$
