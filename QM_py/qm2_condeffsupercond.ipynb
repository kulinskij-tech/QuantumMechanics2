## Надпровідність
<!--  -->
Зрозуміло, що явища надпровідності (зникнення опору) і надплинності (зникнення в'язкості) пов'язані між собою. Надпровідність можна уявляти як надплинність електронної підсистеми у НП. Але електрони є фермі-частинками і відповідно не можуть утворювати бозе-айнштайнівський надплинний конденсат. Ключовим механізмом є утворення пар електронних збуджень які вже є бозонами. Таким механізмом у низькотемпературних надпровідниках є електрон-фононна взаємодія, тобто поляризація кристалічної ґратки при русі електронного збудження. В результаті обміну фононами два збудження утворюють зв'язаний стан - [куперівську пару](https://en.wikipedia.org/wiki/Cooper_pair).

Перехід НП-НМ як фазовий перехід у низькотемпературний впорядкований стан пов'язаний з дією сил взаємодії між частинками. Тому основні зусилля при побудові мікроскопічної теорії були спрямовані на пошук механізму притягнення між ЕЗ у металі. Такий механізм дійсно існує незважаючи на те, що такі збудження є однаково зарядженими. Це механізм поляризації кристалічної ґратки при росповсюдженні ЕЗ. Завдяки великій різниці у масі така взаємодія дуже мала оскільки малі зміщення іонів внаслідок кулонівської взаємодії з квазічастинкою.
Гамільтоніан системи може бути записаний у вигляді:
<!--  -->
\begin{equation}\label{ham}
	\hat{H} = \hat{H}^{(0)}_{el} + \hat{H}^{(0)}_{ph} + \hat{H}_{el-ph}\,.
\end{equation}
<!--  -->
Останній доданок якраз і представляє взаємодію
між зазначеними підсистемами. Вирішальним експериментальним фактом у розумінні мікроскопічного механізму надпровідності було відкриття ізотопічного ефекту. Ізотопічним ефектом називається залежність властивості системи від ізотопічного складу ґратки. Ізотопічний склад змінюється опромінюванням речовини нейтронами відповідної енергії. Ізотопи як відомо хімічно еквівалентні і відрізняються тільки масою ядра. Ядро впливає на електронну підсистему у вигляді кінетичної енергії руху ядер. Зсув обумовлений різницею маси ізотопів дозволяє робити висновки про вплив руху ядер на властивості системи. У випадку надпровідності спостерігається зсув критичної температури від ізотопічного складу. Експериментальні дані відповідають такій залежності:
<!--  -->
\begin{equation}\label{isotopeff}
	T_c \,\sqrt{M} \approx \text{const} \Rightarrow \,\, T_c \sim \frac{1}{\sqrt{M}}\,.
\end{equation}
<!--  -->
Оскільки маса ядер визначає у першу чергу частоти коливань кристалічної ґратки, зокрема дебаївську
частоту $\omega_{D}$, то звідси випливає, що механізм надпровідності обумовлений взаємодією електронів з
фононами. Цей механізм був запропонований незалежно [Х. Фрелихом](https://journals.aps.org/pr/abstract/10.1103/PhysRev.79.845) і [Дж. Бардіном](https://journals.aps.org/pr/abstract/10.1103/PhysRev.79.167.3). Він полягає у слабкій поляризації кристалічної ґратки ЕЗ
(електроном або діркою) із збудженням фонону. Цей фонон може взаємодіяти з іншим ЕЗ. Опис цих процесів зручно зображати у вигляді графічних схем, які називаються діаграмами Фейнмана. Слід зазначити, що таким процесам і відповідно діаграмам можна співставити певні вирази у рамках теорії збурень. Ця теорія збурень будується виходячи з припущення про малість вказаної взаємодії.
<!--  -->

<center><a id='feynmadiag'>Рис. Діаграми Фейнмана які зображують електрон-фононну та ефективну електрон-електронну взаємодії.</a></center>
<!--  -->

Таким чином теорію надпровідності слід будувати на
основі електрон-фононного механізму взаємодії і для цього потрібно конкретизувати члени гамільтоніана \eqref{ham}. Розв'язок задачі про спектр коливань кристалічної ґратки також відомий з курсу стат. фізики (теорія Дебая теплоємності кристалів), де показано, що енергію коливань кристалічної ґратки можна представити у вигляді суми нормальних мод, які є квантами звукового поля і називаються фононами. Вигляд спектру і відповідно гамільтоніанів незбурених систем, таким чином вважається відомим. Це сума енергій (гамільтоніанів) елементарних збуджень: частинок і дірок для електронного газу і фононів для ґратки:
<!--  -->
\begin{equation}\label{h0_electr_phon}
	\hat{H}^{(0)}_{el} = \sum_{\mathbf{p},s_z} \varepsilon(\mathbf{p})\,\hat{n}^{(el)}_{\mathbf{p},s_z}\,,\quad   \hat{H}^{(0)}_{ph} = \sum_{\mathbf{k},\alpha } \hbar \omega (\mathbf{k})\,\hat{n}^{(ph)}_{\mathbf{k},\alpha}\,,
\end{equation}
<!--  -->
Тут $\hat{n}$ т.з. оператори чисел заповнення станів. Стан електрона характеризується імпульсом і проекцією спіна $s_z=\pm 1/2$, стан фонона - хвильовим вектором і поляризацією $\alpha =1,2,3$. Середнє значення числа заповнення це середня кількість частинок у даному стані. Для ферміонів це кількість між 0 і 1 у силу принципу Паулі, для бозонів воно може бути як завгодно великим. По аналогії з відомою задачею про осцилятор, зручно також ввести оператори народження і знищення відповідних елементарних збуджень:
<!--  -->
$$\hat{n}^{(el)}_{\mathbf{p},s_z}  = \hat{a}^\dagger_{\mathbf{p},s_z}\,\hat{a}_{\mathbf{p},s_z}\,,\quad \hat{n}^{(ph)}_{\mathbf{k},\alpha }  = \hat{b}^\dagger_{\mathbf{k},\alpha }\,\hat{b}_{\mathbf{k},\alpha }$$
<!--  -->
Оператори народження і знищення діють на одночастинкові стани відповідно підвищуючи або понижуючи кількість частинок у цьому стані. Звичайно для ферміонних операторів принцип Паулі приводить до заборони народження ще однієї частинки у заповненому стані: 
$\hat{a} \left|\,1\,\right\rangle = 0$. За допомогою таких операторів для елементарних процесів можна будувати більш складні оператори які реалізують обмін частинками між станами за рахунок взаємодії. Наприклад можна записати формальний вираз для оператора взаємодії, співставляючи його [діаграмі](#feynmadiag):
<!--  -->
\begin{equation}\label{ham_elphonint}
	\hat{H}_{el-ph} = \sum_{\mathbf{p},\mathbf{k},s_z,\alpha} V(\mathbf{p},\mathbf{k})\,\hat{a}^\dagger_{\mathbf{p}-\mathbf{k},s_z}\hat{a}_{\mathbf{p},s_z}\,\hat{b}^\dagger_{\mathbf{k},\alpha }+ V^*(\mathbf{p},\mathbf{k})\,\hat{a}^\dagger_{\mathbf{p},s_z} \hat{a}_{\mathbf{p}-\mathbf{k},s_z}\,\hat{b}_{\mathbf{k},\alpha }\,.
\end{equation}
<!--  -->
Дійсно [діаграма](#feynmadiag) ілюструє, що спочатку знищується ЕЗ електронної підсистеми у стані з імпульсом $\mathbf{p}$ і народжується ЕЗ фононної підсистеми у стані імпульсом $\mathbf{k} \,(\hbar=1)$. Зазначимо, що вираз \eqref{ham_elphonint} є ермітово симетричним. Цей вираз складається з двох ермітово спряжених операторів. Оскільки ермітове спряження у квантовій теорії пов'язано з симетрією процесів у часі, можна вважати, що це є сума прямого і оберненого у часі процесу. В зв'язку з цим виникає інтерпретація дірки як частинки, яка рухається ''зворотньо за часом''. Зазначимо, що спочатку така інтерпретація була створена Р. Фейнманом для позитрона, оскільки квантова електродинаміка (КЕД) на понятійному рівні практично еквівалентна мікроскопічній теорії надпровідності. Різниця полягає у тому, що вакуумний основний стан у КЕД це справжній вакуум, у той час як у ТНП основним станом є основний стан НП у якому відлік імпульсів здійснюється від імпульсу Фермі $p_F$. Зрозуміло, середнє значення оператора \eqref{ham_elphonint} у будь-якому стані з певною кількістю ЕЗ дорівнює 0, оскільки цей оператор змінює кількість ЕЗ. Повний розв'язок задачі про обчислення спектра енергії гамільтоніана \eqref{ham} з вказаними доданками є важкою задачею. Однак принциповий результат можна отримати вже у рамках елементарної теорії збурень. Введення взаємодії призведе до зміни спектра незбуреної системи. При цьому можна вважати, що спектр фононної підсистеми практично не змінюється, так само як ми нехтували зміною об'єму або пружних властивостей металу при переході у НП стан. Основна увага приділяється зміні основного стану електронної підсистеми - сфері Фермі. Енергія основного стану у рамках теорії збурень повинна представлятися у вигляді:
<!--  -->
\begin{equation}\label{e0_pert}
E_{0} = E^{(0)}_{0} + \left\langle\, 0\,\right\vert \hat{H}_{ph-el} \left|\,0\,\right\rangle + \sum\limits_{n}\frac{\left\langle\, 0\,\right\vert \hat{H}^2_{ph-el} \left|\,0\,\right\rangle}{E^{(0)}_0 - E^{(0)}_n}+\ldots
\end{equation}
<!--  -->
Перша поправка, як було зазначено, дорівнює 0. Друга поправка до основного стану, як відомо, завжди від'ємна. Ця друга поправка якраз поредставляє ефективну взаємодією між ЕЗ електронного газу, яке зображується [діаграмою](#feynmadiag). Взагалі, зрозуміло, порядок теорії збурень збігається з кількістю вершин взаємодії у діаграмі. Тому можна сподіватись на те, що ця взаємодія ЕЗ електронної підсистеми завдяки фононній приведе до утворення зв'язанного стану ЕЗ з від'ємною енергією. Звичайно тут необгрунтовано припускається що більш високі поправки не змінюють такого висновку. В примитивній теорії збурень більш високі поправки зростають і робити належні кількісні висновки необхідно на основі більш строго підходу. Для нас, проте, важливе якісне уявлення суті.

Завдяки скінченності цієї енергії це веде до кардинальної перебудови основного стану. Дійсно, за можливості утворення зв'язанного стану нескінченно малий програш у кінетичній
енергії для пари таких частинок, які у результаті взаємодії ''вискочують'' з під сфери
Фермі, із надлишком компенсується скінченною енергією зв'язку такої пари. Якщо врахувати що кількість таких пар є макроскопічним, то зрозуміло, що основний стан зміниться якісним чином і ми отримаємо зовсім іншу систему зі певними властивостями і спектром ЕЗ.

### Куперівські пари
Очевидно, що для розгляду утворення зв'язаних станів необхідно обрати тільки ті пари одночастинкових станів між якими утворення зв'язанного стану є найбільш вірогідним. З міркувань симетрії зрозуміло, що основний стан двохферміонної системи повинен бути синглетним, тобто бути утвореним з пар станів з
протилежними проекціями спінів. Крім того повний імпульс такої пари у об'ємному НП, у рівновазі, повинен дорівнювати 0. Звідси випливає, що найбільш ефективно зв'язуються у пари стани $(\mathbf{p},s_z)$ і $(-\mathbf{p},-s_z)$. Картина утворення НП заснована на нестійкості основного стану і спектру ЕЗ електронної підсистеми відносно слабкого притягнення між ЕЗ. В результаті виникає ефект конденсації пар ЕЗ з утворенням зв'язаних станів які є бозе-частинками і відповідно можуть перебувати у одному стані  який описується єдиною хвильовою функцією. Такі стани називаються куперівськими. Можна вважати, що НП це явище надплинності бозе-рідини куперівських пар.

Проста модель, яка демонструє утворення зв'язаного стану для пари збуджень електронного газу над ("електрони") чи під ("дірки") сферою Фермі, була запропонована Л.~Купером.

Для спрощення обчислень, надалі будемо вважати, що $T\to 0$ і можна знехтувати тепловими збудженнями (тобто має місце простий сходинковий розподіл за енергіями ). Рівняння Шредінгера для пари ЕЗ має вигляд:

\begin{equation}\label{eq:ham_couper}
	\left(\, \hat{H}^{(0)}_{1} + \hat{H}^{(0)}_{2}+ \hat{V}(\mathbf{r}_1,\mathbf{r}_2)\,\right)\Psi(\mathbf{r}_1,\mathbf{r}_2) = E\Psi(\mathbf{r}_1,\mathbf{r}_2).
\end{equation}

Будемо шукати розв'язок у вигляді суперпозиції синглетних станів
пари вільних квазічастинок

\begin{equation}\label{eq:psicouperpair}
	\Psi(\mathbf{r}_1,\mathbf{r}_2)   = \sum\limits_{\mathbf{p}} c_{\mathbf{p}} \psi_{\mathbf{p},\uparrow}(\mathbf{r}_1)\psi_{-\mathbf{p},\downarrow}(\mathbf{r}_2)\,.
\end{equation}

Підставляючи цей вираз у \eqref{eq:ham_couper} відповідно маємо:

\begin{equation}\label{shcr_cp0}
	2\epsilon(\mathbf{p})\,c_{\mathbf{p}} + \sum\limits_{\mathbf{p}'}\,
	V_{\mathbf{p},\,\mathbf{p}'}\,c_{\mathbf{p}'} = E\,c_{\mathbf{p}}\,,
\end{equation}

Ефективна взаємодія виникає завдяки довгохвильовим фононам з $k<\omega_{D}$, тому розумно прийняти наступну спрощену модель потенціалу взаємодії\footnote{нагадаємо, що імпульси відлічуються від $p_F$, тобто мова іде про стани у енергетичному шарі шириною $2\omega_D$ у околі рівня Фермі $\epsilon_F$.}:

\begin{equation}\label{cooper_v}
	V_{\mathbf{p},\,\mathbf{p}'} = \begin{cases}
		-\frac{g}{V}\,, & \text{якщо}\,\, |\mathbf{p}|,\,|\mathbf{p}-\mathbf{p}'|< \omega_D/v_F\,;\\
		\,\,\,\,\,0\,, & \text{якщо}\,\, |\mathbf{p}|,\,|\mathbf{p}-\mathbf{p}'|> \omega_D/v_F\,.
	\end{cases}
\end{equation}

З \eqref{shcr_cp0} отримаємо систему рівнянь:

\begin{equation}\label{shcr_cp1}
	2\epsilon(\mathbf{p})\,c_{\mathbf{p}} - \frac{g}{V} \sum\limits_{\mathbf{p}'}\,
	\,c_{\mathbf{p}'} = E\,c_{\mathbf{p}}\,,
\end{equation}

де сума здійснюється за вказаним інтервалом. Далі,

\begin{equation}\label{shcr_cp2}
	c_{\mathbf{p}}  =  \frac{g}{V}\frac{\sum\limits_{\mathbf{p}'}\,c_{\mathbf{p}'}}{2\epsilon(\mathbf{p})\,-E}\,,
\end{equation}

що при обчисленні суми дозволяє отримати:

\begin{equation}\label{shcr_cp3}
	1  = \frac{g}{V} \sum\limits_{\mathbf{p}}\,\frac{1}{2\epsilon(\mathbf{p})\,-E}\,.
\end{equation}

Позначаючи енергію зв'язаного стану такої \textbf{куперівської пари} як
 
\begin{equation}\label{eq:enrgcooperpair}
E_0 = -2\,\Delta_0\,,
\end{equation}
і змінюючи суму на інтеграл\footnote{це стандартна процедура $\sum\limits_{p} \to V\int \frac{d\mathbf{p}}{(2\pi)^3}$ враховуючи \eqref{fermidens_states}}, отримаємо рівняння для $\Delta_0$:

\begin{equation}\label{schel0}
	1 = \frac{g}{2} \int\limits_{0}^{\omega_D} \frac{\rho_{\epsilon}}{\epsilon +\Delta_0}\,d\epsilon = \frac{g\,\rho_{F}}{2}\,\ln\frac{\omega_D}{\Delta_0}
\end{equation}

Тут $\rho_{\epsilon}$ - енергетична густина станів, яку при $T\to 0$ можна вважати сталою, яка дорівнює густині станів ЕЗ на рівні Фермі $\rho_{F}$ (див. \eqref{eq:n_states}). Крім того, при інтегруванні у \eqref{schel0}, ми врахували, що $\Delta_0 \ll \omega_D$\footnote{ми поклали $\hbar = 1$}. Дійсно, температури, що відповідають $\omega_D$ за величиною порядку температури плавлення ґратки ($T_{\text{пл}}\sim 10^2\div 10^3 \,K$), у той  час як   $\Delta_0 \sim T_c \sim 1\,K$. В результаті маємо таке рівняння:

\begin{equation}\label{eq:schel_delta}
	\Delta_0  = \omega_D e^{-\frac{2}{g\,\rho_{F}}}\,,
\end{equation}

Воно демонструє неаналітичну залежність $\Delta_0$ від параметра  електрон-фононної взаємодії $g$, що вказує на неможливість отримання цього результату на основі простої теорії збурень. Величина $\Delta_0$ визначає щілину у спектрі елементарних збуджень НП: $\Delta_0 = \Delta(T\to0)$). Детальне обчислення ми тут робити не будемо (див. Розділ~\ref{subsec:e0scond} або \cite{book_abrikosov_en}), а замість цього наведемо евристичні аргументи  фізичного характеру. Як вже зазначалось, спектр ЕЗ нормального металу - безщілинний,  оскільки потрібна як завгодно мала енергія для збудження системи. У випадку НП це вже не так, завдяки конденсації куперівських пар при $T<T_c$. В результаті, для утворення ЕЗ потрібно розірвати куперівську пару, витративши енергію $2 \Delta $. Тепер можна скористатись аналогією з релятивістським співвідношенням імпульс-енергія:

\begin{equation}\label{eq:epm}
	E = \sqrt{(p\,c)^2 + \left(\,m_0c^2\,\right)^2}.
\end{equation}

Тут роль енергії спокою буде відігравати  $\Delta$, оскільки це енергія яка необхідна для народження частинки з вакуумного стану, в якості якого виступає бозе-конденсат куперівських пар. На основі такої аналогії для спектра ЕЗ у НП можна записати:
<!--  -->
\begin{equation}\label{eq:spectr_sp}
	E(p) = \sqrt{\epsilon(p)^2 + \Delta^2_0}.
\end{equation}

Якщо цю аналогію використовувати "у зворотному напрямку", то наявність маси спокою у певного класу елементарних частинок (калібровочних бозонів), які розглядаються як кванти відповідних полів, можна трактувати, як наслідок наявності певного фонового бозе-поля. У сучасній теорії елементарних частинок таке поле носить назву поля Хіггса. [Власне ідея про введення такого поля і з'явилась у Хіггса на основі ідей теорії надпровідності](https://en.wikipedia.org/wiki/Higgs_mechanism) (наявність зарядженого бозе конденсату куперівських пар призводить до діамагнетизму, тобто до екранування магнітного поля на відстанях лондонівської довжини $\lambda_{L}$, що можна вважати еквівалентним до виникнення ''маси'' у фотона $m\sim \lambda^{-2}_{L}$ і заміни кулонівського потенціалу $1/r \leftrightarrow \mathbf{q}^{-2}$ на експоненційно екранований $e^{-r/\lambda}/r \leftrightarrow (\mathbf{q}^{2}+m^2)^{-1}$). Оскільки маса має динамічне походження, фізично ситуація з появою маси у елементарних частинок може бути пояснена аналогічно як ефект приєднаної маси. Дійсно, безмасову частинку можна розглядати як бульбашку у "рідині" поля Хіггса. Останнє заповнює весь простір і має свій тензор енергії-імпульсу. Взаємодія такої безмасової бульбашки із рідиною призводить до появи в неї маси. Фотон, наприклад, не взаємодіє із полем Хіггса і тому у нього не має маси.

Зрозуміло, що отриманий спектр ЕЗ також задовольняє критерій надплинності Ландау \eqref{eq:landcrit}. Слід також зауважити, що енергетична густина числа
станів ЕЗ надпровідника має сингулярність при $E = \Delta$. Це випливає безпосередньо з \eqref{eq:spectr_sp}:
<!--  -->
\begin{equation}\label{eq:dens_spectr_sp}
	\rho_{\text{нп}}  = \frac{d\,N}{d\,E(p)} = \frac{d\,N}{d\,\epsilon}\,\frac{d\,\epsilon}{d\,E}  = \rho_{\epsilon_{F}}\,\frac{\,E(p)}{\sqrt{E(p)^2 - \Delta^2}}
\end{equation}
<!--  -->
тобто енергетична густина числа станів необмежено зростає поблизу верхньої границі щілини.

Виходячи з такої картини можна оцінити відповідну величину пониження енергії при такій конденсації. Враховуючи, що кількість пар у енергетичному околі ширини $\Delta$ фермі-сфери дорівнює $1/2\,dN/d\epsilon\,\Delta/2$, а енергія кожної пари $-2\,\Delta$, отримаємо, що виграш у енергії від конденсації дорівнює:
<!--  -->
\begin{equation}\label{e_condens}
	E_{{\rm cond}} = -\frac{1}{2}\,\left.\frac{dN}{d\epsilon}
	\right|_{\epsilon_F}\,\Delta^2\,\,.
\end{equation}

## Ефекти Джозефсона
Заряджений конденсат куперівських пар характеризується хвильовою функцією $\Psi = |\Psi|\,e^{i\,\theta}$. Нормуючи таку функцію на повну кількість пар можна вважати, що $|\Psi| = \sqrt{n_s}$, де $n_s$ - густина конденсату. Наявність фази хвильової функції призводить до цікавих ефектів які виникають при контакті двох надпровідників. Між ними може протікати тунельний струм завдяки різниці фаз між хвильовими функціями їх конденсатів! Розглянемо цю ситуацію з точки зору нестаціонарної теорії збурень. Припустимо, що до контакту двох надпровідників (з однакового матеріалу) прикладена стала напруга $V$.  З т.з. енергетичного стану куперівської пари спрощено можна вважати таку систему як дворівневу (різниця в енергії для пари складає $2eV$). Перехід між цими енергетичними станами обумовлений оператором тунельного переходу $V$.
<!--  -->
<img src="./figs/fig_sc12.png" alt="Image" width="300">
<!--  -->
Рівняння для еволюції х.ф. $\ket{\Psi} = c_1\ket{\psi_1}+c_1\ket{\psi_2}$, де $\ket{\psi_i}$ - стаціонарні стани незбуреного гамільтоніана (без тунельного оператора):
$$i\hbar\,\frac{\partial}{\partial t}\ket{\Psi} = \left(\hat{H_0}+\hat{T}\right)\ket{\Psi}$$
відповідно:
$$i\hbar\,\frac{d c_1}{d t}  = eV\,c_1+T_{12}\,c_2\\
i\hbar\,\frac{d c_2}{d t}  = -eV\,c_2+T_{12}\,c_1
$$
Підставимо $c_{1,2} = \sqrt{n_s}\,e^{\theta_{1,2}}$:
$$\frac{d n_s}{d t}  = \frac{2T_{12}}{\hbar}\,n_s\\
\frac{d \theta_1}{d t}  = -\frac{T_{12}}{\hbar}\,\cos(\theta_2-\theta_1)-\frac{eV}{\hbar}\\
\frac{d \theta_2}{d t}  = -\frac{T_{12}}{\hbar}\,\cos(\theta_2-\theta_1)+\frac{eV}{\hbar}
$$

З цієї системи витікає, що:
- навіть за відсутності напруги $V=0$ можливе протікання струму $j \sim d n_s/dt = j_c\,\sin\Delta \theta$ який обумовлений тунелюванням куперівських пар крізь контакт (стаціонарний ефект Джозефсона);
- Якщо значення густини струму перевищує критичне $j_c$, то пари починають руйнуватися, а на контакті виникає падіння напруги внаслідок виникнення струму елементарних збуджень, які народжуються при розпаді куперівських пар. В такому випадку повний струм, який фіксується, складається з дисипативної частини від струму збуджень і надпровідної компоненти:
$I = I_0 + I_s\,,\quad I_0 = \frac{\hbar}{2eR}\frac{d \Delta \theta}{d t}\,,\quad I_s = I_c\,\sin\Delta \theta$
- якщо струм на контакті тримати сталим $I>I_c$ то на ньому генерується змінна напруга
$$	V(t) = R\,\frac{I^2-I^2_c}{I+I_c\cos{\left(\,\omega t\,\right)}}\,,\quad \omega  = 2e\,\frac{R}{\hbar}\sqrt{I^2-I^2_c}$$
(нестаціонарний ефект Джозефсона)
