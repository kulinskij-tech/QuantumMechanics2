## Конденсация Бозе-Айнштайна у ідеальному бозе-газі.
<!--  -->
$\newcommand{\ket}[1]{\left.\left\vert\, #1\,\right. \right \rangle}$
$\newcommand{\bra}[1]{\left. \langle \, #1\, \right\vert}$
$\newcommand{\braket}[2]{\langle \,#1 \vert #2 \, \rangle}$
$\newcommand{\av}[3]{\left\langle \,#1 \left \vert\, #2 \,\right\vert #3 \,\right \rangle}$
<!--  --> 
Явище конденсації у бозе газі обумовлено властивістю бозе-частинок накопичуватись у основному стані при $T\to 0$. При $T = 0 $ всі бозони знаходяться у найнижчому  енергетичному рівні. Такие частинки називаются
бозе-айнштайновским конденсатом (БАк). Для ідеального газа вільних частинок це стан $\varepsilon =0$. Тому очевидно що при
$T=0$: 
<!--  -->
\begin{equation}\label{n_t0}
	n_{0}(\mathbf{p}) = (2\pi \hbar)^3\, \frac{N}{V}\,\delta (\mathbf{p})\,.
\end{equation} 
<!--  -->
C підвищенням температури частина частинок буде
переходить на більш високі енергетичні рівні, однак зрозуміло, що певна доля частинок всі ж буде перебувати у основному стані. При високих температурах зрозуміло що всі квантові ефекти нивелюються і доля частинок з $\varepsilon =0$ дорівнює 0. За непрервністю зрозуміло, що повинна існувати певна температура при якій у стані $\varepsilon =0$ з'являється ненульова доля частинок. Тому кількість заповнення розумно прелставити у вигляді суми відповідних внесків:
<!--  -->
\begin{equation}\label{np_t}
	\left\langle\, n(\mathbf{p}) \,\right\rangle = N\,\frac{(2\pi
		\hbar)^3}{V}\,A(T)\,n_{0}(\mathbf{p}) +
	\frac{1}{\exp\left(\,\frac{\frac{\mathbf{p}^2}{2m} - \mu }{T} \,\right)-1}
\end{equation} 
<!--  -->
Тут функція $A(T)$ визначає долю частинок які знаходяться у конденсаті. Зрозуміло, що при температурі за якої з'являється БА-конденсат хім. потенціал повинен прямувати до 0 з боку відємних значень (нагадаємо, що для ідеального бозе газа у будь якому випадку $\mu\le 0$). Використовуючи нормування на повну кількість частинок 
<!--  -->
$$N = V\int\,\left\langle\, n(\mathbf{p})
\,\right\rangle\,\frac{d^3\mathbf{p}}{(2\pi \hbar)^3}$$ 
<!--  -->
з \eqref{np_t} отримаємо: 
<!--  -->
\begin{equation}\label{n_mu}
	N(1-A(T)) = V\int\,\frac{1}{\exp\left(\,\frac{\frac{\mathbf{p}^2}{2m}}{T}
		\,\right)-1}\,\frac{d^3\mathbf{p}}{(2\pi \hbar)^3}
\end{equation} 
<!--  -->
Умова на температуру конденсації яку було вказано
вище визначається оберненням хім. потенціалу у 0:
\begin{equation}\label{n0_mu}
	N = V\int\,\frac{1}{\exp\left(\,\frac{\frac{\mathbf{p}^2}{2m}}{T}
		\,\right)-1}\,\frac{d^3\mathbf{p}}{(2\pi \hbar)^3} =
	\frac{Vm^{3/2}}{2^{1/2}\,\pi^2\,\hbar^3}\,\int\limits_{0}^{\infty}\,\frac{\sqrt{\varepsilon
	}}{e^{\frac{\varepsilon }{T}}-1}\,d\varepsilon =
	T^{3/2}\,\frac{Vm^{3/2}}{2^{1/2}\,\pi^2\,\hbar^3}\,
	\int\limits_{0}^{\infty}\,\frac{\sqrt{x}}{e^{x}-1}\,dx
\end{equation} 
<!--  -->
з якого чисельним інтегруванням можна отримати
$$T_{BEC}  \approx 3.3\,
\frac{\hbar^2}{m}\,\left(\frac{N}{V}\right)^{2/3}$$ 
<!--  -->
Доля конденсату $A(T)$ визначається з рівняння \eqref{n_mu} при $\mu =0$ і по суті співпадає з обчисленням у \eqref{n0_mu}:
<!--  -->
\begin{equation}\label{a_condens}
	1-A(T) = T^{3/2}\,\frac{Vm^{3/2}}{N\,2^{1/2}\,\pi^2\,\hbar^3}\,
	\int\limits_{0}^{\infty}\,\frac{\sqrt{x}}{e^{x}-1}\,dx =
	\left(\,\frac{T}{T_{BEC}} \,\right)^{3/2}
\end{equation} 
<!--  -->
Звідки отримаємо остаточний результат
<!--  -->
\begin{equation}\label{a_becondens}
	A(T) =1- \left(\,\frac{T}{T_{BEC}} \,\right)^{3/2}\,.
\end{equation} 
<!--  -->
Як видно з \eqref{a_becondens} точка БА конденсації є особою, оскільки друга похідна по температурі від $A(T)$ розбіжна. Цікаво проаналізувати поведінку теплоємності, оскільки зрозуміло, що кількість ступенів вільності задіяних у тепловому русі залежить від долі конденсата. На якісном рівні можна очікувати підвищення теплоємності із зростанням температури вплоть до $T_{BEC}$ у зв'язку із зменшенням долі конденсату і збільшенням числа частинок задіяних у тепловому русі. При високих температурах, зрозуміло що $c_v$ прямує до класичної границі $3/2$. Тому характер поведінки $c_v$ залежить від співвідношення значення теплоємності при $T=T_{BEC}$ і класичного значення 3/2. Якщо $c_v(T_{BEC})>3/2$, то очевидно
що температурна залежність не є монотонною. Саме цей випадок має місце. Дійсно, при температурі
$T_{BEC}$ квантові ефекти тотожності частинок все ще є суттєвими, так як хім. потенціал відрізняється від класичного значення. Квантові властивості
пов'язані з перекриттям хвильових пакетів, тому якісно зрозуміло що ефективна кількість ступенів вільності при $T_{BEC}$ більше такового для класичного числа точкових частинок.
<!--  -->
<img src="./figs/fig_cvbec.png" alt="Image" width="400">
<center>Температурна поведінка теплоємності ідеального Бозе газа</center>
<!--  -->
Тому $c_v(T_{BEC})>3/2$ і залежність
теплоємності від $T$ має вигляд изображенний на Рис.
Наявність злому похідної теплоємності при температурі конденсації є наслідком розбіжності другої похідної у
функції $A(T)$, яка дає долю конденсату, при температурі конденсації. Покажемо це. Внутрішня енергія визначається тільки
збужденними частинками у станах $\mathbf{p} \ne 0$. Тому при температурах нижче $T_{BEC}$: 
<!--  -->
\begin{equation}\label{u_bose}
	U = \frac{V}{(2\pi \hbar)^3}\,\int \,\frac{\mathbf{p}^2}{2m}\,
	\left\langle\, n(\mathbf{p}) \,\right\rangle\, d^3\mathbf{p} =
	T^{5/2}\,\frac{Vm^{3/2}}{2^{1/2}\,\pi^2\,\hbar^3}\,
	\int\limits_{0}^{\infty}\,\frac{x^{3/2}}{e^{x}-1}\,dx\,,
\end{equation} 
<!--  -->
і внутрішня енергія пропорційна $T^{5/2}$. Відповідно теплоємність веде себе як: 
<!--  -->
$$c_v = \left. \frac{\partial\, U}{\partial\, T}\right|_V
= \frac{5}{2}\,\frac{U}{T} \sim T^{3/2}\,.$$ 
<!--  -->
Нагадаємо, що теплове випромінювання (бозе газ фотонів) чи фононний внесок має іншу залежність ($U\sim T^4\,, C_V\sim T^3$) завдяки лінійному закону дисперсії $\varepsilon \sim p$.}
Оскільки теплоємність визначається похідною внутрішньої енергії за температурою то питання про поведінку теплоємності зводиться до розгляду поведінки хім. потенціалу і його похідних при $T\to T_{BEC}+0$, оскільки при $T\to T_{BEC} -0$ хім. потенціал і всі його похідні дорівнюють 0. Диференціювання виразу який пов'язує хім. потенціал з густиною і температурою вище $T_{BEC}$:
<!--  -->
\begin{equation}\label{nmut}
n = \int\limits_{0}^{\infty}\,\frac{\varepsilon^{1/2}}{e^{\frac{\varepsilon-\mu}{T}}-1}\,d\varepsilon
\end{equation}
<!--  -->
при $T_{BEC}$, дає вираз для $\left. \frac{\partial\, \mu }{\partial\, T}\right|_{T=T_{BEC}+0}$
\footnote{ми учли що при $\mu(T_{BEC}) =0$.}
<!--  -->
$$ \frac{ \partial\,
	\mu }{\partial\, T}\,\int\limits_{0}^{\infty}\,\frac{\varepsilon^{1/2}
	e^{\frac{\varepsilon}{T}}}{\left(\,
	e^{\frac{\varepsilon}{T}}-1\,\right)^2}\,d\varepsilon=
-\frac{1}{T}\int\limits_{0}^{\infty}\,\frac{\varepsilon^{3/2}
	e^{\frac{\varepsilon}{T}}}{\left(\,
	e^{\frac{\varepsilon}{T}}-1\,\right)^2}\,d\varepsilon $$
<!--  -->
Т.ч. ми получаєо $\left.\frac{\partial\, \mu }{\partial\,T}\right|_{T=T{BEC}}\ne 0$, 
що і означає наявність стрибка у похідній теплоємності по температурі. З точки зору
классификації Эренфеста точка конденсації є прикладом фазового переходу 3-го роду.

## Надплинність
### Надплинність неідеального Бозе-газу. Крітерій Ландау.
<!--  -->
Властивість бозечастинок утворювати конденсат при низьких температурах є визначальним для утворення надплинності у рідкому ${\rm He}$. При цьому суттєвим є взаємодія між частинками, оскільки ідеальний бозе-газ навіть маючи бозе-конденсат не може бути надплинним. Це видно з простих енергетичних міркувань \cite{sc_bogolyubov_dan1958}. Якщо конденсат утворює стійкий надплинний стан, то поява елементарного збудження (випадання частинок з конденсату), наприклад при взаємодії (тертя) зі стінками повинно супроводжуватися підвищенням енергії системи, так щоб народження збудження було б енергетично невигідне. Але енергія надплинного конденсата ідеального бозе-газау який рухається зі швидкістю $u$ при появі (квазі)частинки зменшується, тому що:
<!--  -->
\begin{equation}\label{eq:bosequasi0}
E_0 =N_0\,\frac{m\,u^2}{2} >  (N_0-1)\,\frac{m\,u^2}{2} + \frac{m\,v^2}{2}\,,\quad v<u
\end{equation}
<!--  -->
Відповідно надплинний рух не є стійким і весь конденсат ''випариться''. 
Тому надплинність з т.з. термодинамічної стійкості повинна бути обумовлена взаємодією між частинками. В рамках теорії збурень можна вважати її слабкою. Завдяки взаємодії у  системі змінюється спектр елементарних збуджень. Із зрозумілих причин неідеальний бозе-газ вже уявляє суцільне середовище у якому найменшу енергію збудження мають колективні збудження  типу звукових хвиль - фонони, із законом дисперсії $\varepsilon (p) = c\,p$, де $c$ - швидкість звуку. Довжина хвилі таких збуджень значно більше ніж міжчастинкова відстань. Із збільшенням енергії (імпульсу) колективний харакетр зникає і збудження все більше набуває характер руху окремої частинки з масою $m$ тобто законом дисперсії $\varepsilon (p) = \frac{p^2}{2\,m}$. З \eqref{eq:bosequasi0} зрозуміло, що для надплинності спектр елементарнх збуджень повинен мати такий характер щоб випадіння частинок з конденсату тобто народження ЕЗ було енергетично не вигідно.
<!--  -->
Нехай конденсат рухається зі швидкістю $\mathbf{u}_0$  і в ньому народжується збудження з імпульсом $\mathbf{p}$, тоді імпульс конденсата:
$$\mathbf{P}_0 = M\,\mathbf{u}_0 = \mathbf{P}'_0 + \mathbf{p}$$
$$E_0'= \frac{\mathbf{P}'^2_0}{2\,M} = E_0 - \mathbf{u}_0\cdot \mathbf{p}  + \frac{\mathbf{p}^2}{2\,M} \approx E_0 - \mathbf{u}_0\cdot \mathbf{p}$$
Народження ЕЗ буде не вигідним, якщо:
<!--  -->
\begin{equation}\label{eq:landcrit}
	E_0' + \varepsilon(p) > E_0 \Rightarrow u_0 < \varepsilon(p)/p
\end{equation}
<!--  -->
тобто функція $\varepsilon(p)/p$ повинна мати додатній мінімум для деякого $p$. Це є знаменитий крітерий надплинності Ландау. 
<!--  -->
<img src="./figs/fig_spectrbose1.png" alt="Image" width="300"/>
<!--  -->
<center> Спектр елементарних збуджень у рідкому гелії</center>
<!--  -->

М.М. Боголюбов показав яким чином в моделі слабко неідеального газу виникає спектр який задовільняє крітерію Ландау \eqref{eq:landcrit}. Наведемо цей простий метод наближенної діагоналізації гамільтоніану. Гамільтоніан слабко неідеального бозе газу у представленні операторів народження-знищення $\hat{b}_{\mathbf{p}}, \hat{b}^\dagger_\mathbf{p}$ станів вільних частинок з імпульсами $\hbar\mathbf{k}$ 
$$\hat{b}^\dagger_\mathbf{p} \ket{N_{\mathbf{p}}} = \sqrt{N_{\mathbf{p}}+1}\ket{N_{\mathbf{p}}}\\
\hat{b}_\mathbf{p} \ket{N_{\mathbf{p}}} = \sqrt{N_{\mathbf{p}}}\ket{N_{\mathbf{p}}-1}$$
має вигляд:
<!--  -->
\begin{equation}\label{hambog0}
	\hat{H} - \mu\,N = \sum_{\mathbf{p}}\left(\frac{\mathbf{p}^2}{2m}-\mu\right)\,\hat{b}^{\dagger}_{\mathbf{p}}\,\hat{b}_{\mathbf{p}}+\frac{1}{2\,V}\,\sum_{\mathbf{p}_1+\mathbf{p}_2-\mathbf{p}'_1-\mathbf{p}'_2 = 0}	\Phi(\mathbf{q})\,\hat{b}^{\dagger}_{\mathbf{p}'_1}\hat{b}^{\dagger}_{\mathbf{p}'_2}\hat{b}_{\mathbf{p}_1}\hat{b}_{\mathbf{p}_2}
\end{equation}
<!--  -->
де $\mathbf{q}$ це вектор зміни імпульсу при взаємодії:
<!--  -->
$$\mathbf{q} = \mathbf{p}'_1-\mathbf{p}_1 = \mathbf{p}_2-\mathbf{p}'_2$$
<!--  -->
Оскільки при низьких темпертурах більшість атомів знаходяться у конденсаті, то найбільший внесок у енергію дають доданки з операторами $b_0\,,b^{\dagger}_0$ які можна вважати комутуючими оскільки $N_0\gg 1$ і просто покласти:
$$\hat{b}_0\sim \hat{b}^{\dagger}_0 = \sqrt{N_0}$$
<!--  -->
Тоді у операторі взаємодії можна обмежитися тільки доданками порядка $N^2_0$ і $N_0$:
<!--  -->
\begin{multline}\label{hambog1}
	\hat{H} - \mu\,N = \frac{N_0^2}{2\,V}\,\Phi(0) - \mu\,N_0 +\\
	\sum_{\mathbf{p}}\left(\frac{\mathbf{p}^2}{2m} + \frac{N_0}{V}\,\Phi(\mathbf{p})+\frac{N_0}{V}\,\Phi(0)-\mu\right)\,\hat{n}_{\mathbf{p}}+\\
	\frac{N_0}{2\,V}\,\sum_{\mathbf{p}}	\Phi(\mathbf{p})\,\left(\hat{b}^{\dagger}_{\mathbf{p}}\hat{b}^{\dagger}_{-\mathbf{p}}+\hat{b}_{-\mathbf{p}}\hat{b}_{\mathbf{p}}\right)
\end{multline}
<!--  -->
хім. потенціал визначаємо з умови мінімума енергії основного стану:
<!--  -->
$$E_0 = \frac{N_0^2}{2\,V}\,\Phi(0) - \mu\,N_0\Rightarrow \frac{\partial \, E_0}{\partial\,N_0} = 0 \Rightarrow \mu = n_0\,\Phi(0)$$
<!--  -->
\begin{equation}\label{hambog2}
	\hat{H} - \mu\,N = -\frac{N_0^2}{2\,V}\,\Phi(0) +
	\sum_{\mathbf{p}}\left(\frac{\mathbf{p}^2}{2m} + \frac{N_0}{V}\,\Phi(\mathbf{p})\right)\,\hat{n}_{\mathbf{p}}+
	\\
	\frac{N_0}{2\,V}\,\sum_{\mathbf{p}}	\Phi(\mathbf{p})\,\left(\hat{b}^{\dagger}_{\mathbf{p}}\hat{b}^{\dagger}_{-\mathbf{p}}+\hat{b}_{-\mathbf{p}}\hat{b}_{\mathbf{p}}\right)
\end{equation}
<!--  -->
Отриманий гамільтоніан має квадратичну структуру за операторами $\hat{b},\hat{b}^\dagger$, Тому його можна діагоналізувати:
<!--  -->
\begin{equation}\label{hambog_diag}
	\hat{H} - \mu\,N = -E_0 +
	\sum_{\mathbf{p}}E(p)\,\hat{c}^{\dagger}_{\mathbf{p}}\,
	\hat{c}_{\mathbf{p}}
\end{equation}
<!--  -->
вводячи оператори:
<!--  -->
\begin{equation}
	\hat{c}_{\mathbf{p}} = u_{\mathbf{p}}\,\hat{b}_{\mathbf{p}} +v_{\mathbf{p}}\,\hat{b}^{\dagger}_{-\mathbf{p}}\,,\quad 
	\hat{c}^{\dagger}_{\mathbf{p}} = u_{\mathbf{p}}\,\hat{b}^{\dagger}_{\mathbf{p}} +v_{\mathbf{p}}\,\hat{b}_{-\mathbf{p}}
\end{equation}
<!--  -->
і накладаючи коммутаційні співвідношення:
<!--  -->
$$ [\hat{c}_{\mathbf{p}},\hat{c}^{\dagger}_{\mathbf{p}}] = 1 \Rightarrow u^2_{\mathbf{p}} - v^2_{\mathbf{p}} =1 $$
<!--  -->
Роблячи необхідні підстановки і прирівнюючи відповідні доданки, отримаємо:
<!--  -->
\begin{gather}
	E_{p}\,u_{\mathbf{p}}\,v_{\mathbf{p}} =\frac{n_0}{2}\,\Phi(\mathbf{p})\,\\
	E_{p}\,\left(u^2_{\mathbf{p}}+
	v^2_{\mathbf{p}}\right) = \frac{\mathbf{p}^2}{2m} + n_0\,\Phi(\mathbf{p})
\end{gather}
<!--  -->
Звідки маємо:
$$
u^2_{\mathbf{p}} =\frac{1}{2\,E_p}\left(
\frac{\mathbf{p}^2}{2m} + n_0\,\Phi(\mathbf{p})+E_p
\right)\,,\quad 
v^2_{\mathbf{p}} =\frac{1}{2\,E_p}\left(
\frac{\mathbf{p}^2}{2m} + n_0\,\Phi(\mathbf{p}) - E_p
\right)$$
$$
E_p = \sqrt{\left(\frac{\mathbf{p}^2}{2m}\right)^2 + \frac{\mathbf{p}^2}{m}\,n_0\,\Phi(\mathbf{p})}
$$
<!--  -->
Неважко бачити, що отриманий спектр задовільняє крітерію Ландау якщо $\Phi(0)>0$.


## Надпровідність
<!--  -->
Зрозуміло що явища надпровідності (зникнення опору) і надплинності (зникнення в'язкості) повязані між собою. Надпровідність можна уявляти як надплинність електронної підсистеми у НП. Але електрони є фермі-частинками і відповідно не можуть утворювати бозе-айнштайнівський надплинний конденсат. Ключовим механізмом є утворювання пар електроних збуджень які вже є бозонами. Таким механіхзмом у низькотемпературних надпровідниках є електрон-фононна взаємодія, тобто поляризація кристалічної гратки при русі електронного збудження. В результаті обміну фононами два збудження утворюють звязаний стан - [куперівську пару](https://en.wikipedia.org/wiki/Cooper_pair).

Перехід НП-НМ як фазовий перехід у низькотемпературний впрорядкований стан пов'язаний з дією сил взаємодії між частинками. Тому основні зусилля при побудові мікроскопічної теорії були спрямовані на пошук механізму притягнення між ЕЗ у металі. Такий механізм дійсно існує незважаючи на те що такі збудження є однаково зарядженими. Це механізм поляризації кристалічної гратки при росповсюдженні ЕЗ. Завдяки великий різиці у масі така взаємодія дуже мала оскільки мали зміщення іонів внаслідок кулоновської взаємодії з квазічастинкою.
Гамільтоніан системи може бути записан у вигляді:
<!--  -->
\begin{equation}\label{ham}
	\hat{H} = \hat{H}^{(0)}_{el} + \hat{H}^{(0)}_{ph} + \hat{H}_{el-ph}\,.
\end{equation}
<!--  -->
Последнее доданок як раз і уявляє взаємодія
між указанними подсистемами. Вирішальним експериментальним фактом у розумінні мікроскопічного механізма надпровідності було відкриття ізотопічного ефекту. Ізотопічним ефектом називається залежність властивості системи від ізотопічного склад гратки. Ізотопічний склад змінюється опромінюванням речовини нейтронами відповідної енергії. Ізотопи як відомо хімічно еквівалентні і відрізняються тільки масою ядра. Ядро впливає на електрону підсистему у вигляді кінетичній енергії руху ядер. Зсув обумовлений різницею маси ізотопів дозволяє робити висновки про вплив руху ядер на властивості системи. У випадку надпровідності спостерігається зсув критичної температури від ізотопічного складу. Експериментальні дані відповідають наступної залежності:
<!--  -->
\begin{equation}\label{isotopeff}
	T_c \,\sqrt{M} \approx \text{const} \Rightarrow \,\, T_c \sim \frac{1}{\sqrt{M}}\,.
\end{equation}
<!--  -->
Оскільки маса ядер визначає у першу чергу частоти коливань кристалічної гратки, зокрема дебаївську
частоту $\omega_{D}$, то звідси витікає що механізм надпровідності обумовлений взаємодією електронів з
фононами. Цей механізм був запропонований незалежно [Х. Фрелихом](https://journals.aps.org/pr/abstract/10.1103/PhysRev.79.845) і [Дж. Бардіном](https://journals.aps.org/pr/abstract/10.1103/PhysRev.79.167.3). Він полягає у слабкій поляризації кристалічної гратки ЕЗ
(електроном або діркою) із збудженням фонону. Цей фонон може взаємодіяти з іншим ЕЗ. Опис цих процесів зручно зображати у вигляді графічних схем, які називаються діаграмами Фейнмана. Слід зазначити, що таким процесам і відповідно діаграмам можна співставити певні вирази у рамках теорії збурень. Ця теорія збурень будується виходячи з припущення про малість вказаної взаємодії.
<!--  -->
<img src="./figs/fig_feynman_elphon.png" alt="Image" width="400">
<center><a id='feynmadiag'>Рис. Діаграми Фейнмана які зображують електрон-фононну та ефективну електрон-електронну взаємодії.</a></center>
<!--  -->

Таким чином теорію надпровідності слід будувати на
основі електрон-фононного механізму взаємодії і для цього потрібно конкретизувати члени гамільтоніана \eqref{ham}. Розв'язок задачі про спектр коливань кристалічної гратки також відомий з курсу стат. фізики (теорія Дебая теплоємності кристалів), де показано, що енергію коливань кристалічної гратки можна представити у вигляді суми нормальних мод, які є квантами звукового поля і називаются фононами. Вигляд спектру і відповідно гамільтоніанів незбурених систем, таким чином вважається відомим. Це сума енергій (гамільтоніанів) елементарних збуджень: частинок і дірок для електронного газа і фононів для граткі:
<!--  -->
\begin{equation}\label{h0_electr_phon}
	\hat{H}^{(0)}_{el} = \sum_{\mathbf{p},s_z} \varepsilon(\mathbf{p})\,\hat{n}^{(el)}_{\mathbf{p},s_z}\,,\quad   \hat{H}^{(0)}_{ph} = \sum_{\mathbf{k},\alpha } \hbar \omega (\mathbf{k})\,\hat{n}^{(ph)}_{\mathbf{k},\alpha}\,,
\end{equation}
<!--  -->
Тут $\hat{n}$ т.з. оператори чисел заповнення станів. Стан електрона характеризується імпульсом і проекцією спіна $s_z=\pm 1/2$, стан фонона - хвильовим вектором і поляризацією $\alpha =1,2,3$. Середнє значення числа заповнення це середнє кількість частинок у данном стані. Для ферміонов це кількість між 0 і 1 у силу принципа Паули, для бозонів воно може бути як завгодно великим. По аналогії з відомою задачею про осцилятор, зручно також ввести оператори народження і знищення відповідних елементарних збуджень:
<!--  -->
$$\hat{n}^{(el)}_{\mathbf{p},s_z}  = \hat{a}^\dagger_{\mathbf{p},s_z}\,\hat{a}_{\mathbf{p},s_z}\,,\quad \hat{n}^{(ph)}_{\mathbf{k},\alpha }  = \hat{b}^\dagger_{\mathbf{k},\alpha }\,\hat{b}_{\mathbf{k},\alpha }$$
<!--  -->
Оператори народження і знищення діють на одночастинкові стани відповідно підвищуючи або понижуючи кількість частинок у цьому стані. Звичайно для ферміонних операторів принцип Паули приводить до заборони народження ще однієї частинки у заповненому стані: 
$\hat{a} \left|\,1\,\right\rangle = 0$. За допомогою таких операторів для елементарних процесів можна будувати більш складні оператори які реализують обмін частинками між станами за рахунок взаємодії. Наприклад можна записати формальний вираз для оператора взаємодії, співставляючи його [діаграмі](#feynmadiag):
<!--  -->
\begin{equation}\label{ham_elphonint}
	\hat{H}_{el-ph} = \sum_{\mathbf{p},\mathbf{k},s_z,\alpha} V(\mathbf{p},\mathbf{k})\,\hat{a}^\dagger_{\mathbf{p}-\mathbf{k},s_z}\hat{a}_{\mathbf{p},s_z}\,\hat{b}^\dagger_{\mathbf{k},\alpha }+ V^*(\mathbf{p},\mathbf{k})\,\hat{a}^\dagger_{\mathbf{p},s_z} \hat{a}_{\mathbf{p}-\mathbf{k},s_z}\,\hat{b}_{\mathbf{k},\alpha }\,.
\end{equation}
<!--  -->
Дійсно [діаграма](#feynmadiag) ілюструє що спочатку знищується ЕЗ електронної підсистеми у стані з імпульсом $\mathbf{p}$ і народжується ЕЗ фононної підсистеми у стані імпульсом $\mathbf{k} \,(\hbar=1)$. Зазначимо, що вираз \eqref{ham_elphonint} є ермітово симетричним. Цей вираз складається з двох ермітово спряжених операторів. Оскільки ермітове спряження у квантовій теорії пов'язано з симетрією процесів у часі, можна вважати що це є сума прямого і оберненого у часі процесу. В зв'язку з цим виникає інтерпретація дірки як частинки яка рухається ''зворотньо за часом''. Зазначимо, що первоначально така інтерпретація була створена Р. Фейнманом для позитрона, оскільки квантова електродинаміка (КЕД) на понятійному рівні практично еквівалентна мікроскопічній теорії надпровідності. Різниця полягає у тому що вакуумний основний стан у КЕД це справжній вакуум, у той час як у ТНП основним станом є основний стан НП у якому відлік імпульсів здійснюється від імпульса Фермі $p_F$. Зрозуміло, середнє значення оператора \eqref{ham_elphonint} у будь якому стані з певною кількістю ЕЗ дорівнює 0, оскільки цей оператор змінює кількість ЕЗ. Повний розв'язок задачі про обчислення спектра енергії гамільтоніана \eqref{ham} з вказанними доданками є важкою задачею. Однак принциповий результат можна отримати вже у рамках елементрної теорії збурень. Введення взаємодії призведе до зміни спектра незбуреної системи. При цьому можна вважати, що спектр фононної підсистеми практично не змінюється, так же як ми нехтували зміною об'єму або пружніх властивостей металу при переході у НП стан. Основна увага приділяється зміні основного стану електронної підсистеми - сфері Фермі. Енергія основного стану у рамках теорії збурень повинна представлятися у вигляді:
<!--  -->
\begin{equation}\label{e0_pert}
E_{0} = E^{(0)}_{0} + \left\langle\, 0\,\right\vert \hat{H}_{ph-el} \left|\,0\,\right\rangle + \sum\limits_{n}\frac{\left\langle\, 0\,\right\vert \hat{H}^2_{ph-el} \left|\,0\,\right\rangle}{E^{(0)}_0 - E^{(0)}_n}+\ldots
\end{equation}
<!--  -->
Перша поправка, як було зазначено, дорівнює 0. Друга поправка до основного стану, як відомо, завжди від'ємна. Ця друга поправка як раз поредставляє ефективну взаємодією між ЕЗ електронного газу, яке зображується [діаграмою](#feynmadiag). Взагалі, зрозуміло, порядок теорії збурень співпадає з кількістю вершин взаємодії у діаграмі. Тому можна сподіватись на те що ця взаємодія ЕЗ електронної підсистеми завдяки фононній приведе до утворення зв'язанного стану ЕЗ з від'ємною енергією. Звичайно тут необгрунтовано припускається що більш високі поправки не змінюють такого висновку. В примитивній теорії збурень більш високі поправки зростають і робити належні кількісні висновки необхідно на основі більш строго підходу. Для нас, тим не менш, важливе якісне уявлення суті.

Завдяки скінченності цієї енергії це веде до кардинальної перебудови основного стану. Дійсно, за можливості утворення зв'язанного стану нескінченно малий програш у кінетичній
енергії для пари таких частинок, які у результаті взаємодії ''вискочують'' з під сфери
Фермі, із надлишком компенсується скінченною енергією зв'язку такої пари. Якщо врахувати що кількість таких пар є макроскопічним, то зрозуміло, що основний стан зміниться якісним чином і ми отримаємо зовсім іншу систему зі певними властивостями і спектром ЕЗ.

### Куперівські пари
Очевидно, що для розгляду утворення зв'язаних станів необхідно обрати тільки ті пари одночастинкових станів між якими утворення зв'язанного стану є найбільш вірогідним. З міркувань симетрії зрозуміло, що основний стан двохферміонної системи повинен бути синглетним, тобто бути утвореним з пар станів з
протилежними проекціями спінів. Крім того повний імпульс такої пари у об'ємному НП, у рівновазі, повинен дорівнювати 0. Звідси витікає що найбільш ефективно зв'язуються у пари стани $(\mathbf{p},s_z)$ і $(-\mathbf{p},-s_z)$. Картина утворення НП заснована на нестійкості основного стану і спектру ЕЗ електронної підсистеми відносно слабкого притягнення між ЕЗ. В результаті виникає ефект конденсації пар ЕЗ з утворенням зв'язаних станів які є бозе-частинками і відповідно можуть перебувати у одному стані  який описується єдиною хвильовоїю функцією. Такі стани називаються куперівськими. Можна вважати, що НП це явище надплинності бозе-рідини куперівських пар.

Проста модель, яка демонструє утворення зав'язаного стану для пари збуджень електронного газу над ("електрони") чи під ("дірки") сферою Фермі, була запропонована Л.~Купером.

Для спрощення обчислень, надалі будемо вважати, що $T\to 0$ і можна знехтувати тепловими збудженнями (тобто має місце простий сходинковий розподіл за енергіями ). Рівняння Шредінгера для пари ЕЗ має вигляд:

\begin{equation}\label{eq:ham_couper}
	\left(\, \hat{H}^{(0)}_{1} + \hat{H}^{(0)}_{2}+ \hat{V}(\mathbf{r}_1,\mathbf{r}_2)\,\right)\Psi(\mathbf{r}_1,\mathbf{r}_2) = E\Psi(\mathbf{r}_1,\mathbf{r}_2).
\end{equation}

Будемо шукати розв'язок у вигляді суперпозиції синглетних станів
пари вільних квазічастинок

\begin{equation}\label{eq:psicouperpair}
	\Psi(\mathbf{r}_1,\mathbf{r}_2)   = \sum\limits_{\mathbf{p}} c_{\mathbf{p}} \psi_{\mathbf{p},\uparrow}(\mathbf{r}_1)\psi_{-\mathbf{p},\downarrow}(\mathbf{r}_2)\,.
\end{equation}

Підставляючи цей вираз у \eqref{eq:ham_couper} відповідно маємо:

\begin{equation}\label{shcr_cp0}
	2\epsilon(\mathbf{p})\,c_{\mathbf{p}} + \sum\limits_{\mathbf{p}'}\,
	V_{\mathbf{p},\,\mathbf{p}'}\,c_{\mathbf{p}'} = E\,c_{\mathbf{p}}\,,
\end{equation}

Ефективна взаємодія виникає завдяки довгохвильовими фононами з $k<\omega_{D}$, тому розумно прийняти наступну спрощену модель потенціалу взаємодії\footnote{нагадаємо, що імпульси відлічуються від $p_F$, тобто мова іде про стани у енергетичному шарі шириною $2\omega_D$ у околі рівня Фермі $\epsilon_F$.}:

\begin{equation}\label{cooper_v}
	V_{\mathbf{p},\,\mathbf{p}'} = \begin{cases}
		-\frac{g}{V}\,, & \text{якщо}\,\, |\mathbf{p}|,\,|\mathbf{p}-\mathbf{p}'|< \omega_D/v_F\,;\\
		\,\,\,\,\,0\,, & \text{якщо}\,\, |\mathbf{p}|,\,|\mathbf{p}-\mathbf{p}'|> \omega_D/v_F\,.
	\end{cases}
\end{equation}

З \eqref{shcr_cp0} отримаємо систему рівнянь:

\begin{equation}\label{shcr_cp1}
	2\epsilon(\mathbf{p})\,c_{\mathbf{p}} - \frac{g}{V} \sum\limits_{\mathbf{p}'}\,
	\,c_{\mathbf{p}'} = E\,c_{\mathbf{p}}\,,
\end{equation}

де сума здійснюється за вказаним інтервалом. Далі,

\begin{equation}\label{shcr_cp2}
	c_{\mathbf{p}}  =  \frac{g}{V}\frac{\sum\limits_{\mathbf{p}'}\,c_{\mathbf{p}'}}{2\epsilon(\mathbf{p})\,-E}\,,
\end{equation}

що при обчисленні суми дозволяє отримати:

\begin{equation}\label{shcr_cp3}
	1  = \frac{g}{V} \sum\limits_{\mathbf{p}}\,\frac{1}{2\epsilon(\mathbf{p})\,-E}\,.
\end{equation}

Позначаючи енергію зв'язаного стану такої \textbf{куперівської пари} як
 
\begin{equation}\label{eq:enrgcooperpair}
E_0 = -2\,\Delta_0\,,
\end{equation}
і змінюючи суму на інтеграл\footnote{це стандартна процедура $\sum\limits_{p} \to V\int \frac{d\mathbf{p}}{(2\pi)^3}$ враховуючи \eqref{fermidens_states}}, отримаємо рівняння для $\Delta_0$:

\begin{equation}\label{schel0}
	1 = \frac{g}{2} \int\limits_{0}^{\omega_D} \frac{\rho_{\epsilon}}{\epsilon +\Delta_0}\,d\epsilon = \frac{g\,\rho_{F}}{2}\,\ln\frac{\omega_D}{\Delta_0}
\end{equation}

Тут $\rho_{\epsilon}$ - енергетична густина станів, яку при $T\to 0$ можна вважати сталою, яка дорівнює густині станів ЕЗ на рівні Фермі $\rho_{F}$ (див. \eqref{eq:n_states}). Крім того, при інтегруванні у \eqref{schel0}, ми врахували, що $\Delta_0 \ll \omega_D$\footnote{ми поклали $\hbar = 1$}. Дійсно, температури, що відповідають $\omega_D$ за величиною порядка температури плавління гратки ($T_{\text{пл}}\sim 10^2\div 10^3 \,K$), у той  час як   $\Delta_0 \sim T_c \sim 1\,K$. В результаті маємо таке рівняння:

\begin{equation}\label{eq:schel_delta}
	\Delta_0  = \omega_D e^{-\frac{2}{g\,\rho_{F}}}\,,
\end{equation}

Воно демонструє неаналітичну залежність $\Delta_0$ від параметру  електрон-фононної взаємодії $g$, що вказує на неможливість отримання цього результату на основі простої теорії збурень. Величина $\Delta_0$ визначає щілину у спектрі елементарних збуджень НП: $\Delta_0 = \Delta(T\to0)$). Детальне обчислення ми тут робити не будемо (див. Розділ~\ref{subsec:e0scond} або \cite{book_abrikosov_en}), а замість цього наведемо евристичні аргументи  фізичного характеру. Як вже зазначалось, спектр ЕЗ нормального металу - безщілинний,  оскільки потрібна як завгодно мала енергія для збудження системи. У випадку НП це вже не так, завдяки конденсації куперівських пар при $T<T_c$. В результаті, для утворення ЕЗ потрібно розірвати куперівську пару, витративши енергію $2 \Delta $. Тепер можна скористатись аналогією з релятивістським співвідношенням імпульс-енергія:

\begin{equation}\label{eq:epm}
	E = \sqrt{(p\,c)^2 + \left(\,m_0c^2\,\right)^2}.
\end{equation}

Тут роль енергії спокою буде відгравати  $\Delta$, оскільки це енергія яка необхідна для народження частинки з вакуумного стану, в якості якого виступає бозе-конденсат куперівських пар. На основі такої аналогії для спектра ЕЗ у НП можна записати:
<!--  -->
\begin{equation}\label{eq:spectr_sp}
	E(p) = \sqrt{\epsilon(p)^2 + \Delta^2_0}.
\end{equation}

<center>Спектр елементарних збуджень у надпровіднику</center>

Якщо цю аналогію використовувати "у зворотному напрямку", то наявність маси спокою у певного класу елементарних частинок (калібровочних бозонів), які розглядаються як кванти відповідних полів, можна трактувати, як наслідок наявності певного фонового бозе-поля. У сучасній теорії елементарних частинок таке поле носить назву поля Хіггса. [Власне ідея про введення такого поля і з'явилась у Хіггса на основі ідей теорії надпровідності](https://en.wikipedia.org/wiki/Higgs_mechanism) (наявність зарядженного бозе конденсату куперівських пар призводить до діамагнетизму, тобто до екранування магнітного поля на відстанях лондонівської довжини $\lambda_{L}$, що можна вважати еквівалентним до виникнення ''маси'' у фотона $m\sim \lambda^{-2}_{L}$ і заміни кулонівського потенціалу $1/r \leftrightarrow \mathbf{q}^{-2}$ на експоненційно екранований $e^{-r/\lambda}/r \leftrightarrow (\mathbf{q}^{2}+m^2)^{-1}$). Оскільки маса має динамічне походження, фізично ситуація з появою маси у елементарних частинок може бути пояснена аналогічно як ефект приєднаної маси. Дійсно, безмасову частинку можна розглядати як бульбашку у "рідині" поля Хіггса. Останнє заповнює весь простір і має свій тензор енергії-імпульсу. Взаємодія такої безмасової бульбашки із рідиною призводить до появи в неї маси. Фотон, наприклад, не взаємодіє із полем Хіггса і тому у нього не має маси.

Зрозуміло, що отриманий спектр ЕЗ також задовільняє критерію надплинності Ландау \eqref{eq:landcrit}. Слід також зауважити, що енергетична густина числа
станів ЕЗ надпровідника має сингулярність при $E = \Delta$. Це випливає безпосередньо з \eqref{eq:spectr_sp}:
<!--  -->
\begin{equation}\label{eq:dens_spectr_sp}
	\rho_{\text{нп}}  = \frac{d\,N}{d\,E(p)} = \frac{d\,N}{d\,\epsilon}\,\frac{d\,\epsilon}{d\,E}  = \rho_{\epsilon_{F}}\,\frac{\,E(p)}{\sqrt{E(p)^2 - \Delta^2}}
\end{equation}
<!--  -->
тобто енергетична густина числа станів необмежено зростає поблизу верхньої границі щілини.

Виходячи з такої картини можна оцінити відповідну величину пониження енергії при такій конденсації. Враховуючи що кількість пар у енергетичному околі ширини $\Delta$ фермі-сфери дорівнює $1/2\,dN/d\epsilon\,\Delta/2$, а енергія кожної пари $-2\,\Delta$, отримаємо що виграш у енергії від конденсації дорівнює:
<!--  -->
\begin{equation}\label{e_condens}
	E_{{\rm cond}} = -\frac{1}{2}\,\left.\frac{dN}{d\epsilon}
	\right|_{\epsilon_F}\,\Delta^2\,\,.
\end{equation}
